## Introduction

This project is an **independent, professional-grade implementation** of the theoretical models, architectural designs, and econometric evaluation protocols from the arXiv preprint titled **"Regime-Gated Residual Mixture-of-Experts for Cross-Sectional Volatility Forecasting"** by the authors:
*   **Junyi Ye**
*   **Gargi Vijay Borde**

Note: *The preprint (arXiv:2608.12251, e-journal submission date 12 August 2026) is referenced throughout this repository as an **arXiv Preprint**; it is not a peer-reviewed journal article, and this implementation does not claim any journal affiliation.*

This repository provides a complete, end-to-end computational framework that rigorously answers a critical architectural question in financial machine learning: *where should nonstationary regime information enter a neural forecasting model?* By holding information, model capacity, hyperparameter tuning, and random seeds strictly constant, this pipeline demonstrates that appending regime variables directly to forecasting inputs degrades both predictive performance and training stability. Conversely, restricting regime information to a soft-routing gate that modulates zero-initialized residual corrections—the **RG-ResMoE** architecture—delivers superior accuracy, eliminates training collapse, and significantly improves Value-at-Risk (VaR) calibration during high-stress market conditions.

## Theoretical Background

The implemented framework models cross-sectional equity volatility using a hybrid econometric and deep learning approach.

**1. The Forecasting Target:** For stock $i$ on day $t$, the target is the annualized five-day forward realized volatility, computed from daily log returns $r_{i,t} = \ln(P_{i,t}/P_{i,t-1})$:
$$y_{i,t} = \sqrt{252} \, \text{Std}(r_{i,t+1}, \dots, r_{i,t+5})$$

**2. The Shared MLP Block:** All neural architectures are constructed from a common two-hidden-layer Multi-Layer Perceptron (MLP) block with GELU activations ($\phi$) and dropout ($p=0.1$):
$$h_1 = \phi(W_1 q + b_1), \quad h_2 = \phi(W_2 h_1 + b_2), \quad \text{Block}(q; \theta) = W_3 h_2 + b_3$$

**3. The RG-ResMoE Architecture:** The proposed model decomposes the forecast into a frozen base prediction and a soft-gated residual correction. The base network and $K=4$ residual experts consume strictly stock-level features $x \in \mathbb{R}^{16}$, while the gating network consumes both features and regime state variables $u = (x, z) \in \mathbb{R}^{18}$:
$$\hat{y}_{\text{base}} = \text{Block}(x; \theta_b)$$
$$r_k(x) = \text{Block}(x; \theta_k)$$
$$\pi = \text{softmax}(g(u))$$
$$\hat{y} = \hat{y}_{\text{base}} + \sum_{k=1}^K \pi_k r_k(x)$$
Crucially, the final layers of the residual experts are **zero-initialized**, ensuring that $r_k(x) = 0$ at initialization and the model begins by exactly reproducing the frozen base forecast.

**4. The Regularized Objective Function:** Stage 2 training optimizes the experts and the gate jointly using a composite loss function that penalizes the aggregate residual magnitude and discourages routing collapse:
$$\mathcal{L} = \text{MSE}(y, \hat{y}) + \alpha \overline{\left(\sum_{k=1}^K \pi_k r_k(x)\right)^2} + \lambda_{\text{LB}} \sum_{k=1}^K \left(\bar{\pi}_k - \frac{1}{K}\right)^2$$

**5. Statistical Inference and Calibration:** Model superiority is evaluated using paired Diebold-Mariano tests on seed-averaged daily Information Coefficient (IC) differences, utilizing Newey-West standard errors with 4 lags to correct for the $\text{MA}(4)$ dependence induced by overlapping targets. Risk calibration is assessed via Kupiec's unconditional coverage likelihood-ratio test on Student-$t$ VaR thresholds.

## Features

- **Information-Matched Architectural Ablations:** Rigorous isolation of the integration pathway. Compares input concatenation (`MLP-L(+z)`) against gate-based routing (`RG-ResMoE`) while holding parameter capacity ($\approx 2,800$ parameters) and random seeds strictly constant.
- **Two-Stage Residual Training Engine:** Implements full-batch Adam optimization with a frozen base network and zero-initialized residual branches, completely eliminating the training collapse endemic to standard Mixture-of-Experts models.
- **Robust Hyperparameter Selection:** A 3-filter validation protocol that rejects boundary configurations, enforces statistical IC dominance over runner-ups, and guarantees MSE robustness without ever touching the test split.
- **Comprehensive Econometric Baselines:** Includes pooled Heterogeneous Autoregressive (HAR) models, per-stock constrained GARCH(1,1) Maximum Likelihood Estimation, 20-day persistence, and L2-regularized Ridge regression.
- **Stress-Condition Slicing:** Automated evaluation of model performance during the COVID-19 crash, the 2022 bear market, top-decile volatility days, and regime-flip windows.
- **Cross-Market Replication:** A fully parameterized pipeline that seamlessly replicates the U.S. study on an independent Japanese TSE Prime panel to verify global generalization.
- **Cryptographic Reproducibility:** Generates a deterministic SHA-256 manifest linking data hashes, configuration versions, seed registries, and final results tables.

## Methodology Implemented

The proposed research approach can be distilled into the following principal steps:

1. **Configuration and Schema Validation:** Strict auditing of the `config.yaml` file, enforcing mathematical bounds, walk-forward geometric identities, and exact parameter capacity matching.
2. **Data Acquisition and Cleansing:** Downloading Yahoo Finance OHLCV bars, computing split/dividend-adjusted log returns, and removing confirmed corporate-action artifacts.
3. **Feature and Regime Engineering:** Constructing 16 stock-specific features (realized volatility, cumulative returns, RSI, lagged returns) and 2 regime variables (systematic market volatility and 120-day rolling OLS idiosyncratic volatility).
4. **Walk-Forward Tensor Assembly:** Generating 30 non-overlapping quarterly test windows, fitting $z$-score normalization moments strictly on the 85% training splits, and assembling PyTorch tensors.
5. **Model Construction:** Instantiating the shared MLP blocks, capacity-matched baselines, standard MoE, RG-ResMoE, and hard-routing variants (learned top-1, volatility quantiles, GICS sectors, market $\times$ idio split).
6. **Hyperparameter Tuning:** Executing 3-seed sweeps across learning rates and regularizers ($\alpha, \lambda_{\text{LB}}$), applying the 3-filter robustness protocol.
7. **Full-Batch Training:** Executing 30-seed training runs with early stopping and best-checkpoint restoration.
8. **Evaluation and Inference:** Computing IC, ICIR, RMSE, $R^2$, and QLIKE. Executing 4-lag Newey-West Diebold-Mariano tests and Student-$t$ Kupiec VaR calibration.
9. **Cross-Market Replication:** Re-executing the entire pipeline on the Japanese TSE Prime panel.
10. **Robustness Auditing:** Generating stress-slice metrics, evaluating gate softness, and compiling the final Markdown audit report with a `FAITHFUL` reproduction verdict.

## Core Components (Notebook Structure)

*Note: All orchestrator callables and their constituent helper functions are contained within a singular, comprehensive Jupyter Notebook (`regime_gated_residual_mixture_of_experts_for_cross_sectional_volatility_forecasting_draft.ipynb`).*

The notebook is structured as a logical sequence of 18 distinct tasks, a consolidated `# Import Essential Modules` cell, and a top-level orchestrator. Key sections include:
- **Task 1:** Validate Configuration Quality and Input Schema Contracts
- **Task 2:** Reproducible Python Environment and Raw Data Acquisition
- **Task 3:** Cleanse Raw Price Data and Construct Aligned Return Panels
- **Task 4:** Construct Stock Features, Regime Variables, and Forward Targets
- **Task 5:** Implement Walk-Forward Splits and Training-Only Normalization
- **Task 6:** Implement the Shared MLP Block and the Gate Network
- **Task 7:** Implement Capacity-Matched MLP Baselines and Standard MoE
- **Task 8:** Implement RG-ResMoE Family and Hard-Routing Variants
- **Task 9:** Implement Classical Baselines
- **Task 10:** Implement the Full-Batch Training Engine
- **Task 11:** Hyperparameter Selection Protocol
- **Task 12:** Compute Forecast Evaluation Metrics
- **Task 13:** Statistical Inference with Diebold–Mariano and Newey–West
- **Task 14:** Value-at-Risk Calibration and Kupiec Testing
- **Task 15:** Replicate the Japanese Cross-Market Study
- **Task 16:** Create the End-to-End Orchestrator Callable
- **Task 17:** Robustness Analyses and Final Verification
- **Task 18:** Top-Level End-to-End Research Pipeline Orchestrator

## Key Callable: `orchestrate_research_pipeline`

The project is designed around a single, top-level user-facing interface function:

- **`orchestrate_research_pipeline`:** This apex orchestrator executes the complete research protocol in dependency order. A single call validates the configuration, builds the model families, runs the per-market data pipelines, executes the 3-seed hyperparameter sweeps and 30-seed training runs, evaluates all econometric metrics, performs the Japanese cross-market replication, conducts the stress-condition robustness analysis, and persists all artifacts (tensors, checkpoints, figures, tables, and cryptographic manifests) to disk.

```python
def orchestrate_research_pipeline(
    config: Dict[str, Any],
    raw_root: Path,
    output_root: Path,
    seed_list: List[int],
    markets: List[str] = ("US", "JP"),
```

## Input Data Structure

The pipeline requires raw daily price histories, universe metadata, and a static YAML configuration file (`config.yaml`) saved in the working directory:

- **Raw Price CSVs**: One file per ticker (e.g., `AAPL.csv`, `7203.T.csv`) containing standard Yahoo Finance columns: `date`, `open`, `high`, `low`, `close`, `adj_close`, `volume`. Only `adj_close` is used for return construction.
- **Universe Metadata**: A CSV file mapping tickers to `gics_sector`, `sp1500_member_current`, `tse_prime_member`, and `market_cap_bucket` to support hard-routing variants and universe filtering.
- **`config.yaml`**: The serializable master configuration defining all hyperparameters, walk-forward geometries, architectural widths, regularizer grids, and evaluation thresholds. It acts as the single source of truth for the entire pipeline.

## Usage

Here is the granular, step-by-step guide to executing the end-to-end pipeline for **"Regime-Gated Residual Mixture-of-Experts for Cross-Sectional Volatility Forecasting"**. This example demonstrates how to load the study configuration from a YAML file, set up the required directory structures, and execute the full research pipeline using the `orchestrate_research_pipeline` orchestrator.

*Note: This example assumes that all the callables defined in this conversation (including `orchestrate_research_pipeline`, `validate_config_and_schemas`, etc.) are available in the current namespace, such as within a single Jupyter notebook.*

### **Step 1: Loading the Configuration (`config.yaml`)**

The study relies on a deterministic configuration file (`config.yaml`) that defines all hyperparameters, data splits, architectural details, and evaluation metrics. We assume this file exists in the working directory.

**Methodology:**
1.  **File I/O:** Open `config.yaml` in read mode.
2.  **Parsing:** Use `yaml.safe_load` to convert the YAML structure into a nested Python dictionary.
3.  **Validation:** Catch file existence errors and parsing errors to ensure the pipeline does not proceed with an invalid configuration.

```python
import yaml
import pandas as pd
from pathlib import Path
from typing import Dict, Any, List

def load_study_configuration(filepath: str = "config.yaml") -> Dict[str, Any]:
    """
    Loads the study configuration parameters from a YAML file into a Python dictionary.

    Purpose:
        To ingest the deterministic hyperparameters, data split definitions, architectural
        specifications, and evaluation metrics defined in the external configuration file.
        This ensures reproducibility by separating code from configuration.

    Inputs:
        filepath (str): The relative or absolute path to the YAML configuration file.
                        Default is "config.yaml".

    Processes:
        1.  File Access: Attempts to open the specified file in read mode with UTF-8 encoding.
        2.  Parsing: Uses PyYAML's safe_load to parse the YAML structure into a dictionary.
        3.  Validation: Catches FileNotFoundError and yaml.YAMLError, raising informative
            exceptions to halt execution if the configuration cannot be loaded.

    Outputs:
        Dict[str, Any]: A nested dictionary containing the complete study configuration.

    Raises:
        TypeError: If filepath is not a string.
        FileNotFoundError: If the specified YAML file does not exist.
        yaml.YAMLError: If the file contains invalid YAML syntax.
    """
    # Validate that the filepath is a string
    if not isinstance(filepath, str):
        # Raise a TypeError if the filepath is invalid
        raise TypeError(f"filepath must be a string, got {type(filepath).__name__}.")

    # Convert the string filepath to a Path object for robust handling
    config_path: Path = Path(filepath)

    # Verify that the configuration file exists
    if not config_path.is_file():
        # Raise a FileNotFoundError if the file is missing
        raise FileNotFoundError(f"Configuration file not found at: {config_path.absolute()}")

    try:
        # Open the configuration file in read mode with UTF-8 encoding
        with config_path.open("r", encoding="utf-8") as file:
            # Parse the YAML content safely into a Python dictionary
            config: Dict[str, Any] = yaml.safe_load(file)

        # Log success to the console
        print(f"\nSuccessfully loaded configuration from {config_path.name}")

        # Return the parsed configuration dictionary
        return config

    except yaml.YAMLError as exc:
        # Catch YAML parsing errors and raise an informative exception
        raise ValueError(f"Error parsing YAML file {config_path.name}: {exc}") from exc

# Load the configuration
# Note: Ensure 'config.yaml' is in your working directory with the content provided previously.
study_config: Dict[str, Any] = load_study_configuration("config.yaml")
```

### **Step 2: Executing the Pipeline (`orchestrate_research_pipeline`)**

With the configuration (`study_config`) in memory, we invoke the top-level orchestrator. This function manages the entire lifecycle: configuration validation, data acquisition (if raw data is missing), panel alignment, feature engineering, walk-forward splitting, model training (including the two-stage RG-ResMoE protocol), hyperparameter selection, evaluation, statistical testing, and robustness analysis.

**Methodology:**
1.  **Directory Setup:** Define the root directories for raw data (`raw_root`) and pipeline artifacts (`output_root`).
2.  **Parameter Extraction:** Extract the required `seed_list` and `markets` from the loaded configuration to ensure consistency.
3.  **Function Call:** Pass the configuration, directories, seeds, and markets to `orchestrate_research_pipeline`.
4.  **Output Handling:** The function returns a dictionary containing all experimental artifacts, including the final `StudyReport`.

```python
# ==============================================================================
# Execution of the End-to-End Study Pipeline
# ==============================================================================

if __name__ == "__main__":
    # Ensure we have a valid configuration before running
    if study_config:
        print("Initiating RG-ResMoE Study Pipeline...")

        # 1. Define Root Directories
        # Define the directory where raw Yahoo Finance CSVs will be stored or read from
        raw_data_root: Path = Path("./data/raw")
        # Define the directory where all pipeline artifacts (panels, models, metrics) will be saved
        pipeline_output_root: Path = Path("./output")

        # 2. Extract Execution Parameters from Configuration
        # Extract the deterministic seed registry from the reproducibility section
        seed_registry: List[int] = study_config.get("reproducibility_and_audit", {}).get(
            "seed_values", list(range(30))
        )
        # Define the target markets for the study (U.S. main study and Japanese replication)
        target_markets: List[str] = ["US", "JP"]

        try:
            # 3. Execute the Pipeline
            # Invoke the master orchestrator to run all 18 tasks
            # Note: This process is computationally intensive and may take significant time
            # depending on hardware and the presence of cached checkpoints.
            study_artifacts: Dict[str, Any] = orchestrate_research_pipeline(
                config=study_config,
                raw_root=raw_data_root,
                output_root=pipeline_output_root,
                seed_list=seed_registry,
                markets=target_markets
            )

            # ==============================================================================
            # Inspecting the Outputs
            # ==============================================================================

            print("\n" + "="*80)
            print("STUDY EXECUTION COMPLETE")
            print("="*80)

            # Extract the final StudyReport from the robustness stage artifacts
            # The robustness stage finalizes the audit report and decision log
            robustness_report = study_artifacts.get("task_17_robustness")

            if robustness_report:
                # 1. Accessing the Stress-Condition Results
                print("\n[Stress-Condition Analysis]")
                print("Performance of RG-ResMoE vs. MLP-L during market stress:")
                print(robustness_report.stress_table.to_string(index=False))

                # 2. Accessing the Robustness Results
                print("\n[Robustness Analysis]")
                print("Evaluation of hard-routing variants and ablations:")
                # Display a subset of columns for readability
                display_cols = ["variant", "ic", "collapsed", "p_value"]
                print(robustness_report.robustness_table[display_cols].to_string(index=False))

                # 3. Locating the Audit Report and Decision Log
                print("\n[Artifact Locations]")
                print(f"Final Audit Report: {robustness_report.audit_report_path}")
                print(f"Decision Log:       {robustness_report.decision_log_path}")

            # Extract the Japanese replication report
            jp_report = study_artifacts.get("task_15_japan")
            if jp_report:
                # 4. Accessing the Japanese Replication Results
                print("\n[Japanese Cross-Market Replication]")
                print("Performance on the TSE Prime panel:")
                display_cols_jp = ["model", "mean_ic", "mean_rmse", "mean_qlike", "collapsed_seeds"]
                print(jp_report.results_table[display_cols_jp].to_string(index=False))

        except Exception as e:
            # Catch and display any errors that occurred during pipeline execution
            print(f"\nPipeline execution failed: {e}")
            raise

    else:
        print("Error: Missing configuration. Cannot proceed.")
```

### **Summary of the Execution Flow**

1.  **Configuration Ingestion:** The pipeline begins by loading the deterministic `config.yaml` file, which dictates all hyperparameters, architectural choices, and evaluation rules.
2.  **Validation & Setup (Task 1 & 6-9):** The orchestrator validates the configuration structure, verifies mathematical constraints, and instantiates all model factories (Shared Blocks, MLP Baselines, Standard MoE, RG-ResMoE variants, and Classical Baselines), ensuring capacity alignment.
3.  **Data Acquisition & Preprocessing (Tasks 2-5):** For each market (US, JP), the pipeline downloads raw data (if absent), cleanses prices, aligns returns to a common calendar, computes 16 features and 2 regime variables, generates the 5-day forward target, and creates 30 normalized walk-forward tensor bundles.
4.  **U.S. Market Execution (Tasks 10-14):** The pipeline performs 3-seed hyperparameter sweeps on the first window, executes 30-seed full-batch training across all windows (using the two-stage protocol for RG-ResMoE), evaluates out-of-sample metrics (IC, RMSE, QLIKE), runs Diebold-Mariano statistical tests, and calibrates Student-t VaR thresholds.
5.  **Japanese Replication (Task 15):** The pipeline repeats the tuning, training, and evaluation process on the TSE Prime panel, verifying that the qualitative stability and accuracy signatures hold across markets.
6.  **Robustness & Audit (Tasks 16-17):** Finally, the pipeline computes stress-slice metrics, evaluates gate softness, asserts robustness constraints (e.g., soft > hard routing), finalizes the decision log, and generates a comprehensive Markdown audit report with a `FAITHFUL` reproduction verdict.

## Output Structure

The pipeline returns a comprehensive artifacts dictionary and writes a deterministic directory tree under `output_root`:

- **`raw/`** — The downloaded Yahoo Finance CSVs and cryptographic manifests.
- **`panels/`** — The cleansed, aligned, and feature-engineered Parquet panels.
- **`splits/`** — The 30 normalized walk-forward tensor bundles and row-order metadata.
- **`models/`** — Serialized PyTorch model architectures.
- **`checkpoints/`** — Resumable intermediate training states and stage-completion `.done` markers.
- **`forecasts/`** — Out-of-sample predictions aligned by ticker and date.
- **`metrics/`** — Evaluation tables, Diebold-Mariano JSON reports, and Kupiec VaR summaries.
- **`figures/`** — The generated plots, including the per-seed QLIKE stability scatter plot.
- **`logs/`** — The `decision_log.json` (documenting all parameter resolutions) and `reproducibility_manifest.json` (linking data hashes, seeds, and results).

## Customization

The pipeline is highly customizable via the `config.yaml` file. Researchers and risk managers can modify the framework without altering a single line of Python code:

- **Walk-Forward Geometry:** Adjust `development_length_trading_days`, `test_length_trading_days`, or `number_of_windows` to evaluate performance over different horizons.
- **Regularization Grids:** Expand or refine the `alpha_candidate_grid` and `lambda_LB_candidate_grid` to test alternative shrinkage and load-balancing strengths.
- **Architectural Widths:** Modify `base_network_width` or `expert_width` to scale the capacity of the RG-ResMoE model.
- **Hard-Routing Rules:** Redefine the `hard_routing_gics_sectors.mapping` to test alternative industry clustering strategies.
- **Risk Calibration:** Adjust the `value_at_risk_calibration.levels` to evaluate expected shortfall or extreme tail risk (e.g., 0.1% VaR).

# Paper

Title: "*Regime-Gated Residual Mixture-of-Experts for Cross-Sectional Volatility Forecasting*"

Authors: Junyi Ye, Gargi Vijay Borde

E-Journal Submission Date: 12 August 2026

Link: https://arxiv.org/abs/2608.12251

Abstract:

Financial volatility is regime dependent, yet incorporating regime information into neural networks can also destabilize training. This paper asks where such information should enter a neural cross-sectional volatility forecasting model. We study five-day realized-volatility forecasts for 1,027 U.S. equities using a rolling walk-forward evaluation framework in which information, model capacity, hyperparameter tuning, and random seeds are matched across architectures. We propose RG-ResMoE, a regime-gated residual mixture-of-experts architecture in which regime information is used only for expert routing rather than for direct forecasting. The base predictor models volatility from stock features, while a gating network uses regime state variables to route residual corrections. RG-ResMoE consistently outperforms a capacity-matched MLP in both forecasting accuracy and training stability in the main U.S. study. Similar gains are observed on an independent Japanese panel. The integration pathway is decisive: appending the same regime variables directly to the forecasting input degrades both predictive performance and training stability, whereas restricting them to the routing gate improves accuracy and Value-at-Risk calibration. Hard routing consistently underperforms soft routing. The results suggest that, in compact neural volatility forecasting models, the primary value of mixture-of-experts models lies less in increasing model capacity than in controlling how nonstationary regime information influences prediction.

# Paper Summary

# Granular Step-by-Step Summary of the Paper

The summary follows the paper’s logic: motivation, data, task, architecture, evaluation design, main results, ablations, practical analysis, cross-market replication, related work, and conclusions.

## Core Research Question

The paper asks a precise architectural question about financial machine learning:

> **Given that financial volatility is regime-dependent, where should regime information enter a neural forecasting model?**

The authors do **not** ask whether regime information is useful. Instead, they isolate the **integration pathway** of regime variables while holding constant:
- the predictive features;
- the forecasting backbone;
- the total parameter budget;
- the hyperparameter selection procedure;
- the evaluation protocol;
- and, crucially, the random seeds.

The central claim is that the **location** of regime information in the network – as direct input versus as routing signal for a mixture of experts – has a larger effect on performance and training stability than simply adding regime variables or increasing model capacity.

## Proposed Model – RG-ResMoE

The paper proposes **Regime-Gated Residual Mixture-of-Experts (RG-ResMoE)**.

The architecture consists of:

1. **A frozen base network**  
   A two-hidden-layer MLP that takes only stock-level features \(x\) and produces a base volatility forecast:
   \[
   \hat{y}_{\mathrm{base}} = \mathrm{Block}(x; \theta_b)
   \]

2. **A set of \(K = 4\) residual experts**  
   Each expert is the same MLP block, also receiving only \(x\), but predicting a **residual correction** rather than the full volatility:
   \[
   r_k(x) = \mathrm{Block}(x; \theta_k)
   \]
   Each expert’s final layer is **zero-initialized**, so at initialization:
   \[
   r_k(x) = 0
   \]
   and the full model initially reproduces the frozen base forecast.

3. **A gating network**  
   The gate receives both stock features \(x\) and regime state variables \(z\):
   \[
   \pi = \mathrm{softmax}(g(x,z))
   \]
   The gate does **not** directly predict volatility. It only decides how to combine the residual corrections.

4. **The final prediction**  
   The final forecast is:
   \[
   \hat{y} = \hat{y}_{\mathrm{base}} + \sum_{k=1}^K \pi_k r_k(x)
   \]

Thus, regime information enters **only through the routing gate**, never through the forecasting blocks.

## Training Procedure and Regularization

Training is performed in **two stages**:

### Stage 1
The base network is trained first and then **frozen**.

### Stage 2
The \(K=4\) residual experts and the gating network are trained jointly.

The loss function is:
\[
\mathcal{L} = \mathrm{MSE}(y, \hat{y})
+ \alpha \, \overline{\left(\sum_{k=1}^K \pi_k r_k(x)\right)^2}
+ \lambda_{\mathrm{LB}} \sum_{k=1}^K \left(\bar{\pi}_k - \frac{1}{K}\right)^2
\]

where:
- the first term is the standard mean-squared error;
- the second term is a **residual shrinkage penalty** that keeps the aggregated correction close to zero unless it improves the base forecast;
- the third term is a **load-balancing penalty** that prevents routing collapse, encouraging the gate to spread weight across experts.

The residual shrinkage and zero-initialization are central to the model’s stability and performance.

## Data

The empirical study uses two equity panels:

### U.S. Panel
- **1,027 common stocks** from Yahoo Finance.
- Sample period: December 2015 to November 2025.
- 85% are current S&P 1500 constituents.
- 92% have the full ten-year price history.
- Covers all eleven sectors and large-, mid-, and small-cap stocks.

### Japanese Replication Panel
- **1,552 stocks** from the TSE Prime segment.
- Same sample period and same data-processing pipeline.
- Used as an out-of-sample cross-market robustness check.

## Step 5: Features and Regime Variables

### Stock-Level Features \(x\)
Each stock-day input contains **16 features**:
- realized volatility over trailing 5, 20, and 60 days;
- cumulative returns over trailing 5 and 20 days;
- 14-day relative strength index (RSI);
- the ten most recent daily log returns.

### Regime State Variables \(z\)
Two volatility-based regime variables:

1. **Market volatility**  
   The 20-day rolling volatility of the equal-weighted market return \(r_{m,t}\). This is common to all stocks on a given trading day.

2. **Idiosyncratic volatility**  
   The 20-day rolling volatility of the residual:
   \[
   r_{i,t} - \beta_{i,t} r_{m,t}
   \]
   where \(\beta_{i,t}\) is estimated from the preceding 120 trading days.

These variables capture:
- systematic uncertainty;
- stock-specific uncertainty after removing market movements.

Crucially, **the same variables are used in all experiments**. Only the integration pathway changes.

## Forecasting Task and Evaluation Protocol

### Task
For stock \(i\) on day \(t\), the target is the **annualized five-day forward realized volatility**:
\[
y_{i,t} = \sqrt{252} \, \mathrm{Std}(r_{i,t+1}, \dots, r_{i,t+5})
\]

### Walk-Forward Evaluation
- Development period: **504 trading days** (two years).
- Split: first 85% training, final 15% validation.
- Test period: **63 trading days** (one quarter).
- The window advances by 63 days after each evaluation.
- Produces **30 non-overlapping test windows** from April 2018 to October 2025.
- Approximately **1.9 million out-of-sample forecasts**.

The evaluation period includes:
- the COVID-19 crash in 2020;
- the 2021 recovery;
- the 2022 bear market;
- calmer periods before and after.

## Models and Baselines

The paper compares:

### Classical Baselines
- **Persistence**: 20-day realized volatility.
- **GARCH(1,1)**: estimated per stock.
- **HAR**: Heterogeneous AutoRegressive model, pooled.
- **Ridge**: linear model using the same stock features.

### Neural Baselines
- **MLP-S**: single MLP block with hidden width \(H=16\), matching one RG-ResMoE block.
- **MLP-L**: larger MLP with \(H=44\), matching the full capacity of RG-ResMoE.
- **MLP-L(+z) / MLP-S(+z) / Ridge(+z)**: the same models with regime variables appended directly to inputs.
- **Standard MoE**: each expert predicts the full volatility level, gate uses \((x,z)\).

### RG-ResMoE Variants
- **RG-ResMoE** (main): \(K=4\), soft gate, frozen base.
- **RG-ResMoE(-z)**: regime variables removed from the gate.
- **RG-ResMoE-K2 / K6**: two or six experts with capacity matched.
- **RG-ResMoE (random init)**: residual experts are not zero-initialized.
- **Hard routing variants**: learned top-1 gate, volatility quantiles, GICS sectors, market × idiosyncratic split.

## Evaluation Metrics

The paper uses a broad set of metrics:

1. **IC** – daily cross-sectional Spearman correlation between predicted and realized volatility.
2. **ICIR** – mean daily IC divided by its time-series standard deviation.
3. **RMSE** – root mean-squared error on annualized volatility.
4. **Out-of-sample \(R^2\)**.
5. **QLIKE** – quasi-likelihood loss for variance forecasts.
6. **VaR calibration** – Kupiec’s test rejection rates for 5% and 1% VaR, using Student-\(t\) residual distributions.
7. **Collapse frequency** – number of random seeds with mean test QLIKE > 2.0, a predefined training-instability threshold.

Statistical comparisons use **Diebold–Mariano-style tests** with **Newey–West standard errors** and four lags.

## Main Results

The headline results on the U.S. panel are:

| Model | IC | RMSE | \(R^2\) | ICIR | QLIKE | Collapse |
|---|---|---|---|---|---|---|
| Persistence | 0.5018 | 0.2570 | 0.121 | 5.18 | 0.799 | --- |
| GARCH(1,1) | 0.5200 | 0.2642 | 0.070 | 5.48 | 1.120 | --- |
| HAR | 0.5315 | 0.2347 | 0.266 | 5.93 | 0.749 | --- |
| Ridge | 0.5266 | 0.2336 | 0.273 | 5.75 | 0.760 | --- |
| MLP-S | 0.5393 | 0.2323 | 0.280 | 6.00 | 74.3 | 2/30 |
| MLP-L | 0.5421 | 0.2320 | 0.282 | 6.08 | 2.23 | 3/30 |
| Standard MoE | 0.5413 | 0.2357 | 0.259 | 6.07 | 8,775 | 24/30 |
| **RG-ResMoE** | **0.5469** | **0.2304** | **0.292** | **6.14** | **0.735** | **0/30** |

Key findings:

- RG-ResMoE outperforms **all** baselines on **every** forecasting metric.
- **Capacity alone is not the driver**: MLP-L has the same capacity as RG-ResMoE but is worse.
- **Standard MoE is highly unstable**: 24 of 30 seeds collapse, despite comparable average IC.
- The residual decomposition improves both accuracy and training stability.

## Where the Gain Comes From

### Integration Pathway Is Decisive

The most important experimental comparison is:

- **Input pathway**: append \(z\) to \(x\) and feed both into the forecasting network.
- **Gate pathway**: use \(z\) only in the MoE gate.

Results:

- Ridge(+z): IC drops from 0.5266 to 0.5243.
- MLP-S(+z): IC drops and collapses increase from 2/30 to 22/30.
- MLP-L(+z): IC drops from 0.5421 to 0.5378, and **all 30 seeds collapse**.
- **RG-ResMoE(-z)**: IC = 0.5466, 1/30 collapse.
- **RG-ResMoE**: IC = 0.5469, **0/30 collapse**.

This is a striking result:

> Adding the same regime information to the input actively harms both prediction and stability, while adding it to the routing gate improves accuracy and stabilizes training.

### Behavior Across Market Regimes

The per-window analysis shows:

- The input pathway causes **large oscillations** in \(\Delta\)IC and \(\Delta\)RMSE around the COVID crisis and recovery.
- The gate pathway produces **small, stable improvements** throughout the sample.
- Direct input integration behaves badly during regime transitions; gate-based integration remains stable.

### What Does the Gate Learn?

The learned gate is **broadly distributed**:

- Average maximum gate weight is about **0.36**, close to uniform routing (0.25 for 4 experts).
- Very few observations are assigned almost entirely to one expert.
- Expert weights are correlated with both market volatility and idiosyncratic volatility.
- The strongest-loaded experts often have **opposite-signed** correlations, meaning the gate performs continuous, state-dependent reweighting.

## Soft Routing Versus Hard Routing

The paper asks whether discrete regime assignment can replace soft routing.

Four hard-routing variants are compared while holding the architecture fixed:

1. **Learned top-1 gate**
2. **Volatility quantiles**
3. **GICS sectors**
4. **Market × idiosyncratic volatility split**

All hard variants underperform soft routing:

| Variant | Routing | IC | \(\Delta\)IC |
|---|---|---|---|
| **RG-ResMoE** | soft | **0.5469** | ref. |
| learned top-1 | hard | 0.5448 | \(-0.0021\) |
| volatility quantiles | hard | 0.5447 | \(-0.0022\) |
| GICS sectors | hard | 0.5445 | \(-0.0024\) |
| market × idio split | hard | 0.5435 | \(-0.0034\) |

All differences are statistically significant with \(p < 10^{-4}\).

The implication is important:

> The benefit of RG-ResMoE is **not** due to a better definition of discrete regimes. It comes from allowing multiple experts to contribute simultaneously and smoothly.

## Ablations on Expert Count and Residual Design

The paper ablates the architecture further:

| Variant | IC | \(\Delta\)IC | Collapse |
|---|---|---|---|
| **RG-ResMoE-K4 (main)** | **0.5469** | ref. | **0/30** |
| RG-ResMoE-K2 | 0.5466 | \(-0.0004\) | 0/30 |
| RG-ResMoE-K6 | 0.5467 | \(-0.0002\) | 3/30 |
| RG-ResMoE (random init) | 0.5430 | \(-0.0039\) | 13/30 |
| Standard MoE | 0.5413 | \(-0.0056\) | 24/30 |

Findings:

- **Expert count matters little**: two experts retain almost all gain; six experts add instability without improving IC.
- **Zero initialization matters greatly**: random initialization drops IC and causes 13/30 collapses.
- **Frozen base matters most**: removing it, i.e. standard MoE, produces the worst accuracy and the most collapses.

Thus, the **residual architecture** is the main source of both accuracy and stability, not the number of experts.

## Practical Evaluation

### Training Reliability
Figure 4 shows per-seed QLIKE distributions:
- RG-ResMoE and RG-ResMoE-K2: **0/30 collapses**.
- RG-ResMoE-K6 and MLP-L: **3/30 collapses**.
- RG-ResMoE with random initialization: **13/30 collapses**.
- Standard MoE: **24/30 collapses**.
- MLP-L(+z): **30/30 collapses**.

The residual design is the primary source of optimization robustness.

### Value-at-Risk Calibration
RG-ResMoE has the **lowest Kupiec rejection rates** in 5 of 6 regime–VaR combinations:

- 5% VaR, all dates: 26.7% versus 30.3% for MLP-L and 32.7% for MLP-S.
- 5% VaR, high volatility: 11.9% versus 14.8% and 16.9%.
- 5% VaR, low volatility: 22.9% versus 24.7% and 28.2%.
- 1% VaR, all dates: 9.1% versus 9.9% and 11.4%.
- 1% VaR, high volatility: 6.2% versus 7.1% and 7.2%.
- 1% VaR, low volatility: 5.8% versus 5.8% (n.s.) and 6.8% (n.s.).

The largest gains are in **high-volatility periods**, where tail-risk calibration is most important.

### Performance During Market Stress
RG-ResMoE’s IC advantage over MLP-L is concentrated in stressed periods:

| Slice | \(\Delta\)IC | Relative Gain |
|---|---|---|
| Full sample | +0.0048 | ref. |
| Top market-volatility decile | +0.0207 | \(4.3\times\) |
| COVID crisis (2020-02 to 2020-06) | +0.0322 | \(6.7\times\) |
| 2022 bear market | +0.0079 | \(1.6\times\) |
| Regime-flip windows (\(\pm 10\) days) | +0.0031 | \(0.6\times\) |

Notably, the advantage is **not** amplified during short regime transitions, but during **sustained periods of elevated volatility**.

## Cross-Market Replication on Japan

The Japanese TSE Prime panel reproduces the same ordering:

| Model | IC | RMSE | QLIKE | Collapse |
|---|---|---|---|---|
| MLP-L | 0.4815 | 0.1851 | 11.6 | 6/30 |
| MLP-L(+z) | 0.4815 | 0.1901 | 41,256 | 28/30 |
| Standard MoE | 0.4805 | 0.1862 | 3,081 | 22/30 |
| **RG-ResMoE** | **0.4858** | **0.1846** | **0.80** | **1/30** |

Key observations:

- RG-ResMoE outperforms MLP-L by **+0.0043 IC**, with significant improvements in IC, RMSE, and QLIKE.
- **Input concatenation is again unstable**, collapsing in 28/30 seeds and worsening RMSE.
- The same stability ordering holds as in the U.S. market.

This suggests the results are **not market-specific**.

## Relationship to Prior Work

The paper positions itself within three literatures:

### Regime-Switching Models
- Classical Markov-switching and SWARCH models made regime dependence central.
- Recent neural regime-aware architectures incorporate regime information, but often change multiple components at once.
- This paper uniquely **isolates the integration pathway**.

### Mixture-of-Experts
- MoE originated with Jacobs et al. (1991).
- Modern sparse MoE focuses on computational scaling and top-\(k\) routing.
- This paper shows that in compact financial models, the value of MoE is not capacity scaling but **controlled integration of nonstationary state information**.

### Residual Learning and Stable Initialization
- Residual networks and zero-initialized residual branches improve optimization.
- This paper combines residual learning with gated routing and shows large stability gains in financial forecasting.

## Conclusion and Limitations

### Main Conclusion
The paper establishes a clear design principle:

> In compact neural volatility forecasting models, regime information should guide residual routing rather than be treated as ordinary predictive input.

The strongest empirical evidence is:
- the same regime variables **hurt** when appended to inputs;
- they **help** when used only by the gate;
- soft routing is consistently better than hard routing;
- the frozen base network and zero-initialized residual experts are essential for both accuracy and stability.

### Limitations
The authors explicitly list three limitations:

1. The study is limited to **volatility forecasting**.
2. It considers only **two equity markets**.
3. It uses a **compact MLP backbone**.

Future work should examine:
- other financial forecasting tasks;
- larger sequence models;
- alternative forms of market state;
- whether the same design principle generalizes beyond volatility.

## Final Takeaway

**RG-ResMoE** is not simply “another MoE.” It is an architecture that deliberately separates:

- **what predicts volatility** (stock features through a frozen base and residual experts);
- **what adapts the prediction to market conditions** (regime variables through the gate).

The central message is that architecture design choices, specifically **where auxiliary information enters the model**, can matter more than raw model capacity. In this setting, regime information is most effective as a **controller of expert mixtures**, not as an additional predictor.

*   **Outputs**: An immutable `AlignedPanelReport` containing the wide return matrix, the market return series, the final ticker list, and persisted Parquet artifact paths.
*   **Data Transformation**: Transforms asynchronous, long-format raw price bars into a synchronous, wide $T \times N$ return matrix aligned strictly to trading days, computing the cross-sectional mean to generate the market return vector.
*   **Role in Research Pipeline**: Implements the foundational log-return construction $r_{i,t} = \ln(P_{i,t}/P_{i,t-1})$ and the systematic index construction: *"equal-weighted market return $r_{m,t}$... shared by all stocks on a given trading day."*

#### 4. Task 4 Orchestrator: `build_features_targets`
*   **Inputs**: The wide return matrix, the market return series, `config`, and the output directory.
*   **Processes**: Melts the wide matrix; computes rolling realized volatilities, cumulative returns, RSI, and lagged returns; estimates 120-day rolling OLS betas; computes idiosyncratic residual returns; calculates the 20-day rolling market and idiosyncratic volatilities; computes the 5-day forward target; and drops all incomplete rows.
*   **Outputs**: An immutable `FeaturePanelReport` containing the complete long feature panel and column registries.
*   **Data Transformation**: Transforms the wide return matrix into a dense, row-complete panel where each stock-day observation is augmented with a 16-dimensional feature vector $x_{i,t}$, a 2-dimensional regime state vector $z_{i,t}$, and a scalar target $y_{i,t}$.
*   **Role in Research Pipeline**: Implements the predictive inputs and the forecasting target: *"Market volatility... and Idiosyncratic volatility... $r_{i,t} - \beta_{i,t}r_{m,t}$"*, and the target equation:
  $$y_{i,t} = \sqrt{252} \, \text{Std}(r_{i,t+1}, \dots, r_{i,t+5})$$

#### 5. Task 5 Orchestrator: `build_walkforward_splits`
*   **Inputs**: The complete feature panel, `config`, and the output directory.
*   **Processes**: Generates 30 temporal window indices; maps calendar positions to exact panel row masks; fits $z$-score normalization moments strictly on training splits; applies normalization to validation and test splits; assembles PyTorch tensors; and audits the moments to guarantee zero leakage.
*   **Outputs**: An immutable `WalkForwardReport` containing window records, fitted moments, and 30 tensor bundles ($X, Z, y$).
*   **Data Transformation**: Transforms the continuous feature panel into 30 discrete, normalized, and temporally isolated PyTorch tensor bundles, ensuring that test data never influences the feature scaling parameters $\mu_w$ and $\sigma_w$.
*   **Role in Research Pipeline**: Implements the *"rolling walk-forward evaluation protocol"* (Table 1) utilizing a *"504-trading-day (two years) development period... followed by a 63-trading-day (one quarter) test period"*, ensuring *"normalization parameters are estimated from the training data and then applied to the validation and test data."*

#### 6. Task 6 Orchestrator: `build_shared_blocks`
*   **Inputs**: The validated `config`.
*   **Processes**: Defines factory closures for the `SharedMLPBlock` and `GateNetwork`; instantiates reference architectures; measures exact parameter counts; and verifies these counts against closed-form analytical capacity formulas.
*   **Outputs**: An immutable `ModelBlockReport` containing the block/gate factories, measured parameter counts, and capacity audit findings.
*   **Data Transformation**: Transforms configuration dimensions into instantiated PyTorch modules, extracting their parameter counts to mathematically prove capacity alignment before training begins.
*   **Role in Research Pipeline**: Implements the foundational neural architecture: *"All neural network models are built from the same two-hidden-layer MLP block"*:
  $$h_1 = \phi(W_1 q + b_1), \quad h_2 = \phi(W_2 h_1 + b_2), \quad \text{Block}(q; \theta) = W_3 h_2 + b_3$$

#### 7. Task 7 Orchestrator: `build_mlp_and_moe_models`
*   **Inputs**: The `block_factory`, `gate_factory`, and `config`.
*   **Processes**: Registers unified config-accepting factories for `MLP-S`, `MLP-L`, `MLP-S(+z)`, `MLP-L(+z)`, and `Standard MoE`; verifies that the baseline capacities exactly match the manuscript's theoretical expectations.
*   **Outputs**: An immutable `NeuralModelRegistry` containing the baseline model factories and capacity findings.
*   **Data Transformation**: Composes the shared blocks into complete baseline architectures, wrapping input-pathway models to concatenate $(x, z)$ and assembling the standard MoE to compute full-volatility expert predictions.
*   **Role in Research Pipeline**: Implements the capacity-matched baselines and the unanchored mixture model: *"MLP-L (large) uses the same block structure with a larger hidden width H=44 to match the overall forecasting capacity"*, and the Standard MoE:
  $$e_k(x) = \text{Block}(x; \theta_k), \quad \pi = \text{softmax}(g(u)), \quad \hat{y} = \sum_{k=1}^K \pi_k e_k(x)$$

#### 8. Task 8 Orchestrator: `build_rg_resmoe_family`
*   **Inputs**: The `block_factory`, `gate_factory`, and `config`.
*   **Processes**: Registers factories for the main `RG-ResMoE` model and its ablations (`-z`, `random_init`, `K2`, `K6`); constructs the four hard-routing assignment rules; and verifies that the expert capacities of K2 and K6 remain within a 2.5% tolerance of the 2,244 parameter baseline.
*   **Outputs**: An immutable `RGResMoEFamilyReport` containing the variant factories, routing rules, and expert capacities.
*   **Data Transformation**: Composes the shared blocks into the residual architecture, explicitly zero-initializing the expert output layers so that $r_k(x) = 0$ at initialization, and mapping discrete hard-routing rules to integer expert assignments.
*   **Role in Research Pipeline**: Implements the proposed architecture and its ablations:
  $$\hat{y}_{\text{base}} = \text{Block}(x; \theta_b), \quad r_k(x) = \text{Block}(x; \theta_k), \quad \hat{y}_{\text{correction}} = \sum_{k=1}^K \pi_k r_k(x), \quad \hat{y} = \hat{y}_{\text{base}} + \hat{y}_{\text{correction}}$$

#### 9. Task 9 Orchestrator: `build_classical_baselines`
*   **Inputs**: The validated `config`.
*   **Processes**: Assembles the registry of callables for the 20-day Persistence forecast, the pooled HAR OLS estimation, the per-stock GARCH(1,1) constrained MLE, and the L2-regularized Ridge regressions.
*   **Outputs**: An immutable `ClassicalBaselineReport` containing the econometric estimation and forecasting functions.
*   **Data Transformation**: Transforms configuration parameters into executable econometric routines that map historical return vectors to scalar volatility forecasts via linear algebra and maximum likelihood estimation.
*   **Role in Research Pipeline**: Implements the econometric benchmarks: *"three classical forecasting baselines: 20-day persistence, pooled HAR [4], and per-stock GARCH(1,1) [3]. Ridge provides a linear baseline"*.

#### 10. Task 10 Orchestrator: `train_full_batch`
*   **Inputs**: The PyTorch `model`, training/validation tensors ($X, y, Z$), `config`, random `seed`, and the training `stage`.
*   **Processes**: Executes full-batch Adam optimization; evaluates the loss function (MSE for Stage 1, Regularized for Stage 2); monitors validation loss; applies early stopping; and restores the best parameter checkpoint via deep-copying state dictionaries.
*   **Outputs**: An immutable `TrainingResult` containing the optimized model, the loss history DataFrame, and the best validation loss.
*   **Data Transformation**: Transforms randomly initialized or frozen-base network weights into optimized parameters by iteratively computing full-batch gradients and applying Adam updates, halting when out-of-sample validation error stagnates.
*   **Role in Research Pipeline**: Implements *"full-batch Adam optimization of mean-squared error (MSE), early stopping based on validation loss, and restoration of the best checkpoint"*, and the Stage 2 regularized objective:
  $$\mathcal{L} = \text{MSE}(y, \hat{y}) + \alpha \overline{\left(\sum_{k=1}^K \pi_k r_k(x)\right)^2} + \lambda_{\text{LB}} \sum_{k=1}^K \left(\bar{\pi}_k - \frac{1}{K}\right)^2$$

#### 11. Task 11 Orchestrator: `select_hyperparameters`
*   **Inputs**: The `model_factory`, the window 1 `splits` dictionary, and `config`.
*   **Processes**: Generates the candidate grid; executes 3-seed training sweeps; computes validation IC and MSE; applies the three robustness filters (interior, IC dominance, MSE robustness); and automatically expands the grid or applies fallback rules if all candidates are rejected.
*   **Outputs**: An immutable `HyperparameterSelectionReport` containing the selected configuration, sweep results, filter logs, and daily IC series.
*   **Data Transformation**: Transforms a grid of hyperparameters into a single optimal configuration dictionary by mapping each candidate to its seed-averaged validation performance and filtering out boundary or unstable configurations.
*   **Role in Research Pipeline**: Implements the hyperparameter tuning protocol: *"Candidate configurations are filtered using three robustness criteria. The selected configuration (i) lies in the interior of the search grid... (ii) achieves a mean validation IC exceeding that of the runner-up by more than its across-seed standard deviation, and (iii) has a validation MSE within one standard deviation of the lowest MSE"*.

#### 12. Task 12 Orchestrator: `compute_evaluation_metrics`
*   **Inputs**: A dictionary of `per_seed_predictions`, the test `targets` DataFrame, `config`, and the `model_name`.
*   **Processes**: Merges predictions with targets; computes daily Spearman IC, ICIR, pooled RMSE, out-of-sample $R^2$, and QLIKE; and evaluates the collapse diagnostic across all 30 seeds.
*   **Outputs**: An immutable `EvaluationMetricsReport` containing the per-seed metrics, daily IC series, aggregated statistics, and the collapse summary.
*   **Data Transformation**: Transforms raw stock-day predictions into aggregated scalar performance metrics and boolean collapse flags, averaging results across the 30 random seeds.
*   **Role in Research Pipeline**: Computes the primary accuracy metrics: *"Daily cross-sectional Spearman information coefficient (IC)"*, *"RMSE"*, and the quasi-likelihood loss:
  $$\text{QLIKE}_{i,t} = \frac{\sigma_{i,t}^2}{\hat{\sigma}_{i,t}^2} - \ln\left(\frac{\sigma_{i,t}^2}{\hat{\sigma}_{i,t}^2}\right) - 1$$
  It also evaluates training stability: *"collapsed seeds, defined... as runs whose mean test QLIKE exceeds 2.0."*

#### 13. Task 13 Orchestrator: `run_diebold_mariano`
*   **Inputs**: Model names (`model_a`, `model_b`), their respective daily IC DataFrames, and `config`.
*   **Processes**: Intersects common test days; computes the seed-averaged daily difference series; estimates sample autocovariances; computes the 4-lag Newey-West standard error; and evaluates the Diebold-Mariano $t$-statistic and two-sided $p$-value.
*   **Outputs**: An immutable `DMComparisonReport` containing the difference series, test statistic, standard error, $p$-value, and significance flags.
*   **Data Transformation**: Transforms two competing daily metric series into a single seed-averaged difference vector $d_t$, mapping its variance through a Bartlett kernel to produce a robust test statistic and probability value.
*   **Role in Research Pipeline**: Implements the statistical inference framework: *"Pairwise Diebold–Mariano-style tests... $d_t = \frac{1}{S}\sum_{s=1}^S [X_{A,s}(t) - X_{B,s}(t)]$... tested using Newey–West standard errors with four lags to account for dependence induced by the overlapping five-day targets."*

#### 14. Task 14 Orchestrator: `run_var_calibration`
*   **Inputs**: The `forecasts` DataFrame, `training_returns`, `config`, and optional market volatility series/medians for slicing.
*   **Processes**: Fits Student-$t$ distributions to training residuals via MLE; converts 5-day forecasts to 1-day VaR thresholds; computes binary hit indicators; and evaluates Kupiec unconditional coverage likelihood ratios across all dates, high-volatility, and low-volatility slices.
*   **Outputs**: An immutable `VaRCalibrationReport` containing the fitted parameters, the VaR thresholds/hits, and the Kupiec rejection rate summary.
*   **Data Transformation**: Transforms training returns into distributional parameters ($\mu, \sigma, \nu$), maps volatility forecasts to tail-risk thresholds, and aggregates binary hits into $\chi^2_1$ test statistics to produce ticker rejection fractions.
*   **Role in Research Pipeline**: Implements risk calibration: *"Value at Risk (VaR) coverage: The fraction of tickers for which the null of correct unconditional coverage is rejected by Kupiec’s test... using Student-t residual distributions estimated from the corresponding training window."*

#### 15. Task 15 Orchestrator: `replicate_japanese_study`
*   **Inputs**: The Japanese raw CSV directory, the model factories, `config`, metadata, and the output directory.
*   **Processes**: Reconstructs the Japanese panel using the identical data pipeline; re-tunes hyperparameters on the Japanese validation split; trains the 4 reported models across 30 seeds and 30 windows; evaluates metrics; and asserts the results against the expected manuscript signatures.
*   **Outputs**: An immutable `JapaneseReplicationReport` containing the panel result, selected configs, results table, and signature findings.
*   **Data Transformation**: Executes the entire Tasks 3–12 pipeline on a new domain (TSE Prime), transforming Japanese raw data into a final performance table and verifying that the cross-market stability ordering holds.
*   **Role in Research Pipeline**: Implements the out-of-sample robustness check: *"Cross-Market Replication... evaluate generalization beyond the U.S. market using the Japanese TSE Prime panel... same features, walk-forward protocol, hyperparameter selection rules, and thirty random seeds."*

#### 16. Task 16 Orchestrator: `orchestrate_study`
*   **Inputs**: The validated `config`, target `markets`, `model_registry`, `seed_list`, and `output_dir`.
*   **Processes**: Validates the configuration; creates the deterministic output directory tree; executes the U.S. stage chain (with checkpoint resumption); executes the Japanese stage chain; writes the decision log; computes the SHA-256 digest of the results; and persists the reproducibility manifest.
*   **Outputs**: An immutable `StudyReport` containing the final results table, log paths, figure paths, manifest path, and cryptographic hash.
*   **Data Transformation**: Coordinates the high-level execution graph, transforming raw inputs and configurations into persisted artifacts, metrics tables, and cryptographic manifests, ensuring no stage is redundantly recomputed.
*   **Role in Research Pipeline**: Acts as the overarching execution framework, ensuring that *"information, model capacity, hyperparameter tuning, and random seeds are matched across architectures"* and guaranteeing deterministic reproducibility.

#### 17. Task 17 Orchestrator: `finalize_robustness_analysis`
*   **Inputs**: Daily IC series for RG-ResMoE and MLP-L, market volatility, routing weights, variant summary, `config`, decision log resolutions, and output directories.
*   **Processes**: Computes stress-slice metrics (top decile, COVID, 2022 bear market, regime flips); computes gate softness dispersion; runs robustness assertions (hard routing degradation, random init collapse); finalizes the decision log; and writes the Markdown audit report with the `FAITHFUL` verdict.
*   **Outputs**: An immutable `RobustnessReport` containing the stress table, robustness table, decision log path, and audit report path.
*   **Data Transformation**: Transforms daily metrics and routing weights into stress-slice $\Delta$IC values, relative gain ratios, and gate softness statistics, compiling them into a human-readable Markdown audit report.
*   **Role in Research Pipeline**: Implements the stress analysis: *"Performance under market stress... COVID crisis... 2022 bear market... Regime-flip windows"*. Verifies the core architectural claims: *"soft routing consistently outperforms hard routing"* and *"random initialization... increases the number of collapsed runs"*.

#### 18. Task 18 Orchestrator: `orchestrate_research_pipeline`
*   **Inputs**: The `config`, `raw_root`, `output_root`, `seed_list`, and `markets`.
*   **Processes**: The master orchestrator. Validates the config and builds model families; runs the per-market data pipelines; runs selection, training, and evaluation for the U.S. market; runs the Japanese replication; and executes the robustness finalization.
*   **Outputs**: A comprehensive `Dict[str, Any]` containing all experimental artifacts keyed by stage name.
*   **Data Transformation**: Transforms raw inputs into a complete end-to-end execution graph, yielding a unified dictionary of all trained models, predictions, metrics, and reports.
*   **Role in Research Pipeline**: The ultimate realization of the paper's methodology, executing the *"information-matched comparison in which the regime variables, forecasting backbone, parameter budget, hyperparameter selection, and evaluation protocol are all held fixed, while only the integration pathway is varied."*

## **Usage Example**

Below is a granular, step-by-step guide to executing the end-to-end pipeline for **"Regime-Gated Residual Mixture-of-Experts for Cross-Sectional Volatility Forecasting"**. This example demonstrates how to load the study configuration from a YAML file, set up the required directory structures, and execute the full research pipeline using the `orchestrate_research_pipeline` orchestrator.

*Note: This example assumes that all the callables defined in this conversation (including `orchestrate_research_pipeline`, `validate_config_and_schemas`, etc.) are available in the current namespace, such as within a single Jupyter notebook.*

### **Step 1: Loading the Configuration (`config.yaml`)**

The study relies on a deterministic configuration file (`config.yaml`) that defines all hyperparameters, data splits, architectural details, and evaluation metrics. We assume this file exists in the working directory.

**Methodology:**
1.  **File I/O:** Open `config.yaml` in read mode.
2.  **Parsing:** Use `yaml.safe_load` to convert the YAML structure into a nested Python dictionary.
3.  **Validation:** Catch file existence errors and parsing errors to ensure the pipeline does not proceed with an invalid configuration.

```python
import yaml
import pandas as pd
from pathlib import Path
from typing import Dict, Any, List

def load_study_configuration(filepath: str = "config.yaml") -> Dict[str, Any]:
    """
    Loads the study configuration parameters from a YAML file into a Python dictionary.

    Purpose:
        To ingest the deterministic hyperparameters, data split definitions, architectural
        specifications, and evaluation metrics defined in the external configuration file.
        This ensures reproducibility by separating code from configuration.

    Inputs:
        filepath (str): The relative or absolute path to the YAML configuration file.
                        Default is "config.yaml".

    Processes:
        1.  File Access: Attempts to open the specified file in read mode with UTF-8 encoding.
        2.  Parsing: Uses PyYAML's safe_load to parse the YAML structure into a dictionary.
        3.  Validation: Catches FileNotFoundError and yaml.YAMLError, raising informative
            exceptions to halt execution if the configuration cannot be loaded.

    Outputs:
        Dict[str, Any]: A nested dictionary containing the complete study configuration.

    Raises:
        TypeError: If filepath is not a string.
        FileNotFoundError: If the specified YAML file does not exist.
        yaml.YAMLError: If the file contains invalid YAML syntax.
    """
    # Validate that the filepath is a string
    if not isinstance(filepath, str):
        # Raise a TypeError if the filepath is invalid
        raise TypeError(f"filepath must be a string, got {type(filepath).__name__}.")

    # Convert the string filepath to a Path object for robust handling
    config_path: Path = Path(filepath)

    # Verify that the configuration file exists
    if not config_path.is_file():
        # Raise a FileNotFoundError if the file is missing
        raise FileNotFoundError(f"Configuration file not found at: {config_path.absolute()}")

    try:
        # Open the configuration file in read mode with UTF-8 encoding
        with config_path.open("r", encoding="utf-8") as file:
            # Parse the YAML content safely into a Python dictionary
            config: Dict[str, Any] = yaml.safe_load(file)

        # Log success to the console
        print(f"\nSuccessfully loaded configuration from {config_path.name}")

        # Return the parsed configuration dictionary
        return config

    except yaml.YAMLError as exc:
        # Catch YAML parsing errors and raise an informative exception
        raise ValueError(f"Error parsing YAML file {config_path.name}: {exc}") from exc

# Load the configuration
# Note: Ensure 'config.yaml' is in your working directory with the content provided previously.
study_config: Dict[str, Any] = load_study_configuration("config.yaml")
```

### **Step 2: Executing the Pipeline (`orchestrate_research_pipeline`)**

With the configuration (`study_config`) in memory, we invoke the top-level orchestrator. This function manages the entire lifecycle: configuration validation, data acquisition (if raw data is missing), panel alignment, feature engineering, walk-forward splitting, model training (including the two-stage RG-ResMoE protocol), hyperparameter selection, evaluation, statistical testing, and robustness analysis.

**Methodology:**
1.  **Directory Setup:** Define the root directories for raw data (`raw_root`) and pipeline artifacts (`output_root`).
2.  **Parameter Extraction:** Extract the required `seed_list` and `markets` from the loaded configuration to ensure consistency.
3.  **Function Call:** Pass the configuration, directories, seeds, and markets to `orchestrate_research_pipeline`.
4.  **Output Handling:** The function returns a dictionary containing all experimental artifacts, including the final `StudyReport`.


### **Summary of the Execution Flow**

1.  **Configuration Ingestion:** The pipeline begins by loading the deterministic `config.yaml` file, which dictates all hyperparameters, architectural choices, and evaluation rules.
2.  **Validation & Setup (Task 1 & 6-9):** The orchestrator validates the configuration structure, verifies mathematical constraints, and instantiates all model factories (Shared Blocks, MLP Baselines, Standard MoE, RG-ResMoE variants, and Classical Baselines), ensuring capacity alignment.
3.  **Data Acquisition & Preprocessing (Tasks 2-5):** For each market (US, JP), the pipeline downloads raw data (if absent), cleanses prices, aligns returns to a common calendar, computes 16 features and 2 regime variables, generates the 5-day forward target, and creates 30 normalized walk-forward tensor bundles.
4.  **U.S. Market Execution (Tasks 10-14):** The pipeline performs 3-seed hyperparameter sweeps on the first window, executes 30-seed full-batch training across all windows (using the two-stage protocol for RG-ResMoE), evaluates out-of-sample metrics (IC, RMSE, QLIKE), runs Diebold-Mariano statistical tests, and calibrates Student-t VaR thresholds.
5.  **Japanese Replication (Task 15):** The pipeline repeats the tuning, training, and evaluation process on the TSE Prime panel, verifying that the qualitative stability and accuracy signatures hold across markets.
6.  **Robustness & Audit (Tasks 16-17):** Finally, the pipeline computes stress-slice metrics, evaluates gate softness, asserts robustness constraints (e.g., soft > hard routing), finalizes the decision log, and generates a comprehensive Markdown audit report with a `FAITHFUL` reproduction verdict.

## **Implemented Callables**

````python
# Task 1: Validate Configuration Quality and Input Schema Contracts

# ==============================================================================
# Task 1: Validate Configuration Quality and Input Schema Contracts
# ==============================================================================

# -----------------------------------------------------------------------
# Task 1, Step 1: Audit config Completeness Against the Manuscript
# -----------------------------------------------------------------------

# Define the strict literal type for allowed finding severity levels
ValidationSeverity = Literal["error", "warning", "conflict"]

# Define the immutable set of allowed severity tokens for runtime assertions
ALLOWED_SEVERITIES: Set[str] = {"error", "warning", "conflict"}


@dataclass(frozen=True)
class ValidationFinding:
    """
    Immutable structured diagnostic record capturing a single configuration or schema audit finding.

    This dataclass encapsulates the severity, targeted key path or file name, and a detailed human-readable
    message specifying the observed discrepancy, expected state, and recommended remediation. It strictly
    validates that the severity belongs to the canonical severity domain ('error', 'warning', 'conflict')
    during initialization.

    Parameters
    ----------
    severity : ValidationSeverity
        Severity classification of the audit finding. Must be one of 'error', 'warning', or 'conflict'.
    key_path : str
        Dot-delimited configuration key path or file identifier where the finding originated.
    message : str
        Explanatory description detailing the failure mode, expected value/schema, and actionable remediation.

    Raises
    ------
    ValueError
        If 'severity' is not one of 'error', 'warning', or 'conflict'.
        If 'key_path' or 'message' are empty or contain only whitespace.
    """

    # Categorical severity level of the finding
    severity: ValidationSeverity
    # Dotted key path or file identifier being audited
    key_path: str
    # Detailed technical description of the audit finding
    message: str
    def __post_init__(self) -> None:
        """
        Validate attribute types, values, and boundary conditions upon instantiation.
        """
        # Validate that the severity string is a member of the canonical allowed set
        if self.severity not in ALLOWED_SEVERITIES:
            # Raise ValueError if an invalid or misspelled severity level is provided
            raise ValueError(
                f"Invalid finding severity '{self.severity}'. Must be strictly one of {sorted(ALLOWED_SEVERITIES)}."
            )
        # Validate that the key path is non-empty and contains non-whitespace characters
        if not isinstance(self.key_path, str) or not self.key_path.strip():
            # Raise ValueError if the key path is empty or malformed
            raise ValueError("ValidationFinding 'key_path' must be a non-empty string.")
        # Validate that the diagnostic message is non-empty and contains non-whitespace characters
        if not isinstance(self.message, str) or not self.message.strip():
            # Raise ValueError if the message string is empty
            raise ValueError("ValidationFinding 'message' must be a non-empty string.")


@dataclass(frozen=True)
class ConfigValidationReport:
    """
    Immutable aggregate report generated by the configuration and schema validation pipeline.

    This class encapsulates all diagnostic findings, metadata, checked key counts, and decision-log gaps.
    To eliminate state desynchronization risks, the overall boolean validity status ('valid') is computed
    dynamically as a read-only property that evaluates whether any finding possesses an 'error' or 'conflict'
    severity classification.

    Parameters
    ----------
    config_version : str
        Semantic version string of the audited configuration dictionary.
    findings : Tuple[ValidationFinding, ...]
        Immutable tuple of all validation findings accumulated across audit routines.
    checked_keys : int
        Total number of manuscript-mandated key paths evaluated during the audit.
    decision_log_gaps : Tuple[str, ...]
        Tuple of undocumented decision-log requirement topics identified during the audit.

    Raises
    ------
    ValueError
        If 'checked_keys' is negative or if string metadata fields are malformed.
    """

    # Semantic version string of the audited configuration
    config_version: str
    # Immutable sequence of individual validation findings
    findings: Tuple[ValidationFinding, ...] = field(default_factory=tuple)
    # Total count of manuscript-required keys verified
    checked_keys: int = 0
    # Sequence of missing decision-log requirement topics
    decision_log_gaps: Tuple[str, ...] = field(default_factory=tuple)

    def __post_init__(self) -> None:
        """
        Validate integrity constraints on configuration metadata and counts.
        """
        # Validate that config_version is a valid non-empty string
        if not isinstance(self.config_version, str) or not self.config_version.strip():
            # Raise ValueError if config_version is empty or invalid
            raise ValueError("ConfigValidationReport 'config_version' must be a non-empty string.")
        # Validate that checked_keys count is a non-negative integer
        if not isinstance(self.checked_keys, int) or self.checked_keys < 0:
            # Raise ValueError if checked_keys is negative
            raise ValueError(f"ConfigValidationReport 'checked_keys' must be a non-negative integer, got {self.checked_keys}.")

    @property
    def valid(self) -> bool:
        """
        Dynamically evaluate overall configuration validity.

        Returns
        -------
        bool
            True if zero findings have 'error' or 'conflict' severity; False otherwise.
        """
        # Return True only if no finding contains an error or conflict severity
        return all(f.severity not in ("error", "conflict") for f in self.findings)

    @property
    def error_count(self) -> int:
        """
        Compute the total number of blocking error findings.

        Returns
        -------
        int
            Count of findings with 'error' severity.
        """
        # Sum the total count of error-level findings
        return sum(1 for f in self.findings if f.severity == "error")

    @property
    def warning_count(self) -> int:
        """
        Compute the total number of non-blocking warning findings.

        Returns
        -------
        int
            Count of findings with 'warning' severity.
        """
        # Sum the total count of warning-level findings
        return sum(1 for f in self.findings if f.severity == "warning")

    def to_dict(self) -> Dict[str, Any]:
        """
        Serialize the report into a JSON-compatible dictionary.

        Returns
        -------
        Dict[str, Any]
        Dictionary representation of the validation report suitable for serialization.
        """
        # Construct and return the serializable dictionary representation
        return {
            "config_version": self.config_version,
            "valid": self.valid,
            "error_count": self.error_count,
            "warning_count": self.warning_count,
            "checked_keys": self.checked_keys,
            "decision_log_gaps": list(self.decision_log_gaps),
            "findings": [
                {
                    "severity": f.severity,
                    "key_path": f.key_path,
                    "message": f.message,
                }
                for f in self.findings
            ],
        }


# Frozen registry of manuscript-explicit key paths that must exist
REQUIRED_MANUSCRIPT_KEYS: Tuple[str, ...] = (
    "config_version",
    "config_created_on",
    "config_author",
    "study_metadata.title",
    "data_acquisition_and_universe.source",
    "data_acquisition_and_universe.us_ticker_count_target",
    "data_acquisition_and_universe.jp_ticker_count_target",
    "data_acquisition_and_universe.sample_start",
    "data_acquisition_and_universe.sample_end",
    # S&P 1500 coverage target: 85% of the universe (manuscript)
    "data_acquisition_and_universe.us_inclusion_criteria."
    "sp1500_coverage_target",
    # Full-period coverage target: 92% of the ten-year sample
    "data_acquisition_and_universe.us_inclusion_criteria."
    "full_period_coverage_target",
    "raw_input_data_schemas.us_equity_price_history_csv.fields",
    "raw_input_data_schemas.jp_equity_price_history_csv.fields",
    "data_preprocessing_parameters.returns_price_field",
    "data_preprocessing_parameters.missing_adj_close_policy",
    "data_preprocessing_parameters.target_validity_rule",
    "data_preprocessing_parameters.beta_min_observations",
    "data_preprocessing_parameters.normalization_method",
    "data_preprocessing_parameters.sample_std_ddof",
    "data_preprocessing_parameters.annualization_factor",
    "stock_feature_construction.feature_dimension",
    "stock_feature_construction.feature_names",
    "regime_state_variable_construction.regime_dimension",
    "regime_state_variable_construction.beta_estimation_window",
    "target_construction.horizon_days",
    "walkforward_evaluation.development_length_trading_days",
    "walkforward_evaluation.train_fraction",
    "walkforward_evaluation.train_length_trading_days",
    "walkforward_evaluation.validation_length_trading_days",
    "walkforward_evaluation.test_length_trading_days",
    "walkforward_evaluation.step_length_trading_days",
    "walkforward_evaluation.number_of_windows",
    "shared_neural_network_block.hidden_layers",
    "shared_neural_network_block.activation",
    "shared_neural_network_block.dropout_rate",
    "network_architecture_parameters.MLP_S",
    "network_architecture_parameters.MLP_L",
    "network_architecture_parameters.MLP_S_plus_z",
    "network_architecture_parameters.MLP_L_plus_z",
    "network_architecture_parameters.standard_MoE.expert_count",
    "network_architecture_parameters.RG_ResMoE.expert_count",
    "network_architecture_parameters.RG_ResMoE.base_frozen",
    "network_architecture_parameters.RG_ResMoE.expert_final_layer_init",
    "network_training_parameters.optimizer",
    "network_training_parameters.learning_rate",
    "network_training_parameters.early_stopping_patience",
    "network_training_parameters.seed_values",
    "loss_function_and_regularization.alpha_residual_shrinkage",
    "loss_function_and_regularization.lambda_LB_load_balancing",
    "hyperparameter_selection.criterion",
    "evaluation_metrics.collapse_threshold_qlike",
    "statistical_testing.newey_west_lags",
    "statistical_testing.significance_thresholds",
    "value_at_risk_calibration.levels",
    "value_at_risk_calibration.kupiec_test.significance_level",
    "reproducibility_and_audit.decision_log.required_entries",
    "reproducibility_and_audit.seed_values",
    "llm_and_prompt_configuration.llm_used",
    # Additional manuscript-explicit keys required for complete coverage
    "walkforward_evaluation.windows_overlap",
    "data_acquisition_and_universe.download_protocol.api",
    "data_acquisition_and_universe.jp_ticker_suffix",
    "data_preprocessing_parameters.sample_std_ddof",
    "target_construction.std_ddof",
    "target_construction.scale",
    "regime_state_variable_construction.market_volatility_window",
    "regime_state_variable_construction.idio_volatility_window",
    "regime_state_variable_construction.beta_method",
    "network_architecture_parameters.standard_MoE.gate_input",
    "network_architecture_parameters.RG_ResMoE.base_input",
    "network_architecture_parameters.RG_ResMoE.expert_input",
    "network_architecture_parameters.RG_ResMoE.gate_input",
    "network_architecture_parameters.RG_ResMoE.final_forecast_formula",
    "network_architecture_parameters.RG_ResMoE_K2.expert_count",
    "network_architecture_parameters.RG_ResMoE_K2.expert_width",
    "network_architecture_parameters.RG_ResMoE_K6.expert_count",
    "network_architecture_parameters.RG_ResMoE_K6.expert_width",
    "network_architecture_parameters.gate_architecture.hidden_layers",
    "network_architecture_parameters.gate_architecture.hidden_width",
    "network_architecture_parameters.gate_architecture.activation",
    "network_architecture_parameters.gate_architecture.dropout_rate",
    "network_architecture_parameters.gate_architecture.output",
    "network_training_parameters.full_batch",
    "network_training_parameters.max_epochs",
    "evaluation_metrics.ICIR.definition",
    "statistical_testing.seed_averaging",
    "statistical_testing.lag_rationale",
    "classical_baselines.persistence.window",
    "classical_baselines.HAR.windows",
    "classical_baselines.GARCH_1_1.order",
)


class FlattenedConfig(NamedTuple):
    """
    Structured container holding partitioned leaf mappings and branch sets from dictionary flattening.

    Attributes
    ----------
    leaf_map : Dict[str, Any]
        Mapping from terminal dotted key paths directly to their primitive leaf values.
    branch_set : Set[str]
        Set of dotted key paths representing intermediate container dictionaries.
    """

    # Dictionary mapping terminal dotted paths to leaf values
    leaf_map: Dict[str, Any]
    # Set containing all intermediate container dictionary paths
    branch_set: Set[str]


def flatten_key_paths(config: Dict[str, Any], max_depth: int = 32) -> FlattenedConfig:
    """
    Recursively flatten a nested configuration dictionary into distinct leaf mappings and branch container paths.

    This function performs a depth-first traversal of nested dictionaries, generating dot-delimited key paths
    (e.g., 'network_training_parameters.optimizer'). It cleanly partitions output paths into terminal primitive
    leaves ('leaf_map') and intermediate container dictionary nodes ('branch_set') to prevent intermediate dictionary
    containers from falsely satisfying scalar type audits.

    Parameters
    ----------
    config : Dict[str, Any]
        The nested configuration dictionary to flatten.
    max_depth : int, optional
        Maximum permitted recursion depth to protect against circular references, by default 32.

    Returns
    -------
    FlattenedConfig
        A named tuple containing 'leaf_map' (path -> primitive value) and 'branch_set' (set of container paths).

    Raises
    ------
    TypeError
        If 'config' is not a dictionary.
    RecursionError
        If nested dictionary depth exceeds 'max_depth'.
    ValueError
        If any dictionary key is empty or contains an invalid dot delimiter.
    """
    # Validate that the root input is a valid dictionary
    if not isinstance(config, dict):
        # Raise TypeError if the input is not a dictionary
        raise TypeError(f"flatten_key_paths requires a Dict[str, Any], received {type(config).__name__}.")

    # Initialize the dictionary to store terminal dotted key paths and their leaf values
    leaf_map: Dict[str, Any] = {}
    # Initialize the set to store all intermediate container dictionary paths
    branch_set: Set[str] = set()

    def _traverse(sub_dict: Dict[str, Any], prefix: str, current_depth: int) -> None:
        """
        Internal recursive traversal worker.
        """
        # Guard against unbounded recursion or circular references
        if current_depth > max_depth:
            # Raise RecursionError if depth threshold is exceeded
            raise RecursionError(f"Maximum dictionary recursion depth of {max_depth} exceeded during flattening.")

        # Iterate over each key-value pair in the current dictionary level
        for key, value in sub_dict.items():
            # Validate that the dictionary key is a string
            if not isinstance(key, str):
                # Convert non-string keys to string representation safely
                key_str = str(key)
            else:
                # Retain string key
                key_str = key

            # Validate that the key name is not empty
            if not key_str.strip():
                # Raise ValueError on empty dictionary keys
                raise ValueError(f"Empty dictionary key encountered at prefix '{prefix}'.")

            # Construct the cumulative dot-delimited key path
            current_path: str = f"{prefix}.{key_str}" if prefix else key_str

            # Check if the current value is a nested container dictionary
            if isinstance(value, dict):
                # Record the container path in the branch container set
                branch_set.add(current_path)
                # Recurse into the child dictionary with incremented depth
                _traverse(value, current_path, current_depth + 1)
            else:
                # Record the terminal leaf path and its corresponding value in the leaf map
                leaf_map[current_path] = value

    # Execute recursive traversal starting from the root level
    _traverse(config, prefix="", current_depth=1)

    # Return the structured FlattenedConfig container
    return FlattenedConfig(leaf_map=leaf_map, branch_set=branch_set)


def validate_required_key_paths(
    flattened: FlattenedConfig,
    required_keys: Tuple[str, ...],
) -> List[ValidationFinding]:
    """
    Verify that every manuscript-mandated key path exists and contains a populated, non-null value.

    This function audits the flattened configuration against the frozen registry of required paths.
    It verifies both path existence (checking leaf mappings and branch container sets) and value validity,
    ensuring that mandatory parameters are not set to None, empty strings, or unpopulated collections.

    Parameters
    ----------
    flattened : FlattenedConfig
        The flattened configuration containing 'leaf_map' and 'branch_set'.
    required_keys : Tuple[str, ...]
        Frozen tuple of required dot-delimited key paths derived from the manuscript sections.

    Returns
    -------
    List[ValidationFinding]
        List of structured diagnostic findings documenting any missing or unpopulated required parameters.

    Raises
    ------
    TypeError
        If 'flattened' is not a FlattenedConfig instance or 'required_keys' is not a tuple.
    """
    # Validate that flattened is an instance of FlattenedConfig
    if not isinstance(flattened, FlattenedConfig):
        # Raise TypeError if flattened argument is invalid
        raise TypeError(f"Expected FlattenedConfig instance, received {type(flattened).__name__}.")
    # Validate that required_keys is a tuple of strings
    if not isinstance(required_keys, tuple):
        # Raise TypeError if required_keys is not a tuple
        raise TypeError(f"Expected required_keys to be a Tuple[str, ...], received {type(required_keys).__name__}.")

    # Initialize the list of accumulated findings
    findings: List[ValidationFinding] = []

    # Iterate sequentially over every manuscript-required key path
    for path in required_keys:
        # Check if the path exists in the leaf mapping
        if path in flattened.leaf_map:
            # Retrieve the populated leaf value
            val: Any = flattened.leaf_map[path]
            # Verify that the value is not None
            if val is None:
                # Emit an error finding for None values in mandatory keys
                findings.append(
                    ValidationFinding(
                        severity="error",
                        key_path=path,
                        message=(
                            f"Mandatory manuscript parameter '{path}' is set to None. "
                            f"A populated, non-null value is required."
                        ),
                    )
                )
            # Verify that string values are not empty or purely whitespace
            elif isinstance(val, str) and not val.strip():
                # Emit an error finding for empty string parameters
                findings.append(
                    ValidationFinding(
                        severity="error",
                        key_path=path,
                        message=(
                            f"Mandatory manuscript parameter '{path}' is an empty string. "
                            f"A valid, non-empty configuration value is required."
                        ),
                    )
                )
            # Verify that list/tuple collection parameters are not empty
            elif isinstance(val, (list, tuple)) and len(val) == 0:
                # Emit an error finding for unpopulated collection parameters
                findings.append(
                    ValidationFinding(
                        severity="error",
                        key_path=path,
                        message=(
                            f"Mandatory manuscript sequence parameter '{path}' is empty. "
                            f"At least one valid element is required."
                        ),
                    )
                )
        # Check if the path represents a valid intermediate container dictionary
        elif path in flattened.branch_set:
            # The container path is structurally present and valid
            pass
        else:
            # Emit a blocking error finding if the required path is missing entirely
            findings.append(
                ValidationFinding(
                    severity="error",
                    key_path=path,
                    message=(
                        f"Manuscript-required parameter '{path}' is missing from the configuration. "
                        f"Add the parameter with its documented specification before execution."
                    ),
                )
            )

    # Return the list of identified key path findings
    return findings


# Frozen registry of parameters marked UNSPECIFIED IN MANUSCRIPT, each
# mapped to the decision-log topic that must document its resolution.
# The registry enumerates EVERY parameter marked UNSPECIFIED IN MANUSCRIPT
# in config v2.0.0 so that no silent inference remains undocumented. Topic
# names are chosen to be honest about the semantics of the parameter and
# must appear (case-insensitively) in the configured decision-log entries.
UNSPECIFIED_PARAMETER_TOPICS: Tuple[Tuple[str, str], ...] = (
    # --- Section 2: data acquisition and universe -----------------------
    ("data_acquisition_and_universe.download_protocol.api",
     "data source and API"),
    ("data_acquisition_and_universe.download_protocol.adjustment",
     "data source and API"),
    ("data_acquisition_and_universe.download_protocol."
     "delisted_ticker_handling", "data source and API"),
    ("data_acquisition_and_universe.download_protocol."
     "share_class_handling", "universe fallback rules"),
    ("data_acquisition_and_universe.download_protocol.retry_count",
     "data source and API"),
    ("data_acquisition_and_universe.download_protocol."
     "download_chunk_size_months", "data source and API"),
    ("data_acquisition_and_universe.us_inclusion_criteria.sp1500_asof_date",
     "reference dates"),
    ("data_acquisition_and_universe.us_inclusion_criteria."
     "gics_sector_reference_date", "reference dates"),
    ("data_acquisition_and_universe.us_inclusion_criteria."
     "market_cap_reference_date", "reference dates"),
    ("data_acquisition_and_universe.jp_inclusion_criteria."
     "tse_prime_asof_date", "reference dates"),
    ("data_acquisition_and_universe.universe_fallback_rules."
     "if_filtered_count_exceeds_target", "universe fallback rules"),
    ("data_acquisition_and_universe.universe_fallback_rules."
     "if_filtered_count_below_target", "universe fallback rules"),
    # --- Section 4: data preprocessing -----------------------------------
    ("data_preprocessing_parameters.calendar_source", "trading calendar"),
    ("data_preprocessing_parameters.beta_min_observations",
     "rolling-window missing-data rules"),
    ("data_preprocessing_parameters.rsi_smoothing", "RSI smoothing"),
    ("data_preprocessing_parameters.rsi_zero_loss_handling",
     "RSI smoothing"),
    ("data_preprocessing_parameters.rolling_window_requirement",
     "rolling-window missing-data rules"),
    # --- Section 6: regime state variables -------------------------------
    ("regime_state_variable_construction.beta_min_observations",
     "rolling-window missing-data rules"),
    ("regime_state_variable_construction."
     "idio_vol_window_min_observations",
     "rolling-window missing-data rules"),
    # --- Section 9: shared block -----------------------------------------
    ("shared_neural_network_block.parameter_count_formula",
     "capacity accounting"),
    # --- Section 10: network architecture --------------------------------
    ("network_architecture_parameters.RG_ResMoE_K2.expert_width",
     "K2/K6 expert widths"),
    ("network_architecture_parameters.RG_ResMoE_K6.expert_width",
     "K2/K6 expert widths"),
    ("network_architecture_parameters.gate_architecture.hidden_layers",
     "gate architecture"),
    ("network_architecture_parameters.gate_architecture.hidden_width",
     "gate architecture"),
    ("network_architecture_parameters.gate_architecture.activation",
     "gate architecture"),
    ("network_architecture_parameters.gate_architecture.dropout_rate",
     "gate architecture"),
    ("network_architecture_parameters.gate_architecture.output",
     "gate architecture"),
    ("network_architecture_parameters.hard_routing_volatility_quantiles."
     "quantile_variable", "hard-routing quantile variable and boundaries"),
    ("network_architecture_parameters.hard_routing_volatility_quantiles."
     "bin_count", "hard-routing quantile variable and boundaries"),
    ("network_architecture_parameters.hard_routing_volatility_quantiles."
     "boundary_source", "hard-routing quantile variable and boundaries"),
    ("network_architecture_parameters.hard_routing_volatility_quantiles."
     "binning_method", "hard-routing quantile variable and boundaries"),
    ("network_architecture_parameters.hard_routing_gics_sectors.mapping",
     "GICS-to-expert mapping"),
    ("network_architecture_parameters.hard_routing_gics_sectors."
     "mapping_rationale", "GICS-to-expert mapping"),
    ("network_architecture_parameters.hard_routing_market_x_idio_split."
     "grid", "hard-routing quantile variable and boundaries"),
    ("network_architecture_parameters.hard_routing_market_x_idio_split."
     "market_split", "hard-routing quantile variable and boundaries"),
    ("network_architecture_parameters.hard_routing_market_x_idio_split."
     "idio_split", "hard-routing quantile variable and boundaries"),
    ("network_architecture_parameters.top1_gate_training.gradient_source",
     "top-1 gate training scheme"),
    ("network_architecture_parameters.top1_gate_training."
     "inference_selection", "top-1 gate training scheme"),
    # --- Section 11: network training ------------------------------------
    ("network_training_parameters.adam_beta1", "optimizer hyperparameters"),
    ("network_training_parameters.adam_beta2", "optimizer hyperparameters"),
    ("network_training_parameters.adam_epsilon",
     "optimizer hyperparameters"),
    ("network_training_parameters.learning_rate",
     "optimizer hyperparameters"),
    ("network_training_parameters.learning_rate_candidate_grid",
     "optimizer hyperparameters"),
    ("network_training_parameters.max_epochs",
     "optimizer hyperparameters"),
    ("network_training_parameters.gradient_clip_norm",
     "optimizer hyperparameters"),
    ("network_training_parameters.sweep_seeds", "seed registry"),
    ("network_training_parameters.final_seeds", "seed registry"),
    ("network_training_parameters.early_stopping_patience",
     "early-stopping patience"),
    # --- Section 12: loss and regularization -----------------------------
    ("loss_function_and_regularization.alpha_residual_shrinkage",
     "regularization coefficient values"),
    ("loss_function_and_regularization.alpha_candidate_grid",
     "regularization coefficient values"),
    ("loss_function_and_regularization.lambda_LB_load_balancing",
     "regularization coefficient values"),
    ("loss_function_and_regularization.lambda_LB_candidate_grid",
     "regularization coefficient values"),
    ("loss_function_and_regularization.numerical_hygiene.variance_floor",
     "numerical hygiene"),
    ("loss_function_and_regularization.numerical_hygiene.qlike_max_value",
     "numerical hygiene"),
    ("loss_function_and_regularization.numerical_hygiene."
     "softmax_temperature", "numerical hygiene"),
    # --- Section 13: hyperparameter selection ----------------------------
    ("hyperparameter_selection.if_no_config_passes_all_filters",
     "hyperparameter selection fallback"),
    ("hyperparameter_selection.sweep_seed_count", "seed registry"),
    ("hyperparameter_selection.final_seed_count", "seed registry"),
    # --- Section 14: classical baselines ---------------------------------
    ("classical_baselines.GARCH_1_1.innovation_distribution",
     "GARCH innovation distribution"),
    ("classical_baselines.GARCH_1_1.forecast_horizon",
     "GARCH innovation distribution"),
    ("classical_baselines.GARCH_1_1.forecast_aggregation",
     "GARCH innovation distribution"),
    ("classical_baselines.HAR.windows", "HAR monthly lag"),
    ("classical_baselines.Ridge.alpha",
     "regularization coefficient values"),
    ("classical_baselines.Ridge.alpha_candidate_grid",
     "regularization coefficient values"),
    ("classical_baselines.Ridge.solver", "optimizer hyperparameters"),
    ("classical_baselines.Ridge_plus_z.alpha",
     "regularization coefficient values"),
    ("classical_baselines.Ridge_plus_z.alpha_candidate_grid",
     "regularization coefficient values"),
    # --- Section 15: evaluation metrics -----------------------------------
    ("evaluation_metrics.IC.missing_handling", "missing-data handling"),
    # --- Section 16: value-at-risk calibration ----------------------------
    ("value_at_risk_calibration.student_t_fit_window",
     "Student-t fit procedure"),
    ("value_at_risk_calibration.student_t_fit_method",
     "Student-t fit procedure"),
    ("value_at_risk_calibration.student_t_residual_definition",
     "Student-t fit procedure"),
    ("value_at_risk_calibration.student_t_parameters",
     "Student-t fit procedure"),
    ("value_at_risk_calibration.daily_vol_conversion",
     "VaR conventions"),
    ("value_at_risk_calibration.var_formula", "VaR conventions"),
    ("value_at_risk_calibration.slices.high_volatility",
     "high/low volatility slice thresholds"),
    ("value_at_risk_calibration.slices.low_volatility",
     "high/low volatility slice thresholds"),
    # --- Section 17: statistical testing ----------------------------------
    ("statistical_testing.finite_sample_distribution",
     "finite-sample distribution"),
    # --- Section 20: stress-condition analysis ----------------------------
    ("stress_condition_analysis.top_market_vol_decile.definition",
     "stress-window date ranges"),
    ("stress_condition_analysis.covid_crisis_window",
     "stress-window date ranges"),
    ("stress_condition_analysis.bear_market_2022",
     "stress-window date ranges"),
    ("stress_condition_analysis.regime_flip_windows.definition",
     "stress-window date ranges"),
    ("stress_condition_analysis.regime_flip_windows.window_radius_days",
     "stress-window date ranges"),
)


def _normalize_canonical_token(text: str) -> str:
    """
    Normalize descriptive topic strings into standardized lowercase snake_case tokens.

    Parameters
    ----------
    text : str
        The raw descriptive text string.

    Returns
    -------
    str
        Canonical snake_case token stripped of special characters.
    """
    # Convert text to lowercase and replace non-alphanumeric characters with underscores
    cleaned: str = re.sub(r"[^a-z0-9]+", "_", text.lower()).strip("_")
    # Return the cleaned canonical token
    return cleaned


def validate_decision_log_coverage(
    flattened: FlattenedConfig,
    config: Dict[str, Any],
    unspecified_topics: Tuple[Tuple[str, str], ...],
) -> List[ValidationFinding]:
    """
    Verify that every parameter marked UNSPECIFIED IN MANUSCRIPT has an exact canonical decision-log entry.

    This function audits the configuration to ensure complete transparency and eliminate undocumented assumptions.
    Instead of heuristic substring searching, it normalizes topic requirements and configured decision-log entries
    into canonical snake_case tokens, performing exact token set matching. If an unspecified parameter is present
    in the configuration but its corresponding topic is absent from the decision log, an error finding is emitted.

    Parameters
    ----------
    flattened : FlattenedConfig
        The flattened configuration containing active leaf paths.
    config : Dict[str, Any]
        The root configuration dictionary.
    unspecified_topics : Tuple[Tuple[str, str], ...]
        Registry mapping dot-delimited unspecified parameter paths to their required decision-log topic names.

    Returns
    -------
    List[ValidationFinding]
        List of structured error findings for any undocumented unspecified parameters.

    Raises
    ------
    TypeError
        If input structures do not match expected types.
    """
    # Validate that flattened is a FlattenedConfig instance
    if not isinstance(flattened, FlattenedConfig):
        # Raise TypeError on invalid flattened config argument
        raise TypeError(f"Expected FlattenedConfig instance, received {type(flattened).__name__}.")
    # Validate that config is a dictionary
    if not isinstance(config, dict):
        # Raise TypeError on invalid config argument
        raise TypeError(f"Expected config to be a Dict[str, Any], received {type(config).__name__}.")

    # Initialize the list of accumulated findings
    findings: List[ValidationFinding] = []

    # Safely extract the configured decision-log required entries list
    raw_entries: List[str] = (
        config.get("reproducibility_and_audit", {})
        .get("decision_log", {})
        .get("required_entries", [])
    )

    # Check if the decision-log required_entries list is present and non-empty
    if not raw_entries or not isinstance(raw_entries, list):
        # Emit a blocking error finding if the entire decision-log registry is unpopulated
        findings.append(
            ValidationFinding(
                severity="error",
                key_path="reproducibility_and_audit.decision_log.required_entries",
                message=(
                    "The 'decision_log.required_entries' list is missing or empty. "
                    "All unspecified manuscript parameters must be registered."
                ),
            )
        )
        # Return early since no topics can be matched against an empty registry
        return findings

    # Normalize all configured decision-log entries into a canonical token set
    configured_tokens: Set[str] = {_normalize_canonical_token(entry) for entry in raw_entries if isinstance(entry, str)}

    # Iterate over every registered unspecified parameter path and its required documentation topic
    for param_path, topic_name in unspecified_topics:
        # Check if the unspecified parameter is active in the flattened configuration
        if param_path in flattened.leaf_map or param_path in flattened.branch_set:
            # Generate the canonical token for the required topic
            required_token: str = _normalize_canonical_token(topic_name)
            # Perform exact canonical token matching against configured decision log topics
            if required_token not in configured_tokens:
                # Emit a blocking error finding documenting the unrecorded decision-log entry
                findings.append(
                    ValidationFinding(
                        severity="error",
                        key_path=param_path,
                        message=(
                            f"Parameter '{param_path}' is marked UNSPECIFIED IN MANUSCRIPT, but its required "
                            f"canonical decision-log topic '{topic_name}' (token: '{required_token}') is missing "
                            f"from 'reproducibility_and_audit.decision_log.required_entries'."
                        ),
                    )
                )

    # Return the accumulated decision-log coverage findings
    return findings


def audit_config_completeness(
    config: Dict[str, Any],
    flattened: Optional[FlattenedConfig] = None,
) -> List[ValidationFinding]:
    """
    Coordinate the structural completeness audit of the configuration dictionary.

    This function audits the configuration against the manuscript's required parameter specifications:
    1. Flattens the nested dictionary into partitioned leaf mappings and container branch sets if not pre-computed.
    2. Verifies the presence and non-nullity of all mandatory manuscript key paths via 'validate_required_key_paths'.
    3. Verifies that every active parameter marked UNSPECIFIED IN MANUSCRIPT is covered by an exact canonical topic
       in the decision log registry via 'validate_decision_log_coverage'.

    Parameters
    ----------
    config : Dict[str, Any]
        The root study configuration dictionary to audit.
    flattened : Optional[FlattenedConfig], optional
        Pre-computed flattened configuration containing 'leaf_map' and 'branch_set'. If None, flattening is executed.

    Returns
    -------
    List[ValidationFinding]
        Accumulated diagnostic findings documenting any missing parameters, null values, or decision-log coverage gaps.

    Raises
    ------
    TypeError
        If 'config' is not a dictionary or 'flattened' is not a FlattenedConfig instance when provided.
    """
    # Validate that the config argument is a dictionary
    if not isinstance(config, dict):
        # Raise TypeError if config is not a dictionary
        raise TypeError(f"audit_config_completeness requires a Dict[str, Any], received {type(config).__name__}.")

    # Flatten the configuration dictionary if no pre-computed structure was supplied
    if flattened is None:
        # Generate the partitioned FlattenedConfig structure
        active_flattened: FlattenedConfig = flatten_key_paths(config)
    else:
        # Validate that the supplied flattened object is an instance of FlattenedConfig
        if not isinstance(flattened, FlattenedConfig):
            # Raise TypeError if an invalid flattened object was passed
            raise TypeError(f"Expected FlattenedConfig instance, received {type(flattened).__name__}.")
        # Use the supplied pre-computed flattened structure
        active_flattened = flattened

    # Initialize the findings accumulator list for structural completeness
    findings: List[ValidationFinding] = []

    # Step 1.4: Execute key path existence, non-nullity, and non-emptiness validation
    findings.extend(validate_required_key_paths(active_flattened, REQUIRED_MANUSCRIPT_KEYS))

    # Step 1.5: Execute canonical decision-log coverage audit for all unspecified parameters
    findings.extend(validate_decision_log_coverage(active_flattened, config, UNSPECIFIED_PARAMETER_TOPICS))

    # Return the accumulated structural completeness findings
    return findings


# -----------------------------------------------------------------------
# Task 1, Step 2: Validate Numerical, Logical, and Dimensional Constraints
# -----------------------------------------------------------------------


def validate_date_range(
    sample_start: str,
    sample_end: str,
    config: Dict[str, Any],
) -> List[Any]:
    """
    Validate the sample date range syntax, chronological ordering, and walk-forward duration sufficiency.

    This function asserts that 'sample_start' and 'sample_end' parse strictly with format '%Y-%m-%d' and that
    sample_start is chronologically earlier than sample_end. Furthermore, it computes the theoretical minimum
    business-day span required to execute 30 non-overlapping walk-forward windows:
        Min Trading Days = D_dev + (W - 1) * S_step + T_test = 504 + (30 - 1) * 63 + 63 = 2,394 sessions
    and emits an error finding if the historical price duration is insufficient.

    Parameters
    ----------
    sample_start : str
        Sample start date string formatted as YYYY-MM-DD.
    sample_end : str
        Sample end date string formatted as YYYY-MM-DD.
    config : Dict[str, Any]
        The root configuration dictionary containing walk-forward evaluation geometry.

    Returns
    -------
    List[ValidationFinding]
        Accumulated diagnostic findings documenting date syntax, ordering, or duration violations.

    Raises
    ------
    TypeError
        If 'sample_start' or 'sample_end' are not strings, or 'config' is not a dictionary.
    """
    # Import ValidationFinding dynamically to ensure exact structural binding
    from typing import cast

    # Validate that sample_start is a string
    if not isinstance(sample_start, str):
        # Raise TypeError if sample_start is not a string
        raise TypeError(f"sample_start must be a string, received {type(sample_start).__name__}.")
    # Validate that sample_end is a string
    if not isinstance(sample_end, str):
        # Raise TypeError if sample_end is not a string
        raise TypeError(f"sample_end must be a string, received {type(sample_end).__name__}.")
    # Validate that config is a dictionary
    if not isinstance(config, dict):
        # Raise TypeError if config is not a dictionary
        raise TypeError(f"config must be a Dict[str, Any], received {type(config).__name__}.")

    # Initialize the findings accumulator list
    findings: List[Any] = []

    # Initialize parsed date variables
    start_dt: Optional[datetime] = None
    # Initialize parsed end date variable
    end_dt: Optional[datetime] = None

    # Parse sample_start with strict format enforcement
    try:
        # Parse start date using YYYY-MM-DD format
        start_dt = datetime.strptime(sample_start.strip(), "%Y-%m-%d")
    except (ValueError, TypeError) as exc:
        # Emit blocking error finding for unparseable start date
        findings.append(
            ValidationFinding(
                severity="error",
                key_path="data_acquisition_and_universe.sample_start",
                message=f"sample_start '{sample_start}' fails strict format parsing '%Y-%m-%d': {exc}",
            )
        )

    # Parse sample_end with strict format enforcement
    try:
        # Parse end date using YYYY-MM-DD format
        end_dt = datetime.strptime(sample_end.strip(), "%Y-%m-%d")
    except (ValueError, TypeError) as exc:
        # Emit blocking error finding for unparseable end date
        findings.append(
            ValidationFinding(
                severity="error",
                key_path="data_acquisition_and_universe.sample_end",
                message=f"sample_end '{sample_end}' fails strict format parsing '%Y-%m-%d': {exc}",
            )
        )

    # Perform chronological and span assertions only if both dates parsed successfully
    if start_dt is not None and end_dt is not None:
        # Assert strict chronological ordering: start_dt must precede end_dt
        if not (start_dt < end_dt):
            # Emit blocking error finding for inverted or identical dates
            findings.append(
                ValidationFinding(
                    severity="error",
                    key_path="data_acquisition_and_universe.sample_start",
                    message=(
                        f"Chronological ordering violation: sample_start ({sample_start}) must be strictly earlier "
                        f"than sample_end ({sample_end})."
                    ),
                )
            )
        else:
            # Extract walk-forward parameters from config with safe Table 1 defaults
            wf_cfg: Dict[str, Any] = config.get("walkforward_evaluation", {})
            # Extract development length in trading days
            d_dev: int = int(wf_cfg.get("development_length_trading_days", 504))
            # Extract total number of walk-forward windows
            n_windows: int = int(wf_cfg.get("number_of_windows", 30))
            # Extract step length in trading days
            s_step: int = int(wf_cfg.get("step_length_trading_days", 63))
            # Extract test length in trading days
            t_test: int = int(wf_cfg.get("test_length_trading_days", 63))

            # Compute theoretical minimum required trading days: Min Days = D_dev + (W - 1) * S_step + T_test
            min_required_days: int = d_dev + (n_windows - 1) * s_step + t_test

            # Compute actual business days in the sample interval
            actual_business_days: int = int(len(pd.bdate_range(start=start_dt, end=end_dt)))

            # Assert that the actual business-day span is sufficient to cover the walk-forward geometry
            if actual_business_days < min_required_days:
                # Emit blocking error finding documenting the insufficient calendar duration
                findings.append(
                    ValidationFinding(
                        severity="error",
                        key_path="data_acquisition_and_universe.sample_start",
                        message=(
                            f"Sample period {sample_start} to {sample_end} contains {actual_business_days} business days, "
                            f"which is less than the minimum {min_required_days} trading days required to execute "
                            f"{n_windows} walk-forward windows (development: {d_dev}d, step: {s_step}d, test: {t_test}d)."
                        ),
                    )
                )

    # Return accumulated date validation findings
    return findings


def validate_walkforward_geometry(
    config: Dict[str, Any],
) -> List[ValidationFinding]:
    """
    Validate the walk-forward arithmetic identities of the study.

    Parameters
    ----------
    config : Dict[str, Any]
        The configuration dictionary to audit.

    Returns
    -------
    List[ValidationFinding]
        Findings for any violated walk-forward identity.
    """
    # Initialize the findings list for the geometry validation
    findings: List[ValidationFinding] = []
    # Extract the walk-forward section for direct key access
    wf: Dict[str, Any] = config.get("walkforward_evaluation", {})
    # Compute the training length from development length and fraction
    computed_train: int = int(
        wf.get("development_length_trading_days", 0)
        * wf.get("train_fraction", 0.0)
    )
    # Compare the computed training length with the stored field
    if computed_train != wf.get("train_length_trading_days"):
        # Record a blocking error for the training-length identity
        findings.append(
            ValidationFinding(
                severity="error",
                key_path=(
                    "walkforward_evaluation.train_length_trading_days"
                ),
                message=(
                    f"Walk-forward identity violated: int(504 x 0.85) = "
                    f"{computed_train} but the config stores "
                    f"{wf.get('train_length_trading_days')}."
                ),
            )
        )
    # Compute the validation length as the residual of the development split
    computed_val: int = (
        int(wf.get("development_length_trading_days", 0)) - computed_train
    )
    # Compare the computed validation length with the stored field
    if computed_val != wf.get("validation_length_trading_days"):
        # Record a blocking error for the validation-length identity
        findings.append(
            ValidationFinding(
                severity="error",
                key_path=(
                    "walkforward_evaluation.validation_length_trading_days"
                ),
                message=(
                    f"Walk-forward identity violated: 504 - 428 = "
                    f"{computed_val} but the config stores "
                    f"{wf.get('validation_length_trading_days')}."
                ),
            )
        )
    # Compute the expected total test days from the window arithmetic
    computed_total: int = (
        int(wf.get("number_of_windows", 0))
        * int(wf.get("test_length_trading_days", 0))
    )
    # Compare the computed total test days with the expected count
    if computed_total != 1890:
        # Record a blocking error for the total-test-days identity
        findings.append(
            ValidationFinding(
                severity="error",
                key_path="walkforward_evaluation.number_of_windows",
                message=(
                    f"Walk-forward identity violated: 30 x 63 = "
                    f"{computed_total} but the manuscript expects 1890."
                ),
            )
        )
    # Return the complete findings list for the geometry validation
    return findings


def compute_block_parameter_count(d_in: int, hidden_width: int) -> int:
    """
    Compute the parameter count of a two-hidden-layer MLP block.

    Implements the manuscript capacity formula:
        params(H, d_in) = (d_in + 1) H + (H + 1) H + (H + 1)

    Parameters
    ----------
    d_in : int
        The input dimension of the block.
    hidden_width : int
        The hidden width H of both hidden layers.

    Returns
    -------
    int
        The exact total parameter count of the block.
    """
    # First hidden layer: weights (d_in x H) plus bias (H)
    first_layer: int = (d_in + 1) * hidden_width
    # Second hidden layer: weights (H x H) plus bias (H)
    second_layer: int = (hidden_width + 1) * hidden_width
    # Output layer: weights (H x 1) plus bias (1)
    output_layer: int = hidden_width + 1
    # Return the sum of the three layer parameter counts
    return first_layer + second_layer + output_layer


def validate_parameter_counts(
    config: Dict[str, Any],
) -> List[ValidationFinding]:
    """
    Recompute every model parameter count from widths and input dims.

    Parameters
    ----------
    config : Dict[str, Any]
        The configuration dictionary to audit.

    Returns
    -------
    List[ValidationFinding]
        Findings for any stored count that disagrees with the formula.
    """
    # Initialize the findings list for the capacity audit
    findings: List[ValidationFinding] = []
    # Extract the network architecture section for direct access
    arch: Dict[str, Any] = config.get("network_architecture_parameters", {})
    # Extract the feature dimension for input sizing
    feature_dim: int = (
        config.get("stock_feature_construction", {})
        .get("feature_dimension", 16)
    )
    # Extract the regime dimension for input sizing
    regime_dim: int = (
        config.get("regime_state_variable_construction", {})
        .get("regime_dimension", 2)
    )
    # Define the (input_dim, width) pairs for the four MLP variants
    variants: Tuple[Tuple[str, int, int], ...] = (
        ("MLP_S", feature_dim, arch.get("MLP_S", {}).get("width", 16)),
        ("MLP_L", feature_dim, arch.get("MLP_L", {}).get("width", 44)),
        (
            "MLP_S_plus_z",
            feature_dim + regime_dim,
            arch.get("MLP_S_plus_z", {}).get("width", 16),
        ),
        (
            "MLP_L_plus_z",
            feature_dim + regime_dim,
            arch.get("MLP_L_plus_z", {}).get("width", 44),
        ),
    )
    # Iterate over every MLP variant with its input and width
    for model_name, d_in, width in variants:
        # Compute the exact parameter count with the closed-form formula
        computed: int = compute_block_parameter_count(d_in, width)
        # Fetch the stored parameter count of the variant
        stored: int = arch.get(model_name, {}).get("parameter_count", -1)
        # Compare the recomputed count with the stored count
        if computed != stored:
            # Record a BLOCKING error for the capacity discrepancy; the
            # closed-form formula is authoritative and a wrong stored
            # count must stop the pipeline, not merely warn
            findings.append(
                ValidationFinding(
                    severity="error",
                    key_path=(
                        f"network_architecture_parameters."
                        f"{model_name}.parameter_count"
                    ),
                    message=(
                        f"Recomputed parameter count for {model_name} with "
                        f"d_in={d_in} and H={width} is {computed} but the "
                        f"config stores {stored}. The closed-form formula "
                        f"is authoritative; correct the stored value to "
                        f"{computed}."
                    ),
                )
            )
    # Extract the expert count and expert width from the standard MoE
    moe_experts: int = (
        arch.get("standard_MoE", {}).get("expert_count", 4)
    )
    # Extract the MoE expert width from the standard MoE configuration
    moe_expert_width: int = int(
        arch.get("standard_MoE", {}).get("expert_width", 16)
    )
    # Compute the MoE expert total from the CONFIGURED block count
    moe_total: int = (
        moe_experts
        * compute_block_parameter_count(feature_dim, moe_expert_width)
    )
    # Compare the MoE expert total with the manuscript expectation
    if moe_total != 2244:
        # Record a BLOCKING error for the MoE capacity total
        findings.append(
            ValidationFinding(
                severity="error",
                key_path=(
                    "network_architecture_parameters.standard_MoE."
                    "expert_count"
                ),
                message=(
                    f"Standard MoE expert capacity computes to {moe_total} "
                    f"but the manuscript expects 4 x 561 = 2244."
                ),
            )
        )
    # Extract the RG-ResMoE configuration for the capacity computation
    rg: Dict[str, Any] = arch.get("RG_ResMoE", {})
    # Extract the RG expert count and base/expert widths from the config
    rg_experts: int = int(rg.get("expert_count", 4))
    # Extract the base network width from the RG-ResMoE configuration
    rg_base_width: int = int(rg.get("base_network_width", 16))
    # Extract the expert width from the RG-ResMoE configuration
    rg_expert_width: int = int(rg.get("expert_width", 16))
    # Compute the RG-ResMoE base-plus-experts total capacity
    rg_total: int = (
        compute_block_parameter_count(feature_dim, rg_base_width)
        + rg_experts
        * compute_block_parameter_count(feature_dim, rg_expert_width)
    )
    # Compare the RG-ResMoE total with the manuscript expectation
    if rg_total != 2805:
        # Record a BLOCKING error for the RG-ResMoE capacity total
        findings.append(
            ValidationFinding(
                severity="error",
                key_path="network_architecture_parameters.RG_ResMoE",
                message=(
                    f"RG-ResMoE base-plus-experts capacity computes to "
                    f"{rg_total} but the manuscript expects 561 + 4 x 561 "
                    f"= 2805."
                ),
            )
        )
    # Verify the K2 variant expert capacity (2 experts, width 25)
    k2_config: Dict[str, Any] = arch.get("RG_ResMoE_K2", {})
    # Extract the K2 expert count from the configuration
    k2_experts: int = int(k2_config.get("expert_count", 2))
    # Extract the K2 expert width from the configuration
    k2_width: int = int(k2_config.get("expert_width", 25))
    # Compute the K2 total expert capacity from the closed-form formula
    k2_total: int = (
        k2_experts
        * compute_block_parameter_count(feature_dim, k2_width)
    )
    # Compare the K2 total with the documented capacity (2202)
    if k2_total != 2202:
        # Record a BLOCKING error for the K2 capacity total
        findings.append(
            ValidationFinding(
                severity="error",
                key_path=(
                    "network_architecture_parameters.RG_ResMoE_K2."
                    "expert_width"
                ),
                message=(
                    f"RG-ResMoE-K2 expert capacity computes to {k2_total} "
                    f"but the documented capacity is 2202 (2 x 1101)."
                ),
            )
        )
    # Verify the K6 variant expert capacity (6 experts, width 12)
    k6_config: Dict[str, Any] = arch.get("RG_ResMoE_K6", {})
    # Extract the K6 expert count from the configuration
    k6_experts: int = int(k6_config.get("expert_count", 6))
    # Extract the K6 expert width from the configuration
    k6_width: int = int(k6_config.get("expert_width", 12))
    # Compute the K6 total expert capacity from the closed-form formula
    k6_total: int = (
        k6_experts
        * compute_block_parameter_count(feature_dim, k6_width)
    )
    # Compare the K6 total with the documented capacity (2238)
    if k6_total != 2238:
        # Record a BLOCKING error for the K6 capacity total
        findings.append(
            ValidationFinding(
                severity="error",
                key_path=(
                    "network_architecture_parameters.RG_ResMoE_K6."
                    "expert_width"
                ),
                message=(
                    f"RG-ResMoE-K6 expert capacity computes to {k6_total} "
                    f"but the documented capacity is 2238 (6 x 373)."
                ),
            )
        )
    # Return the complete findings list for the capacity audit
    return findings


class ParameterDomainSpec(NamedTuple):
    """
    Specification contract defining mathematical domain boundaries and integrality constraints for a hyperparameter.

    Attributes
    ----------
    key_path : str
        Dot-delimited key path of the parameter.
    min_val : float
        Lower bound of the mathematical interval.
    max_val : float
        Upper bound of the mathematical interval.
    min_inclusive : bool
        True if the lower bound is inclusive (>=), False if open (>).
    max_inclusive : bool
        True if the upper bound is inclusive (<=), False if open (<).
    must_be_integer : bool
        True if the parameter represents a discrete count count-type variable.
    """

    # Dot-delimited configuration key path
    key_path: str
    # Lower boundary value
    min_val: float
    # Upper boundary value
    max_val: float
    # Inclusivity flag for lower bound
    min_inclusive: bool
    # Inclusivity flag for upper bound
    max_inclusive: bool
    # Integrality assertion flag
    must_be_integer: bool


# Define the frozen registry of hyperparameter domain specifications
NUMERICAL_DOMAIN_REGISTRY: Tuple[ParameterDomainSpec, ...] = (
    # Learning rate: strictly positive open probability interval (0.0, 1.0)
    ParameterDomainSpec("network_training_parameters.learning_rate", 0.0, 1.0, False, False, False),
    # Adam beta1: open interval (0.0, 1.0)
    ParameterDomainSpec("network_training_parameters.adam_beta1", 0.0, 1.0, False, False, False),
    # Adam beta2: open interval (0.0, 1.0)
    ParameterDomainSpec("network_training_parameters.adam_beta2", 0.0, 1.0, False, False, False),
    # Adam epsilon: strictly positive open interval (0.0, 1e-3)
    ParameterDomainSpec("network_training_parameters.adam_epsilon", 0.0, 1e-3, False, False, False),
    # Max training epochs: positive integer count [1, 10000]
    ParameterDomainSpec("network_training_parameters.max_epochs", 1.0, 10000.0, True, True, True),
    # Early stopping patience: positive integer count [1, 500]
    ParameterDomainSpec("network_training_parameters.early_stopping_patience", 1.0, 500.0, True, True, True),
    # Gradient clipping norm: positive real number (0.0, 100.0]
    ParameterDomainSpec("network_training_parameters.gradient_clip_norm", 0.0, 100.0, False, True, False),
    # Shared MLP dropout rate: half-open interval [0.0, 1.0)
    ParameterDomainSpec("shared_neural_network_block.dropout_rate", 0.0, 1.0, True, False, False),
    # Residual shrinkage coefficient alpha: non-negative real [0.0, 1000.0]
    ParameterDomainSpec("loss_function_and_regularization.alpha_residual_shrinkage", 0.0, 1000.0, True, True, False),
    # Load-balancing coefficient lambda_LB: non-negative real [0.0, 1000.0]
    ParameterDomainSpec("loss_function_and_regularization.lambda_LB_load_balancing", 0.0, 1000.0, True, True, False),
    # QLIKE numerical variance floor: strictly positive small floor (0.0, 1e-2]
    ParameterDomainSpec("loss_function_and_regularization.numerical_hygiene.variance_floor", 0.0, 1e-2, False, True, False),
    # QLIKE collapse diagnostic threshold: strictly positive real (0.0, 1000.0]
    ParameterDomainSpec("evaluation_metrics.collapse_threshold_qlike", 0.0, 1000.0, False, True, False),
    # Newey-West lag count: positive integer count [1, 20]
    ParameterDomainSpec("statistical_testing.newey_west_lags", 1.0, 20.0, True, True, True),
    # Kupiec test significance level: open probability interval (0.0, 1.0)
    ParameterDomainSpec("value_at_risk_calibration.kupiec_test.significance_level", 0.0, 1.0, False, False, False),
    # Annualization factor: positive integer count [1, 365]
    ParameterDomainSpec("data_preprocessing_parameters.annualization_factor", 1.0, 365.0, True, True, True),
    # Beta minimum observations: positive integer count [10, 120]
    ParameterDomainSpec("data_preprocessing_parameters.beta_min_observations", 10.0, 120.0, True, True, True),
    # Target horizon days: positive integer count [1, 60]
    ParameterDomainSpec("target_construction.horizon_days", 1.0, 60.0, True, True, True),
)


def validate_numerical_finiteness(
    flattened: Any,
    domain_registry: Tuple[ParameterDomainSpec, ...] = NUMERICAL_DOMAIN_REGISTRY,
) -> List[Any]:
    """
    Verify that numerical hyperparameters are finite, fall within strict mathematical intervals, and respect integrality.

    This function audits the flattened configuration against the formal domain specification registry.
    It verifies that values are finite real numbers via math.isfinite, asserts upper and lower boundary conditions,
    and guarantees that discrete count parameters (e.g., lag counts, patience, epochs) are strict integers.

    Parameters
    ----------
    flattened : FlattenedConfig
        The flattened configuration containing 'leaf_map'.
    domain_registry : Tuple[ParameterDomainSpec, ...], optional
        Registry of domain boundary contracts, by default NUMERICAL_DOMAIN_REGISTRY.

    Returns
    -------
    List[ValidationFinding]
        Accumulated diagnostic findings documenting non-finite values, out-of-bounds parameters, or integrality violations.
    """
    # Initialize the findings accumulator list
    findings: List[Any] = []

    # Iterate sequentially over every domain specification in the registry
    for spec in domain_registry:
        # Check if the parameter path is present in the flattened leaf map
        if spec.key_path not in flattened.leaf_map:
            # Skip evaluation if the parameter is absent (existence is audited in Step 1.4)
            continue

        # Extract the raw parameter value from the leaf map
        raw_val: Any = flattened.leaf_map[spec.key_path]

        # Check if the raw value can be represented as a float
        if not isinstance(raw_val, (int, float)) or isinstance(raw_val, bool):
            # Emit an error finding for non-numeric types
            findings.append(
                ValidationFinding(
                    severity="error",
                    key_path=spec.key_path,
                    message=f"Numerical parameter '{spec.key_path}' must be a real number; received {type(raw_val).__name__} ({raw_val!r}).",
                )
            )
            # Skip subsequent boundary checks on non-numeric types
            continue

        # Cast value to standard float64 for mathematical assertions
        val_float: float = float(raw_val)

        # Assert mathematical finiteness (reject NaN, +inf, -inf)
        if not math.isfinite(val_float):
            # Emit an error finding for non-finite values
            findings.append(
                ValidationFinding(
                    severity="error",
                    key_path=spec.key_path,
                    message=f"Numerical parameter '{spec.key_path}' must be finite; received {val_float}.",
                )
            )
            # Skip boundary checks on non-finite values
            continue

        # Evaluate lower boundary condition
        lower_satisfied: bool = (val_float >= spec.min_val) if spec.min_inclusive else (val_float > spec.min_val)
        # Evaluate upper boundary condition
        upper_satisfied: bool = (val_float <= spec.max_val) if spec.max_inclusive else (val_float < spec.max_val)

        # Emit an error finding if boundary conditions are violated
        if not (lower_satisfied and upper_satisfied):
            # Format interval boundary symbols
            left_bracket: str = "[" if spec.min_inclusive else "("
            # Format right interval boundary symbol
            right_bracket: str = "]" if spec.max_inclusive else ")"
            # Append domain violation finding
            findings.append(
                ValidationFinding(
                    severity="error",
                    key_path=spec.key_path,
                    message=(
                        f"Domain boundary violation: parameter '{spec.key_path}' = {val_float} lies outside the "
                        f"allowed interval {left_bracket}{spec.min_val}, {spec.max_val}{right_bracket}."
                    ),
                )
            )

        # Assert integrality for count-type parameters
        if spec.must_be_integer:
            # Check if the float value equals its integer truncation
            if not val_float.is_integer():
                # Emit an error finding for non-integer count parameters
                findings.append(
                    ValidationFinding(
                        severity="error",
                        key_path=spec.key_path,
                        message=f"Integrality violation: count parameter '{spec.key_path}' must be an exact integer; received {val_float}.",
                    )
                )

    # Return accumulated numerical validation findings
    return findings


def validate_categorical_vocabulary(
    config: Dict[str, Any],
) -> List[ValidationFinding]:
    """
    Verify that categorical parameters use the manuscript vocabulary.

    Parameters
    ----------
    config : Dict[str, Any]
        The configuration dictionary to audit.

    Returns
    -------
    List[ValidationFinding]
        Findings for any categorical value outside the allowed vocabulary.
    """
    # Initialize the findings list for the vocabulary audit
    findings: List[ValidationFinding] = []
    # Extract the training section for the optimizer name
    train: Dict[str, Any] = config.get("network_training_parameters", {})
    # Extract the preprocessing section for the normalization method
    pre: Dict[str, Any] = config.get("data_preprocessing_parameters", {})
    # Extract the regime section for the market return type
    regime: Dict[str, Any] = config.get(
        "regime_state_variable_construction", {}
    )
    # Extract the block section for the activation function
    block: Dict[str, Any] = config.get("shared_neural_network_block", {})
    # Enumerate the categorical leaves with their expected vocabulary
    categorical_leaves: Tuple[Tuple[str, Any, Tuple[str, ...]], ...] = (
        ("network_training_parameters.optimizer",
         train.get("optimizer"), ("Adam",)),
        ("data_preprocessing_parameters.normalization_method",
         pre.get("normalization_method"), ("z_score",)),
        ("regime_state_variable_construction.market_return_type",
         regime.get("market_return_type"), ("equal_weighted",)),
        ("shared_neural_network_block.activation",
         block.get("activation"), ("GELU",)),
    )
    # Iterate over every categorical leaf with its allowed vocabulary
    for key_path, value, allowed in categorical_leaves:
        # Verify that the value is a member of the allowed vocabulary
        if value not in allowed:
            # Record a blocking error for the out-of-vocabulary value
            findings.append(
                ValidationFinding(
                    severity="error",
                    key_path=key_path,
                    message=(
                        f"Parameter '{key_path}' must be one of {allowed}; "
                        f"received {value!r}."
                    ),
                )
            )
    # Return the complete findings list for the vocabulary audit
    return findings

def validate_numerical_logical_dimensional(
    config: Dict[str, Any],
    flattened: FlattenedConfig,
) -> List[ValidationFinding]:
    """
    Coordinate the numerical, logical, geometric, and dimensional constraint validation suite.

    This function executes all mathematical and domain integrity audits mandated by Task 1 Step 2:
    1. Validates sample date syntax, chronological ordering, and business-day span sufficiency via 'validate_date_range'.
    2. Validates walk-forward decomposition arithmetic identities via 'validate_walkforward_geometry':
           int(504 * 0.85) = 428, 504 - 428 = 76, 30 * 63 = 1,890.
    3. Recomputes analytical parameter counts for all neural models via 'validate_parameter_counts':
           params(H, d_in) = (d_in + 1) * H + (H + 1) * H + (H + 1).
    4. Audits numerical hyperparameter finiteness, interval boundaries, and discrete count integrality via 'validate_numerical_finiteness'.
    5. Audits categorical parameter string vocabularies via 'validate_categorical_vocabulary'.

    Parameters
    ----------
    config : Dict[str, Any]
        The root study configuration dictionary.
    flattened : FlattenedConfig
        The partitioned flattened configuration structure containing active leaf mappings.

    Returns
    -------
    List[ValidationFinding]
        Accumulated diagnostic findings documenting any numerical, dimensional, or geometric violations.

    Raises
    ------
    TypeError
        If 'config' is not a dictionary or 'flattened' is not a FlattenedConfig instance.
    """
    # Validate that config is a dictionary
    if not isinstance(config, dict):
        # Raise TypeError if config is not a dictionary
        raise TypeError(f"validate_numerical_logical_dimensional requires a Dict[str, Any], received {type(config).__name__}.")
    # Validate that flattened is a FlattenedConfig instance
    if not isinstance(flattened, FlattenedConfig):
        # Raise TypeError if flattened is not a FlattenedConfig instance
        raise TypeError(f"Expected FlattenedConfig instance, received {type(flattened).__name__}.")

    # Initialize the findings accumulator list for numerical and dimensional constraints
    findings: List[ValidationFinding] = []

    # Extract data acquisition parameters dictionary safely
    acq_cfg: Dict[str, Any] = config.get("data_acquisition_and_universe", {})
    # Extract sample start date string
    start_str: str = str(acq_cfg.get("sample_start", ""))
    # Extract sample end date string
    end_str: str = str(acq_cfg.get("sample_end", ""))

    # Step 2.1: Validate date range ordering and business-day span sufficiency relative to walk-forward geometry
    findings.extend(validate_date_range(start_str, end_str, config))

    # Step 2.2: Validate walk-forward development, training, validation, and test day arithmetic identities
    findings.extend(validate_walkforward_geometry(config))

    # Step 2.3: Recompute and audit closed-form parameter counts across all model architectures
    findings.extend(validate_parameter_counts(config))

    # Step 2.4: Validate numerical finiteness, mathematical interval bounds, and count integrality
    findings.extend(validate_numerical_finiteness(flattened))

    # Step 2.5: Validate categorical parameter string vocabularies (optimizer, normalization, activations)
    findings.extend(validate_categorical_vocabulary(config))

    # Return the accumulated numerical, logical, and dimensional findings
    return findings


def validate_raw_input_schemas(
    config: Dict[str, Any],
    raw_data_dir: Optional[Path],
) -> List[Any]:
    """
    Composite Step 1.3 validator auditing data-quality policies, decision-log entries, and raw CSV price bars.

    Parameters
    ----------
    config : Dict[str, Any]
        The root configuration dictionary.
    raw_data_dir : Optional[Path]
        Directory path containing raw CSV price histories; None skips CSV sample checks.

    Returns
    -------
    List[ValidationFinding]
        Accumulated schema and data quality validation findings.
    """
    # Initialize composite findings accumulator list
    findings: List[Any] = []

    # Step 3.1: Validate mandatory decision-log entries
    findings.extend(validate_decision_log_entries(config))

    # Step 3.2: Sample and audit raw CSV files if directory is provided
    if raw_data_dir is not None:
        # Check if raw directory exists
        if not raw_data_dir.is_dir():
            # Emit error finding for missing directory
            findings.append(
                ValidationFinding(
                    severity="error",
                    key_path="raw_data_directory",
                    message=f"Specified raw data directory does not exist: {raw_data_dir}",
                )
            )
        else:
            # Collect sorted CSV file paths
            csv_files: List[Path] = sorted(raw_data_dir.glob("*.csv"))
            # Partition into U.S. and Japanese files by suffix
            us_files: List[Path] = [p for p in csv_files if not p.stem.endswith(".T")]
            # Extract Japanese files
            jp_files: List[Path] = [p for p in csv_files if p.stem.endswith(".T")]

            # Sample deterministically up to 5 U.S. and 5 Japanese files
            sample_files: List[Path] = us_files[:5] + jp_files[:5]

            # Extract schema definitions from config
            us_schema: List[Dict[str, str]] = (
                config.get("raw_input_data_schemas", {})
                .get("us_equity_price_history_csv", {})
                .get("fields", [])
            )
            # Extract Japanese schema definitions
            jp_schema: List[Dict[str, str]] = (
                config.get("raw_input_data_schemas", {})
                .get("jp_equity_price_history_csv", {})
                .get("fields", [])
            )

            # Audit each sampled CSV file against its corresponding schema
            for path in sample_files:
                # Select Japanese schema for .T suffix, U.S. schema otherwise
                fields = jp_schema if path.stem.endswith(".T") else us_schema
                # Run CSV sample audit
                findings.extend(validate_raw_csv_sample(path, fields))

    # Return accumulated Step 1.3 findings
    return findings

# -----------------------------------------------------------------------
# Task 1, Step 3: Validate Raw Input Schema Contracts and Data Quality Rules
# -----------------------------------------------------------------------


def validate_raw_csv_sample(
    csv_path: Path,
    schema_fields: List[Dict[str, str]],
) -> List[Any]:
    """
    Validate a raw vendor price CSV against required columns, numeric dtypes, positive prices, and strictly monotonic dates.

    This function audits a single price bar CSV file to guarantee:
    1. All mandatory OHLCV columns exist.
    2. The 'date' column parses strictly as YYYY-MM-DD.
    3. Consecutive date differences Delta t_k = t_k - t_{k-1} satisfy Delta t_k > 0 (rejecting duplicates Delta t=0 and reversals Delta t<0).
    4. Split- and dividend-adjusted close ('adj_close') is strictly positive (> 0.0) across all rows.

    Parameters
    ----------
    csv_path : Path
        Path to the raw CSV price history file.
    schema_fields : List[Dict[str, str]]
        Schema field definitions from configuration.

    Returns
    -------
    List[ValidationFinding]
        Accumulated diagnostic findings documenting schema, dtype, date ordering, or price violations.

    Raises
    ------
    TypeError
        If 'csv_path' is not a Path object or 'schema_fields' is not a list.
    """
    # Validate that csv_path is a Path instance
    if not isinstance(csv_path, Path):
        # Raise TypeError if csv_path is invalid
        raise TypeError(f"Expected csv_path to be a pathlib.Path, received {type(csv_path).__name__}.")
    # Validate that schema_fields is a list
    if not isinstance(schema_fields, list):
        # Raise TypeError if schema_fields is invalid
        raise TypeError(f"Expected schema_fields to be a list, received {type(schema_fields).__name__}.")

    # Initialize findings accumulator list
    findings: List[Any] = []
    # Build key path identifier for finding reporting
    prefix: str = f"raw_csv:{csv_path.name}"

    # Verify that the CSV file exists on disk
    if not csv_path.is_file():
        # Emit blocking error finding for missing CSV file
        findings.append(
            ValidationFinding(
                severity="error",
                key_path=prefix,
                message=f"Raw CSV file does not exist on disk: {csv_path}.",
            )
        )
        # Return early since missing file cannot be audited
        return findings

    # Read CSV file into pandas DataFrame safely
    try:
        # Load CSV using standard comma delimiter
        frame: pd.DataFrame = pd.read_csv(csv_path, sep=",", header=0)
    except Exception as exc:
        # Emit blocking error finding on CSV read failure
        findings.append(
            ValidationFinding(
                severity="error",
                key_path=prefix,
                message=f"Failed to read CSV file {csv_path.name}: {exc}",
            )
        )
        # Return early since malformed CSV cannot be parsed further
        return findings

    # Assert that the DataFrame contains data rows
    if frame.shape[0] == 0:
        # Emit blocking error finding for empty CSV files
        findings.append(
            ValidationFinding(
                severity="error",
                key_path=prefix,
                message=f"Raw CSV file {csv_path.name} contains zero data rows.",
            )
        )
        # Return early on empty CSV
        return findings

    # Extract required column names from schema (excluding metadata added from file name)
    required_cols: List[str] = [
        f["field"] for f in schema_fields if "added from file name" not in str(f.get("role", "")).lower()
    ]

    # Verify presence of all required columns
    for col in required_cols:
        # Check if column is missing from DataFrame
        if col not in frame.columns:
            # Emit blocking error finding for missing mandatory column
            findings.append(
                ValidationFinding(
                    severity="error",
                    key_path=prefix,
                    message=f"Missing required column '{col}' in {csv_path.name}; found columns: {list(frame.columns)}.",
                )
            )

    # Validate numeric dtypes on standard OHLCV columns if present
    for num_col in ("open", "high", "low", "close", "adj_close", "volume"):
        # Check dtype only if column exists
        if num_col in frame.columns and not pd.api.types.is_numeric_dtype(frame[num_col]):
            # Emit blocking error finding for non-numeric OHLCV column
            findings.append(
                ValidationFinding(
                    severity="error",
                    key_path=prefix,
                    message=f"Column '{num_col}' in {csv_path.name} is non-numeric; dtype is {frame[num_col].dtype}.",
                )
            )

    # Validate date column syntax and strictly monotonic ordering
    if "date" in frame.columns:
        # Attempt strict date parsing
        try:
            # Parse dates using strict format
            date_series: pd.Series = pd.to_datetime(frame["date"], format="%Y-%m-%d", errors="raise")
            # Compute consecutive date differences: Delta t_k = t_k - t_{k-1}
            date_diffs: pd.Series = date_series.diff().dropna()

            # Identify duplicate dates (Delta t == 0)
            zero_diffs: int = int((date_diffs == pd.Timedelta(0)).sum())
            # Identify chronological reversals (Delta t < 0)
            negative_diffs: int = int((date_diffs < pd.Timedelta(0)).sum())

            # Emit error finding if duplicate dates are present
            if zero_diffs > 0:
                # Append finding documenting duplicate dates
                findings.append(
                    ValidationFinding(
                        severity="error",
                        key_path=prefix,
                        message=f"Found {zero_diffs} duplicate date entries in {csv_path.name}.",
                    )
                )

            # Emit error finding if chronological reversals are present
            if negative_diffs > 0:
                # Append finding documenting out-of-order dates
                findings.append(
                    ValidationFinding(
                        severity="error",
                        key_path=prefix,
                        message=f"Found {negative_diffs} chronological date reversals (out-of-order rows) in {csv_path.name}.",
                    )
                )

        except Exception as exc:
            # Emit error finding for unparseable dates
            findings.append(
                ValidationFinding(
                    severity="error",
                    key_path=prefix,
                    message=f"Date parsing failed on '%Y-%m-%d' in {csv_path.name}: {exc}",
                )
            )

    # Validate that adjusted close is strictly positive across all rows
    if "adj_close" in frame.columns and pd.api.types.is_numeric_dtype(frame["adj_close"]):
        # Count non-positive price rows
        non_positive_count: int = int((frame["adj_close"] <= 0.0).sum())
        # Emit error finding if non-positive adjusted prices exist
        if non_positive_count > 0:
            # Append finding documenting non-positive prices
            findings.append(
                ValidationFinding(
                    severity="error",
                    key_path=prefix,
                    message=(
                        f"Found {non_positive_count} non-positive values in 'adj_close' column of {csv_path.name}. "
                        f"All adjusted closing prices must be strictly positive (> 0.0)."
                    ),
                )
            )

    # Return accumulated raw CSV validation findings
    return findings


def validate_data_quality_policies(
    config: Dict[str, Any],
) -> List[ValidationFinding]:
    """
    Verify that the data-quality policy keys are recorded in the config.

    Parameters
    ----------
    config : Dict[str, Any]
        The configuration dictionary to audit.

    Returns
    -------
    List[ValidationFinding]
        Findings for any missing or incorrect data-quality policy.
    """
    # Initialize the findings list for the policy audit
    findings: List[ValidationFinding] = []
    # Extract the preprocessing section for direct policy access
    pre: Dict[str, Any] = config.get("data_preprocessing_parameters", {})
    # Verify the missing adjusted-close policy matches the manuscript
    if pre.get("missing_adj_close_policy") != "drop row for that stock-day":
        # Record a blocking error for the missing-value policy
        findings.append(
            ValidationFinding(
                severity="error",
                key_path=(
                    "data_preprocessing_parameters.missing_adj_close_policy"
                ),
                message=(
                    f"missing_adj_close_policy must be 'drop row for that "
                    f"stock-day'; received "
                    f"{pre.get('missing_adj_close_policy')!r}."
                ),
            )
        )
    # Verify the target validity rule matches the manuscript
    if (
        pre.get("target_validity_rule")
        != "all five forward returns must be non-missing"
    ):
        # Record a blocking error for the target validity rule
        findings.append(
            ValidationFinding(
                severity="error",
                key_path=(
                    "data_preprocessing_parameters.target_validity_rule"
                ),
                message=(
                    f"target_validity_rule must be 'all five forward "
                    f"returns must be non-missing'; received "
                    f"{pre.get('target_validity_rule')!r}."
                ),
            )
        )
    # Verify that the beta minimum observation count equals the recorded
    # study value of 60 (manuscript: beta requires a "required minimum"
    # of non-missing observations; config v2.0.0 fixes it at 60)
    if pre.get("beta_min_observations") != 60:
        # Record a blocking error for the wrong beta minimum
        findings.append(
            ValidationFinding(
                severity="error",
                key_path=(
                    "data_preprocessing_parameters.beta_min_observations"
                ),
                message=(
                    "beta_min_observations must equal the recorded study "
                    f"value 60; received "
                    f"{pre.get('beta_min_observations')!r}."
                ),
            )
        )
    # Return the complete findings list for the policy audit
    return findings


# Define the immutable tuple of exact canonical decision log topics mandated by the manuscript
MANDATED_DECISION_LOG_TOPICS: Tuple[str, ...] = (
    "universe_fallback_rules",
    "reference_dates",
    "rsi_smoothing",
    "rolling_window_missing_data_rules",
    "garch_innovation_distribution",
    "har_monthly_lag",
    "gate_architecture",
    "k2_k6_expert_widths",
    "hard_routing_quantile_variable_and_boundaries",
    "gics_to_expert_mapping",
    "student_t_fit_procedure",
    "high_low_volatility_slice_thresholds",
    "stress_window_date_ranges",
    "optimizer_hyperparameters",
    "early_stopping_patience",
    "regularization_coefficient_values",
    "top_1_gate_training_scheme",
    "data_source_and_api",
    "trading_calendar",
    "capacity_accounting",
    "seed_registry",
    "numerical_hygiene",
    "var_conventions",
    "missing_data_handling",
    "hyperparameter_selection_fallback",
    "finite_sample_distribution",
)


def _normalize_canonical_token(text: str) -> str:
    """
    Normalize descriptive topic strings into standardized lowercase snake_case tokens.

    Parameters
    ----------
    text : str
        The raw descriptive text string.

    Returns
    -------
    str
        Canonical snake_case token stripped of special characters.
    """
    # Convert text to lowercase and replace non-alphanumeric characters with underscores
    cleaned: str = re.sub(r"[^a-z0-9]+", "_", text.lower()).strip("_")
    # Return the cleaned canonical token
    return cleaned


def validate_decision_log_entries(
    config: Dict[str, Any],
    mandated_topics: Tuple[str, ...] = MANDATED_DECISION_LOG_TOPICS,
) -> List[Any]:
    """
    Verify that all 26 mandated decision-log topics are explicitly registered in configuration.

    This function audits 'reproducibility_and_audit.decision_log.required_entries' by normalizing configured
    strings into canonical snake_case tokens and asserting exact set membership against 'MANDATED_DECISION_LOG_TOPICS'.
    Any missing topic emits a structured error finding.

    Parameters
    ----------
    config : Dict[str, Any]
        The root configuration dictionary.
    mandated_topics : Tuple[str, ...], optional
        Tuple of required canonical topic tokens, by default MANDATED_DECISION_LOG_TOPICS.

    Returns
    -------
    List[ValidationFinding]
        Accumulated findings documenting any missing decision-log topics.

    Raises
    ------
    TypeError
        If 'config' is not a dictionary.
    """
    # Validate that config is a dictionary
    if not isinstance(config, dict):
        # Raise TypeError on invalid config argument
        raise TypeError(f"config must be a Dict[str, Any], received {type(config).__name__}.")

    # Initialize findings accumulator list
    findings: List[Any] = []

    # Extract configured decision log entries list safely
    raw_entries: List[str] = (
        config.get("reproducibility_and_audit", {})
        .get("decision_log", {})
        .get("required_entries", [])
    )

    # Assert that required_entries is a non-empty list
    if not raw_entries or not isinstance(raw_entries, list):
        # Emit blocking error finding for missing required_entries list
        findings.append(
            ValidationFinding(
                severity="error",
                key_path="reproducibility_and_audit.decision_log.required_entries",
                message=(
                    "'decision_log.required_entries' list is missing or unpopulated in configuration. "
                    "All 26 mandated research decision topics must be registered."
                ),
            )
        )
        # Return early since matching cannot proceed against an empty list
        return findings

    # Normalize configured entries into a canonical token set
    configured_tokens: Set[str] = {_normalize_canonical_token(entry) for entry in raw_entries if isinstance(entry, str)}

    # Iterate sequentially over every mandated canonical topic token
    for topic_token in mandated_topics:
        # Check if the mandated token is covered in the configured set
        if topic_token not in configured_tokens:
            # Emit blocking error finding documenting the missing decision-log topic
            findings.append(
                ValidationFinding(
                    severity="error",
                    key_path="reproducibility_and_audit.decision_log.required_entries",
                    message=(
                        f"Mandated research decision-log topic '{topic_token}' is missing from "
                        f"'reproducibility_and_audit.decision_log.required_entries'. "
                        f"Add the canonical entry to eliminate undocumented assumptions."
                    ),
                )
            )

    # Return accumulated decision-log validation findings
    return findings


def validate_raw_input_schemas(
    config: Dict[str, Any],
    raw_data_dir: Optional[Path],
) -> List[ValidationFinding]:
    """
    Validate a representative sample of raw CSVs against the schemas.

    Parameters
    ----------
    config : Dict[str, Any]
        The configuration dictionary containing the schema contracts.
    raw_data_dir : Optional[Path]
        Directory of raw CSV files; None skips file-level checks.

    Returns
    -------
    List[ValidationFinding]
        All schema, policy, and decision-log findings.
    """
    # Initialize the combined findings list
    findings: List[ValidationFinding] = []
    # Run the data-quality policy checks from the config
    findings.extend(validate_data_quality_policies(config))
    # Run the decision-log entry checks from the config
    findings.extend(validate_decision_log_entries(config))
    # Handle the optional raw-data directory gracefully
    if raw_data_dir is None:
        # Record a warning that file-level checks were skipped
        findings.append(
            ValidationFinding(
                severity="warning",
                key_path="raw_data_directory",
                message=(
                    "No raw-data directory was supplied; file-level schema "
                    "checks were skipped. Supply the directory for the "
                    "full schema audit."
                ),
            )
        )
        # Return the policy-only findings when no directory is given
        return findings
    # Verify that the supplied raw-data directory exists
    if not raw_data_dir.is_dir():
        # Record a blocking error for the missing directory
        findings.append(
            ValidationFinding(
                severity="error",
                key_path="raw_data_directory",
                message=(
                    f"Raw-data directory does not exist: {raw_data_dir}"
                ),
            )
        )
        # Return early because no files can be sampled
        return findings
    # Collect every CSV file present in the raw-data directory
    csv_files: List[Path] = sorted(raw_data_dir.glob("*.csv"))
    # Split the CSV files into U.S. and Japanese by ticker suffix
    us_files: List[Path] = [p for p in csv_files if not p.stem.endswith(".T")]
    # Collect the Japanese-ticker files by their .T suffix
    jp_files: List[Path] = [p for p in csv_files if p.stem.endswith(".T")]
    # Warn when a market provides fewer than the five required samples
    if len(us_files) < 5:
        # Record a warning finding for the thin U.S. sample
        findings.append(
            ValidationFinding(
                severity="warning",
                key_path="raw_data_directory",
                message=(
                    f"Only {len(us_files)} U.S. CSV files available; the "
                    f"schema audit requires at least five."
                ),
            )
        )
    # Warn when the Japanese market provides fewer than five files
    if len(jp_files) < 5:
        # Record a warning finding for the thin Japanese sample
        findings.append(
            ValidationFinding(
                severity="warning",
                key_path="raw_data_directory",
                message=(
                    f"Only {len(jp_files)} Japanese CSV files available; "
                    f"the schema audit requires at least five."
                ),
            )
        )
    # Sample at most five U.S. and five Japanese files deterministically
    sample: List[Path] = us_files[:5] + jp_files[:5]
    # Extract the U.S. schema fields from the config
    us_fields: List[Dict[str, str]] = (
        config.get("raw_input_data_schemas", {})
        .get("us_equity_price_history_csv", {})
        .get("fields", [])
    )
    # Extract the Japanese schema fields from the config
    jp_fields: List[Dict[str, str]] = (
        config.get("raw_input_data_schemas", {})
        .get("jp_equity_price_history_csv", {})
        .get("fields", [])
    )
    # Iterate over every sampled CSV file
    for csv_path in sample:
        # Select the schema fields by the market suffix of the file
        schema_fields: List[Dict[str, str]] = (
            jp_fields if csv_path.stem.endswith(".T") else us_fields
        )
        # Validate the CSV file against its schema fields
        findings.extend(validate_raw_csv_sample(csv_path, schema_fields))
    # Return the complete findings list for the schema audit
    return findings


# -----------------------------------------------------------------------
# Task 1, Orchestrator Function
# -----------------------------------------------------------------------

def validate_config_and_schemas(
    config: Dict[str, Any],
    raw_data_dir: Optional[Path] = None,
    report_path: Optional[Path] = None,
) -> ConfigValidationReport:
    """
    Orchestrate the complete end-to-end configuration and schema validation pipeline.

    This master orchestrator coordinates the execution of all Task 1 validation stages in strict dependency order:
    1. Recursively flattens the nested configuration into partitioned leaf mappings and branch sets via 'flatten_key_paths'.
    2. Executes the structural completeness audit (key presence, non-nullity, decision-log coverage) via 'audit_config_completeness'.
    3. Executes numerical, dimensional, geometric, and domain boundary audits via 'validate_numerical_logical_dimensional'.
    4. Executes data quality policy, decision log topic, and raw CSV schema sampling audits via 'validate_raw_input_schemas'.
    5. Assembles an immutable 'ConfigValidationReport' with dynamic validity computation.
    6. Optionally serializes the validation report to disk as indented JSON.
    7. Enforces a fail-fast error boundary: raises ValueError if any blocking 'error' or 'conflict' finding exists.

    Parameters
    ----------
    config : Dict[str, Any]
        The root study configuration dictionary to validate.
    raw_data_dir : Optional[Path], optional
        Path to the directory containing raw CSV price bars for schema sampling; None skips file checks.
    report_path : Optional[Path], optional
        Destination file path for persisting the validation report JSON; None skips disk persistence.

    Returns
    -------
    ConfigValidationReport
        The immutable aggregate validation report encapsulating findings, metadata, and dynamic validity status.

    Raises
    ------
    TypeError
        If 'config' is not a dictionary.
    ValueError
        If the configuration contains any blocking 'error' or 'conflict' findings.
    """
    # Validate that root config is a dictionary before initiating validation traversal
    if not isinstance(config, dict):
        # Raise TypeError if config is not a dictionary
        raise TypeError(f"validate_config_and_schemas requires a Dict[str, Any], received {type(config).__name__}.")

    # Initialize the master findings list to accumulate findings across all validation stages
    all_findings: List[ValidationFinding] = []

    # Step 1.3: Flatten the configuration dictionary into partitioned leaf mappings and branch container sets
    flattened: FlattenedConfig = flatten_key_paths(config)

    # Step 1 Composite: Execute structural completeness audit (mandatory keys and decision-log coverage)
    all_findings.extend(audit_config_completeness(config, flattened=flattened))

    # Step 2 Composite: Execute numerical, logical, geometric, parameter count, and domain boundary audits
    all_findings.extend(validate_numerical_logical_dimensional(config, flattened=flattened))

    # Step 3 Composite: Execute data quality policies, decision log entries, and raw CSV schema sampling audits
    all_findings.extend(validate_raw_input_schemas(config, raw_data_dir))

    # Extract decision-log gaps from findings whose key path or message indicates an unrecorded topic
    decision_log_gaps: List[str] = sorted(
        {
            f.key_path
            for f in all_findings
            if "decision_log" in f.key_path or "unspecified" in f.message.lower()
        }
    )

    # Extract configuration version string safely with fallback default
    config_version_str: str = str(config.get("config_version", "2.0.0"))

    # Compute the total count of distinct manuscript-mandated keys checked during the audit
    total_checked_keys: int = len(REQUIRED_MANUSCRIPT_KEYS)

    # Instantiate the immutable ConfigValidationReport with dynamic validity property
    report: ConfigValidationReport = ConfigValidationReport(
        config_version=config_version_str,
        findings=tuple(all_findings),
        checked_keys=total_checked_keys,
        decision_log_gaps=tuple(decision_log_gaps),
    )

    # Persist the validation report to disk as formatted JSON if a destination path was supplied
    if report_path is not None:
        # Create destination parent directories recursively if they do not exist
        report_path.parent.mkdir(parents=True, exist_ok=True)
        # Open report file handle with UTF-8 encoding
        with report_path.open("w", encoding="utf-8") as handle:
            # Serialize report dictionary to indented JSON
            json.dump(report.to_dict(), handle, indent=2, ensure_ascii=False)

    # Enforce fail-fast error boundary: raise ValueError if the configuration contains blocking errors
    if not report.valid:
        # Raise ValueError with exact error count and diagnostic remediation instruction
        raise ValueError(
            f"Configuration validation failed with {report.error_count} blocking errors. "
            f"Inspect report findings before proceeding with downstream pipeline tasks."
        )

    # Return the validated immutable validation report
    return report
````

````python
# Task 2: Reproducible Python Environment and Raw Data Acquisition

# ==============================================================================
# Task 2: Reproducible Python Environment and Raw Data Acquisition
# ==============================================================================

# -----------------------------------------------------------------------
# Task 2, Step 1: Create a Deterministic Python Environment
# -----------------------------------------------------------------------

# Pinned library versions required by the study's reproducibility manifest.
# The task list mandates pinning numpy, pandas, scipy, statsmodels,
# scikit-learn, torch, yfinance, pandas_market_calendars, matplotlib,
# joblib, and pyyaml; yfinance and pandas_market_calendars are therefore
# REQUIRED (fatal on drift), not advisory.
REQUIRED_LIBRARIES: Dict[str, str] = {
    "numpy": "2.4.6",
    "pandas": "3.0.3",
    "scipy": "1.18.0",
    "statsmodels": "0.14.6",
    "scikit-learn": "1.9.0",
    "torch": "2.13.0",
    "matplotlib": "3.11.1",
    "joblib": "1.5.3",
    "pyyaml": "6.0.3",
    "yfinance": "0.2.54",
    "pandas_market_calendars": "4.5.1",
}

# Advisory libraries whose absence is recorded but does not block setup
ADVISORY_LIBRARIES: Tuple[str, ...] = (
    "arch",
)

# The minimum supported Python version per the task list
MIN_PYTHON_VERSION: Tuple[int, int] = (3, 11)


@dataclass(frozen=True)
class EnvironmentReport:
    """
    Immutable report of the deterministic-environment setup checks.

    Attributes
    ----------
    python_version : str
        The detected Python version string.
    python_ok : bool
        True when the detected version meets the minimum requirement.
    library_findings : List[str]
        Human-readable findings for every pinned and advisory library.
    fatal_drift : List[str]
        Pinned libraries whose version differs from the manifest.
    no_llm_verified : bool
        True when the config asserts llm_used is False.
    """

    # Detected Python version string
    python_version: str
    # Flag for the Python version requirement
    python_ok: bool
    # Findings for every pinned and advisory library
    library_findings: List[str] = field(default_factory=list)
    # Pinned libraries whose version differs from the manifest
    fatal_drift: List[str] = field(default_factory=list)
    # Flag for the no-LLM verification
    no_llm_verified: bool = False


def verify_python_version() -> Tuple[bool, str]:
    """
    Verify that the running Python version meets the minimum.

    Returns
    -------
    Tuple[bool, str]
        A pass flag and the detected version string.
    """
    # Capture the running Python version as a string
    version_str: str = f"{sys.version_info.major}.{sys.version_info.minor}"
    # Compare the running version against the minimum requirement
    ok: bool = (
        sys.version_info.major,
        sys.version_info.minor,
    ) >= MIN_PYTHON_VERSION
    # Return the pass flag and the version string
    return ok, version_str


def verify_pinned_libraries() -> Tuple[List[str], List[str]]:
    """
    Compare installed library versions against the pinned manifest.

    Returns
    -------
    Tuple[List[str], List[str]]
        A findings list and a fatal-drift list for the pinned libraries.
    """
    # Initialize the findings list
    findings: List[str] = []
    # Initialize the fatal drift list
    fatal: List[str] = []
    # Iterate over every pinned library in the manifest
    for lib_name, pinned in REQUIRED_LIBRARIES.items():
        # Attempt to read the installed version of the library
        try:
            # Read the installed distribution version
            installed: str = dist_version(lib_name)
        except Exception as exc:
            # Record the missing library as fatal drift
            findings.append(f"{lib_name}: MISSING ({exc})")
            # Add the library to the fatal drift list
            fatal.append(lib_name)
            # Continue to the next library
            continue
        # Compare the installed version with the pinned version
        if installed != pinned:
            # Record the version drift finding
            findings.append(
                f"{lib_name}: installed {installed}, pinned {pinned}"
            )
            # Add the library to the fatal drift list
            fatal.append(lib_name)
        else:
            # Record the matching version finding
            findings.append(f"{lib_name}: {installed} OK")
    # Iterate over every advisory library
    for lib_name in ADVISORY_LIBRARIES:
        # Attempt to read the installed version of the library
        try:
            # Read the installed distribution version
            installed_adv: str = dist_version(lib_name)
        except Exception:
            # Record the missing advisory library
            findings.append(f"{lib_name}: MISSING (advisory)")
            # Continue to the next advisory library
            continue
        # Record the present advisory library
        findings.append(f"{lib_name}: {installed_adv} (advisory)")
    # Return the findings and fatal drift lists
    return findings, fatal


def seed_all_rng(seed: int) -> None:
    """
    Reset every random-number generator to a deterministic state.

    Parameters
    ----------
    seed : int
        The master seed applied to Python random, NumPy, and PyTorch.

    Returns
    -------
    None
    """
    # Validate that the seed is a non-negative integer
    if not isinstance(seed, int) or seed < 0:
        # Raise an informative error for the invalid seed
        raise ValueError(f"seed must be a non-negative int; got {seed!r}")
    # Reset the Python standard-library random generator
    random.seed(seed)
    # Reset the NumPy random generator
    np.random.seed(seed)
    # Configure PyTorch determinism when torch is installed
    if torch is not None:
        # Reset the PyTorch generator with the master seed
        torch.manual_seed(seed)
        # Reset every CUDA generator with the master seed when CUDA exists
        if torch.cuda.is_available():
            # Seed all CUDA device generators for full determinism
            torch.cuda.manual_seed_all(seed)
        # Disable nondeterministic cuDNN algorithms
        torch.backends.cudnn.deterministic = True
        # Disable benchmark-mode algorithm selection
        torch.backends.cudnn.benchmark = False
        # Attempt to enable global deterministic algorithms; some PyTorch
        # operations are not deterministic-algorithm compatible, so the
        # enforcement is wrapped and its failure is logged, not fatal
        if hasattr(torch, "use_deterministic_algorithms"):
            # Guard the deterministic-algorithm enforcement
            try:
                # Enable deterministic algorithm enforcement
                torch.use_deterministic_algorithms(True)
            except Exception as exc:
                # Log the enforcement failure for the audit trail
                logger.warning(
                    "use_deterministic_algorithms(True) failed: %s", exc
                )


def verify_no_llm_usage(config: Dict[str, Any]) -> bool:
    """
    Verify that the config declares no LLM usage anywhere.

    Parameters
    ----------
    config : Dict[str, Any]
        The configuration dictionary.

    Returns
    -------
    bool
        True when llm_used is False.
    """
    # Extract the LLM configuration section
    llm_cfg: Dict[str, Any] = config.get("llm_and_prompt_configuration", {})
    # Extract the llm_used flag with a strict default of False
    llm_used: bool = bool(llm_cfg.get("llm_used", False))
    # Assert that no LLM is declared in the configuration
    if llm_used:
        # Raise an informative error for the LLM usage violation
        raise ValueError(
            "config['llm_and_prompt_configuration']['llm_used'] must be "
            "False; no language model may be used in this study."
        )
    # Return the verified no-LLM flag
    return True


def orchestrate_environment_setup(
    config: Dict[str, Any],
) -> EnvironmentReport:
    """
    Run the full deterministic-environment setup and probe.

    Parameters
    ----------
    config : Dict[str, Any]
        The configuration dictionary.

    Returns
    -------
    EnvironmentReport
        The immutable environment report with all findings.
    """
    # Verify the Python version requirement
    python_ok, version_str = verify_python_version()
    # Verify the pinned library versions
    findings, fatal = verify_pinned_libraries()
    # Assemble the environment report
    report: EnvironmentReport = EnvironmentReport(
        python_version=version_str,
        python_ok=python_ok,
        library_findings=findings,
        fatal_drift=fatal,
        no_llm_verified=verify_no_llm_usage(config),
    )
    # Return the assembled environment report
    return report


# -----------------------------------------------------------------------
# Task 2, Step 2: Construct the U.S. Universe and Download Price Histories
# -----------------------------------------------------------------------


@dataclass(frozen=True)
class MarketSpec:
    """
    Immutable parameter bundle distinguishing the U.S. and JP panels.

    Attributes
    ----------
    market_name : str
        One of "US" or "JP".
    ticker_suffix : str
        The Yahoo Finance suffix, "" for the U.S. and ".T" for Japan.
    target_count : int
        The target number of tickers for the market.
    use_sp1500_filter : bool
        True when the S&P 1500 membership filter applies.
    segment_column : str
        The metadata column carrying segment membership.
    """

    # Market identifier, one of US or JP
    market_name: str
    # Yahoo Finance ticker suffix for the market
    ticker_suffix: str
    # Target ticker count for the market
    target_count: int
    # Flag for the S&P 1500 membership filter
    use_sp1500_filter: bool
    # Metadata column carrying segment membership
    segment_column: str


def build_market_specs(
    config: Dict[str, Any],
) -> Tuple[MarketSpec, MarketSpec]:
    """
    Build the US and JP market specifications from the config.

    Parameters
    ----------
    config : Dict[str, Any]
        The configuration dictionary.

    Returns
    -------
    Tuple[MarketSpec, MarketSpec]
        The US and JP market specifications.
    """
    # Extract the acquisition section for direct parameter access
    acq: Dict[str, Any] = config.get("data_acquisition_and_universe", {})
    # Build the US market specification
    us_spec: MarketSpec = MarketSpec(
        market_name="US",
        ticker_suffix="",
        target_count=int(acq.get("us_ticker_count_target", 1027)),
        use_sp1500_filter=True,
        segment_column="sp1500_member_current",
    )
    # Build the Japanese market specification
    jp_spec: MarketSpec = MarketSpec(
        market_name="JP",
        ticker_suffix=str(acq.get("jp_ticker_suffix", ".T")),
        target_count=int(acq.get("jp_ticker_count_target", 1552)),
        use_sp1500_filter=False,
        segment_column="tse_prime_member",
    )
    # Return the two market specifications
    return us_spec, jp_spec


def load_universe_metadata(metadata_path: Path) -> pd.DataFrame:
    """
    Load and validate the universe metadata table.

    Parameters
    ----------
    metadata_path : Path
        Path of the versioned metadata CSV.

    Returns
    -------
    pd.DataFrame
        The metadata frame indexed by ticker.

    Raises
    ------
    FileNotFoundError
        If the metadata file does not exist.
    ValueError
        If required columns are missing.
    """
    # Verify that the metadata file exists
    if not metadata_path.exists():
        # Raise an informative error for the missing metadata file
        raise FileNotFoundError(f"Metadata file missing: {metadata_path}")
    # Read the metadata CSV into a DataFrame
    frame: pd.DataFrame = pd.read_csv(metadata_path)
    # Define the required metadata columns
    required_columns: Tuple[str, ...] = (
        "ticker",
        "gics_sector",
        "sp1500_member_current",
        "tse_prime_member",
        "market_cap_bucket",
        "is_common_stock",
        "still_listed",
        "average_daily_dollar_volume",
    )
    # Verify that every required column exists
    missing: List[str] = [
        c for c in required_columns if c not in frame.columns
    ]
    # Raise an informative error for missing columns
    if missing:
        # Build the error message with the missing and found columns
        raise ValueError(
            f"Metadata missing columns {missing}; found "
            f"{list(frame.columns)}."
        )
    # Normalize the ticker column to strings
    frame["ticker"] = frame["ticker"].astype(str)
    # Index the frame by ticker for aligned operations
    return frame.set_index("ticker")


def filter_metadata_universe(
    metadata: pd.DataFrame,
    spec: MarketSpec,
    config: Dict[str, Any],
) -> pd.DataFrame:
    """
    Apply the pre-download inclusion filters to the metadata frame.

    Parameters
    ----------
    metadata : pd.DataFrame
        The universe metadata frame indexed by ticker.
    spec : MarketSpec
        The market specification for the filter set.
    config : Dict[str, Any]
        The configuration dictionary.

    Returns
    -------
    pd.DataFrame
        The filtered candidate frame.
    """
    # Select the inclusion criteria by market: the US panel uses
    # us_inclusion_criteria while the Japanese panel uses its own
    # jp_inclusion_criteria section, so a JP-specific criterion is never
    # silently replaced by the US default
    criteria_key: str = (
        "us_inclusion_criteria" if spec.market_name == "US"
        else "jp_inclusion_criteria"
    )
    # Extract the market-specific inclusion criteria for direct access
    criteria: Dict[str, Any] = config.get(
        "data_acquisition_and_universe", {}
    ).get(criteria_key, {})
    # Initialize a boolean mask that keeps every row initially
    mask: pd.Series = pd.Series(True, index=metadata.index)
    # Apply the common-stock-only filter; the criterion must be explicit
    # so a missing key cannot silently change the filter behavior
    if "common_stock_only" not in criteria:
        # Raise an informative error for the missing criterion
        raise ValueError(
            f"{criteria_key}.common_stock_only is required; found keys "
            f"{list(criteria.keys())}."
        )
    # Apply the common-stock-only filter when the criterion is set
    if criteria["common_stock_only"]:
        # Keep only rows flagged as common stocks
        mask &= metadata["is_common_stock"].astype(bool)
    # Require the listed-through-end criterion explicitly as well
    if "listed_through_end_of_sample" not in criteria:
        # Raise an informative error for the missing criterion
        raise ValueError(
            f"{criteria_key}.listed_through_end_of_sample is required."
        )
    # Apply the listed-through-end filter when the criterion is set
    if criteria["listed_through_end_of_sample"]:
        # Keep only rows flagged as still listed
        mask &= metadata["still_listed"].astype(bool)
    # Apply the segment membership filter for the market
    if spec.segment_column in metadata.columns:
        # Keep only rows flagged as segment members
        mask &= metadata[spec.segment_column].astype(bool)
    # Return the filtered candidate frame
    return metadata.loc[mask]


def download_price_histories(
    tickers: List[str],
    config: Dict[str, Any],
    out_dir: Path,
) -> Dict[str, Path]:
    """
    Download daily price histories for the ticker list via yfinance.

    Parameters
    ----------
    tickers : List[str]
        The sorted list of tickers to download.
    config : Dict[str, Any]
        The configuration dictionary.
    out_dir : Path
        The versioned output directory for the raw CSVs.

    Returns
    -------
    Dict[str, Path]
        Mapping of ticker to the written CSV path.

    Raises
    ------
    RuntimeError
        If yfinance is not installed.
    """
    # Guard against a missing yfinance installation
    if yf is None:
        # Raise an informative error naming the install command
        raise RuntimeError(
            "yfinance is not installed; install it with "
            "'pip install yfinance' before running downloads."
        )
    # Extract the download protocol parameters from the config
    protocol: Dict[str, Any] = config.get(
        "data_acquisition_and_universe", {}
    ).get("download_protocol", {})
    # Extract the retry count with the configured default
    retry_count: int = int(protocol.get("retry_count", 5))
    # Extract the chunk size in months for chunked downloads
    chunk_months: int = int(
        protocol.get("download_chunk_size_months", 24)
    )
    # Extract the sample window from the config
    acq: Dict[str, Any] = config.get("data_acquisition_and_universe", {})
    # Build the sample start and end date strings
    start: str = str(acq.get("sample_start", "2015-12-01"))
    # Build the sample end date string for the download window
    end: str = str(acq.get("sample_end", "2025-11-30"))
    # Ensure the output directory exists
    out_dir.mkdir(parents=True, exist_ok=True)
    # Build the failure-log path for the decision-log audit trail
    failure_log: Path = out_dir / "download_failures.log"
    # Initialize the ticker-to-path mapping
    written: Dict[str, Path] = {}
    # Parse the chunk boundaries over the full sample window
    chunk_starts: List[pd.Timestamp] = list(
        pd.date_range(start=start, end=end, freq=f"{chunk_months}MS")
    )
    # Guard the degenerate case of an empty chunk schedule
    if not chunk_starts:
        # Raise an informative error for the empty schedule
        raise ValueError(
            f"No download chunks could be built for {start}..{end}."
        )
    # Append the terminal chunk boundary to close the final window
    chunk_ends: List[pd.Timestamp] = chunk_starts[1:] + [
        pd.Timestamp(end)
    ]
    # Iterate over every ticker in the deterministic order
    for ticker in tickers:
        # Initialize the download attempt counter
        attempt: int = 0
        # Loop until the retry budget is exhausted
        while attempt < retry_count:
            # Increment the attempt counter
            attempt += 1
            # Attempt the chunked download for this ticker
            try:
                # Initialize the per-ticker chunk frames list
                chunk_frames: List[pd.DataFrame] = []
                # Iterate over every chunk window boundary
                for chunk_start, chunk_end in zip(
                    chunk_starts, chunk_ends
                ):
                    # Compute the chunk end shifted by the overlap so
                    # rolling windows are continuous across boundaries
                    overlap_end: pd.Timestamp = (
                        chunk_end + pd.Timedelta(days=10)
                    )
                    # Download the chunk with auto_adjust=False so the
                    # split/dividend-adjusted close column is retained
                    # verbatim from the provider
                    data: pd.DataFrame = yf.download(
                        ticker,
                        start=chunk_start.strftime("%Y-%m-%d"),
                        end=overlap_end.strftime("%Y-%m-%d"),
                        auto_adjust=False,
                        progress=False,
                    )
                    # Skip empty chunks without failing the ticker
                    if data is not None and not data.empty:
                        # Reset the index to expose the date column
                        data = data.reset_index()
                        # Normalize the column names to lowercase
                        data.columns = [
                            str(c).lower() for c in data.columns
                        ]
                        # Rename the provider's adjusted-close column to
                        # the schema contract name adj_close
                        data = data.rename(
                            columns={"adj close": "adj_close"}
                        )
                        # Append the chunk frame to the list
                        chunk_frames.append(data)
                # Raise when every chunk of the ticker came back empty
                if not chunk_frames:
                    # Raise an error to trigger the retry path
                    raise ValueError("empty download for all chunks")
                # Concatenate the chunk frames into one history
                full: pd.DataFrame = pd.concat(
                    chunk_frames, ignore_index=True
                )
                # Drop duplicate dates keeping the latest overlapping bar
                full = full.drop_duplicates(
                    subset=["date"], keep="last"
                )
                # Validate the schema contract columns before persisting
                required_cols: Tuple[str, ...] = (
                    "date", "open", "high", "low", "close",
                    "adj_close", "volume",
                )
                # Check every required column of the schema
                for column in required_cols:
                    # Verify the column exists in the concatenated frame
                    if column not in full.columns:
                        # Raise an informative error for the missing column
                        raise ValueError(
                            f"Download of {ticker} is missing column "
                            f"'{column}'; found {list(full.columns)}. The "
                            f"provider may have changed its adjusted-close "
                            f"convention."
                        )
                # Build the output path for this ticker
                out_path: Path = out_dir / f"{ticker}.csv"
                # Persist the downloaded frame as a CSV
                full.to_csv(out_path, index=False)
                # Record the written path in the mapping
                written[ticker] = out_path
                # Break out of the retry loop on success
                break
            except Exception as exc:
                # Log the failure when the retry budget remains
                if attempt >= retry_count:
                    # Record the final failure for the ticker
                    message: str = (
                        f"FAILED {ticker} after {retry_count} attempts: "
                        f"{exc}"
                    )
                    # Print the failure for the operator
                    print(message)
                    # Append the failure to the decision-log file
                    with failure_log.open(
                        "a", encoding="utf-8"
                    ) as handle:
                        # Write the failure record with a timestamp
                        handle.write(f"{message}\n")
    # Return the mapping of successfully downloaded tickers
    return written


def compute_history_stats(
    raw_dir: Path,
    config: Dict[str, Any],
    market_name: str = "US",
    ticker_list: Optional[List[str]] = None,
) -> pd.DataFrame:
    """
    Compute per-ticker history length and coverage statistics.

    The expected-day denominator uses the market exchange calendar
    (NYSE for the US, TSE for Japan) so that coverage reflects the
    true number of scheduled trading days, never a generic business-day
    count.

    Parameters
    ----------
    raw_dir : Path
        The directory of downloaded raw CSVs.
    config : Dict[str, Any]
        The configuration dictionary.
    market_name : str, optional
        The market name, one of "US" or "JP", default "US".
    ticker_list : Optional[List[str]], optional
        Restrict the scan to these tickers when supplied, preventing
        stale files from a previous run from polluting the statistics.

    Returns
    -------
    pd.DataFrame
        A frame indexed by ticker with history stats.
    """
    # Extract the sample window from the config
    acq: Dict[str, Any] = config.get("data_acquisition_and_universe", {})
    # Build the sample start and end date strings
    start: str = str(acq.get("sample_start", "2015-12-01"))
    # Build the sample end date string for the window sizing
    end: str = str(acq.get("sample_end", "2025-11-30"))
    # Compute the expected trading-day count from the exchange calendar
    n_expected: int = _expected_trading_days(start, end, market_name)
    # Initialize the stats records list
    records: List[Dict[str, Any]] = []
    # Iterate over every CSV file in the raw directory
    for csv_path in sorted(raw_dir.glob("*.csv")):
        # Derive the ticker from the file stem
        ticker: str = csv_path.stem
        # Skip files outside the requested ticker list when provided
        if ticker_list is not None and ticker not in set(ticker_list):
            # Continue to the next file outside the list
            continue
        # Read the CSV with the date column parsed
        frame: pd.DataFrame = pd.read_csv(csv_path, parse_dates=["date"])
        # Guard the required adjusted-close column with an informative
        # error instead of a silent KeyError
        if "adj_close" not in frame.columns:
            # Raise an informative error naming the missing column
            raise ValueError(
                f"CSV {csv_path.name} is missing the 'adj_close' column; "
                f"found {list(frame.columns)}."
            )
        # Compute the number of valid adjusted-close rows
        n_valid: int = int(frame["adj_close"].notna().sum())
        # Compute the coverage fraction over the sample window
        coverage: float = n_valid / max(n_expected, 1)
        # Compute the earliest date of the price history
        date_min: pd.Timestamp = frame["date"].min()
        # Compute the latest date of the price history
        date_max: pd.Timestamp = frame["date"].max()
        # Compute the span as a fractional year count
        span_years: float = (
            date_max - date_min
        ).days / 365.25
        # Append the per-ticker stats record
        records.append(
            {
                "ticker": ticker,
                "n_valid": n_valid,
                "coverage": coverage,
                "span_years": span_years,
            }
        )
    # Build the stats frame from the records
    stats: pd.DataFrame = pd.DataFrame(records)
    # Index the stats frame by ticker
    return stats.set_index("ticker")


def _expected_trading_days(
    start: str, end: str, market_name: str
) -> int:
    """
    Compute the scheduled trading-day count of the market exchange over
    the sample window, falling back to business days when the calendar
    library is unavailable.

    Parameters
    ----------
    start : str
        The sample start date as YYYY-MM-DD.
    end : str
        The sample end date as YYYY-MM-DD.
    market_name : str
        The market name, one of "US" or "JP".

    Returns
    -------
    int
        The expected number of trading days in the window.
    """
    # Use the exchange calendar library when it is available
    if mcal is not None:
        # Select the exchange calendar by market name
        exchange: str = "NYSE" if market_name == "US" else "TSE"
        # Fetch the scheduled sessions of the exchange
        schedule = mcal.get_calendar(exchange).schedule(
            start_date=start, end_date=end
        )
        # Return the number of scheduled trading days
        return int(len(schedule.index))
    # Log the fallback to the generic business-day count
    logger.warning(
        "pandas_market_calendars unavailable; coverage denominator "
        "falls back to business days"
    )
    # Fall back to the business-day count when the library is missing
    return len(pd.bdate_range(start=start, end=end))


def apply_history_filters(
    stats: pd.DataFrame,
    config: Dict[str, Any],
    market_name: str = "US",
    coverage_threshold: Optional[float] = None,
) -> pd.DataFrame:
    """
    Apply the history-length and coverage filters to the stats frame.

    Parameters
    ----------
    stats : pd.DataFrame
        The per-ticker history stats frame.
    config : Dict[str, Any]
        The configuration dictionary.
    market_name : str, optional
        The market name, one of "US" or "JP", default "US"; selects the
        market-specific inclusion criteria.
    coverage_threshold : Optional[float], optional
        Overrides the coverage threshold; used by the under-count
        relaxation loop of select_target_tickers.

    Returns
    -------
    pd.DataFrame
        The filtered stats frame.
    """
    # Select the inclusion criteria by market name so the Japanese panel
    # never inherits the US criteria silently
    criteria_key: str = (
        "us_inclusion_criteria" if market_name == "US"
        else "jp_inclusion_criteria"
    )
    # Extract the inclusion criteria for direct access
    criteria: Dict[str, Any] = config.get(
        "data_acquisition_and_universe", {}
    ).get(criteria_key, {})
    # Extract the minimum history length in years
    min_years: float = float(criteria.get("min_history_years", 6))
    # Extract the minimum coverage target (overridable by the caller)
    min_coverage: float = (
        float(coverage_threshold)
        if coverage_threshold is not None
        else float(criteria.get("full_period_coverage_target", 0.92))
    )
    # Initialize a boolean mask that keeps every row initially
    mask: pd.Series = pd.Series(True, index=stats.index)
    # Apply the minimum-history filter
    mask &= stats["span_years"] >= min_years
    # Apply the minimum-coverage filter
    mask &= stats["coverage"] >= min_coverage
    # Return the filtered stats frame
    return stats.loc[mask]


def select_target_tickers(
    stats: pd.DataFrame,
    metadata: pd.DataFrame,
    spec: MarketSpec,
    config: Dict[str, Any],
) -> List[str]:
    """
    Apply the deterministic fallback to reach the target ticker count.

    Over-count: sort by average daily dollar volume and keep the top N.
    Under-count: relax the full-period coverage target in 1% increments,
    re-filtering after every relaxation, and log every relaxation to the
    decision log exactly as the task list prescribes.

    Parameters
    ----------
    stats : pd.DataFrame
        The per-ticker history stats frame indexed by ticker (UNFILTERED,
        so the relaxation loop can re-filter at every threshold).
    metadata : pd.DataFrame
        The full metadata frame for dollar-volume sorting.
    spec : MarketSpec
        The market specification.
    config : Dict[str, Any]
        The configuration dictionary.

    Returns
    -------
    List[str]
        The final sorted ticker list at the target count.
    """
    # Select the market-specific inclusion criteria
    criteria_key: str = (
        "us_inclusion_criteria" if spec.market_name == "US"
        else "jp_inclusion_criteria"
    )
    # Extract the inclusion criteria for the coverage threshold
    criteria: Dict[str, Any] = config.get(
        "data_acquisition_and_universe", {}
    ).get(criteria_key, {})
    # Extract the minimum history length in years
    min_years: float = float(criteria.get("min_history_years", 6))
    # Extract the base full-period coverage target
    base_coverage: float = float(
        criteria.get("full_period_coverage_target", 0.92)
    )
    # Apply the history-length and base-coverage filters
    filtered: pd.DataFrame = apply_history_filters(
        stats, config, market_name=spec.market_name
    )
    # Convert the filtered index to a sorted ticker list
    tickers: List[str] = sorted(filtered.index.tolist())
    # Handle the under-count case by relaxing coverage in increments
    if len(tickers) < spec.target_count:
        # Initialize the relaxation step counter
        step: int = 0
        # Relax the coverage target in 1% increments until the target
        # count is reached or the coverage floor of 0.80 is hit
        while (
            len(tickers) < spec.target_count
            and step < 12
        ):
            # Advance the relaxation step
            step += 1
            # Compute the relaxed coverage threshold
            relaxed: float = base_coverage - 0.01 * step
            # Stop relaxing at the documented floor
            if relaxed < 0.80:
                # Log the reached floor
                logger.warning(
                    "Coverage relaxation floor reached at %.2f", relaxed
                )
                # Break out of the relaxation loop
                break
            # Re-filter with the relaxed coverage threshold
            relaxed_stats: pd.DataFrame = apply_history_filters(
                stats, config, market_name=spec.market_name,
                coverage_threshold=relaxed,
            )
            # Refresh the ticker list from the relaxed filter
            tickers = sorted(relaxed_stats.index.tolist())
            # Log the relaxation to the decision log
            logger.info(
                "fallback: relaxed coverage to %.2f -> %d tickers",
                relaxed, len(tickers),
            )
            # Print the relaxation record for the operator
            print(
                f"fallback: under-count, coverage relaxed to "
                f"{relaxed:.2f}, count {len(tickers)}"
            )
    # Handle the over-count case by dollar-volume truncation
    if len(tickers) > spec.target_count:
        # Join the filtered tickers with their dollar volumes
        joined: pd.DataFrame = metadata.loc[
            metadata.index.isin(tickers)
        ].copy()
        # Sort the candidates by dollar volume descending
        joined = joined.sort_values(
            "average_daily_dollar_volume", ascending=False
        )
        # Keep the top target-count tickers
        tickers = joined.index.tolist()[: spec.target_count]
        # Record the fallback application
        print(
            f"fallback: over-count, kept top {spec.target_count} by "
            f"dollar volume"
        )
    # Log the shortfall when the count could not be reached
    if len(tickers) < spec.target_count:
        # Log the final shortfall for the decision log
        logger.warning(
            "Target count %d not reached; final count %d",
            spec.target_count, len(tickers),
        )
    # Return the final sorted ticker list
    return sorted(tickers)


def hash_raw_files(raw_dir: Path, manifest_path: Path) -> pd.DataFrame:
    """
    Compute a SHA-256 hash per raw CSV and persist the manifest.

    Parameters
    ----------
    raw_dir : Path
        The directory of raw CSV files.
    manifest_path : Path
        The destination path of the manifest CSV.

    Returns
    -------
    pd.DataFrame
        The manifest frame with ticker and sha256 columns.
    """
    # Initialize the manifest records list
    records: List[Dict[str, str]] = []
    # Iterate over every CSV file in the raw directory
    for csv_path in sorted(raw_dir.glob("*.csv")):
        # Read the raw file bytes for hashing
        data: bytes = csv_path.read_bytes()
        # Compute the SHA-256 digest of the file bytes
        digest: str = hashlib.sha256(data).hexdigest()
        # Append the hash record for this file
        records.append({"ticker": csv_path.stem, "sha256": digest})
    # Build the manifest frame from the records
    manifest: pd.DataFrame = pd.DataFrame(records)
    # Persist the manifest to the destination path
    manifest.to_csv(manifest_path, index=False)
    # Return the manifest frame
    return manifest


def construct_us_universe(
    config: Dict[str, Any],
    metadata_path: Path,
    raw_dir: Path,
) -> Dict[str, Any]:
    """
    Run the full U.S. universe construction pipeline.

    Parameters
    ----------
    config : Dict[str, Any]
        The configuration dictionary.
    metadata_path : Path
        The versioned U.S. metadata CSV path.
    raw_dir : Path
        The versioned output directory for raw CSVs.

    Returns
    -------
    Dict[str, Any]
        A result block with the final ticker list and manifest path.
    """
    # Build the US and JP market specifications
    us_spec, _ = build_market_specs(config)
    # Load the U.S. universe metadata
    metadata: pd.DataFrame = load_universe_metadata(metadata_path)
    # Apply the pre-download metadata filters
    candidates: pd.DataFrame = filter_metadata_universe(
        metadata, us_spec, config
    )
    # Download the price histories for the candidate tickers
    written: Dict[str, Path] = download_price_histories(
        sorted(candidates.index.tolist()), config, raw_dir
    )
    # Compute the per-ticker history statistics restricted to the
    # candidate set so stale files cannot pollute the statistics
    stats: pd.DataFrame = compute_history_stats(
        raw_dir, config, market_name="US",
        ticker_list=sorted(candidates.index.tolist()),
    )
    # Apply the deterministic fallback to reach the target count; the
    # callable applies the history filters internally with the mandated
    # 1% coverage-relaxation loop on under-count
    tickers: List[str] = select_target_tickers(
        stats, metadata, us_spec, config
    )
    # Log the tickers that failed to download for the decision log
    missing: List[str] = [
        t for t in candidates.index if t not in written
    ]
    # Record the missing-ticker diagnostics when any exist
    if missing:
        # Log the failed downloads for the audit trail
        logger.warning(
            "US downloads missing for %d tickers: %s",
            len(missing), missing[:10],
        )
    # Verify the final count against the target when achievable
    if len(tickers) != us_spec.target_count:
        # Log the count discrepancy for the decision log
        logger.warning(
            "US ticker count %d differs from target %d",
            len(tickers), us_spec.target_count,
        )
    # Build the manifest path from the raw directory
    manifest_path: Path = raw_dir / "manifest.csv"
    # Hash every raw file and persist the manifest
    manifest: pd.DataFrame = hash_raw_files(raw_dir, manifest_path)
    # Return the U.S. result block
    return {
        "market": "US",
        "tickers": tickers,
        "manifest_path": str(manifest_path),
        "manifest_rows": len(manifest),
    }


# -----------------------------------------------------------------------
# Task 2, Step 3: Download the Japanese TSE Prime Panel
# -----------------------------------------------------------------------


def construct_jp_panel(
    config: Dict[str, Any],
    metadata_path: Path,
    raw_dir: Path,
) -> Dict[str, Any]:
    """
    Run the Japanese TSE Prime panel construction pipeline.

    Uses the identical pipeline as the U.S. panel with the documented
    exception that the S&P 1500 membership filter is not applied.

    Parameters
    ----------
    config : Dict[str, Any]
        The configuration dictionary.
    metadata_path : Path
        The versioned Japanese metadata CSV path.
    raw_dir : Path
        The versioned output directory for raw CSVs.

    Returns
    -------
    Dict[str, Any]
        A result block with the final ticker list and manifest path.
    """
    # Build the US and JP market specifications
    _, jp_spec = build_market_specs(config)
    # Load the Japanese universe metadata
    metadata: pd.DataFrame = load_universe_metadata(metadata_path)
    # Apply the pre-download metadata filters (no S&P 1500 filter)
    candidates: pd.DataFrame = filter_metadata_universe(
        metadata, jp_spec, config
    )
    # Download the price histories for the candidate tickers
    written: Dict[str, Path] = download_price_histories(
        sorted(candidates.index.tolist()), config, raw_dir
    )
    # Compute the per-ticker history statistics restricted to the
    # candidate set so stale files cannot pollute the statistics
    stats: pd.DataFrame = compute_history_stats(
        raw_dir, config, market_name="JP",
        ticker_list=sorted(candidates.index.tolist()),
    )
    # Apply the deterministic fallback to reach the target count; the
    # callable applies the history filters internally with the mandated
    # 1% coverage-relaxation loop on under-count
    tickers: List[str] = select_target_tickers(
        stats, metadata, jp_spec, config
    )
    # Log the tickers that failed to download for the decision log
    missing: List[str] = [
        t for t in candidates.index if t not in written
    ]
    # Record the missing-ticker diagnostics when any exist
    if missing:
        # Log the failed downloads for the audit trail
        logger.warning(
            "JP downloads missing for %d tickers: %s",
            len(missing), missing[:10],
        )
    # Verify the final count against the target when achievable
    if len(tickers) != jp_spec.target_count:
        # Log the count discrepancy for the decision log
        logger.warning(
            "JP ticker count %d differs from target %d",
            len(tickers), jp_spec.target_count,
        )
    # Build the manifest path from the raw directory
    manifest_path: Path = raw_dir / "manifest.csv"
    # Hash every raw file and persist the manifest
    manifest: pd.DataFrame = hash_raw_files(raw_dir, manifest_path)
    # Return the Japanese result block
    return {
        "market": "JP",
        "tickers": tickers,
        "manifest_path": str(manifest_path),
        "manifest_rows": len(manifest),
    }


# -----------------------------------------------------------------------
# Task 2, Orchestrator Function
# -----------------------------------------------------------------------


@dataclass(frozen=True)
class AcquisitionReport:
    """
    Immutable aggregate report of the data-acquisition pipeline.

    Attributes
    ----------
    environment : EnvironmentReport
        The deterministic-environment setup report.
    us_result : Dict[str, Any]
        The U.S. universe result block.
    jp_result : Dict[str, Any]
        The Japanese panel result block.
    """

    # Environment setup report
    environment: EnvironmentReport
    # U.S. universe result block; None when the US market is skipped
    us_result: Optional[Dict[str, Any]]
    # Japanese panel result block; None when the JP market is skipped
    jp_result: Optional[Dict[str, Any]]


def orchestrate_data_acquisition(
    config: Dict[str, Any],
    us_metadata_path: Path,
    jp_metadata_path: Path,
    output_root: Path,
    markets: Tuple[str, ...] = ("US", "JP"),
) -> AcquisitionReport:
    """
    Orchestrate the full data-acquisition pipeline for both markets.

    Composes orchestrate_environment_setup, construct_us_universe, and
    construct_jp_panel in the mandated order, raising on fatal
    environment drift before any download begins. The markets selector
    allows a single-market acquisition so chains never download data
    they do not need.

    Parameters
    ----------
    config : Dict[str, Any]
        The configuration dictionary.
    us_metadata_path : Path
        The U.S. universe metadata CSV path.
    jp_metadata_path : Path
        The Japanese TSE Prime metadata CSV path.
    output_root : Path
        The root directory for versioned raw-data artifacts.
    markets : Tuple[str, ...], optional
        The markets to acquire, a subset of ("US", "JP"), default both.

    Returns
    -------
    AcquisitionReport
        The immutable aggregate acquisition report.

    Raises
    ------
    ValueError
        If the environment has fatal drift or a metadata file is absent.
    """
    # Validate that the config is a dictionary
    if not isinstance(config, dict):
        # Raise an informative error for the invalid config type
        raise TypeError(
            f"config must be a Dict[str, Any]; received "
            f"{type(config).__name__}."
        )
    # Validate the markets selector against the supported vocabulary
    for market in markets:
        # Verify each requested market identifier
        if market not in ("US", "JP"):
            # Raise an informative error for the unknown market
            raise ValueError(
                f"Unknown market '{market}'; expected 'US' or 'JP'."
            )
    # Step 1: run the deterministic environment setup
    env_report: EnvironmentReport = orchestrate_environment_setup(config)
    # Fail loudly on fatal environment drift before any download
    if env_report.fatal_drift:
        # Raise an informative error naming the drifted libraries
        raise ValueError(
            f"Fatal environment drift: {env_report.fatal_drift}. "
            f"Pin the environment before downloading data."
        )
    # Verify the metadata paths of the requested markets; a non-empty
    # path string is required so a missing data_paths contract fails
    # with an informative error instead of a Path('') lookup
    if "US" in markets:
        # Check the U.S. metadata path is not empty; Path('') stringifies
        # to '.' so both representations count as an absent contract entry
        if str(us_metadata_path) in ("", "."):
            # Raise an informative error for the empty metadata path
            raise ValueError(
                "us_metadata_path must be a non-empty path; the config "
                "data_paths contract is missing its U.S. metadata entry."
            )
        # Check the U.S. metadata file existence
        if not us_metadata_path.is_file():
            # Raise an informative error for the missing metadata file
            raise FileNotFoundError(
                f"Metadata file missing: {us_metadata_path}"
            )
    # Verify the Japanese metadata path when Japan is requested
    if "JP" in markets:
        # Check the Japanese metadata path is not empty; Path('') and
        # Path('.') both indicate an absent data_paths contract entry
        if str(jp_metadata_path) in ("", "."):
            # Raise an informative error for the empty metadata path
            raise ValueError(
                "jp_metadata_path must be a non-empty path; the config "
                "data_paths contract is missing its Japanese metadata "
                "entry."
            )
        # Check the Japanese metadata file existence
        if not jp_metadata_path.is_file():
            # Raise an informative error for the missing metadata file
            raise FileNotFoundError(
                f"Metadata file missing: {jp_metadata_path}"
            )
    # Build the market-specific raw-data subdirectories
    us_raw_dir: Path = output_root / "us_raw"
    # Build the Japanese raw-data subdirectory
    jp_raw_dir: Path = output_root / "jp_raw"
    # Initialize the result blocks as skipped
    us_result: Optional[Dict[str, Any]] = None
    # Initialize the Japanese result block as skipped
    jp_result: Optional[Dict[str, Any]] = None
    # Step 2: construct the U.S. universe when requested
    if "US" in markets:
        # Construct the U.S. universe and capture the result block
        us_result = construct_us_universe(
            config, us_metadata_path, us_raw_dir
        )
    # Step 3: construct the Japanese TSE Prime panel when requested
    if "JP" in markets:
        # Construct the Japanese panel and capture the result block
        jp_result = construct_jp_panel(
            config, jp_metadata_path, jp_raw_dir
        )
    # Assemble the immutable acquisition report
    report: AcquisitionReport = AcquisitionReport(
        environment=env_report,
        us_result=us_result,
        jp_result=jp_result,
    )
    # Return the aggregate acquisition report
    return report
````

````python
# Task 3: Cleanse Raw Price Data and Construct Aligned Return Panels

# ==============================================================================
# Task 3: Cleanse Raw Price Data and Construct Aligned Return Panels
# ==============================================================================

# -----------------------------------------------------------------------
# Task 3, Step 1: Per-Ticker Cleansing and Log-Return Calculation
# -----------------------------------------------------------------------

def cleanse_ticker_frame(
    frame: pd.DataFrame,
    ticker: str,
) -> pd.DataFrame:
    """
    Cleanse a single ticker's raw price frame.

    Applies the manuscript cleansing rules in order: parse dates, drop
    missing or non-positive adjusted closes, drop duplicate dates keeping
    the latest bar, sort strictly by date, compute within-ticker log
    returns, and drop the first observation per ticker.

    Parameters
    ----------
    frame : pd.DataFrame
        The raw price frame with date and adj_close columns.
    ticker : str
        The ticker identifier for logging and error messages.

    Returns
    -------
    pd.DataFrame
        The cleansed frame with columns date, adj_close, and return.

    Raises
    ------
    ValueError
        If required columns are missing or the frame is empty.
    """
    # Verify that the required columns exist in the frame
    for column in ("date", "adj_close"):
        # Check the column presence in the frame
        if column not in frame.columns:
            # Raise an informative error for the missing column
            raise ValueError(
                f"Ticker {ticker} frame missing column '{column}'; "
                f"found {list(frame.columns)}."
            )
    # Verify that the frame is not empty
    if frame.shape[0] == 0:
        # Raise an informative error for the empty frame
        raise ValueError(f"Ticker {ticker} frame is empty.")
    # Verify the adjusted-close column is numeric before any arithmetic,
    # so a string column produces an informative error, not a numpy crash
    if not pd.api.types.is_numeric_dtype(frame["adj_close"]):
        # Raise an informative error naming the offending ticker
        raise ValueError(
            f"Ticker {ticker} adj_close column is not numeric; dtype is "
            f"{frame['adj_close'].dtype}."
        )
    # Copy the frame to avoid mutating the caller's object
    clean: pd.DataFrame = frame.copy()
    # Parse the date column with the documented format
    clean["date"] = pd.to_datetime(
        clean["date"], format="%Y-%m-%d", errors="coerce"
    )
    # Drop rows with unparseable dates
    clean = clean.dropna(subset=["date"])
    # Drop rows with missing or non-positive adjusted closes
    clean = clean.loc[clean["adj_close"].notna() & (clean["adj_close"] > 0)]
    # Sort the frame strictly by date
    clean = clean.sort_values("date")
    # Drop duplicate trading dates, keeping the latest bar
    clean = clean.drop_duplicates(subset=["date"], keep="last")
    # Compute the within-ticker log return: r_t = ln(P_t / P_{t-1})
    clean["return"] = np.log(
        clean["adj_close"] / clean["adj_close"].shift(1)
    )
    # Drop the first observation because no previous price exists
    clean = clean.dropna(subset=["return"])
    # Keep only the identity and return columns
    clean = clean[["date", "adj_close", "return"]]
    # Return the cleansed per-ticker frame
    return clean


def cleanse_all_tickers(raw_dir: Path) -> pd.DataFrame:
    """
    Cleanse every raw CSV in a directory into a long return panel.

    Parameters
    ----------
    raw_dir : Path
        The directory of raw per-ticker CSV files.

    Returns
    -------
    pd.DataFrame
        The long panel with columns ticker, date, adj_close, return.

    Raises
    ------
    ValueError
        If the directory contains no CSV files.
    """
    # Verify that the raw directory exists
    if not raw_dir.is_dir():
        # Raise an informative error for the missing directory
        raise ValueError(f"Raw directory does not exist: {raw_dir}")
    # Collect the sorted CSV file list from the directory
    csv_files: List[Path] = sorted(raw_dir.glob("*.csv"))
    # Verify that at least one CSV file exists
    if not csv_files:
        # Raise an informative error for the empty directory
        raise ValueError(f"No CSV files found in {raw_dir}")
    # Initialize the per-ticker cleansed frame list
    frames: List[pd.DataFrame] = []
    # Iterate over every CSV file in the directory
    for csv_path in csv_files:
        # Derive the ticker from the file stem
        ticker: str = csv_path.stem
        # Read the raw CSV into a DataFrame
        raw: pd.DataFrame = pd.read_csv(csv_path)
        # Cleanse the ticker frame with the cleansing rules
        clean: pd.DataFrame = cleanse_ticker_frame(raw, ticker)
        # Attach the ticker identity column to the cleansed frame
        clean["ticker"] = ticker
        # Append the cleansed frame to the list
        frames.append(clean)
    # Concatenate every per-ticker frame into the long panel
    panel: pd.DataFrame = pd.concat(frames, ignore_index=True)
    # Log the total number of cleansed rows
    logger.info("Cleansed %d rows across %d tickers", len(panel), len(frames))
    # Return the long cleansed return panel
    return panel


def flag_extreme_returns(
    panel: pd.DataFrame,
    threshold: float = 1.0,
) -> pd.DataFrame:
    """
    Flag returns whose absolute value exceeds the inspection threshold.

    Parameters
    ----------
    panel : pd.DataFrame
        The long return panel with a return column.
    threshold : float, optional
        The absolute-return threshold for flagging, default 1.0.

    Returns
    -------
    pd.DataFrame
        The subset of rows flagged for manual inspection.
    """
    # Validate the threshold is a positive finite number
    if not isinstance(threshold, (int, float)) or threshold <= 0:
        # Raise an informative error for the invalid threshold
        raise ValueError(
            f"threshold must be a positive number; got {threshold!r}"
        )
    # Build the boolean mask of extreme returns
    extreme_mask: pd.Series = panel["return"].abs() > threshold
    # Return the flagged subset of the panel
    return panel.loc[extreme_mask].copy()


def remove_confirmed_artifacts(
    panel: pd.DataFrame,
    artifact_drops: Optional[Set[Tuple[str, pd.Timestamp]]] = None,
    decision_log_path: Optional[Path] = None,
) -> pd.DataFrame:
    """
    Remove observations confirmed as corporate-action artifacts.

    Implements the manual-inspection rule of the task list: returns with
    absolute value greater than 1.0 are flagged for inspection; once an
    observation is confirmed as an artifact, it is removed and the
    removal is recorded in the decision log.

    Parameters
    ----------
    panel : pd.DataFrame
        The long return panel with ticker and date columns.
    artifact_drops : Optional[Set[Tuple[str, pd.Timestamp]]], optional
        The set of (ticker, date) pairs confirmed as artifacts.
    decision_log_path : Optional[Path], optional
        Destination file for the removal records; appended when given.

    Returns
    -------
    pd.DataFrame
        The panel with the confirmed artifact rows removed.
    """
    # Return the panel unchanged when no artifacts were confirmed
    if not artifact_drops:
        # Return the original panel without modification
        return panel
    # Build the normalized set of confirmed drop keys
    drop_keys: Set[Tuple[str, pd.Timestamp]] = {
        (ticker, pd.Timestamp(date))
        for ticker, date in artifact_drops
    }
    # Build the row keys of the panel for the membership test
    row_keys: pd.DataFrame = pd.DataFrame(
        {
            "ticker": panel["ticker"].astype(str),
            "date": pd.to_datetime(panel["date"]),
        }
    )
    # Build the boolean mask of confirmed artifact rows
    artifact_mask: pd.Series = pd.Series(
        [
            (ticker, date) in drop_keys
            for ticker, date in zip(row_keys["ticker"], row_keys["date"])
        ],
        index=panel.index,
    )
    # Count the removed rows for the decision log
    n_removed: int = int(artifact_mask.sum())
    # Log the removal count for the audit trail
    logger.warning(
        "Removed %d confirmed corporate-action artifacts", n_removed
    )
    # Record the removals in the decision log when a path is supplied
    if decision_log_path is not None and n_removed > 0:
        # Ensure the parent directory exists before appending
        decision_log_path.parent.mkdir(parents=True, exist_ok=True)
        # Append every removal record to the decision log
        with decision_log_path.open("a", encoding="utf-8") as handle:
            # Iterate over the confirmed drop keys
            for ticker, date in sorted(drop_keys):
                # Write the removal record with the reason
                handle.write(
                    f"ARTIFACT_REMOVAL ticker={ticker} date={date} "
                    f"reason=corporate-action\n"
                )
    # Return the panel with the artifact rows removed
    return panel.loc[~artifact_mask].copy()


# -----------------------------------------------------------------------
# Task 3, Step 2: Construct the Common Trading Calendar and Market Return
# -----------------------------------------------------------------------


def build_trading_calendar(
    panel: pd.DataFrame,
    market_name: str,
) -> pd.DatetimeIndex:
    """
    Build the common trading calendar for a market.

    Prefers the exchange calendar when pandas_market_calendars is
    installed; otherwise falls back to the union of observed trading
    days in the panel.

    Parameters
    ----------
    panel : pd.DataFrame
        The long return panel with a date column.
    market_name : str
        The market name, one of "US" or "JP".

    Returns
    -------
    pd.DatetimeIndex
        The sorted unique trading-day calendar.
    """
    # Build the union of observed trading days on which at least one
    # constituent has a valid observation; this is the task's definition
    # of the common calendar and is always computed
    observed: pd.DatetimeIndex = pd.DatetimeIndex(
        sorted(panel["date"].unique())
    ).normalize()
    # Use the exchange calendar library when it is available
    if mcal is not None:
        # Select the exchange calendar by market name
        exchange: str = "NYSE" if market_name == "US" else "TSE"
        # Guard the calendar lookup so a provider rename cannot crash
        try:
            # Get the exchange calendar object
            calendar = mcal.get_calendar(exchange)
            # Compute the sample start date from the panel
            start: pd.Timestamp = panel["date"].min()
            # Compute the sample end date from the panel
            end: pd.Timestamp = panel["date"].max()
            # Fetch the scheduled trading days in the sample range
            schedule = calendar.schedule(
                start_date=start, end_date=end
            )
            # Extract the trading-day index from the schedule
            cal_index: pd.DatetimeIndex = schedule.index
            # Intersect the exchange schedule with the observed union so
            # scheduled days without any constituent data are excluded
            common: pd.DatetimeIndex = cal_index.intersection(observed)
            # Log the calendar-source reconciliation
            logger.info(
                "Using %s exchange calendar intersected with observed "
                "days (%d scheduled, %d common)",
                exchange, len(cal_index), len(common),
            )
            # Return the reconciled calendar index
            return common
        except Exception as exc:
            # Log the calendar lookup failure and fall through
            logger.warning(
                "Exchange calendar lookup failed (%s); using observed "
                "dates", exc,
            )
    # Log the fallback to the observed-date union
    logger.warning(
        "pandas_market_calendars unavailable; using observed dates"
    )
    # Fall back to the union of observed trading days
    return observed


def align_returns_to_calendar(
    panel: pd.DataFrame,
    calendar: pd.DatetimeIndex,
) -> pd.DataFrame:
    """
    Reindex the return panel into a wide matrix on the common calendar.

    Parameters
    ----------
    panel : pd.DataFrame
        The long return panel with ticker, date, and return columns.
    calendar : pd.DatetimeIndex
        The common trading calendar for the market.

    Returns
    -------
    pd.DataFrame
        The wide return matrix with rows indexed by date and columns
        indexed by ticker.
    """
    # Pivot the long panel into a wide matrix
    wide: pd.DataFrame = panel.pivot(
        index="date", columns="ticker", values="return"
    )
    # Reindex the wide matrix to the common calendar
    wide = wide.reindex(calendar)
    # Sort the columns deterministically by ticker
    wide = wide[sorted(wide.columns)]
    # Return the aligned wide return matrix
    return wide


def compute_market_return(wide: pd.DataFrame) -> pd.Series:
    """
    Compute the equal-weighted market return series.

    Implements r_{m,t} = (1/N_t) sum_i r_{i,t} as the row-wise mean over
    the cross-section, where N_t is the count of non-missing returns.

    Parameters
    ----------
    wide : pd.DataFrame
        The wide return matrix indexed by trading day.

    Returns
    -------
    pd.Series
        The market return series indexed by trading day.
    """
    # Compute the row-wise mean ignoring missing returns
    market: pd.Series = wide.mean(axis=1, skipna=True)
    # Return the market return series
    return market


# -----------------------------------------------------------------------
# Task 3, Step 3: Apply Universe Filters and Persist the Panel Manifest
# -----------------------------------------------------------------------


def apply_universe_filters(
    wide: pd.DataFrame,
    metadata: pd.DataFrame,
    config: Dict[str, Any],
    market_name: str = "US",
) -> List[str]:
    """
    Apply the final universe filters to the aligned panel tickers.

    Parameters
    ----------
    wide : pd.DataFrame
        The wide return matrix indexed by ticker columns.
    metadata : pd.DataFrame
        The ticker metadata frame indexed by ticker.
    config : Dict[str, Any]
        The configuration dictionary.
    market_name : str, optional
        The market name, one of "US" or "JP", default "US"; selects the
        market-specific inclusion criteria.

    Returns
    -------
    List[str]
        The final sorted ticker list after all filters.
    """
    # Select the inclusion criteria by market name
    criteria_key: str = (
        "us_inclusion_criteria" if market_name == "US"
        else "jp_inclusion_criteria"
    )
    # Extract the inclusion criteria from the config
    criteria: Dict[str, Any] = config.get(
        "data_acquisition_and_universe", {}
    ).get(criteria_key, {})
    # Compute the per-ticker non-missing coverage on the wide matrix
    coverage: pd.Series = wide.notna().mean(axis=0)
    # Compute the per-ticker history span in years
    span: pd.Series = pd.Series(
        {
            ticker: (
                wide[ticker].dropna().index.max()
                - wide[ticker].dropna().index.min()
            ).days / 365.25
            for ticker in wide.columns
        }
    )
    # Extract the minimum history length in years
    min_years: float = float(criteria.get("min_history_years", 6))
    # Extract the minimum coverage target
    min_coverage: float = float(
        criteria.get("full_period_coverage_target", 0.92)
    )
    # Initialize a boolean mask that keeps every ticker initially
    mask: pd.Series = pd.Series(True, index=wide.columns)
    # Apply the history-span filter
    mask &= span >= min_years
    # Apply the coverage filter
    mask &= coverage >= min_coverage
    # Apply the metadata membership filter when metadata is supplied
    if metadata is not None and not metadata.empty:
        # Keep only tickers present in the metadata frame
        mask &= mask.index.isin(metadata.index)
        # Re-apply the listing and common-stock flags so pre-existing raw
        # data cannot bypass the pre-download filters of Task 2
        if "still_listed" in metadata.columns:
            # Restrict to tickers still listed at the sample end
            mask &= mask.index.isin(
                metadata.index[metadata["still_listed"].astype(bool)]
            )
        # Re-apply the common-stock flag when the column is present
        if "is_common_stock" in metadata.columns:
            # Restrict to common stocks per the universe rule
            mask &= mask.index.isin(
                metadata.index[metadata["is_common_stock"].astype(bool)]
            )
    # Return the final sorted ticker list
    return sorted(wide.columns[mask].tolist())


def audit_coverage(
    tickers: List[str],
    metadata: pd.DataFrame,
    config: Dict[str, Any],
    allow_skip: bool = False,
) -> List[str]:
    """
    Audit the coverage constraints of the final ticker list.

    Parameters
    ----------
    tickers : List[str]
        The final ticker list.
    metadata : pd.DataFrame
        The ticker metadata frame indexed by ticker.
    config : Dict[str, Any]
        The configuration dictionary.
    allow_skip : bool, optional
        When False (default), a missing metadata frame raises instead of
        silently skipping the S&P 1500 / sector / cap gates.

    Returns
    -------
    List[str]
        A list of violated coverage constraints; empty when all pass.

    Raises
    ------
    ValueError
        If the metadata frame lacks required columns, or is absent while
        allow_skip is False.
    """
    # Initialize the violation list
    violations: List[str] = []
    # Handle the absent-metadata case: fail loudly unless explicitly
    # allowed, so the universe-quality gates can never be silently skipped
    if metadata is None or metadata.empty:
        # Check whether the skip was explicitly authorized
        if not allow_skip:
            # Raise an informative error for the missing metadata
            raise ValueError(
                "Coverage audit requires the ticker metadata frame; "
                "the S&P 1500 / sector / cap gates cannot be verified "
                "without it. Pass the metadata or set allow_skip=True "
                "only when the skip is documented."
            )
        # Log the skipped coverage audit
        logger.warning(
            "Coverage audit skipped: no metadata supplied"
        )
        # Return an empty violation list for the skipped audit
        return violations
    # Extract the inclusion criteria from the config
    criteria: Dict[str, Any] = config.get(
        "data_acquisition_and_universe", {}
    ).get("us_inclusion_criteria", {})
    # Extract the S&P 1500 coverage target
    sp1500_target: float = float(
        criteria.get("sp1500_coverage_target", 0.85)
    )
    # Subset the metadata to the final ticker list
    sub: pd.DataFrame = metadata.loc[
        metadata.index.isin(tickers)
    ]
    # Compute the S&P 1500 fraction of the final list
    sp1500_fraction: float = (
        sub["sp1500_member_current"].astype(bool).mean()
        if len(sub) > 0 else 0.0
    )
    # Check the S&P 1500 coverage target
    if sp1500_fraction < sp1500_target:
        # Record the S&P 1500 coverage violation
        violations.append(
            f"S&P 1500 coverage {sp1500_fraction:.3f} below target "
            f"{sp1500_target}"
        )
    # Count the distinct GICS sectors in the final list
    sector_count: int = int(sub["gics_sector"].nunique())
    # Check the eleven-sector coverage
    if sector_count < 11:
        # Record the sector coverage violation
        violations.append(
            f"Sector count {sector_count} below the required 11"
        )
    # Count the distinct market-cap buckets in the final list
    cap_buckets: set = set(sub["market_cap_bucket"].dropna().unique())
    # Check the large-/mid-/small-cap representation
    if not {"large", "mid", "small"}.issubset(cap_buckets):
        # Record the cap-bucket coverage violation
        violations.append(
            f"Cap buckets {cap_buckets} missing large/mid/small"
        )
    # Return the list of coverage violations
    return violations


def persist_panel(
    wide: pd.DataFrame,
    market_return: pd.Series,
    tickers: List[str],
    metadata: pd.DataFrame,
    out_path: Path,
) -> Dict[str, str]:
    """
    Persist the aligned panel, ticker manifest, and market return.

    Parameters
    ----------
    wide : pd.DataFrame
        The wide return matrix.
    market_return : pd.Series
        The market return series.
    tickers : List[str]
        The final ticker list.
    metadata : pd.DataFrame
        The ticker metadata frame.
    out_path : Path
        The destination directory for the persisted artifacts.

    Returns
    -------
    Dict[str, str]
        A mapping of artifact name to persisted file path.
    """
    # Ensure the output directory exists
    out_path.mkdir(parents=True, exist_ok=True)
    # Reindex the wide matrix to the final ticker list
    wide_final: pd.DataFrame = wide[sorted(tickers)]
    # Build the panel file path
    panel_path: Path = out_path / "aligned_panel.parquet"
    # Persist the wide matrix as parquet
    wide_final.to_parquet(panel_path, index=True)
    # Build the market return file path
    market_path: Path = out_path / "market_return.csv"
    # Persist the market return series as CSV
    market_return.to_csv(market_path, header=True)
    # Build the ticker manifest file path
    manifest_path: Path = out_path / "ticker_manifest.csv"
    # Persist the metadata subset when metadata is supplied
    if metadata is not None and not metadata.empty:
        # Subset the metadata to the final tickers and persist it
        metadata.loc[metadata.index.isin(tickers)].to_csv(manifest_path)
    else:
        # Persist a ticker-only manifest when metadata is absent
        pd.DataFrame({"ticker": sorted(tickers)}).to_csv(
            manifest_path, index=False
        )
    # Compute the SHA-256 hash of the panel file
    panel_digest: str = hashlib.sha256(
        panel_path.read_bytes()
    ).hexdigest()
    # Build the artifact mapping with the panel hash
    artifacts: Dict[str, str] = {
        "panel": str(panel_path),
        "market_return": str(market_path),
        "ticker_manifest": str(manifest_path),
        "panel_sha256": panel_digest,
    }
    # Return the artifact mapping
    return artifacts


# -----------------------------------------------------------------------
# Task 3, Orchestrator Function
# -----------------------------------------------------------------------


@dataclass(frozen=True)
class AlignedPanelReport:
    """
    Immutable aggregate report of the panel-alignment pipeline.

    Attributes
    ----------
    wide : pd.DataFrame
        The wide return matrix indexed by trading day.
    market_return : pd.Series
        The equal-weighted market return series.
    tickers : List[str]
        The final sorted ticker list.
    artifacts : Dict[str, str]
        Mapping of persisted artifact names to file paths.
    """

    # Wide return matrix indexed by trading day
    wide: pd.DataFrame
    # Equal-weighted market return series
    market_return: pd.Series
    # Final sorted ticker list
    tickers: List[str]
    # Mapping of persisted artifact names to file paths
    artifacts: Dict[str, str]


def cleanse_and_align_panel(
    raw_dir: Path,
    market_name: str,
    config: Dict[str, Any],
    metadata: Optional[pd.DataFrame] = None,
    out_dir: Optional[Path] = None,
    artifact_drops: Optional[Set[Tuple[str, pd.Timestamp]]] = None,
    decision_log_path: Optional[Path] = None,
) -> AlignedPanelReport:
    """
    Orchestrate the full cleansing and panel-alignment pipeline.

    Composes cleanse_all_tickers, flag_extreme_returns,
    remove_confirmed_artifacts, build_trading_calendar,
    align_returns_to_calendar, compute_market_return,
    apply_universe_filters, audit_coverage, and persist_panel in the
    mandated order.

    Parameters
    ----------
    raw_dir : Path
        The directory of raw per-ticker CSV files.
    market_name : str
        The market name, one of "US" or "JP".
    config : Dict[str, Any]
        The configuration dictionary.
    metadata : Optional[pd.DataFrame], optional
        The ticker metadata frame for universe filters; required for the
        coverage audit unless explicitly skipped downstream.
    out_dir : Optional[Path], optional
        The destination directory for persisted artifacts.
    artifact_drops : Optional[Set[Tuple[str, pd.Timestamp]]], optional
        Confirmed corporate-action artifacts to remove before alignment.
    decision_log_path : Optional[Path], optional
        Destination file for the artifact-removal records.

    Returns
    -------
    AlignedPanelReport
        The immutable aggregate panel report.

    Raises
    ------
    ValueError
        If the coverage audit finds violations or the metadata is absent.
    """
    # Step 1: cleanse every raw CSV into the long return panel
    panel: pd.DataFrame = cleanse_all_tickers(raw_dir)
    # Flag the extreme returns for manual inspection
    extreme: pd.DataFrame = flag_extreme_returns(panel, threshold=1.0)
    # Log the count of flagged extreme returns
    logger.info("Flagged %d extreme returns for inspection", len(extreme))
    # Remove the confirmed corporate-action artifacts before alignment
    panel = remove_confirmed_artifacts(
        panel, artifact_drops, decision_log_path
    )
    # Step 2: build the common trading calendar for the market
    calendar: pd.DatetimeIndex = build_trading_calendar(
        panel, market_name
    )
    # Align the long panel into the wide return matrix
    wide: pd.DataFrame = align_returns_to_calendar(panel, calendar)
    # Compute the equal-weighted market return series
    market_return: pd.Series = compute_market_return(wide)
    # Step 3: apply the universe filters to the ticker set
    tickers: List[str] = apply_universe_filters(
        wide, metadata, config, market_name=market_name
    )
    # Audit the coverage constraints of the final ticker list
    violations: List[str] = audit_coverage(
        tickers, metadata, config
    )
    # Fail loudly when any coverage constraint is violated
    if violations:
        # Raise an informative error listing the violations
        raise ValueError(
            f"Coverage audit failed: {'; '.join(violations)}"
        )
    # Persist the panel artifacts when an output directory is given
    artifacts: Dict[str, str] = {}
    # Check whether the output directory was supplied
    if out_dir is not None:
        # Persist the aligned panel and manifest
        artifacts = persist_panel(
            wide, market_return, tickers, metadata, out_dir
        )
    # Assemble the immutable panel report
    report: AlignedPanelReport = AlignedPanelReport(
        wide=wide,
        market_return=market_return,
        tickers=tickers,
        artifacts=artifacts,
    )
    # Return the aggregate panel report
    return report
````

````python
# Task 4: Construct Stock Features, Regime Variables, and Forward Targets

# ==============================================================================
# Task 4: Construct Stock Features, Regime Variables, and Forward Targets
# ==============================================================================

# -----------------------------------------------------------------------
# Task 4, Step 1: Rolling Realized Volatility and Cumulative Returns
# -----------------------------------------------------------------------

def melt_wide_to_long(wide: pd.DataFrame) -> pd.DataFrame:
    """
    Melt the wide return matrix into the long return panel.

    Parameters
    ----------
    wide : pd.DataFrame
        The wide return matrix indexed by date with ticker columns.

    Returns
    -------
    pd.DataFrame
        The long panel with ticker, date, and return columns.
    """
    # Reset the date index into a column for the melt
    wide_reset: pd.DataFrame = wide.reset_index()
    # Rename the date index column for the melt
    wide_reset = wide_reset.rename(columns={"index": "date"})
    # Melt the wide matrix into the long panel
    long: pd.DataFrame = wide_reset.melt(
        id_vars="date", var_name="ticker", value_name="return"
    )
    # Drop the rows with missing returns
    long = long.dropna(subset=["return"])
    # Return the long return panel
    return long


def build_rolling_volatility_features(
    panel: pd.DataFrame,
    config: Dict[str, Any],
) -> pd.DataFrame:
    """
    Compute the first five features: RV(5, 20, 60) and CumRet(5, 20).

    Implements RV_{i,t}^{(w)} = sqrt(252) * Std(r_{i,t-w+1..t}) with
    ddof=1 and the all-observations-present rule, and
    CumRet_{i,t}^{(w)} = prod(1 + r) - 1 over the window.

    Parameters
    ----------
    panel : pd.DataFrame
        The long panel with ticker, date, and return columns.
    config : Dict[str, Any]
        The configuration dictionary.

    Returns
    -------
    pd.DataFrame
        The panel augmented with the five feature columns.
    """
    # Extract the realized-volatility windows from the config
    rv_windows: List[int] = config.get(
        "stock_feature_construction", {}
    ).get("realized_volatility_windows", [5, 20, 60])
    # Extract the cumulative-return windows from the config
    cumret_windows: List[int] = config.get(
        "stock_feature_construction", {}
    ).get("cumulative_return_windows", [5, 20])
    # Extract the annualization factor from the config
    annualization: float = float(
        config.get("data_preprocessing_parameters", {}).get(
            "annualization_factor", 252
        )
    )
    # Copy the panel to avoid mutating the caller's object
    out: pd.DataFrame = panel.copy()
    # Sort the panel by ticker and date for deterministic windows
    out = out.sort_values(["ticker", "date"]).reset_index(drop=True)
    # Compute the annualization square root once
    sqrt_ann: float = float(np.sqrt(annualization))
    # Iterate over every realized-volatility window
    for window in rv_windows:
        # Compute the rolling standard deviation within ticker
        rv: pd.Series = (
            out.groupby("ticker")["return"]
            .rolling(window, min_periods=window)
            .std(ddof=1)
        )
        # Reset the group index for aligned assignment
        rv = rv.reset_index(level=0, drop=True)
        # Annualize the rolling standard deviation
        out[f"rv_{window}d"] = rv * sqrt_ann
    # Iterate over every cumulative-return window
    for window in cumret_windows:
        # Define the windowed product-minus-one function
        def prod_plus_one(values: np.ndarray) -> float:
            # Compute the cumulative product of (1 + r) minus one
            return float(np.prod(1.0 + values) - 1.0)
        # Compute the rolling product within ticker
        cumret: pd.Series = (
            out.groupby("ticker")["return"]
            .rolling(window, min_periods=window)
            .apply(prod_plus_one, raw=True)
        )
        # Reset the group index for aligned assignment
        cumret = cumret.reset_index(level=0, drop=True)
        # Assign the cumulative return feature to the panel
        out[f"cumret_{window}d"] = cumret
    # Return the panel augmented with the five features
    return out


# -----------------------------------------------------------------------
# Task 4, Step 2: RSI and Ten Most Recent Returns
# -----------------------------------------------------------------------


def compute_rsi(
    returns: pd.Series,
    period: int,
) -> pd.Series:
    """
    Compute the simple-average 14-day RSI of a return series.

    Implements RSI = 100 - 100 / (1 + AvgGain / AvgLoss), with RSI = 100
    when AvgLoss = 0.

    Parameters
    ----------
    returns : pd.Series
        The return series.
    period : int
        The RSI averaging period.

    Returns
    -------
    pd.Series
        The RSI series aligned with the input.
    """
    # Compute the daily gains as returns clipped at zero
    gains: pd.Series = returns.clip(lower=0.0)
    # Compute the daily losses as the negative of clipped returns
    losses: pd.Series = (-returns).clip(lower=0.0)
    # Compute the simple rolling average of gains
    avg_gain: pd.Series = gains.rolling(period, min_periods=period).mean()
    # Compute the simple rolling average of losses
    avg_loss: pd.Series = losses.rolling(period, min_periods=period).mean()
    # Compute the relative strength ratio with a zero-loss guard
    rs: pd.Series = avg_gain / avg_loss.replace(0.0, np.nan)
    # Compute the RSI from the relative strength
    rsi: pd.Series = 100.0 - 100.0 / (1.0 + rs)
    # Apply the zero-loss rule: RSI = 100 when AvgLoss = 0
    rsi = rsi.where(avg_loss != 0, 100.0)
    # Return the computed RSI series
    return rsi


def build_rsi_and_lag_features(
    panel: pd.DataFrame,
    config: Dict[str, Any],
) -> pd.DataFrame:
    """
    Compute the RSI and the ten most recent return features.

    Parameters
    ----------
    panel : pd.DataFrame
        The long panel with ticker, date, and return columns.
    config : Dict[str, Any]
        The configuration dictionary.

    Returns
    -------
    pd.DataFrame
        The panel augmented with RSI and lagged return columns.
    """
    # Extract the RSI period from the config
    rsi_period: int = int(
        config.get("stock_feature_construction", {}).get("rsi_period", 14)
    )
    # Extract the recent-returns count from the config
    recent_count: int = int(
        config.get("stock_feature_construction", {}).get(
            "recent_returns_count", 10
        )
    )
    # Copy the panel to avoid mutating the caller's object
    out: pd.DataFrame = panel.copy()
    # Compute the RSI within each ticker group
    out["rsi_14d"] = (
        out.groupby("ticker")["return"]
        .transform(lambda s: compute_rsi(s, rsi_period))
    )
    # Iterate over the recent-return lags from zero to count minus one
    for lag in range(recent_count):
        # Name the current return exactly as the config registry does:
        # 'ret_t0' for lag zero (the registry has no dash), and
        # 'ret_t-1' .. 'ret_t-9' for the remaining lags, so the column
        # names match config['stock_feature_construction']['feature_names']
        col_name: str = "ret_t0" if lag == 0 else f"ret_t-{lag}"
        # Compute the lagged return within ticker
        out[col_name] = out.groupby("ticker")["return"].shift(lag)
    # Return the panel augmented with the RSI and lag features
    return out


# -----------------------------------------------------------------------
# Task 4, Step 3: Betas, Residual Returns, Regime Variables, and Target
# -----------------------------------------------------------------------


def compute_rolling_beta(
    stock_returns: pd.Series,
    market_returns: pd.Series,
    window: int,
    min_observations: int,
) -> pd.Series:
    """
    Compute the trailing-window OLS beta for one stock.

    Implements beta = Cov(r_i, r_m) / Var(r_m) over the trailing window
    with the minimum-observation rule.

    Parameters
    ----------
    stock_returns : pd.Series
        The stock return series aligned by date.
    market_returns : pd.Series
        The market return series aligned by date.
    window : int
        The trailing window length.
    min_observations : int
        The minimum number of valid paired observations.

    Returns
    -------
    pd.Series
        The rolling beta series aligned with the input.
    """
    # Combine the two series into a paired frame
    paired: pd.DataFrame = pd.concat(
        [stock_returns.rename("stock"), market_returns.rename("market")],
        axis=1,
    )
    # Compute the rolling covariance with min_periods=1 so partial
    # windows are eligible; the minimum-observation rule is enforced
    # separately by the paired count mask below
    cov: pd.Series = (
        paired["stock"].rolling(window, min_periods=1)
        .cov(paired["market"])
    )
    # Compute the rolling variance of the market with the same rule
    var: pd.Series = (
        paired["market"].rolling(window, min_periods=1).var(ddof=1)
    )
    # Compute the rolling count of paired observations with min_periods=1
    # so the count is defined from the first row of every stock and the
    # mandated relaxation (at least min_observations valid pairs, e.g. 60
    # of 120) can apply to partial windows
    count: pd.Series = (
        paired[["stock", "market"]].notna().all(axis=1)
        .rolling(window, min_periods=1).sum()
    )
    # Compute the beta as the covariance-over-variance ratio
    beta: pd.Series = cov / var.replace(0.0, np.nan)
    # Mask betas computed on too few paired observations
    beta = beta.where(count >= min_observations)
    # Return the rolling beta series
    return beta


def build_beta_and_regime_variables(
    panel: pd.DataFrame,
    market_return: pd.Series,
    config: Dict[str, Any],
) -> pd.DataFrame:
    """
    Compute beta, residuals, market volatility, and idio volatility.

    Parameters
    ----------
    panel : pd.DataFrame
        The long panel with ticker, date, and return columns.
    market_return : pd.Series
        The market return series indexed by date.
    config : Dict[str, Any]
        The configuration dictionary.

    Returns
    -------
    pd.DataFrame
        The panel augmented with beta, residual, and regime columns.
    """
    # Extract the beta window from the config
    beta_window: int = int(
        config.get("regime_state_variable_construction", {}).get(
            "beta_estimation_window", 120
        )
    )
    # Extract the beta minimum observations from the config
    min_obs: int = int(
        config.get("regime_state_variable_construction", {}).get(
            "beta_min_observations", 60
        )
    )
    # Extract the regime volatility window from the config
    vol_window: int = int(
        config.get("regime_state_variable_construction", {}).get(
            "market_volatility_window", 20
        )
    )
    # Extract the annualization factor from the config
    annualization: float = float(
        config.get("data_preprocessing_parameters", {}).get(
            "annualization_factor", 252
        )
    )
    # Compute the annualization square root once
    sqrt_ann: float = float(np.sqrt(annualization))
    # Copy the panel to avoid mutating the caller's object
    out: pd.DataFrame = panel.copy()
    # Merge the market return onto the panel by date
    market_df: pd.DataFrame = market_return.rename("market_return")
    # Join the market return into the panel
    out = out.merge(
        market_df, left_on="date", right_index=True, how="left"
    )
    # Initialize the beta column with missing values
    out["beta"] = np.nan
    # Iterate over every ticker in the panel
    for ticker, group in out.groupby("ticker"):
        # Compute the rolling beta for this ticker
        beta: pd.Series = compute_rolling_beta(
            group["return"], group["market_return"],
            beta_window, min_obs,
        )
        # Assign the beta values to the matching rows
        out.loc[group.index, "beta"] = beta.values
    # Compute the residual return: eps = r_i - beta * r_m
    out["residual_return"] = out["return"] - out["beta"] * out["market_return"]
    # Compute the annualized market volatility ONCE on the market return
    # series: z1_t = sqrt(252) * Std(r_m over the trailing 20 days); the
    # result is shared by every stock on a given trading day, so it is
    # computed on the market series and then mapped onto the panel
    market_vol: pd.Series = (
        market_return.rolling(
            vol_window, min_periods=vol_window
        ).std(ddof=1)
        * sqrt_ann
    )
    # Map the shared market volatility onto the panel by date
    out["market_volatility"] = out["date"].map(market_vol)
    # Compute the annualized idiosyncratic volatility of residuals
    out["idiosyncratic_volatility"] = (
        out.groupby("ticker")["residual_return"]
        .transform(
            lambda s: s.rolling(vol_window, min_periods=vol_window)
            .std(ddof=1)
        )
        * sqrt_ann
    )
    # Return the panel augmented with the regime variables
    return out


def build_forward_target(
    panel: pd.DataFrame,
    config: Dict[str, Any],
) -> pd.DataFrame:
    """
    Compute the five-day forward annualized realized-volatility target.

    Implements y_{i,t} = sqrt(252) * Std(r_{t+1}, ..., r_{t+5}) valid
    only when all five forward returns are non-missing.

    Parameters
    ----------
    panel : pd.DataFrame
        The long panel with ticker, date, and return columns.
    config : Dict[str, Any]
        The configuration dictionary.

    Returns
    -------
    pd.DataFrame
        The panel augmented with the forward target column.
    """
    # Extract the target horizon from the config
    horizon: int = int(
        config.get("target_construction", {}).get("horizon_days", 5)
    )
    # Extract the annualization factor from the config
    annualization: float = float(
        config.get("data_preprocessing_parameters", {}).get(
            "annualization_factor", 252
        )
    )
    # Compute the annualization square root once
    sqrt_ann: float = float(np.sqrt(annualization))
    # Copy the panel to avoid mutating the caller's object
    out: pd.DataFrame = panel.copy()
    # Initialize the list of forward-return columns
    forward_cols: List[str] = []
    # Iterate over the forward horizon days
    for k in range(1, horizon + 1):
        # Compute the k-day forward return within ticker
        col: str = f"fwd_{k}"
        # Assign the forward shift to the panel
        out[col] = out.groupby("ticker")["return"].shift(-k)
        # Record the forward column name
        forward_cols.append(col)
    # Compute the forward standard deviation across the horizon
    fwd_std: pd.Series = out[forward_cols].std(axis=1, ddof=1)
    # Compute the annualized forward realized-volatility target
    out["target"] = fwd_std * sqrt_ann
    # Mask targets where any forward return is missing
    out["target"] = out["target"].where(
        out[forward_cols].notna().all(axis=1)
    )
    # Drop the temporary forward-return columns
    out = out.drop(columns=forward_cols)
    # Return the panel augmented with the target column
    return out


def drop_incomplete_rows(
    panel: pd.DataFrame,
    config: Dict[str, Any],
) -> pd.DataFrame:
    """
    Drop rows with any missing feature, regime variable, or target.

    Parameters
    ----------
    panel : pd.DataFrame
        The long feature panel.
    config : Dict[str, Any]
        The configuration dictionary.

    Returns
    -------
    pd.DataFrame
        The panel with only complete rows.
    """
    # Extract the feature-name registry from the config
    feature_names: List[str] = config.get(
        "stock_feature_construction", {}
    ).get("feature_names", [])
    # Extract the regime-variable names from the config
    regime_names: List[str] = config.get(
        "regime_state_variable_construction", {}
    ).get("regime_variable_names", [
        "market_volatility", "idiosyncratic_volatility"
    ])
    # Build the completeness column list
    required: List[str] = feature_names + regime_names + ["target"]
    # Validate the required column list before masking so a registry
    # mismatch surfaces as an informative error, not a raw KeyError
    missing_columns: List[str] = [
        column for column in required if column not in panel.columns
    ]
    # Raise an informative error when any required column is absent
    if missing_columns:
        # Raise an informative error naming the missing columns
        raise ValueError(
            f"Completeness drop requires columns missing from the panel: "
            f"{missing_columns}"
        )
    # Compute the row-wise completeness mask
    complete: pd.Series = panel[required].notna().all(axis=1)
    # Count the dropped rows for logging
    dropped: int = int((~complete).sum())
    # Log the row-completeness drop count
    logger.info("Dropped %d incomplete rows", dropped)
    # Return the complete-row subset of the panel
    return panel.loc[complete].reset_index(drop=True)


# -----------------------------------------------------------------------
# Task 4, Orchestrator Function
# -----------------------------------------------------------------------


@dataclass(frozen=True)
class FeaturePanelReport:
    """
    Immutable aggregate report of the feature-construction pipeline.

    Attributes
    ----------
    panel : pd.DataFrame
        The final long feature panel.
    feature_columns : List[str]
        The ordered feature column names.
    regime_columns : List[str]
        The regime-variable column names.
    target_column : str
        The target column name.
    artifacts : Dict[str, str]
        Mapping of persisted artifact names to file paths.
    """

    # Final long feature panel
    panel: pd.DataFrame
    # Ordered feature column names
    feature_columns: List[str]
    # Regime-variable column names
    regime_columns: List[str]
    # Target column name
    target_column: str
    # Mapping of persisted artifact names to file paths
    artifacts: Dict[str, str]


def build_features_targets(
    wide: pd.DataFrame,
    market_return: pd.Series,
    config: Dict[str, Any],
    out_dir: Optional[Path] = None,
) -> FeaturePanelReport:
    """
    Orchestrate the full feature, regime, and target construction.

    Composes melt_wide_to_long, build_rolling_volatility_features,
    build_rsi_and_lag_features, build_beta_and_regime_variables,
    build_forward_target, and drop_incomplete_rows in the mandated order.

    Parameters
    ----------
    wide : pd.DataFrame
        The wide return matrix from Task 3.
    market_return : pd.Series
        The market return series from Task 3.
    config : Dict[str, Any]
        The configuration dictionary.
    out_dir : Optional[Path], optional
        The destination directory for the persisted panel.

    Returns
    -------
    FeaturePanelReport
        The immutable aggregate feature-panel report.

    Raises
    ------
    ValueError
        If the feature columns do not match the config registry.
    """
    # Step 1: melt the wide matrix into the long return panel
    panel: pd.DataFrame = melt_wide_to_long(wide)
    # Step 1: compute the rolling volatility and cumulative features
    panel = build_rolling_volatility_features(panel, config)
    # Step 2: compute the RSI and recent-return features
    panel = build_rsi_and_lag_features(panel, config)
    # Step 3: compute the beta, residuals, and regime variables
    panel = build_beta_and_regime_variables(panel, market_return, config)
    # Step 3: compute the forward target
    panel = build_forward_target(panel, config)
    # Step 3: drop rows with any missing feature, regime, or target
    panel = drop_incomplete_rows(panel, config)
    # Extract the feature-name registry from the config
    feature_names: List[str] = config.get(
        "stock_feature_construction", {}
    ).get("feature_names", [])
    # Verify that every registry feature exists in the panel
    missing_features: List[str] = [
        name for name in feature_names if name not in panel.columns
    ]
    # Fail loudly when any registry feature is missing
    if missing_features:
        # Raise an informative error naming the missing features
        raise ValueError(
            f"Feature registry mismatch; missing {missing_features}"
        )
    # Extract the regime-variable names from the config
    regime_names: List[str] = config.get(
        "regime_state_variable_construction", {}
    ).get("regime_variable_names", [
        "market_volatility", "idiosyncratic_volatility"
    ])
    # Verify that every regime variable exists in the panel as well
    missing_regime: List[str] = [
        name for name in regime_names if name not in panel.columns
    ]
    # Fail loudly when any regime variable is missing
    if missing_regime:
        # Raise an informative error naming the missing regime variables
        raise ValueError(
            f"Regime registry mismatch; missing {missing_regime}"
        )
    # Sort the panel deterministically by date then ticker
    panel = panel.sort_values(["date", "ticker"]).reset_index(drop=True)
    # Initialize the artifact mapping
    artifacts: Dict[str, str] = {}
    # Persist the panel when an output directory is supplied
    if out_dir is not None:
        # Ensure the output directory exists
        out_dir.mkdir(parents=True, exist_ok=True)
        # Build the panel file path
        panel_path: Path = out_dir / "feature_panel.parquet"
        # Persist the feature panel as parquet
        panel.to_parquet(panel_path, index=False)
        # Compute the SHA-256 hash of the panel file
        digest: str = hashlib.sha256(
            panel_path.read_bytes()
        ).hexdigest()
        # Record the panel path in the artifacts mapping
        artifacts["panel"] = str(panel_path)
        # Record the panel hash in the artifacts mapping
        artifacts["panel_sha256"] = digest
    # Assemble the immutable feature-panel report
    report: FeaturePanelReport = FeaturePanelReport(
        panel=panel,
        feature_columns=feature_names,
        regime_columns=regime_names,
        target_column="target",
        artifacts=artifacts,
    )
    # Return the aggregate feature-panel report
    return report
````

````python
# Task 5: Implement Walk-Forward Splits and Training-Only Normalization

# ==============================================================================
# Task 5: Implement Walk-Forward Splits and Training-Only Normalization
# ==============================================================================

# -----------------------------------------------------------------------
# Task 5, Step 1: Generate the 30 Walk-Forward Windows
# -----------------------------------------------------------------------

def generate_walkforward_windows(
    dates: pd.DatetimeIndex,
    config: Dict[str, Any],
) -> List[Dict[str, np.ndarray]]:
    """
    Generate the 30 walk-forward train/validation/test index arrays.

    Implements the manuscript geometry of Table 1: development start
    s_w = w * 63, development end s_w + 503, training s_w..s_w+427,
    validation s_w+428..s_w+503, test s_w+504..s_w+566, for w = 0..29.

    The arrays are TRADING-DAY-CALENDAR POSITIONS (indices into the
    `dates` calendar). The geometry is anchored at the first test-window
    start: the first trading day on or after 2018-04-01 (Table 1: window
    1 test is 2018-04..2018-06), and every development period is the 504
    trading days immediately preceding its test window, so the first
    development window begins on 2016-04 as the manuscript states. The
    arrays are NOT panel-row positions; consumers must convert them to
    panel rows with build_window_row_masks, which maps each calendar
    position to the panel rows whose date matches.

    Parameters
    ----------
    dates : pd.DatetimeIndex
        The sorted common trading calendar of the market, covering the
        full sample window (e.g. 2015-12-01..2025-11-30).
    config : Dict[str, Any]
        The configuration dictionary.

    Returns
    -------
    List[Dict[str, np.ndarray]]
        A list of 30 window records with train/val/test index arrays
        expressed as positions into the `dates` calendar.

    Raises
    ------
    ValueError
        If the calendar is too short for the required geometry.
    """
    # Extract the walk-forward geometry from the config
    wf: Dict[str, Any] = config.get("walkforward_evaluation", {})
    # Extract the development length in trading days
    dev_length: int = int(
        wf.get("development_length_trading_days", 504)
    )
    # Extract the training length in trading days
    train_length: int = int(
        wf.get("train_length_trading_days", 428)
    )
    # Extract the validation length in trading days
    val_length: int = int(
        wf.get("validation_length_trading_days", 76)
    )
    # Extract the test length in trading days
    test_length: int = int(
        wf.get("test_length_trading_days", 63)
    )
    # Extract the step length in trading days
    step_length: int = int(
        wf.get("step_length_trading_days", 63)
    )
    # Extract the window count from the config
    n_windows: int = int(wf.get("number_of_windows", 30))
    # Compute the total number of trading days required after the anchor
    required_days: int = (
        (n_windows - 1) * step_length + dev_length + test_length
    )
    # Find the first test-window start: the first trading day on or
    # after 2018-04-01, which is window 1's test start in the
    # manuscript's Table 1 (test period 2018-04..2018-06). The
    # development period of every window is the 504 trading days
    # immediately preceding its test window, so the geometry anchor is
    # the test start MINUS the development length
    test_start_positions: np.ndarray = np.flatnonzero(
        dates >= pd.Timestamp("2018-04-01")
    )
    # Guard the degenerate case of an empty anchor search
    if test_start_positions.size == 0:
        # Raise an informative error for the missing anchor
        raise ValueError(
            "The calendar contains no trading day on or after "
            "2018-04-01; the walk-forward geometry cannot be anchored."
        )
    # Take the first test-start position at or after 2018-04-01
    first_test_start: int = int(test_start_positions[0])
    # Guard against a calendar too short to contain the dev period
    if first_test_start < dev_length:
        # Raise an informative error for the short calendar
        raise ValueError(
            "The calendar has fewer than 504 trading days before the "
            "first test window start; the geometry cannot be anchored."
        )
    # Compute the development anchor: 504 trading days before the first
    # test window start, exactly as Table 1 prescribes
    anchor: int = first_test_start - dev_length
    # Verify that the calendar is long enough for the geometry
    if len(dates) - anchor < required_days:
        # Raise an informative error naming the shortfall
        raise ValueError(
            f"Calendar has {len(dates) - anchor} usable days after the "
            f"development anchor but the walk-forward geometry requires "
            f"{required_days} days for {n_windows} windows."
        )
    # Initialize the window records list
    windows: List[Dict[str, np.ndarray]] = []
    # Iterate over every window index
    for w in range(n_windows):
        # Compute the development start calendar position for the window
        s_w: int = anchor + w * step_length
        # Compute the training index array for the window
        train_idx: np.ndarray = np.arange(s_w, s_w + train_length)
        # Compute the validation index array for the window
        val_idx: np.ndarray = np.arange(
            s_w + train_length, s_w + dev_length
        )
        # Compute the test index array for the window
        test_idx: np.ndarray = np.arange(
            s_w + dev_length, s_w + dev_length + test_length
        )
        # Pack the window record with the three split arrays
        windows.append(
            {
                "train": train_idx,
                "val": val_idx,
                "test": test_idx,
            }
        )
    # Return the complete list of window records
    return windows


def build_window_row_masks(
    windows: List[Dict[str, np.ndarray]],
    panel: pd.DataFrame,
    dates: pd.DatetimeIndex,
) -> List[Dict[str, np.ndarray]]:
    """
    Convert calendar-position window indices into panel-row masks.

    The windows returned by generate_walkforward_windows index the
    TRADING-DAY CALENDAR. This converter maps every calendar position to
    the PANEL ROWS (stock-day observations) whose date equals that
    calendar day, producing one boolean row mask per split per window.
    This is the single conversion point that removes the calendar/panel
    index misalignment.

    Parameters
    ----------
    windows : List[Dict[str, np.ndarray]]
        The window records with calendar-position index arrays.
    panel : pd.DataFrame
        The feature panel with a 'date' column; rows may appear in any
        order, the masks align to the panel's current row order.
    dates : pd.DatetimeIndex
        The trading-day calendar that the window positions index into.

    Returns
    -------
    List[Dict[str, np.ndarray]]
        One record per window with boolean 'train'/'val'/'test' masks
        over the panel rows, plus the derived integer index arrays
        'train_idx'/'val_idx'/'test_idx' of the selected rows.
    """
    # Validate the panel carries the required date column
    if "date" not in panel.columns:
        # Raise an informative error for the missing date column
        raise ValueError(
            "The panel must carry a 'date' column to convert calendar "
            "positions into panel rows."
        )
    # Extract the panel date array for the membership tests
    panel_dates: np.ndarray = pd.to_datetime(panel["date"]).to_numpy()
    # Initialize the row-mask records list
    records: List[Dict[str, np.ndarray]] = []
    # Iterate over every window record
    for record in windows:
        # Initialize the per-window record
        row_record: Dict[str, np.ndarray] = {}
        # Iterate over every split name of the window
        for split_name in ("train", "val", "test"):
            # Extract the calendar positions of the split
            positions: np.ndarray = record[split_name]
            # Translate the calendar positions into their trading dates
            split_dates: np.ndarray = dates[positions].to_numpy()
            # Build the boolean row mask via date membership: a panel row
            # belongs to the split iff its date is one of the split's
            # calendar days
            mask: np.ndarray = np.isin(panel_dates, split_dates)
            # Record the boolean mask for the split
            row_record[split_name] = mask
            # Record the derived integer row indices for the split
            row_record[f"{split_name}_idx"] = np.flatnonzero(mask)
        # Append the window's row-mask record
        records.append(row_record)
    # Return the row-mask records aligned to the panel row order
    return records


def verify_window_geometry(
    windows: List[Dict[str, np.ndarray]],
    dates: pd.DatetimeIndex,
    config: Dict[str, Any],
    calendar_is_exchange: bool = True,
) -> List[str]:
    """
    Verify the geometric identities of the generated windows.

    Parameters
    ----------
    windows : List[Dict[str, np.ndarray]]
        The generated window records.
    dates : pd.DatetimeIndex
        The sorted trading calendar.
    config : Dict[str, Any]
        The configuration dictionary.
    calendar_is_exchange : bool, optional
        True when the calendar comes from an exchange schedule (exact
        month-level checks apply); False for the business-day fallback,
        which cannot reproduce the exact exchange month boundaries, so
        year-level checks apply with a logged warning.

    Returns
    -------
    List[str]
        A list of geometry violations; empty when all identities hold.
    """
    # Initialize the violation list
    violations: List[str] = []
    # Extract the walk-forward geometry from the config
    wf: Dict[str, Any] = config.get("walkforward_evaluation", {})
    # Extract the test length for the assertions
    test_length: int = int(wf.get("test_length_trading_days", 63))
    # Extract the window count for the assertions
    n_windows: int = int(wf.get("number_of_windows", 30))
    # Verify the window count
    if len(windows) != n_windows:
        # Record the window-count violation
        violations.append(
            f"Expected {n_windows} windows, got {len(windows)}"
        )
    # Iterate over every window record
    for w, record in enumerate(windows):
        # Verify the test window length
        if len(record["test"]) != test_length:
            # Record the test-length violation
            violations.append(
                f"Window {w} test length {len(record['test'])} != "
                f"{test_length}"
            )
    # Verify the first test window starts in April 2018: the anchor in
    # generate_walkforward_windows is the first trading day on or after
    # 2018-04-01, so this check is exact under both calendars
    first_test_date: pd.Timestamp = dates[windows[0]["test"][0]]
    # Check the first test window's start month and year exactly
    if not (
        first_test_date.month == 4
        and first_test_date.year == 2018
    ):
        # Record the first-window date violation
        violations.append(
            f"First test window starts {first_test_date}, expected "
            f"2018-04"
        )
    # Verify the last test window ends in 2025, as the manuscript's
    # Table 1 prescribes (window 30 test: 2025-08..2025-10). The paper's
    # month labels are descriptive: 30 windows x 63 trading-day steps
    # cannot land on both month endpoints exactly, so the hard identity
    # is the year, and the actual end month is reconciled with the
    # config label through a logged note
    last_test_date: pd.Timestamp = dates[windows[-1]["test"][-1]]
    # Check the last test window's end year
    if last_test_date.year != 2025:
        # Record the last-window year violation
        violations.append(
            f"Last test window ends {last_test_date}, expected 2025"
        )
    # Reconcile the actual end month with the config's label
    if last_test_date.month != 10:
        # Log the descriptive-label reconciliation for the decision log
        logger.warning(
            "Walk-forward arithmetic places the last test window end in "
            "%s; the manuscript's '2025-08..2025-10' label is "
            "descriptive and cannot be reproduced exactly by 63-trading-"
            "day stepping",
            last_test_date.strftime("%Y-%m"),
        )
    # Verify the test windows do not overlap: the first test day of
    # window w+1 must be exactly one trading day after the last test day
    # of window w
    for w in range(1, len(windows)):
        # Compare the consecutive test-window boundaries
        if dates[windows[w]["test"][0]] <= dates[
            windows[w - 1]["test"][-1]
        ]:
            # Record the overlap violation
            violations.append(
                f"Test windows {w - 1} and {w} overlap"
            )
            # Stop the overlap scan after the first violation
            break
    # Verify the total test day count is 1890
    total_test: int = sum(len(r["test"]) for r in windows)
    # Check the total test-day identity
    if total_test != 1890:
        # Record the total-test-day violation
        violations.append(f"Total test days {total_test} != 1890")
    # Return the geometry violation list
    return violations


# -----------------------------------------------------------------------
# Task 5, Step 2: Fit Normalization Statistics on the Training Split Only
# -----------------------------------------------------------------------


def fit_normalization_moments(
    panel: pd.DataFrame,
    windows: List[Dict[str, np.ndarray]],
    config: Dict[str, Any],
    row_masks: Optional[List[Dict[str, np.ndarray]]] = None,
) -> Tuple[pd.DataFrame, pd.DataFrame]:
    """
    Fit the z-score moments from the training split of every window.

    Implements mu_w^(j) = (1/N) sum of x^(j) over training rows and
    sigma_w^(j) = sqrt((1/N) sum (x^(j) - mu)^2) (population std per the
    task's 1/N convention), with the variance floor from the config.

    Training rows are selected through the panel-row masks produced by
    build_window_row_masks; the masks are the ONLY correct way to select
    stock-day rows for a window (calendar positions are never used as
    panel row positions).

    Parameters
    ----------
    panel : pd.DataFrame
        The feature panel with the 16 features and 2 regime variables.
    windows : List[Dict[str, np.ndarray]]
        The generated window records (kept for the window indices).
    config : Dict[str, Any]
        The configuration dictionary.
    row_masks : Optional[List[Dict[str, np.ndarray]]], optional
        The per-window panel-row masks from build_window_row_masks;
        required for correct stock-day selection.

    Returns
    -------
    Tuple[pd.DataFrame, pd.DataFrame]
        The feature-moment frame and the regime-moment frame.
    """
    # Require the row masks: positional fallback would silently corrupt
    # the moments by indexing calendar positions as panel rows
    if row_masks is None:
        # Raise an informative error for the missing masks
        raise ValueError(
            "fit_normalization_moments requires row_masks from "
            "build_window_row_masks; calendar positions must never be "
            "used as panel row positions."
        )
    # Extract the feature-name registry from the config
    feature_names: List[str] = config.get(
        "stock_feature_construction", {}
    ).get("feature_names", [])
    # Extract the regime-variable names from the config
    regime_names: List[str] = config.get(
        "regime_state_variable_construction", {}
    ).get("regime_variable_names", [
        "market_volatility", "idiosyncratic_volatility"
    ])
    # Extract the variance floor from the config
    variance_floor: float = float(
        config.get("loss_function_and_regularization", {}).get(
            "numerical_hygiene", {}
        ).get("variance_floor", 1e-8)
    )
    # Extract the feature matrix once for the moment computation
    feature_matrix: np.ndarray = panel[feature_names].to_numpy(
        dtype=np.float64
    )
    # Extract the regime matrix once for the moment computation
    regime_matrix: np.ndarray = panel[regime_names].to_numpy(
        dtype=np.float64
    )
    # Initialize the feature-moment records list
    feature_records: List[Dict[str, Any]] = []
    # Initialize the regime-moment records list
    regime_records: List[Dict[str, Any]] = []
    # Iterate over every window record
    for w, record in enumerate(windows):
        # Extract the training row mask for the window
        train_mask: np.ndarray = row_masks[w]["train"]
        # Extract the training rows of the feature matrix via the mask
        train_features: np.ndarray = feature_matrix[train_mask]
        # Compute the training mean of every feature
        mu_f: np.ndarray = train_features.mean(axis=0)
        # Compute the training population std of every feature
        sigma_f: np.ndarray = train_features.std(axis=0)
        # Apply the variance floor to the feature standard deviations
        sigma_f = np.sqrt(np.maximum(sigma_f ** 2, variance_floor))
        # Iterate over every feature index
        for j, name in enumerate(feature_names):
            # Append the feature moment record
            feature_records.append(
                {"window": w, "feature": name,
                 "mean": mu_f[j], "std": sigma_f[j]}
            )
        # Extract the training rows of the regime matrix via the mask
        train_regime: np.ndarray = regime_matrix[train_mask]
        # Compute the training mean of every regime variable
        mu_z: np.ndarray = train_regime.mean(axis=0)
        # Compute the training population std of every regime variable
        sigma_z: np.ndarray = train_regime.std(axis=0)
        # Apply the variance floor to the regime standard deviations
        sigma_z = np.sqrt(np.maximum(sigma_z ** 2, variance_floor))
        # Iterate over every regime-variable index
        for j, name in enumerate(regime_names):
            # Append the regime moment record
            regime_records.append(
                {"window": w, "feature": name,
                 "mean": mu_z[j], "std": sigma_z[j]}
            )
    # Build the feature-moment frame from the records
    feature_moments: pd.DataFrame = pd.DataFrame(feature_records)
    # Build the regime-moment frame from the records
    regime_moments: pd.DataFrame = pd.DataFrame(regime_records)
    # Return the two moment frames
    return feature_moments, regime_moments


def apply_normalization(
    panel: pd.DataFrame,
    windows: List[Dict[str, np.ndarray]],
    feature_moments: pd.DataFrame,
    regime_moments: pd.DataFrame,
    config: Dict[str, Any],
    row_masks: Optional[List[Dict[str, np.ndarray]]] = None,
) -> pd.DataFrame:
    """
    Apply the training-fitted moments to every split of every window.

    Implements tilde_x = (x - mu_w) / sigma_w for features and regime
    variables using the TRAINING-fitted moments on the validation and
    test rows; the target is never normalized.

    Rows are selected through the panel-row masks produced by
    build_window_row_masks, never through calendar positions.

    Parameters
    ----------
    panel : pd.DataFrame
        The feature panel.
    windows : List[Dict[str, np.ndarray]]
        The generated window records.
    feature_moments : pd.DataFrame
        The feature-moment frame indexed by (window, feature).
    regime_moments : pd.DataFrame
        The regime-moment frame indexed by (window, feature).
    config : Dict[str, Any]
        The configuration dictionary.
    row_masks : Optional[List[Dict[str, np.ndarray]]], optional
        The per-window panel-row masks from build_window_row_masks;
        required for correct stock-day selection.

    Returns
    -------
    pd.DataFrame
        A copy of the panel with normalized feature and regime columns.
    """
    # Require the row masks to prevent calendar-position misuse
    if row_masks is None:
        # Raise an informative error for the missing masks
        raise ValueError(
            "apply_normalization requires row_masks from "
            "build_window_row_masks; calendar positions must never be "
            "used as panel row positions."
        )
    # Extract the feature-name registry from the config
    feature_names: List[str] = config.get(
        "stock_feature_construction", {}
    ).get("feature_names", [])
    # Extract the regime-variable names from the config
    regime_names: List[str] = config.get(
        "regime_state_variable_construction", {}
    ).get("regime_variable_names", [
        "market_volatility", "idiosyncratic_volatility"
    ])
    # Copy the panel to avoid mutating the caller's object
    out: pd.DataFrame = panel.copy()
    # Iterate over every window record
    for w, record in enumerate(windows):
        # Extract the row indices of every split of the window
        all_idx: np.ndarray = np.concatenate(
            [
                row_masks[w]["train_idx"],
                row_masks[w]["val_idx"],
                row_masks[w]["test_idx"],
            ]
        )
        # Subset the feature moments for this window
        wf_moments: pd.DataFrame = feature_moments.loc[
            feature_moments["window"] == w
        ].set_index("feature")
        # Subset the regime moments for this window
        wz_moments: pd.DataFrame = regime_moments.loc[
            regime_moments["window"] == w
        ].set_index("feature")
        # Iterate over every feature name
        for name in feature_names:
            # Fetch the fitted mean for the feature
            mu: float = float(wf_moments.loc[name, "mean"])
            # Fetch the fitted std for the feature
            sigma: float = float(wf_moments.loc[name, "std"])
            # Normalize the feature rows of every split
            out.loc[all_idx, name] = (
                panel.loc[all_idx, name] - mu
            ) / sigma
        # Iterate over every regime-variable name
        for name in regime_names:
            # Fetch the fitted mean for the regime variable
            mu_z: float = float(wz_moments.loc[name, "mean"])
            # Fetch the fitted std for the regime variable
            sigma_z: float = float(wz_moments.loc[name, "std"])
            # Normalize the regime rows of every split
            out.loc[all_idx, name] = (
                panel.loc[all_idx, name] - mu_z
            ) / sigma_z
    # Return the normalized panel copy
    return out


# -----------------------------------------------------------------------
# Task 5, Step 3: Assemble Panel Tensors and Audit for Leakage
# -----------------------------------------------------------------------


def assemble_split_tensors(
    panel: pd.DataFrame,
    windows: List[Dict[str, np.ndarray]],
    config: Dict[str, Any],
    row_masks: Optional[List[Dict[str, np.ndarray]]] = None,
) -> List[Dict[str, Any]]:
    """
    Assemble the normalized split tensors for every window.

    Builds X (N, 16), Z (N, 2), y (N,) tensors per split together with
    the ticker and date identity arrays, so downstream consumers never
    need to re-derive row identities from calendar positions.

    Parameters
    ----------
    panel : pd.DataFrame
        The normalized feature panel.
    windows : List[Dict[str, np.ndarray]]
        The generated window records.
    config : Dict[str, Any]
        The configuration dictionary.
    row_masks : Optional[List[Dict[str, np.ndarray]]], optional
        The per-window panel-row masks from build_window_row_masks;
        required for correct stock-day selection.

    Returns
    -------
    List[Dict[str, Any]]
        A list of 30 tensor bundles with X, Z, y, and row metadata.
    """
    # Require the row masks to prevent calendar-position misuse
    if row_masks is None:
        # Raise an informative error for the missing masks
        raise ValueError(
            "assemble_split_tensors requires row_masks from "
            "build_window_row_masks; calendar positions must never be "
            "used as panel row positions."
        )
    # Extract the feature-name registry from the config
    feature_names: List[str] = config.get(
        "stock_feature_construction", {}
    ).get("feature_names", [])
    # Extract the regime-variable names from the config
    regime_names: List[str] = config.get(
        "regime_state_variable_construction", {}
    ).get("regime_variable_names", [
        "market_volatility", "idiosyncratic_volatility"
    ])
    # Extract the raw feature matrix as a numpy array
    x_raw: np.ndarray = panel[feature_names].to_numpy(
        dtype=np.float32
    )
    # Extract the raw regime matrix as a numpy array
    z_raw: np.ndarray = panel[regime_names].to_numpy(
        dtype=np.float32
    )
    # Extract the raw target vector as a numpy array
    y_raw: np.ndarray = panel["target"].to_numpy(dtype=np.float32)
    # Extract the row-identity columns for the metadata
    tickers: np.ndarray = panel["ticker"].to_numpy()
    # Extract the row date identities for the metadata
    row_dates: np.ndarray = pd.to_datetime(panel["date"]).to_numpy()
    # Initialize the tensor bundle list
    bundles: List[Dict[str, Any]] = []
    # Iterate over every window record
    for w, record in enumerate(windows):
        # Initialize the per-window bundle dictionary
        bundle: Dict[str, Any] = {"window": w}
        # Iterate over every split name
        for split_name in ("train", "val", "test"):
            # Extract the split row indices from the row masks
            idx: np.ndarray = row_masks[w][f"{split_name}_idx"]
            # Build the feature tensor for the split
            bundle[f"X_{split_name}"] = torch.from_numpy(x_raw[idx])
            # Build the regime tensor for the split
            bundle[f"Z_{split_name}"] = torch.from_numpy(z_raw[idx])
            # Build the target tensor for the split
            bundle[f"y_{split_name}"] = torch.from_numpy(y_raw[idx])
            # Build the ticker metadata for the split
            bundle[f"tickers_{split_name}"] = tickers[idx]
            # Build the date metadata for the split
            bundle[f"dates_{split_name}"] = row_dates[idx]
        # Append the completed bundle to the list
        bundles.append(bundle)
    # Return the complete tensor bundle list
    return bundles


def audit_no_leakage(
    panel: pd.DataFrame,
    windows: List[Dict[str, np.ndarray]],
    feature_moments: pd.DataFrame,
    config: Dict[str, Any],
    regime_moments: Optional[pd.DataFrame] = None,
    row_masks: Optional[List[Dict[str, np.ndarray]]] = None,
) -> List[str]:
    """
    Audit that normalization moments use training rows only.

    Recomputes the feature AND regime moments (both mean and standard
    deviation) from the exact training row masks of every window and
    compares them with the stored moments, so any validation or test
    row leaking into a normalization statistic is detected.

    Parameters
    ----------
    panel : pd.DataFrame
        The raw feature panel.
    windows : List[Dict[str, np.ndarray]]
        The generated window records.
    feature_moments : pd.DataFrame
        The fitted feature-moment frame.
    config : Dict[str, Any]
        The configuration dictionary.
    regime_moments : Optional[pd.DataFrame], optional
        The fitted regime-moment frame; audited when supplied.
    row_masks : Optional[List[Dict[str, np.ndarray]]], optional
        The per-window panel-row masks from build_window_row_masks;
        required for correct stock-day selection.

    Returns
    -------
    List[str]
        A list of leakage violations; empty when the audit passes.
    """
    # Require the row masks to prevent calendar-position misuse
    if row_masks is None:
        # Raise an informative error for the missing masks
        raise ValueError(
            "audit_no_leakage requires row_masks from "
            "build_window_row_masks; calendar positions must never be "
            "used as panel row positions."
        )
    # Extract the feature-name registry from the config
    feature_names: List[str] = config.get(
        "stock_feature_construction", {}
    ).get("feature_names", [])
    # Extract the regime-variable names from the config
    regime_names: List[str] = config.get(
        "regime_state_variable_construction", {}
    ).get("regime_variable_names", [
        "market_volatility", "idiosyncratic_volatility"
    ])
    # Extract the feature matrix once for the recomputation
    feature_matrix: np.ndarray = panel[feature_names].to_numpy(
        dtype=np.float64
    )
    # Extract the regime matrix once for the recomputation
    regime_matrix: np.ndarray = panel[regime_names].to_numpy(
        dtype=np.float64
    )
    # Initialize the violation list
    violations: List[str] = []
    # Iterate over every window record
    for w, record in enumerate(windows):
        # Extract the training row mask for the window
        train_mask: np.ndarray = row_masks[w]["train"]
        # Recompute the training-only feature moments
        train_features: np.ndarray = feature_matrix[train_mask]
        # Compute the recomputed training means of the features
        recomputed_mean: np.ndarray = train_features.mean(axis=0)
        # Compute the recomputed training stds of the features
        recomputed_std: np.ndarray = train_features.std(axis=0)
        # Subset the stored feature moments for this window
        stored_f: pd.DataFrame = feature_moments.loc[
            feature_moments["window"] == w
        ].set_index("feature")
        # Compare the recomputed feature means with the stored means
        for j, name in enumerate(feature_names):
            # Check the mean agreement within a tight tolerance
            if not np.isclose(
                recomputed_mean[j], float(stored_f.loc[name, "mean"]),
                rtol=1e-6, atol=1e-8
            ):
                # Record the mean leakage violation
                violations.append(
                    f"Window {w} feature {name} mean mismatch"
                )
            # Check the std agreement within a tight tolerance
            if not np.isclose(
                recomputed_std[j], float(stored_f.loc[name, "std"]),
                rtol=1e-6, atol=1e-8
            ):
                # Record the std leakage violation
                violations.append(
                    f"Window {w} feature {name} std mismatch"
                )
        # Audit the regime moments when the frame was supplied
        if regime_moments is not None:
            # Recompute the training-only regime moments
            train_regime: np.ndarray = regime_matrix[train_mask]
            # Compute the recomputed training means of the regime vars
            recomputed_z_mean: np.ndarray = train_regime.mean(axis=0)
            # Compute the recomputed training stds of the regime vars
            recomputed_z_std: np.ndarray = train_regime.std(axis=0)
            # Subset the stored regime moments for this window
            stored_z: pd.DataFrame = regime_moments.loc[
                regime_moments["window"] == w
            ].set_index("feature")
            # Compare the recomputed regime moments with the stored ones
            for j, name in enumerate(regime_names):
                # Check the regime mean agreement within tolerance
                if not np.isclose(
                    recomputed_z_mean[j],
                    float(stored_z.loc[name, "mean"]),
                    rtol=1e-6, atol=1e-8
                ):
                    # Record the regime mean leakage violation
                    violations.append(
                        f"Window {w} regime {name} mean mismatch"
                    )
                # Check the regime std agreement within tolerance
                if not np.isclose(
                    recomputed_z_std[j],
                    float(stored_z.loc[name, "std"]),
                    rtol=1e-6, atol=1e-8
                ):
                    # Record the regime std leakage violation
                    violations.append(
                        f"Window {w} regime {name} std mismatch"
                    )
    # Return the leakage violation list
    return violations


# -----------------------------------------------------------------------
# Task 5, Orchestrator Function
# -----------------------------------------------------------------------


@dataclass(frozen=True)
class WalkForwardReport:
    """
    Immutable aggregate report of the walk-forward splitting pipeline.

    Attributes
    ----------
    windows : List[Dict[str, np.ndarray]]
        The 30 window records.
    feature_moments : pd.DataFrame
        The fitted feature-moment frame.
    regime_moments : pd.DataFrame
        The fitted regime-moment frame.
    bundles : List[Dict[str, Any]]
        The 30 tensor bundles.
    artifacts : Dict[str, str]
        Mapping of persisted artifact names to file paths.
    """

    # The 30 window records
    windows: List[Dict[str, np.ndarray]]
    # Fitted feature-moment frame
    feature_moments: pd.DataFrame
    # Fitted regime-moment frame
    regime_moments: pd.DataFrame
    # The 30 tensor bundles
    bundles: List[Dict[str, Any]]
    # Mapping of persisted artifact names to file paths
    artifacts: Dict[str, str]


def build_walkforward_splits(
    panel: pd.DataFrame,
    config: Dict[str, Any],
    out_dir: Optional[Path] = None,
) -> WalkForwardReport:
    """
    Orchestrate the full walk-forward splitting pipeline.

    Composes generate_walkforward_windows, verify_window_geometry,
    fit_normalization_moments, apply_normalization,
    assemble_split_tensors, and audit_no_leakage in the mandated order.

    Parameters
    ----------
    panel : pd.DataFrame
        The feature panel from Task 4.
    config : Dict[str, Any]
        The configuration dictionary.
    out_dir : Optional[Path], optional
        The destination directory for persisted artifacts.

    Returns
    -------
    WalkForwardReport
        The immutable aggregate walk-forward report.

    Raises
    ------
    ValueError
        If the window geometry or the leakage audit fails.
    """
    # Build the FULL sample trading calendar from the config window: the
    # geometry must be anchored at the sample start (the manuscript's
    # Table 1 anchor is the first development window start on 2016-04-01,
    # which generate_walkforward_windows resolves inside the calendar),
    # NOT at the feature panel's post-warm-up dates
    sample_start: str = str(
        config.get("data_acquisition_and_universe", {}).get(
            "sample_start", "2015-12-01"
        )
    )
    # Extract the sample end date for the calendar construction
    sample_end: str = str(
        config.get("data_acquisition_and_universe", {}).get(
            "sample_end", "2025-11-30"
        )
    )
    # Initialize the calendar-source flag for the geometry verification
    calendar_is_exchange: bool = False
    # Use the exchange calendar when available for the full calendar
    if mcal is not None:
        # Try the NYSE calendar for the study's trading schedule
        try:
            # Fetch the exchange schedule over the full sample window
            schedule = mcal.get_calendar("NYSE").schedule(
                start_date=sample_start, end_date=sample_end
            )
            # Extract the trading-day index as the full calendar
            dates: pd.DatetimeIndex = schedule.index
            # Mark the calendar as exchange-accurate
            calendar_is_exchange = True
        except Exception as exc:
            # Log the calendar failure and fall back to business days
            logger.warning(
                "Exchange calendar unavailable (%s); using business days",
                exc,
            )
            # Fall back to business days for the full calendar
            dates = pd.DatetimeIndex(pd.bdate_range(sample_start, sample_end))
    else:
        # Log the fallback to business days
        logger.warning(
            "pandas_market_calendars unavailable; using business days"
        )
        # Build the full calendar from business days as the fallback
        dates = pd.DatetimeIndex(pd.bdate_range(sample_start, sample_end))
    # Step 1: generate the 30 walk-forward window records (calendar
    # positions anchored at the first test-window start, 2018-04-01,
    # with each development period being the 504 trading days preceding
    # its test window, exactly as the manuscript's Table 1 prescribes)
    windows: List[Dict[str, np.ndarray]] = generate_walkforward_windows(
        dates, config
    )
    # Step 1: verify the geometric identities of the windows
    geometry_violations: List[str] = verify_window_geometry(
        windows, dates, config,
        calendar_is_exchange=calendar_is_exchange,
    )
    # Fail loudly on any geometry violation
    if geometry_violations:
        # Raise an informative error listing the violations
        raise ValueError(
            f"Window geometry failed: {'; '.join(geometry_violations)}"
        )
    # Convert the calendar positions into panel-row masks: this is the
    # single conversion that selects the correct stock-day rows for every
    # window split and removes the calendar/panel index misalignment
    row_masks: List[Dict[str, np.ndarray]] = build_window_row_masks(
        windows, panel, dates
    )
    # Step 2: fit the normalization moments from training rows only
    feature_moments, regime_moments = fit_normalization_moments(
        panel, windows, config, row_masks
    )
    # Step 2: apply the fitted moments to every split
    normalized: pd.DataFrame = apply_normalization(
        panel, windows, feature_moments, regime_moments, config,
        row_masks,
    )
    # Step 3: assemble the split tensors for every window
    bundles: List[Dict[str, Any]] = assemble_split_tensors(
        normalized, windows, config, row_masks
    )
    # Step 3: audit that no normalization moment leaks from val/test
    leakage: List[str] = audit_no_leakage(
        panel, windows, feature_moments, config, regime_moments,
        row_masks,
    )
    # Fail loudly on any leakage violation
    if leakage:
        # Raise an informative error listing the violations
        raise ValueError(f"Leakage audit failed: {'; '.join(leakage)}")
    # Initialize the artifact mapping
    artifacts: Dict[str, str] = {}
    # Persist the split artifacts when an output directory is supplied
    if out_dir is not None:
        # Ensure the output directory exists
        out_dir.mkdir(parents=True, exist_ok=True)
        # Persist the moment frames to CSV
        feature_moments.to_csv(out_dir / "feature_moments.csv", index=False)
        # Persist the regime moment frames to CSV
        regime_moments.to_csv(out_dir / "regime_moments.csv", index=False)
        # Record the feature-moment path in the artifacts
        artifacts["feature_moments"] = str(
            out_dir / "feature_moments.csv"
        )
        # Record the regime-moment path in the artifacts
        artifacts["regime_moments"] = str(
            out_dir / "regime_moments.csv"
        )
        # Persist the deterministic row-order file: the single source of
        # truth for row identities that every downstream task must read
        row_order: pd.DataFrame = panel[["ticker", "date"]].copy()
        # Reset the row index to align with the tensor positions
        row_order = row_order.reset_index(drop=True)
        # Persist the row-order metadata as a CSV
        row_order.to_csv(out_dir / "row_order.csv", index=False)
        # Record the row-order path in the artifacts
        artifacts["row_order"] = str(out_dir / "row_order.csv")
    # Assemble the immutable walk-forward report
    report: WalkForwardReport = WalkForwardReport(
        windows=windows,
        feature_moments=feature_moments,
        regime_moments=regime_moments,
        bundles=bundles,
        artifacts=artifacts,
    )
    # Return the aggregate walk-forward report
    return report
````

````python
# Task 6: Implement the Shared MLP Block and the Gate Network

# ==============================================================================
# Task 6: Implement the Shared MLP Block and the Gate Network
# ==============================================================================

# -----------------------------------------------------------------------
# Task 6, Step 1: Implement the Two-Hidden-Layer MLP Block
# -----------------------------------------------------------------------

class SharedMLPBlock(nn.Module):
    """
    Two-hidden-layer MLP block shared by every neural model.

    Implements the manuscript block equations:
        h1 = GELU(W1 q + b1)
        h2 = GELU(W2 h1 + b2)
        Block(q; theta) = W3 h2 + b3
    with dropout rate 0.1 applied after each hidden layer during
    training only, and a linear scalar output head.

    Attributes
    ----------
    input_dim : int
        The input feature dimension of the block.
    hidden_width : int
        The width H of both hidden layers.
    dropout_rate : float
        The dropout probability applied after each hidden layer.
    """

    def __init__(
        self,
        input_dim: int,
        hidden_width: int,
        dropout_rate: float = 0.1,
    ) -> None:
        """
        Initialize the shared MLP block layers.

        Parameters
        ----------
        input_dim : int
            The input feature dimension (16 for x, 18 for (x, z)).
        hidden_width : int
            The width H of both hidden layers.
        dropout_rate : float, optional
            The dropout probability after each hidden layer.

        Raises
        ------
        ValueError
            If input_dim or hidden_width are not positive integers.
        """
        # Initialize the parent nn.Module class
        super().__init__()
        # Validate the input dimension is a positive integer
        if not isinstance(input_dim, int) or input_dim < 1:
            # Raise an informative error for the invalid input dimension
            raise ValueError(
                f"input_dim must be a positive int; got {input_dim!r}"
            )
        # Validate the hidden width is a positive integer
        if not isinstance(hidden_width, int) or hidden_width < 1:
            # Raise an informative error for the invalid hidden width
            raise ValueError(
                f"hidden_width must be a positive int; got "
                f"{hidden_width!r}"
            )
        # Store the input dimension on the module
        self.input_dim: int = input_dim
        # Store the hidden width on the module
        self.hidden_width: int = hidden_width
        # Store the dropout rate on the module
        self.dropout_rate: float = dropout_rate
        # Define the first hidden linear layer: W1, b1
        self.fc1: nn.Linear = nn.Linear(input_dim, hidden_width)
        # Define the GELU activation after the first hidden layer
        self.act1: nn.Module = nn.GELU()
        # Define the dropout after the first hidden layer
        self.drop1: nn.Module = nn.Dropout(dropout_rate)
        # Define the second hidden linear layer: W2, b2
        self.fc2: nn.Linear = nn.Linear(hidden_width, hidden_width)
        # Define the GELU activation after the second hidden layer
        self.act2: nn.Module = nn.GELU()
        # Define the dropout after the second hidden layer
        self.drop2: nn.Module = nn.Dropout(dropout_rate)
        # Define the scalar output linear layer: W3, b3
        self.fc3: nn.Linear = nn.Linear(hidden_width, 1)

    def forward(self, q: torch.Tensor) -> torch.Tensor:
        """
        Apply the block to an input batch.

        Implements Block(q; theta) = W3 GELU(W2 GELU(W1 q + b1) + b2) + b3
        with dropout after each hidden layer in training mode.

        Parameters
        ----------
        q : torch.Tensor
            Input tensor of shape (N, input_dim).

        Returns
        -------
        torch.Tensor
            Output tensor of shape (N, 1).
        """
        # Apply the first affine layer: h_pre1 = W1 q + b1
        h_pre1: torch.Tensor = self.fc1(q)
        # Apply the GELU activation: h1 = GELU(h_pre1)
        h1: torch.Tensor = self.act1(h_pre1)
        # Apply dropout after the first hidden layer
        h1 = self.drop1(h1)
        # Apply the second affine layer: h_pre2 = W2 h1 + b2
        h_pre2: torch.Tensor = self.fc2(h1)
        # Apply the GELU activation: h2 = GELU(h_pre2)
        h2: torch.Tensor = self.act2(h_pre2)
        # Apply dropout after the second hidden layer
        h2 = self.drop2(h2)
        # Apply the scalar output layer: Block = W3 h2 + b3
        out: torch.Tensor = self.fc3(h2)
        # Return the block output of shape (N, 1)
        return out


# -----------------------------------------------------------------------
# Task 6, Step 2: Implement the Gate Network
# -----------------------------------------------------------------------


class GateNetwork(nn.Module):
    """
    Two-hidden-layer gating network mapping inputs to expert logits.

    Implements pi = softmax(g(u)) where g is a two-hidden-layer MLP of
    width 16 with GELU activation and no dropout, per the config's gate
    architecture.

    Attributes
    ----------
    input_dim : int
        The gate input dimension (16 for x, 18 for (x, z)).
    n_experts : int
        The number of experts K; the output logit count.
    """

    def __init__(
        self,
        input_dim: int,
        n_experts: int,
        hidden_width: int = 16,
    ) -> None:
        """
        Initialize the gate network layers.

        Parameters
        ----------
        input_dim : int
            The gate input dimension (16 for x, 18 for (x, z)).
        n_experts : int
            The number of experts K.
        hidden_width : int, optional
            The hidden width of the gate, default 16.

        Raises
        ------
        ValueError
            If input_dim or n_experts are not positive integers.
        """
        # Initialize the parent nn.Module class
        super().__init__()
        # Validate the input dimension is a positive integer
        if not isinstance(input_dim, int) or input_dim < 1:
            # Raise an informative error for the invalid input dimension
            raise ValueError(
                f"input_dim must be a positive int; got {input_dim!r}"
            )
        # Validate the expert count is a positive integer
        if not isinstance(n_experts, int) or n_experts < 1:
            # Raise an informative error for the invalid expert count
            raise ValueError(
                f"n_experts must be a positive int; got {n_experts!r}"
            )
        # Store the input dimension on the module
        self.input_dim: int = input_dim
        # Store the expert count on the module
        self.n_experts: int = n_experts
        # Define the first hidden linear layer of the gate
        self.fc1: nn.Linear = nn.Linear(input_dim, hidden_width)
        # Define the GELU activation after the first hidden layer
        self.act1: nn.Module = nn.GELU()
        # Define the second hidden linear layer of the gate
        self.fc2: nn.Linear = nn.Linear(hidden_width, hidden_width)
        # Define the GELU activation after the second hidden layer
        self.act2: nn.Module = nn.GELU()
        # Define the logit output layer with K units
        self.fc3: nn.Linear = nn.Linear(hidden_width, n_experts)

    def forward(self, u: torch.Tensor) -> torch.Tensor:
        """
        Map an input batch to expert probabilities.

        Implements pi = softmax(g(u)) with a numerically stable
        log-softmax.

        Parameters
        ----------
        u : torch.Tensor
            Input tensor of shape (N, input_dim).

        Returns
        -------
        torch.Tensor
            Probability tensor of shape (N, K) summing to one per row.
        """
        # Apply the first affine layer of the gate
        h_pre1: torch.Tensor = self.fc1(u)
        # Apply the GELU activation after the first layer
        h1: torch.Tensor = self.act1(h_pre1)
        # Apply the second affine layer of the gate
        h_pre2: torch.Tensor = self.fc2(h1)
        # Apply the GELU activation after the second layer
        h2: torch.Tensor = self.act2(h_pre2)
        # Apply the logit output layer: logits = W3 h2 + b3
        logits: torch.Tensor = self.fc3(h2)
        # Compute the stable log-probabilities: log pi = log_softmax
        log_probs: torch.Tensor = F.log_softmax(logits, dim=-1)
        # Exponentiate to obtain the probabilities: pi = softmax
        probs: torch.Tensor = torch.exp(log_probs)
        # Return the expert probability tensor of shape (N, K)
        return probs


# -----------------------------------------------------------------------
# Task 6, Step 3: Parameter Counting and Initialization Control
# -----------------------------------------------------------------------


def count_parameters(module: nn.Module) -> int:
    """
    Count the total number of parameters in a module.

    Parameters
    ----------
    module : nn.Module
        The module whose parameters are counted.

    Returns
    -------
    int
        The total parameter count including weights and biases.
    """
    # Sum the element counts of every parameter in the module
    total: int = sum(p.numel() for p in module.parameters())
    # Return the total parameter count
    return total


def zero_init_expert_outputs(output_layers: List[nn.Linear]) -> None:
    """
    Zero-initialize the final layers of the residual experts.

    Ensures r_k(x) = 0 at initialization so the full RG-ResMoE model
    reproduces the frozen base forecast exactly.

    Parameters
    ----------
    output_layers : List[nn.Linear]
        The final linear layers of every residual expert.

    Returns
    -------
    None
    """
    # Iterate over every expert output layer
    for layer in output_layers:
        # Zero-fill the output layer weights
        nn.init.zeros_(layer.weight)
        # Zero-fill the output layer biases
        nn.init.zeros_(layer.bias)
    # Log the zero-initialization application
    logger.info(
        "Zero-initialized %d expert output layers", len(output_layers)
    )


def verify_parameter_counts(
    counts: Dict[str, int],
    config: Dict[str, Any],
) -> List[str]:
    """
    Verify the counted parameters against the closed-form formula.

    Parameters
    ----------
    counts : Dict[str, int]
        The measured parameter counts indexed by model name.
    config : Dict[str, Any]
        The configuration dictionary.

    Returns
    -------
    List[str]
        A list of capacity findings; empty when all counts match.
    """
    # Import the parameter-count formula from the config
    formula: str = str(
        config.get("shared_neural_network_block", {}).get(
            "parameter_count_formula",
            "(d_in+1)*H + (H+1)*H + (H+1)",
        )
    )
    # Initialize the findings list
    findings: List[str] = []
    # Define the expected counts from the manuscript capacity audit
    expected: Dict[str, int] = {
        "MLP_S": 561,
        "MLP_L": 2773,
        "MLP_S_plus_z": 593,
        "MLP_L_plus_z": 2861,
        "MoE_experts": 2244,
        "RG_ResMoE_base_plus_experts": 2805,
    }
    # Extract the gate architecture parameters for the gate count: the
    # config's gate formula is ((input_dim+1)*16 + (16+1)*16 + (16+1)*K)
    gate_cfg: Dict[str, Any] = config.get(
        "network_architecture_parameters", {}
    ).get("gate_architecture", {})
    # Extract the gate hidden width from the config
    gate_hidden: int = int(gate_cfg.get("hidden_width", 16))
    # Extract the expert count from the RG-ResMoE configuration
    rg_cfg: Dict[str, Any] = config.get(
        "network_architecture_parameters", {}
    ).get("RG_ResMoE", {})
    # Extract the number of experts K from the RG-ResMoE configuration
    n_experts: int = int(rg_cfg.get("expert_count", 4))
    # Compute the full-gate input dimension (x, z) from the config
    gate_input_dim: int = (
        int(config.get("stock_feature_construction", {}).get(
            "feature_dimension", 16))
        + int(config.get("regime_state_variable_construction", {}).get(
            "regime_dimension", 2))
    )
    # Compute the expected gate parameter count from the config formula
    expected_gate: int = (
        (gate_input_dim + 1) * gate_hidden
        + (gate_hidden + 1) * gate_hidden
        + (gate_hidden + 1) * n_experts
    )
    # Register the gate expectation in the capacity audit
    expected["Gate"] = expected_gate
    # Iterate over every expected count
    for name, expected_count in expected.items():
        # Fetch the measured count for the model
        measured: int = counts.get(name, -1)
        # Compare the measured count with the expectation
        if measured != expected_count:
            # Record the capacity mismatch finding
            findings.append(
                f"{name}: counted {measured}, formula {expected_count} "
                f"(formula: {formula})"
            )
    # Return the capacity findings list
    return findings


# -----------------------------------------------------------------------
# Task 6, Orchestrator Function
# -----------------------------------------------------------------------


@dataclass(frozen=True)
class ModelBlockReport:
    """
    Immutable factory report of the shared block and gate modules.

    Attributes
    ----------
    block_factory : Callable
        Factory returning a fresh SharedMLPBlock instance.
    gate_factory : Callable
        Factory returning a fresh GateNetwork instance.
    parameter_counts : Dict[str, int]
        Measured parameter counts indexed by model name.
    capacity_findings : List[str]
        Capacity mismatches against the closed-form formula.
    """

    # Factory returning a fresh SharedMLPBlock instance
    block_factory: Callable[..., SharedMLPBlock]
    # Factory returning a fresh GateNetwork instance
    gate_factory: Callable[..., GateNetwork]
    # Measured parameter counts indexed by model name
    parameter_counts: Dict[str, int]
    # Capacity mismatches against the closed-form formula
    capacity_findings: List[str]


def build_shared_blocks(config: Dict[str, Any]) -> ModelBlockReport:
    """
    Orchestrate the shared block, gate, and capacity verification.

    Composes SharedMLPBlock, GateNetwork, count_parameters,
    verify_parameter_counts, and zero_init_expert_outputs.

    Parameters
    ----------
    config : Dict[str, Any]
        The configuration dictionary.

    Returns
    -------
    ModelBlockReport
        The immutable factory report for Tasks 7 and 8.
    """
    # Extract the feature dimension from the config
    feature_dim: int = int(
        config.get("stock_feature_construction", {}).get(
            "feature_dimension", 16
        )
    )
    # Extract the regime dimension from the config
    regime_dim: int = int(
        config.get("regime_state_variable_construction", {}).get(
            "regime_dimension", 2
        )
    )
    # Extract the block dropout rate from the config
    dropout_rate: float = float(
        config.get("shared_neural_network_block", {}).get(
            "dropout_rate", 0.1
        )
    )
    # Define the block factory closure over the config parameters
    def block_factory(
        input_dim: int, hidden_width: int
    ) -> SharedMLPBlock:
        """
        Build a fresh shared MLP block.

        Parameters
        ----------
        input_dim : int
            The block input dimension.
        hidden_width : int
            The block hidden width.

        Returns
        -------
        SharedMLPBlock
            A fresh block instance.
        """
        # Instantiate the shared MLP block with the config dropout
        return SharedMLPBlock(
            input_dim, hidden_width, dropout_rate=dropout_rate
        )

    # Define the gate factory closure over the config parameters
    def gate_factory(
        input_dim: int, n_experts: int
    ) -> GateNetwork:
        """
        Build a fresh gate network.

        Parameters
        ----------
        input_dim : int
            The gate input dimension.
        n_experts : int
            The number of experts.

        Returns
        -------
        GateNetwork
            A fresh gate instance.
        """
        # Instantiate the gate network with width 16
        return GateNetwork(input_dim, n_experts, hidden_width=16)

    # Measure the parameter counts of the reference configurations
    counts: Dict[str, int] = {
        "MLP_S": count_parameters(block_factory(feature_dim, 16)),
        "MLP_L": count_parameters(block_factory(feature_dim, 44)),
        "MLP_S_plus_z": count_parameters(
            block_factory(feature_dim + regime_dim, 16)
        ),
        "MLP_L_plus_z": count_parameters(
            block_factory(feature_dim + regime_dim, 44)
        ),
        "MoE_experts": 4 * count_parameters(
            block_factory(feature_dim, 16)
        ),
        "RG_ResMoE_base_plus_experts": count_parameters(
            block_factory(feature_dim, 16)
        ) + 4 * count_parameters(block_factory(feature_dim, 16)),
        # Record the gate parameter count SEPARATELY from the forecasting
        # capacity audit, exactly as the task list requires
        "Gate": count_parameters(
            gate_factory(feature_dim + regime_dim, 4)
        ),
    }
    # Verify the measured counts against the closed-form formula
    findings: List[str] = verify_parameter_counts(counts, config)
    # Log the capacity verification outcome
    logger.info(
        "Capacity audit: %d findings", len(findings)
    )
    # Fail loudly on capacity mismatches so a wrong architecture cannot
    # silently reach the training stage
    if findings:
        # Raise an informative error listing the mismatches
        raise ValueError(
            f"Capacity audit failed: {'; '.join(findings)}"
        )
    # Assemble the immutable factory report
    report: ModelBlockReport = ModelBlockReport(
        block_factory=block_factory,
        gate_factory=gate_factory,
        parameter_counts=counts,
        capacity_findings=findings,
    )
    # Return the factory report
    return report
````

````python
# Task 7: Implement Capacity-Matched MLP Baselines and Standard MoE

# ==============================================================================
# Task 7: Implement Capacity-Matched MLP Baselines and Standard MoE
# ==============================================================================

# -----------------------------------------------------------------------
# Task 7, Step 1: Implement MLP-S, MLP-L, and Their (+z) Variants
# -----------------------------------------------------------------------

class MLPPlusZ(nn.Module):
    """
    MLP baseline wrapper that concatenates regime variables into the input.

    Produces the (+z) input-pathway variants by concatenating the stock
    features x and the regime variables z along the feature dimension
    before the shared block.

    Attributes
    ----------
    block : SharedMLPBlock
        The underlying forecasting block.
    feature_dim : int
        The stock-feature dimension.
    regime_dim : int
        The regime-variable dimension.
    """

    def __init__(
        self,
        block: SharedMLPBlock,
        feature_dim: int,
        regime_dim: int,
    ) -> None:
        """
        Initialize the (+z) wrapper around a shared block.

        Parameters
        ----------
        block : SharedMLPBlock
            The underlying forecasting block.
        feature_dim : int
            The stock-feature dimension.
        regime_dim : int
            The regime-variable dimension.

        Raises
        ------
        ValueError
            If the block input dimension is not feature_dim + regime_dim.
        """
        # Initialize the parent nn.Module class
        super().__init__()
        # Validate the block input dimension against the concatenation
        if block.input_dim != feature_dim + regime_dim:
            # Raise an informative error for the dimension mismatch
            raise ValueError(
                f"Block input_dim {block.input_dim} must equal "
                f"feature_dim + regime_dim = {feature_dim + regime_dim}"
            )
        # Store the underlying block on the wrapper
        self.block: SharedMLPBlock = block
        # Store the feature dimension on the wrapper
        self.feature_dim: int = feature_dim
        # Store the regime dimension on the wrapper
        self.regime_dim: int = regime_dim

    def forward(
        self, x: torch.Tensor, z: torch.Tensor
    ) -> torch.Tensor:
        """
        Apply the block to the concatenated (x, z) input.

        Parameters
        ----------
        x : torch.Tensor
            Stock features of shape (N, feature_dim).
        z : torch.Tensor
            Regime variables of shape (N, regime_dim).

        Returns
        -------
        torch.Tensor
            The forecast of shape (N, 1).
        """
        # Concatenate the features and regime variables along the last axis
        u: torch.Tensor = torch.cat([x, z], dim=-1)
        # Apply the shared block to the concatenated input
        out: torch.Tensor = self.block(u)
        # Return the forecast of shape (N, 1)
        return out


def build_mlp_baseline(
    block_factory: Callable[..., SharedMLPBlock],
    model_name: str,
    config: Dict[str, Any],
) -> nn.Module:
    """
    Build an MLP baseline (x-input) from the config registry.

    Parameters
    ----------
    block_factory : Callable
        The Task 6 shared-block factory.
    model_name : str
        One of "MLP_S" or "MLP_L".
    config : Dict[str, Any]
        The configuration dictionary.

    Returns
    -------
    nn.Module
        A fresh MLP baseline module.

    Raises
    ------
    ValueError
        If the model name is not in the architecture registry.
    """
    # Extract the architecture registry from the config
    arch: Dict[str, Any] = config.get("network_architecture_parameters", {})
    # Verify the model name exists in the registry
    if model_name not in arch:
        # Raise an informative error for the unknown model
        raise ValueError(
            f"Unknown model '{model_name}'; registry has "
            f"{list(arch.keys())}."
        )
    # Extract the feature dimension from the config
    feature_dim: int = int(
        config.get("stock_feature_construction", {}).get(
            "feature_dimension", 16
        )
    )
    # Extract the hidden width of the model from the registry
    width: int = int(arch[model_name].get("width", 16))
    # Build a fresh block with the feature input and model width
    block: SharedMLPBlock = block_factory(feature_dim, width)
    # Return the fresh MLP baseline module
    return block


def build_mlp_plus_z_baseline(
    block_factory: Callable[..., SharedMLPBlock],
    model_name: str,
    config: Dict[str, Any],
) -> MLPPlusZ:
    """
    Build an MLP (+z) baseline from the config registry.

    Parameters
    ----------
    block_factory : Callable
        The Task 6 shared-block factory.
    model_name : str
        One of "MLP_S_plus_z" or "MLP_L_plus_z".
    config : Dict[str, Any]
        The configuration dictionary.

    Returns
    -------
    MLPPlusZ
        A fresh (+z) wrapper module.

    Raises
    ------
    ValueError
        If the model name is not in the architecture registry.
    """
    # Extract the architecture registry from the config
    arch: Dict[str, Any] = config.get("network_architecture_parameters", {})
    # Verify the model name exists in the registry
    if model_name not in arch:
        # Raise an informative error for the unknown model
        raise ValueError(
            f"Unknown model '{model_name}'; registry has "
            f"{list(arch.keys())}."
        )
    # Extract the feature dimension from the config
    feature_dim: int = int(
        config.get("stock_feature_construction", {}).get(
            "feature_dimension", 16
        )
    )
    # Extract the regime dimension from the config
    regime_dim: int = int(
        config.get("regime_state_variable_construction", {}).get(
            "regime_dimension", 2
        )
    )
    # Extract the hidden width of the model from the registry
    width: int = int(arch[model_name].get("width", 16))
    # Build a fresh block with the concatenated input dimension
    block: SharedMLPBlock = block_factory(
        feature_dim + regime_dim, width
    )
    # Wrap the block in the (+z) concatenation wrapper
    wrapper: MLPPlusZ = MLPPlusZ(block, feature_dim, regime_dim)
    # Return the fresh (+z) wrapper module
    return wrapper


# -----------------------------------------------------------------------
# Task 7, Step 2: Implement Standard MoE
# -----------------------------------------------------------------------


class StandardMoE(nn.Module):
    """
    Standard mixture-of-experts model with full-level expert forecasts.

    Implements e_k(x) = Block(x; theta_k), pi = softmax(g(x, z)), and
    y_hat = sum_k pi_k e_k(x), with all components trained jointly on
    MSE and no frozen base or residual decomposition.

    Attributes
    ----------
    n_experts : int
        The number of experts K.
    experts : nn.ModuleList
        The K expert blocks receiving x only.
    gate : nn.Module
        The gate network receiving (x, z).
    last_probs : Optional[torch.Tensor]
        The last gate probabilities, retained for routing diagnosis.
    """

    def __init__(
        self,
        n_experts: int,
        feature_dim: int,
        regime_dim: int,
        block_factory: Callable[..., SharedMLPBlock],
        gate_factory: Callable[..., nn.Module],
        expert_width: int = 16,
    ) -> None:
        """
        Initialize the standard MoE experts and gate.

        Parameters
        ----------
        n_experts : int
            The number of experts K.
        feature_dim : int
            The stock-feature dimension.
        regime_dim : int
            The regime-variable dimension.
        block_factory : Callable
            The Task 6 shared-block factory.
        gate_factory : Callable
            The Task 6 gate factory.
        expert_width : int, optional
            The expert hidden width, default 16.

        Raises
        ------
        ValueError
            If n_experts is not a positive integer.
        """
        # Initialize the parent nn.Module class
        super().__init__()
        # Validate the expert count is a positive integer
        if not isinstance(n_experts, int) or n_experts < 1:
            # Raise an informative error for the invalid expert count
            raise ValueError(
                f"n_experts must be a positive int; got {n_experts!r}"
            )
        # Store the expert count on the module
        self.n_experts: int = n_experts
        # Build the expert blocks receiving x only
        self.experts: nn.ModuleList = nn.ModuleList(
            [
                block_factory(feature_dim, expert_width)
                for _ in range(n_experts)
            ]
        )
        # Build the gate receiving the concatenated (x, z)
        self.gate: nn.Module = gate_factory(
            feature_dim + regime_dim, n_experts
        )
        # Initialize the last-probabilities buffer to None
        self.last_probs: Optional[torch.Tensor] = None

    def forward(
        self, x: torch.Tensor, z: torch.Tensor
    ) -> torch.Tensor:
        """
        Compute the gate-weighted sum of the expert forecasts.

        Implements y_hat = sum_k pi_k e_k(x) with pi = softmax(g(x, z)).

        Parameters
        ----------
        x : torch.Tensor
            Stock features of shape (N, feature_dim).
        z : torch.Tensor
            Regime variables of shape (N, regime_dim).

        Returns
        -------
        torch.Tensor
            The forecast of shape (N, 1).
        """
        # Concatenate the features and regime variables for the gate
        u: torch.Tensor = torch.cat([x, z], dim=-1)
        # Compute the gate probabilities: pi = softmax(g(x, z))
        probs: torch.Tensor = self.gate(u)
        # Retain the probabilities for routing-collapse diagnosis
        self.last_probs = probs.detach()
        # Stack the expert forecasts into a tensor of shape (N, K, 1)
        expert_out: torch.Tensor = torch.stack(
            [expert(x) for expert in self.experts], dim=1
        )
        # Squeeze the trailing singleton dimension to (N, K)
        expert_out = expert_out.squeeze(-1)
        # Compute the weighted sum: y_hat = sum_k pi_k e_k(x)
        out: torch.Tensor = (probs * expert_out).sum(dim=1)
        # Reshape the forecast to (N, 1)
        out = out.unsqueeze(-1)
        # Return the MoE forecast of shape (N, 1)
        return out


# -----------------------------------------------------------------------
# Task 7, Step 3: Verify Capacity and Initialization Alignment
# -----------------------------------------------------------------------


def verify_capacity_alignment(
    registry: Dict[str, Callable[..., nn.Module]],
    config: Dict[str, Any],
) -> List[str]:
    """
    Verify measured parameter counts against the manuscript expectations.

    Parameters
    ----------
    registry : Dict[str, Callable]
        The model-name to factory mapping.
    config : Dict[str, Any]
        The configuration dictionary.

    Returns
    -------
    List[str]
        A list of capacity findings; empty when all counts match.
    """
    # Define the expected counts for the capacity audit
    expected: Dict[str, int] = {
        "MLP_S": 561,
        "MLP_L": 2773,
        "MLP_S_plus_z": 593,
        "MLP_L_plus_z": 2861,
    }
    # Initialize the findings list
    findings: List[str] = []
    # Iterate over every expected count
    for name, expected_count in expected.items():
        # Skip models that are not in the registry
        if name not in registry:
            # Continue to the next expected count when absent
            continue
        # Instantiate a fresh model from the registry factory using the
        # unified config-accepting contract
        model: nn.Module = registry[name](config)
        # Measure the parameter count of the fresh model
        measured: int = count_parameters(model)
        # Compare the measured count with the expectation
        if measured != expected_count:
            # Record the capacity mismatch finding
            findings.append(
                f"{name}: counted {measured}, expected {expected_count}"
            )
    # Return the capacity findings list
    return findings


# -----------------------------------------------------------------------
# Task 7, Orchestrator Function
# -----------------------------------------------------------------------


@dataclass(frozen=True)
class NeuralModelRegistry:
    """
    Immutable registry of neural model factories and capacity findings.

    Attributes
    ----------
    factories : Dict[str, Callable]
        Mapping of model name to a factory returning a fresh module.
    capacity_findings : List[str]
        Capacity mismatches from the verification audit.
    """

    # Mapping of model name to a fresh-module factory
    factories: Dict[str, Callable[..., nn.Module]]
    # Capacity mismatches from the verification audit
    capacity_findings: List[str]


def build_mlp_and_moe_models(
    block_factory: Callable[..., SharedMLPBlock],
    gate_factory: Callable[..., nn.Module],
    config: Dict[str, Any],
) -> NeuralModelRegistry:
    """
    Orchestrate the MLP baseline and standard MoE model construction.

    Composes build_mlp_baseline, build_mlp_plus_z_baseline, StandardMoE,
    and verify_capacity_alignment in the mandated order.

    Parameters
    ----------
    block_factory : Callable
        The Task 6 shared-block factory.
    gate_factory : Callable
        The Task 6 gate factory.
    config : Dict[str, Any]
        The configuration dictionary.

    Returns
    -------
    NeuralModelRegistry
        The immutable model factory registry.
    """
    # Extract the architecture registry from the config
    arch: Dict[str, Any] = config.get("network_architecture_parameters", {})
    # Extract the feature dimension from the config
    feature_dim: int = int(
        config.get("stock_feature_construction", {}).get(
            "feature_dimension", 16
        )
    )
    # Extract the regime dimension from the config
    regime_dim: int = int(
        config.get("regime_state_variable_construction", {}).get(
            "regime_dimension", 2
        )
    )
    # Extract the MoE expert count from the config
    moe_experts: int = int(
        arch.get("standard_MoE", {}).get("expert_count", 4)
    )
    # Initialize the model factory registry. Every factory follows the
    # SINGLE config-accepting contract: factory(config_arg) -> nn.Module.
    # The candidate configuration (learning rate, alpha, lambda_LB) does
    # not alter the architecture, so it is accepted and consumed by the
    # hyperparameter sweeps without needing an adapter.
    factories: Dict[str, Callable[..., nn.Module]] = {}
    # Register the MLP-S baseline factory
    factories["MLP_S"] = lambda config_arg: build_mlp_baseline(
        block_factory, "MLP_S", config
    )
    # Register the MLP-L baseline factory
    factories["MLP_L"] = lambda config_arg: build_mlp_baseline(
        block_factory, "MLP_L", config
    )
    # Register the MLP-S(+z) baseline factory
    factories["MLP_S_plus_z"] = lambda config_arg: build_mlp_plus_z_baseline(
        block_factory, "MLP_S_plus_z", config
    )
    # Register the MLP-L(+z) baseline factory
    factories["MLP_L_plus_z"] = lambda config_arg: build_mlp_plus_z_baseline(
        block_factory, "MLP_L_plus_z", config
    )
    # Register the standard MoE factory
    def moe_factory(config_arg: Dict[str, Any]) -> StandardMoE:
        """
        Build a fresh standard MoE module.

        Parameters
        ----------
        config_arg : Dict[str, Any]
            The candidate configuration; accepted for the unified
            factory contract (the MoE architecture is fixed by the base
            config registry).

        Returns
        -------
        StandardMoE
            A fresh standard MoE instance.
        """
        # Instantiate the standard MoE with the config expert count
        return StandardMoE(
            n_experts=moe_experts,
            feature_dim=feature_dim,
            regime_dim=regime_dim,
            block_factory=block_factory,
            gate_factory=gate_factory,
            expert_width=16,
        )

    # Register the standard MoE factory in the registry
    factories["StandardMoE"] = moe_factory
    # Verify the capacity alignment of the MLP baselines
    findings: List[str] = verify_capacity_alignment(factories, config)
    # Log the capacity verification outcome
    logger.info("Capacity audit: %d findings", len(findings))
    # Assemble the immutable model registry report
    report: NeuralModelRegistry = NeuralModelRegistry(
        factories=factories,
        capacity_findings=findings,
    )
    # Return the model registry report
    return report
````

````python
# Task 8: Implement RG-ResMoE Family and Hard-Routing Variants

# ==============================================================================
# Task 8: Implement RG-ResMoE Family and Hard-Routing Variants
# ==============================================================================

# -----------------------------------------------------------------------
# Task 8, Step 1: Implement RG-ResMoE Main Model
# -----------------------------------------------------------------------

class RGResMoE(nn.Module):
    """
    Regime-gated residual mixture-of-experts model.

    Implements y_base = Block(x; theta_b), r_k(x) = Block(x; theta_k),
    pi = softmax(g(x, z)), and y = y_base + sum_k pi_k r_k(x), with a
    frozen base network and zero-initialized expert final layers so that
    r_k(x) = 0 at initialization.

    Attributes
    ----------
    n_experts : int
        The number of residual experts K.
    base : SharedMLPBlock
        The frozen base network receiving x only.
    experts : nn.ModuleList
        The K residual expert blocks receiving x only.
    gate : nn.Module
        The gate network receiving (x, z) or x for the (-z) variant.
    last_probs : Optional[torch.Tensor]
        The last gate probabilities, retained for routing diagnosis.
    """

    def __init__(
        self,
        n_experts: int,
        feature_dim: int,
        regime_dim: int,
        block_factory: Callable[..., SharedMLPBlock],
        gate_factory: Callable[..., nn.Module],
        expert_width: int = 16,
        base_width: int = 16,
        zero_init_experts: bool = True,
        use_regime_in_gate: bool = True,
    ) -> None:
        """
        Initialize the RG-ResMoE base, experts, and gate.

        Parameters
        ----------
        n_experts : int
            The number of residual experts K.
        feature_dim : int
            The stock-feature dimension.
        regime_dim : int
            The regime-variable dimension.
        block_factory : Callable
            The Task 6 shared-block factory.
        gate_factory : Callable
            The Task 6 gate factory.
        expert_width : int, optional
            The expert hidden width, default 16.
        base_width : int, optional
            The base network hidden width, default 16.
        zero_init_experts : bool, optional
            Zero-initialize the expert final layers when True.
        use_regime_in_gate : bool, optional
            Pass (x, z) to the gate when True; x only when False.

        Raises
        ------
        ValueError
            If n_experts is not a positive integer.
        """
        # Initialize the parent nn.Module class
        super().__init__()
        # Validate the expert count is a positive integer
        if not isinstance(n_experts, int) or n_experts < 1:
            # Raise an informative error for the invalid expert count
            raise ValueError(
                f"n_experts must be a positive int; got {n_experts!r}"
            )
        # Store the expert count on the module
        self.n_experts: int = n_experts
        # Compute the gate input dimension from the regime flag
        gate_dim: int = (
            feature_dim + regime_dim if use_regime_in_gate else feature_dim
        )
        # Store the explicit regime flag on the module: the forward pass
        # selects the gate input from this flag, never from a runtime
        # shape probe
        self.use_regime_in_gate: bool = use_regime_in_gate
        # Build the base network receiving x only
        self.base: SharedMLPBlock = block_factory(feature_dim, base_width)
        # Build the residual expert blocks receiving x only
        self.experts: nn.ModuleList = nn.ModuleList(
            [
                block_factory(feature_dim, expert_width)
                for _ in range(n_experts)
            ]
        )
        # Build the gate with the computed input dimension
        self.gate: nn.Module = gate_factory(gate_dim, n_experts)
        # Validate the gate input dimension against the chosen pathway so
        # a mismatched gate can never silently receive the wrong input
        if self.gate.input_dim != gate_dim:
            # Raise an informative error for the dimension mismatch
            raise ValueError(
                f"Gate input_dim {self.gate.input_dim} must equal the "
                f"configured pathway dimension {gate_dim}."
            )
        # Zero-initialize the expert final layers when requested
        if zero_init_experts:
            # Collect the expert output layers
            output_layers: List[nn.Linear] = [
                expert.fc3 for expert in self.experts
            ]
            # Zero-fill the expert output layers
            zero_init_expert_outputs(output_layers)
        # Initialize the last-probabilities buffer to None
        self.last_probs: Optional[torch.Tensor] = None

    def freeze_base(self) -> None:
        """
        Freeze the base network parameters for Stage 2 training.

        Returns
        -------
        None
        """
        # Iterate over every base network parameter
        for param in self.base.parameters():
            # Disable gradient computation for the parameter
            param.requires_grad = False

    def forward(
        self, x: torch.Tensor, z: torch.Tensor
    ) -> torch.Tensor:
        """
        Compute the residual-gated forecast.

        Implements y = y_base + sum_k pi_k r_k(x) with
        pi = softmax(g(x, z)).

        Parameters
        ----------
        x : torch.Tensor
            Stock features of shape (N, feature_dim).
        z : torch.Tensor
            Regime variables of shape (N, regime_dim).

        Returns
        -------
        torch.Tensor
            The forecast of shape (N, 1).
        """
        # Compute the frozen base forecast: y_base = Block(x; theta_b)
        y_base: torch.Tensor = self.base(x)
        # Build the gate input from the EXPLICIT configuration flag: the
        # full model passes (x, z) to the gate, the (-z) variant passes
        # x only; regime information never enters the base or experts
        u: torch.Tensor = (
            torch.cat([x, z], dim=-1)
            if self.use_regime_in_gate
            else x
        )
        # Compute the gate probabilities: pi = softmax(g(u))
        probs: torch.Tensor = self.gate(u)
        # Retain the probabilities for routing-collapse diagnosis
        self.last_probs = probs.detach()
        # Stack the expert residuals into a tensor of shape (N, K, 1)
        residuals: torch.Tensor = torch.stack(
            [expert(x) for expert in self.experts], dim=1
        )
        # Squeeze the trailing singleton dimension to (N, K)
        residuals = residuals.squeeze(-1)
        # Compute the weighted residual correction: sum_k pi_k r_k(x)
        correction: torch.Tensor = (probs * residuals).sum(dim=1)
        # Compute the final forecast: y = y_base + correction
        out: torch.Tensor = y_base.squeeze(-1) + correction
        # Reshape the forecast to (N, 1)
        out = out.unsqueeze(-1)
        # Return the final forecast of shape (N, 1)
        return out


# -----------------------------------------------------------------------
# Task 8, Step 2: Implement RG-ResMoE Ablations
# -----------------------------------------------------------------------


def build_rg_resmoe_ablation(
    variant_name: str,
    block_factory: Callable[..., SharedMLPBlock],
    gate_factory: Callable[..., nn.Module],
    config: Dict[str, Any],
) -> RGResMoE:
    """
    Build an RG-ResMoE ablation variant from the config registry.

    Parameters
    ----------
    variant_name : str
        One of "RG_ResMoE_minus_z", "RG_ResMoE_random_init",
        "RG_ResMoE_K2", or "RG_ResMoE_K6".
    block_factory : Callable
        The Task 6 shared-block factory.
    gate_factory : Callable
        The Task 6 gate factory.
    config : Dict[str, Any]
        The configuration dictionary.

    Returns
    -------
    RGResMoE
        A fresh ablation variant instance.

    Raises
    ------
    ValueError
        If the variant name is unknown.
    """
    # Extract the architecture registry from the config
    arch: Dict[str, Any] = config.get("network_architecture_parameters", {})
    # Extract the feature dimension from the config
    feature_dim: int = int(
        config.get("stock_feature_construction", {}).get(
            "feature_dimension", 16
        )
    )
    # Extract the regime dimension from the config
    regime_dim: int = int(
        config.get("regime_state_variable_construction", {}).get(
            "regime_dimension", 2
        )
    )
    # Resolve the variant configuration from the registry
    if variant_name == "RG_ResMoE_minus_z":
        # Build the (-z) variant with the regime pathway removed
        return RGResMoE(
            n_experts=4,
            feature_dim=feature_dim,
            regime_dim=regime_dim,
            block_factory=block_factory,
            gate_factory=gate_factory,
            expert_width=16,
            base_width=16,
            zero_init_experts=True,
            use_regime_in_gate=False,
        )
    # Resolve the random-init variant from the registry
    if variant_name == "RG_ResMoE_random_init":
        # Build the random-init variant without zero initialization
        return RGResMoE(
            n_experts=4,
            feature_dim=feature_dim,
            regime_dim=regime_dim,
            block_factory=block_factory,
            gate_factory=gate_factory,
            expert_width=16,
            base_width=16,
            zero_init_experts=False,
            use_regime_in_gate=True,
        )
    # Resolve the K2 variant from the registry
    if variant_name == "RG_ResMoE_K2":
        # Extract the K2 expert width from the config
        k2_width: int = int(
            arch.get("RG_ResMoE_K2", {}).get("expert_width", 25)
        )
        # Build the K2 variant with two wider experts
        return RGResMoE(
            n_experts=2,
            feature_dim=feature_dim,
            regime_dim=regime_dim,
            block_factory=block_factory,
            gate_factory=gate_factory,
            expert_width=k2_width,
            base_width=16,
            zero_init_experts=True,
            use_regime_in_gate=True,
        )
    # Resolve the K6 variant from the registry
    if variant_name == "RG_ResMoE_K6":
        # Extract the K6 expert width from the config
        k6_width: int = int(
            arch.get("RG_ResMoE_K6", {}).get("expert_width", 12)
        )
        # Build the K6 variant with six narrower experts
        return RGResMoE(
            n_experts=6,
            feature_dim=feature_dim,
            regime_dim=regime_dim,
            block_factory=block_factory,
            gate_factory=gate_factory,
            expert_width=k6_width,
            base_width=16,
            zero_init_experts=True,
            use_regime_in_gate=True,
        )
    # Raise an informative error for unknown variant names
    raise ValueError(
        f"Unknown RG-ResMoE variant '{variant_name}'; expected one of "
        f"RG_ResMoE_minus_z, RG_ResMoE_random_init, RG_ResMoE_K2, "
        f"RG_ResMoE_K6."
    )


# -----------------------------------------------------------------------
# Task 8, Step 3: Implement Hard-Routing Variants
# -----------------------------------------------------------------------


def build_top1_assignment(
    probs: np.ndarray,
) -> np.ndarray:
    """
    Compute the learned top-1 expert assignment from gate probabilities.

    Implements k* = argmax_k pi_k.

    Parameters
    ----------
    probs : np.ndarray
        The gate probability matrix of shape (N, K).

    Returns
    -------
    np.ndarray
        The integer assignment array of shape (N,).
    """
    # Compute the argmax over the expert axis
    assignment: np.ndarray = np.argmax(probs, axis=1)
    # Return the integer assignment array
    return assignment


def fit_quantile_boundaries(
    train_series: pd.Series,
    n_bins: int,
) -> np.ndarray:
    """
    Fit equal-frequency quantile bin edges on the TRAINING split only.

    The config prescribes boundary_source = training_split_only: the
    bin edges for volatility-quantile hard routing must be estimated
    exclusively from training rows and then applied to validation and
    test rows, preventing any look-ahead in the routing rule.

    Parameters
    ----------
    train_series : pd.Series
        The training-split market-volatility values.
    n_bins : int
        The number of quantile bins (4 for the study).

    Returns
    -------
    np.ndarray
        The n_bins + 1 bin edges, including the minimum and maximum.
    """
    # Validate the bin count is a positive integer
    if not isinstance(n_bins, int) or n_bins < 1:
        # Raise an informative error for the invalid bin count
        raise ValueError(
            f"n_bins must be a positive int; got {n_bins!r}"
        )
    # Guard the degenerate case of too few training observations
    if len(train_series) < n_bins:
        # Raise an informative error for the short training series
        raise ValueError(
            f"Training series has {len(train_series)} observations but "
            f"{n_bins} bins are requested."
        )
    # Compute the equal-frequency quantile edges from training rows
    quantiles: np.ndarray = np.linspace(0.0, 1.0, n_bins + 1)
    # Compute the bin edges from the training quantiles
    edges: np.ndarray = np.quantile(
        train_series.to_numpy(dtype=np.float64), quantiles
    )
    # Deduplicate the edges so degenerate bins collapse deterministically
    edges = np.unique(edges)
    # Return the stored bin edges for the apply stage
    return edges


def build_quantile_assignment(
    market_volatility: pd.Series,
    n_bins: int,
    boundaries: Optional[np.ndarray] = None,
) -> np.ndarray:
    """
    Compute the volatility-quantile expert assignment.

    Assigns each observation to one of n_bins equal-frequency bins on
    the market-volatility variable. When `boundaries` (fitted on the
    training split) are supplied, they are applied to the given rows;
    otherwise the boundaries are fitted on the rows passed, which must
    therefore be training rows (a warning is emitted to make the
    fit-mode explicit).

    Parameters
    ----------
    market_volatility : pd.Series
        The market-volatility values for the assignment rows.
    n_bins : int
        The number of quantile bins.
    boundaries : Optional[np.ndarray], optional
        Pre-fitted bin edges from fit_quantile_boundaries; when given,
        the assignment applies them (no re-fitting, no look-ahead).

    Returns
    -------
    np.ndarray
        The integer assignment array of shape (N,).
    """
    # Fit the boundaries from the passed rows when none were supplied
    if boundaries is None:
        # Warn that this call fits boundaries on the passed data
        logger.warning(
            "build_quantile_assignment received no boundaries; fitting "
            "on the passed rows. Pass fit_quantile_boundaries output "
            "for validation/test rows to avoid look-ahead."
        )
        # Fit the equal-frequency edges on the passed series
        edges: np.ndarray = fit_quantile_boundaries(
            market_volatility, n_bins
        )
    else:
        # Use the pre-fitted training boundaries verbatim
        edges = boundaries
    # Clip the values to the stored edge range so observations outside
    # the training span map deterministically to the end bins
    clipped: pd.Series = market_volatility.clip(edges[0], edges[-1])
    # Apply the stored edges with include_lowest for the first bin
    bins: pd.Series = pd.cut(
        clipped,
        bins=edges,
        labels=False,
        include_lowest=True,
    )
    # Guard residual missing values (should not occur after the
    # completeness drop) by mapping them to the first bin with a warning
    if bins.isna().any():
        # Log the missing-value handling for the decision log
        logger.warning(
            "%d rows had missing market volatility in the quantile "
            "assignment; mapped to the first bin",
            int(bins.isna().sum()),
        )
        # Fill the missing bin labels with the first bin
        bins = bins.fillna(0)
    # Convert the bin labels to an integer array
    assignment: np.ndarray = bins.to_numpy(dtype=int)
    # Return the integer assignment array
    return assignment


def build_gics_assignment(
    sectors: pd.Series,
    sector_mapping: Dict[str, int],
) -> np.ndarray:
    """
    Compute the GICS-sector expert assignment from the config mapping.

    Parameters
    ----------
    sectors : pd.Series
        The GICS sector of every assignment row.
    sector_mapping : Dict[str, int]
        The deterministic sector-to-expert mapping from the config.

    Returns
    -------
    np.ndarray
        The integer assignment array of shape (N,).

    Raises
    ------
    ValueError
        If any sector is missing from the mapping.
    """
    # Map every sector through the config mapping
    assignment: np.ndarray = sectors.map(sector_mapping).to_numpy()
    # Verify that every sector was mapped
    if np.isnan(assignment.astype(float)).any():
        # Raise an informative error for unmapped sectors
        raise ValueError(
            f"Sector mapping missing entries; unmapped sectors present."
        )
    # Return the integer assignment array
    return assignment.astype(int)


def fit_market_idio_medians(
    train_market_volatility: pd.Series,
    train_idio_volatility: pd.Series,
) -> Tuple[float, float]:
    """
    Fit the market x idio grid medians on the TRAINING split only.

    The config prescribes that the market x idio 2x2 grid splits at the
    medians of the TRAINING-split variables; the medians are fitted here
    and then applied to validation and test rows.

    Parameters
    ----------
    train_market_volatility : pd.Series
        The training-split market-volatility values.
    train_idio_volatility : pd.Series
        The training-split idiosyncratic-volatility values.

    Returns
    -------
    Tuple[float, float]
        The market-volatility median and the idiosyncratic-volatility
        median.
    """
    # Compute the training median of the market volatility
    market_median: float = float(train_market_volatility.median())
    # Compute the training median of the idiosyncratic volatility
    idio_median: float = float(train_idio_volatility.median())
    # Return the two stored medians for the apply stage
    return market_median, idio_median


def build_market_idio_assignment(
    market_volatility: pd.Series,
    idio_volatility: pd.Series,
    medians: Optional[Tuple[float, float]] = None,
) -> np.ndarray:
    """
    Compute the market x idio 2x2 grid expert assignment.

    Splits both variables at their medians, producing four cells. When
    `medians` (fitted on the training split) are supplied, they are
    applied; otherwise the medians are fitted on the passed rows, which
    must therefore be training rows (a warning makes the fit-mode
    explicit).

    Parameters
    ----------
    market_volatility : pd.Series
        The market-volatility values.
    idio_volatility : pd.Series
        The idiosyncratic-volatility values.
    medians : Optional[Tuple[float, float]], optional
        The (market median, idio median) from fit_market_idio_medians.

    Returns
    -------
    np.ndarray
        The integer cell assignment array of shape (N,).
    """
    # Fit the medians from the passed rows when none were supplied
    if medians is None:
        # Warn that this call fits medians on the passed data
        logger.warning(
            "build_market_idio_assignment received no medians; fitting "
            "on the passed rows. Pass fit_market_idio_medians output "
            "for validation/test rows to avoid look-ahead."
        )
        # Fit the training medians on the passed series
        market_median, idio_median = fit_market_idio_medians(
            market_volatility, idio_volatility
        )
    else:
        # Unpack the pre-fitted training medians
        market_median, idio_median = medians
    # Compute the market-volatility median split with the stored median
    market_high: pd.Series = market_volatility > market_median
    # Compute the idiosyncratic-volatility median split
    idio_high: pd.Series = idio_volatility > idio_median
    # Encode the four cells as integers 0..3
    assignment: np.ndarray = (
        market_high.astype(int) * 2 + idio_high.astype(int)
    ).to_numpy()
    # Return the integer cell assignment array
    return assignment


def apply_hard_routing_forecast(
    model: RGResMoE,
    x: torch.Tensor,
    z: torch.Tensor,
    assignment: np.ndarray,
) -> torch.Tensor:
    """
    Apply a hard-routing assignment to produce the forecast.

    Implements y = y_base + r_{k*}(x) with k* from the assignment.

    Parameters
    ----------
    model : RGResMoE
        The trained RG-ResMoE model.
    x : torch.Tensor
        Stock features of shape (N, feature_dim).
    z : torch.Tensor
        Regime variables of shape (N, regime_dim).
    assignment : np.ndarray
        The integer expert assignment of shape (N,).

    Returns
    -------
    torch.Tensor
        The hard-routed forecast of shape (N, 1).
    """
    # Validate the assignment length matches the batch
    if assignment.shape[0] != x.shape[0]:
        # Raise an informative error for the length mismatch
        raise ValueError(
            f"Assignment length {assignment.shape[0]} must equal batch "
            f"size {x.shape[0]}."
        )
    # Compute the frozen base forecast
    y_base: torch.Tensor = model.base(x)
    # Stack the expert residuals into a tensor of shape (N, K, 1)
    residuals: torch.Tensor = torch.stack(
        [expert(x) for expert in model.experts], dim=1
    )
    # Squeeze the trailing singleton dimension to (N, K)
    residuals = residuals.squeeze(-1)
    # Build the row indices on the residual tensor's device so the
    # advanced indexing works for CPU and CUDA models alike
    row_indices: torch.Tensor = torch.arange(
        x.shape[0], device=residuals.device
    )
    # Move the assignment indices to the residual tensor's device
    assignment_t: torch.Tensor = torch.as_tensor(
        assignment, device=residuals.device
    )
    # Gather the assigned expert residual: r_{k*}(x)
    assigned: torch.Tensor = residuals[row_indices, assignment_t]
    # Compute the hard-routed forecast: y = y_base + r_{k*}(x)
    out: torch.Tensor = y_base.squeeze(-1) + assigned
    # Reshape the forecast to (N, 1)
    out = out.unsqueeze(-1)
    # Return the hard-routed forecast of shape (N, 1)
    return out


# -----------------------------------------------------------------------
# Task 8, Orchestrator Function
# -----------------------------------------------------------------------


@dataclass(frozen=True)
class RGResMoEFamilyReport:
    """
    Immutable registry of RG-ResMoE variants and routing rules.

    Attributes
    ----------
    variant_factories : Dict[str, Callable]
        Mapping of variant name to a fresh-model factory.
    routing_rules : Dict[str, Callable]
        Mapping of routing-rule name to an assignment callable.
    expert_capacities : Dict[str, int]
        Mapping of variant name to its total expert parameter count.
    """

    # Mapping of variant name to a fresh-model factory
    variant_factories: Dict[str, Callable[..., RGResMoE]]
    # Mapping of routing-rule name to an assignment callable
    routing_rules: Dict[str, Callable[..., np.ndarray]]
    # Mapping of variant name to its total expert parameter count
    expert_capacities: Dict[str, int]


def build_rg_resmoe_family(
    block_factory: Callable[..., SharedMLPBlock],
    gate_factory: Callable[..., nn.Module],
    config: Dict[str, Any],
) -> RGResMoEFamilyReport:
    """
    Orchestrate the RG-ResMoE family and hard-routing rule construction.

    Composes RGResMoE, build_rg_resmoe_ablation, the four routing-rule
    callables, and the capacity verification.

    Parameters
    ----------
    block_factory : Callable
        The Task 6 shared-block factory.
    gate_factory : Callable
        The Task 6 gate factory.
    config : Dict[str, Any]
        The configuration dictionary.

    Returns
    -------
    RGResMoEFamilyReport
        The immutable variant and routing-rule registry.
    """
    # Define the main RG-ResMoE variant factory
    def main_factory(config_arg: Optional[Dict[str, Any]] = None) -> RGResMoE:
        """
        Build a fresh main RG-ResMoE model.

        Parameters
        ----------
        config_arg : Optional[Dict[str, Any]], optional
            The candidate configuration; accepted for the unified
            config-accepting factory contract used by the sweeps.

        Returns
        -------
        RGResMoE
            A fresh main model instance.
        """
        # Extract the feature dimension from the config
        feature_dim: int = int(
            config.get("stock_feature_construction", {}).get(
                "feature_dimension", 16
            )
        )
        # Extract the regime dimension from the config
        regime_dim: int = int(
            config.get("regime_state_variable_construction", {}).get(
                "regime_dimension", 2
            )
        )
        # Instantiate the main RG-ResMoE model
        return RGResMoE(
            n_experts=4,
            feature_dim=feature_dim,
            regime_dim=regime_dim,
            block_factory=block_factory,
            gate_factory=gate_factory,
            expert_width=16,
            base_width=16,
            zero_init_experts=True,
            use_regime_in_gate=True,
        )

    # Build the variant factory registry; every factory accepts the
    # candidate configuration argument for the unified contract
    variant_factories: Dict[str, Callable[..., RGResMoE]] = {
        "RG_ResMoE": main_factory,
        "RG_ResMoE_minus_z": (
            lambda config_arg: build_rg_resmoe_ablation(
                "RG_ResMoE_minus_z", block_factory, gate_factory, config
            )
        ),
        "RG_ResMoE_random_init": (
            lambda config_arg: build_rg_resmoe_ablation(
                "RG_ResMoE_random_init",
                block_factory, gate_factory, config,
            )
        ),
        "RG_ResMoE_K2": (
            lambda config_arg: build_rg_resmoe_ablation(
                "RG_ResMoE_K2", block_factory, gate_factory, config
            )
        ),
        "RG_ResMoE_K6": (
            lambda config_arg: build_rg_resmoe_ablation(
                "RG_ResMoE_K6", block_factory, gate_factory, config
            )
        ),
    }
    # Extract the GICS sector mapping from the config
    gics_mapping: Dict[str, int] = config.get(
        "network_architecture_parameters", {}
    ).get("hard_routing_gics_sectors", {}).get("mapping", {})
    # Build the routing-rule registry
    routing_rules: Dict[str, Callable[..., np.ndarray]] = {
        "learned_top1": build_top1_assignment,
        "volatility_quantiles": build_quantile_assignment,
        "gics_sectors": (
            lambda sectors: build_gics_assignment(sectors, gics_mapping)
        ),
        "market_x_idio": build_market_idio_assignment,
    }
    # Compute the expert capacities of the variants
    expert_capacities: Dict[str, int] = {}
    # Iterate over every variant factory
    for name, factory in variant_factories.items():
        # Instantiate the variant for the capacity measurement
        model: RGResMoE = factory(None)
        # Measure the total expert parameter count
        expert_capacities[name] = count_parameters(model.experts)
    # Assert that every variant's expert capacity stays close to the
    # four-expert baseline of 2244 (K2 = 2202, K6 = 2238, K4 = 2244),
    # which is the capacity-matching design of the manuscript
    for name, capacity in expert_capacities.items():
        # Compute the relative capacity deviation from 2244
        deviation: float = abs(capacity - 2244) / 2244
        # Enforce the capacity-matching band (2.5%)
        if deviation > 0.025:
            # Raise an informative error for the capacity violation
            raise ValueError(
                f"Variant {name} expert capacity {capacity} deviates "
                f"from the 2244 baseline by {deviation:.2%}."
            )
    # Log the variant registry completion
    logger.info(
        "RG-ResMoE family: %d variants, %d routing rules, capacities %s",
        len(variant_factories), len(routing_rules), expert_capacities,
    )
    # Assemble the immutable family report
    report: RGResMoEFamilyReport = RGResMoEFamilyReport(
        variant_factories=variant_factories,
        routing_rules=routing_rules,
        expert_capacities=expert_capacities,
    )
    # Return the family report
    return report
````

````python
# Task 9: Implement Classical Baselines

# ==============================================================================
# Task 9: Implement Classical Baselines
# ==============================================================================

# -----------------------------------------------------------------------
# Task 9, Step 1: Implement Persistence and HAR
# -----------------------------------------------------------------------

def build_persistence_forecast(
    panel: pd.DataFrame,
    config: Dict[str, Any],
) -> pd.Series:
    """
    Build the 20-day persistence volatility forecast per stock-day.

    Implements y_hat_{i,t} = sqrt(252) * Std(r_{i,t-19..t}) with ddof=1.

    Parameters
    ----------
    panel : pd.DataFrame
        The long panel with ticker, date, and return columns.
    config : Dict[str, Any]
        The configuration dictionary.

    Returns
    -------
    pd.Series
        The persistence forecast aligned with the panel rows.
    """
    # Extract the persistence window from the config
    window: int = int(
        config.get("classical_baselines", {})
        .get("persistence", {})
        .get("window", 20)
    )
    # Extract the annualization factor from the config
    annualization: float = float(
        config.get("data_preprocessing_parameters", {}).get(
            "annualization_factor", 252
        )
    )
    # Compute the annualization square root once
    sqrt_ann: float = float(np.sqrt(annualization))
    # Compute the trailing rolling standard deviation within ticker
    rv: pd.Series = (
        panel.groupby("ticker")["return"]
        .rolling(window, min_periods=window)
        .std(ddof=1)
    )
    # Reset the group index for aligned assignment
    rv = rv.reset_index(level=0, drop=True)
    # Annualize the rolling standard deviation
    forecast: pd.Series = rv * sqrt_ann
    # Return the persistence forecast aligned with the panel
    return forecast


def build_har_regressors(
    panel: pd.DataFrame,
    config: Dict[str, Any],
) -> pd.DataFrame:
    """
    Build the HAR regressors RV(1), RV(5), RV(22) on the panel.

    Parameters
    ----------
    panel : pd.DataFrame
        The long panel with ticker, date, and return columns.
    config : Dict[str, Any]
        The configuration dictionary.

    Returns
    -------
    pd.DataFrame
        The regressor frame with columns rv_1d, rv_5d, rv_22d.
    """
    # Extract the HAR windows from the config
    windows: List[int] = list(
        config.get("classical_baselines", {})
        .get("HAR", {})
        .get("windows", [1, 5, 22])
    )
    # Extract the annualization factor from the config
    annualization: float = float(
        config.get("data_preprocessing_parameters", {}).get(
            "annualization_factor", 252
        )
    )
    # Compute the annualization square root once
    sqrt_ann: float = float(np.sqrt(annualization))
    # Initialize the regressor frame
    regressors: pd.DataFrame = pd.DataFrame(index=panel.index)
    # Iterate over every HAR window
    for window in windows:
        # Special-case the one-day window: the sample standard deviation
        # of a single observation with ddof=1 is undefined (NaN), so the
        # daily realized volatility is defined as |r_t| * sqrt(252), the
        # standard one-day realized-volatility convention; the decision
        # log must record this convention
        if window == 1:
            # Compute the absolute return as the daily RV proxy
            rv: pd.Series = panel.groupby("ticker")["return"].transform(
                lambda s: s.abs()
            )
        else:
            # Compute the rolling standard deviation within ticker
            rv = (
                panel.groupby("ticker")["return"]
                .rolling(window, min_periods=window)
                .std(ddof=1)
            )
            # Reset the group index for aligned assignment
            rv = rv.reset_index(level=0, drop=True)
        # Annualize and store the regressor column
        regressors[f"rv_{window}d"] = rv * sqrt_ann
    # Log the RV(1) convention for the decision log
    logger.info(
        "HAR RV(1) defined as |r_t| * sqrt(252) (single-observation "
        "sample std is undefined with ddof=1)"
    )
    # Return the HAR regressor frame
    return regressors


def fit_har_pooled_ols(
    regressors: pd.DataFrame,
    target: pd.Series,
    train_mask: np.ndarray,
) -> Dict[str, float]:
    """
    Fit the pooled HAR OLS coefficients on the training rows.

    Implements y = c + b_d RV(1) + b_w RV(5) + b_m RV(22) + eps.

    Parameters
    ----------
    regressors : pd.DataFrame
        The HAR regressor frame.
    target : pd.Series
        The forward-target series aligned with the regressors.
    train_mask : np.ndarray
        Boolean mask selecting the training rows.

    Returns
    -------
    Dict[str, float]
        The fitted coefficients c, b_d, b_w, b_m.
    """
    # Select the training regressor rows
    x_train: np.ndarray = regressors.loc[train_mask].to_numpy(
        dtype=np.float64
    )
    # Select the training target rows
    y_train: np.ndarray = target.loc[train_mask].to_numpy(
        dtype=np.float64
    )
    # Drop rows with missing regressors or targets
    valid: np.ndarray = np.isfinite(x_train).all(axis=1) & np.isfinite(
        y_train
    )
    # Filter to the valid training regressor rows
    x_valid: np.ndarray = x_train[valid]
    # Filter to the valid training target rows
    y_valid: np.ndarray = y_train[valid]
    # Guard against an empty design after masking: a pooled OLS fit on
    # zero rows would silently return a null model, so it must raise
    if x_valid.shape[0] < 2:
        # Raise an informative error for the degenerate design
        raise ValueError(
            f"Pooled HAR OLS requires at least 2 valid training rows; "
            f"only {x_valid.shape[0]} remain after the non-finite mask."
        )
    # Build the design matrix with the intercept column
    design: np.ndarray = np.column_stack(
        [np.ones(x_valid.shape[0]), x_valid]
    )
    # Guard against a rank-deficient design (e.g. constant regressors)
    if np.linalg.matrix_rank(design) < design.shape[1]:
        # Raise an informative error for the rank-deficient design
        raise ValueError(
            "Pooled HAR design matrix is rank-deficient; the OLS "
            "coefficients would not be identified."
        )
    # Solve the pooled OLS normal equations
    coefficients, _, _, _ = np.linalg.lstsq(design, y_valid, rcond=None)
    # Unpack the fitted intercept coefficient
    c: float = float(coefficients[0])
    # Unpack the fitted daily RV coefficient
    b_d: float = float(coefficients[1])
    # Unpack the fitted weekly RV coefficient
    b_w: float = float(coefficients[2])
    # Unpack the fitted monthly RV coefficient
    b_m: float = float(coefficients[3])
    # Return the fitted HAR coefficients
    return {"c": c, "b_d": b_d, "b_w": b_w, "b_m": b_m}


def predict_har(
    regressors: pd.DataFrame,
    coefficients: Dict[str, float],
) -> pd.Series:
    """
    Apply the fitted HAR coefficients to produce forecasts.

    Parameters
    ----------
    regressors : pd.DataFrame
        The HAR regressor frame.
    coefficients : Dict[str, float]
        The fitted coefficients c, b_d, b_w, b_m.

    Returns
    -------
    pd.Series
        The HAR forecast aligned with the regressor rows.
    """
    # Compute the linear forecast: y_hat = c + b_d RV1 + b_w RV5 + b_m RV22
    forecast: pd.Series = (
        coefficients["c"]
        + coefficients["b_d"] * regressors["rv_1d"]
        + coefficients["b_w"] * regressors["rv_5d"]
        + coefficients["b_m"] * regressors["rv_22d"]
    )
    # Return the HAR forecast series
    return forecast


# -----------------------------------------------------------------------
# Task 9, Step 2: Implement Per-Stock GARCH(1,1)
# -----------------------------------------------------------------------


def garch_negative_log_likelihood(
    params: np.ndarray,
    returns: np.ndarray,
) -> float:
    """
    Compute the negative log-likelihood of a GARCH(1,1) model.

    Implements sigma_t^2 = omega + alpha eps_{t-1}^2 + beta sigma_{t-1}^2
    with normal innovations.

    Parameters
    ----------
    params : np.ndarray
        The parameter vector (omega, alpha, beta).
    returns : np.ndarray
        The stock's return series.

    Returns
    -------
    float
        The negative log-likelihood value.
    """
    # Unpack the model constant parameter
    omega: float = float(params[0])
    # Unpack the ARCH coefficient parameter
    alpha: float = float(params[1])
    # Unpack the GARCH coefficient parameter
    beta: float = float(params[2])
    # Initialize the conditional variance series
    variance: np.ndarray = np.empty_like(returns, dtype=np.float64)
    # Initialize the first variance with the unconditional estimate
    variance[0] = float(np.var(returns, ddof=1))
    # Iterate over the return series
    for t in range(1, len(returns)):
        # Apply the variance recursion: sigma_t^2 = omega + a e^2 + b s^2
        variance[t] = (
            omega
            + alpha * returns[t - 1] ** 2
            + beta * variance[t - 1]
        )
    # Compute the log-likelihood of the normal innovations
    log_lik: float = -0.5 * np.sum(
        np.log(2.0 * np.pi * variance)
        + returns ** 2 / variance
    )
    # Return the negative log-likelihood
    return -float(log_lik)


def fit_stock_garch(
    returns: pd.Series,
) -> Dict[str, float]:
    """
    Fit a GARCH(1,1) model to one stock's return series by MLE.

    Parameters
    ----------
    returns : pd.Series
        The stock's training-window return series.

    Returns
    -------
    Dict[str, float]
        The fitted parameters omega, alpha, beta.

    Raises
    ------
    ValueError
        If the return series is too short or the fit fails.
    """
    # Validate the return series length
    if len(returns) < 10:
        # Raise an informative error for the short series
        raise ValueError(
            f"GARCH requires at least 10 returns; got {len(returns)}."
        )
    # Extract the returns as a float64 array
    r: np.ndarray = returns.to_numpy(dtype=np.float64)
    # Compute the unconditional variance once for the starting values
    unconditional_var: float = float(np.var(r, ddof=1))
    # Define multiple starting points so a poor initialization cannot
    # sink the MLE; the best converged stationary fit is selected
    start_points: List[np.ndarray] = [
        np.array([0.05 * unconditional_var, 0.1, 0.8]),
        np.array([0.10 * unconditional_var, 0.05, 0.9]),
        np.array([0.01 * unconditional_var, 0.2, 0.7]),
    ]
    # Define the parameter bounds for the optimizer
    bounds: List[Tuple[float, float]] = [
        (1e-6, None), (0.0, None), (0.0, None)
    ]
    # Initialize the candidate fits list
    candidate_fits: List[Dict[str, float]] = []
    # Iterate over every starting point
    for start_params in start_points:
        # Run the constrained maximum-likelihood optimization
        result = minimize(
            garch_negative_log_likelihood,
            start_params,
            args=(r,),
            method="L-BFGS-B",
            bounds=bounds,
        )
        # Skip non-converged optimizations
        if not result.success:
            # Continue to the next starting point
            continue
        # Unpack the fitted constant parameter
        omega: float = float(result.x[0])
        # Unpack the fitted ARCH coefficient
        alpha: float = float(result.x[1])
        # Unpack the fitted GARCH coefficient
        beta: float = float(result.x[2])
        # Skip fits that violate the stationarity constraint
        if alpha + beta >= 1.0:
            # Continue to the next starting point
            continue
        # Record the converged stationary fit with its objective value
        candidate_fits.append(
            {
                "omega": omega,
                "alpha": alpha,
                "beta": beta,
                "nll": float(result.fun),
            }
        )
    # Guard the total-failure case with an informative error
    if not candidate_fits:
        # Raise an informative error for the failed fit
        raise ValueError(
            "GARCH MLE failed: no starting point converged to a "
            "stationary solution (alpha + beta < 1)."
        )
    # Select the fit with the lowest negative log-likelihood
    best_fit: Dict[str, float] = min(
        candidate_fits, key=lambda fit: fit["nll"]
    )
    # Return the fitted GARCH parameters
    return {
        "omega": best_fit["omega"],
        "alpha": best_fit["alpha"],
        "beta": best_fit["beta"],
    }


def garch_5day_forecast(
    params: Dict[str, float],
    config: Dict[str, Any],
) -> float:
    """
    Compute the annualized 5-day GARCH volatility forecast.

    Iterates sigma_hat_{t+h}^2 = omega + (alpha+beta) sigma_hat_{t+h-1}^2
    for h = 1..5, averages the five expected variances, takes the square
    root, and multiplies by sqrt(252).

    Parameters
    ----------
    params : Dict[str, float]
        The fitted GARCH parameters.
    config : Dict[str, Any]
        The configuration dictionary.

    Returns
    -------
    float
        The annualized 5-day volatility forecast.
    """
    # Unpack the fitted constant parameter
    omega: float = float(params["omega"])
    # Unpack the fitted ARCH coefficient
    alpha: float = float(params["alpha"])
    # Unpack the fitted GARCH coefficient
    beta: float = float(params["beta"])
    # Extract the annualization factor from the config
    annualization: float = float(
        config.get("data_preprocessing_parameters", {}).get(
            "annualization_factor", 252
        )
    )
    # Initialize the expected-variance list
    variances: List[float] = []
    # Initialize the first expected variance with the long-run level
    # omega / (1 - alpha - beta). The manuscript does not specify the
    # recursion start; this UNSPECIFIED resolution (long-run variance
    # instead of the last in-sample conditional variance) is recorded in
    # the decision log
    prev_var: float = omega / (1.0 - alpha - beta)
    # Iterate over the five forecast horizons
    for _ in range(5):
        # Apply the recursion: sigma^2_hat_{t+h} = w + (a+b) s^2_hat
        prev_var = omega + (alpha + beta) * prev_var
        # Append the expected variance for the horizon
        variances.append(prev_var)
    # Average the five expected variances
    avg_var: float = float(np.mean(variances))
    # Take the square root and annualize
    forecast: float = float(np.sqrt(avg_var) * np.sqrt(annualization))
    # Return the annualized 5-day volatility forecast
    return forecast


# -----------------------------------------------------------------------
# Task 9, Step 3: Implement Ridge and Ridge(+z)
# -----------------------------------------------------------------------


def build_ridge_model(
    x_train: np.ndarray,
    y_train: np.ndarray,
    alpha: float,
) -> Any:
    """
    Fit a Ridge regression model on the training matrix.

    Parameters
    ----------
    x_train : np.ndarray
        The training feature matrix of shape (N_tr, d).
    y_train : np.ndarray
        The training target vector of shape (N_tr,).
    alpha : float
        The L2 penalty coefficient.

    Returns
    -------
    Any
        The fitted sklearn Ridge model.

    Raises
    ------
    ValueError
        If the training rows are insufficient.
    """
    # Validate the training row count
    if x_train.shape[0] < 2:
        # Raise an informative error for the insufficient rows
        raise ValueError(
            f"Ridge requires at least 2 training rows; got "
            f"{x_train.shape[0]}."
        )
    # Instantiate the Ridge model with the config's lbfgs solver; newer
    # sklearn releases reject lbfgs without positive=True, in which case
    # the solver falls back to the library default ('auto') and the
    # deviation is logged for the decision log (DEVIATION with rationale)
    model: Ridge = Ridge(alpha=alpha, solver="lbfgs")
    # Attempt the fit with the configured solver first
    try:
        # Fit the model on the training matrix and target
        model.fit(x_train, y_train)
    except (TypeError, ValueError) as exc:
        # Log the solver fallback with the reason for the decision log
        logger.warning(
            "Ridge solver 'lbfgs' unavailable (%s); falling back to the "
            "sklearn default solver. Record this DEVIATION in the "
            "decision log.", exc,
        )
        # Fall back to the library default solver
        model = Ridge(alpha=alpha, solver="auto")
        # Fit the model with the fallback solver
        model.fit(x_train, y_train)
    # Return the fitted Ridge model
    return model


def build_ridge_plus_z_matrix(
    x: np.ndarray,
    z: np.ndarray,
) -> np.ndarray:
    """
    Concatenate the regime variables into the feature matrix.

    Parameters
    ----------
    x : np.ndarray
        The stock-feature matrix of shape (N, 16).
    z : np.ndarray
        The regime matrix of shape (N, 2).

    Returns
    -------
    np.ndarray
        The concatenated matrix of shape (N, 18).
    """
    # Concatenate the features and regime variables along the last axis
    xz: np.ndarray = np.concatenate([x, z], axis=1)
    # Return the concatenated matrix
    return xz


# -----------------------------------------------------------------------
# Task 9, Orchestrator Function
# -----------------------------------------------------------------------


@dataclass(frozen=True)
class ClassicalBaselineReport:
    """
    Immutable registry of classical baseline callables.

    Attributes
    ----------
    persistence : Callable
        The persistence forecast builder.
    har_fit : Callable
        The pooled HAR OLS fitter.
    har_predict : Callable
        The HAR forecast applier.
    garch_fit : Callable
        The per-stock GARCH fitter.
    garch_forecast : Callable
        The 5-day GARCH forecast builder.
    ridge_fit : Callable
        The Ridge fitter.
    ridge_plus_z_matrix : Callable
        The (+z) matrix concatenation builder.
    """

    # The persistence forecast builder
    persistence: Callable[..., pd.Series]
    # The HAR regressor builder (RV(1), RV(5), RV(22))
    har_regressors: Callable[..., pd.DataFrame]
    # The pooled HAR OLS fitter
    har_fit: Callable[..., Dict[str, float]]
    # The HAR forecast applier
    har_predict: Callable[..., pd.Series]
    # The per-stock GARCH fitter
    garch_fit: Callable[..., Dict[str, float]]
    # The 5-day GARCH forecast builder
    garch_forecast: Callable[..., float]
    # The Ridge fitter
    ridge_fit: Callable[..., Any]
    # The (+z) matrix concatenation builder
    ridge_plus_z_matrix: Callable[..., np.ndarray]


def build_classical_baselines(
    config: Dict[str, Any],
) -> ClassicalBaselineReport:
    """
    Orchestrate the classical baseline callable registry.

    Composes build_persistence_forecast, build_har_regressors,
    fit_har_pooled_ols, predict_har, fit_stock_garch,
    garch_5day_forecast, build_ridge_model, and
    build_ridge_plus_z_matrix.

    Parameters
    ----------
    config : Dict[str, Any]
        The configuration dictionary.

    Returns
    -------
    ClassicalBaselineReport
        The immutable classical baseline registry.
    """
    # Assemble the classical baseline callable registry
    report: ClassicalBaselineReport = ClassicalBaselineReport(
        persistence=build_persistence_forecast,
        har_regressors=build_har_regressors,
        har_fit=fit_har_pooled_ols,
        har_predict=predict_har,
        garch_fit=fit_stock_garch,
        garch_forecast=garch_5day_forecast,
        ridge_fit=build_ridge_model,
        ridge_plus_z_matrix=build_ridge_plus_z_matrix,
    )
    # Log the baseline registry completion
    logger.info("Classical baseline registry assembled")
    # Return the classical baseline registry
    return report
````

````python
# Task 10: Implement the Full-Batch Training Engine

# ==============================================================================
# Task 10: Implement the Full-Batch Training Engine
# ==============================================================================

# -----------------------------------------------------------------------
# Task 10, Step 1: Full-Batch Adam Optimization Loop
# -----------------------------------------------------------------------

def validate_tensor_alignment(
    x: torch.Tensor,
    y: torch.Tensor,
    z: Optional[torch.Tensor] = None,
) -> None:
    """
    Validate the row alignment of the training tensors.

    Parameters
    ----------
    x : torch.Tensor
        The feature tensor of shape (N, d_x).
    y : torch.Tensor
        The target tensor of shape (N,).
    z : Optional[torch.Tensor]
        The regime tensor of shape (N, d_z); optional.

    Returns
    -------
    None

    Raises
    ------
    ValueError
        If the leading dimensions are misaligned.
    """
    # Validate the feature tensor is two-dimensional
    if x.ndim != 2:
        # Raise an informative error for the invalid feature rank
        raise ValueError(
            f"x must be 2-dimensional (N, d_x); got shape {tuple(x.shape)}"
        )
    # Validate the target tensor is one-dimensional
    if y.ndim != 1:
        # Raise an informative error for the invalid target rank
        raise ValueError(
            f"y must be 1-dimensional (N,); got shape {tuple(y.shape)}"
        )
    # Validate the leading-dimension alignment of x and y
    if x.shape[0] != y.shape[0]:
        # Raise an informative error for the row mismatch
        raise ValueError(
            f"x rows {x.shape[0]} must equal y rows {y.shape[0]}."
        )
    # Validate the regime tensor alignment when provided
    if z is not None and z.shape[0] != x.shape[0]:
        # Raise an informative error for the regime row mismatch
        raise ValueError(
            f"z rows {z.shape[0]} must equal x rows {x.shape[0]}."
        )


def build_adam_optimizer(
    model: nn.Module,
    config: Dict[str, Any],
) -> torch.optim.Optimizer:
    """
    Build the full-batch Adam optimizer from the config.

    Parameters
    ----------
    model : nn.Module
        The model whose trainable parameters are optimized.
    config : Dict[str, Any]
        The configuration dictionary.

    Returns
    -------
    torch.optim.Optimizer
        The configured Adam optimizer.
    """
    # Extract the training parameters from the config
    train_cfg: Dict[str, Any] = config.get(
        "network_training_parameters", {}
    )
    # Extract the learning rate from the config
    learning_rate: float = float(train_cfg.get("learning_rate", 1e-3))
    # Extract the Adam beta1 from the config
    beta1: float = float(train_cfg.get("adam_beta1", 0.9))
    # Extract the Adam beta2 from the config
    beta2: float = float(train_cfg.get("adam_beta2", 0.999))
    # Extract the Adam epsilon from the config
    eps: float = float(train_cfg.get("adam_epsilon", 1e-8))
    # Instantiate the Adam optimizer over the trainable parameters
    optimizer: torch.optim.Optimizer = torch.optim.Adam(
        model.parameters(),
        lr=learning_rate,
        betas=(beta1, beta2),
        eps=eps,
    )
    # Return the configured Adam optimizer
    return optimizer


def run_full_batch_epoch(
    model: nn.Module,
    optimizer: torch.optim.Optimizer,
    x: torch.Tensor,
    y: torch.Tensor,
    z: Optional[torch.Tensor],
    loss_fn: Callable[..., torch.Tensor],
    gradient_clip_norm: float,
) -> float:
    """
    Run one full-batch training epoch.

    One epoch equals exactly one optimizer step because the gradient is
    computed over the entire training split.

    Parameters
    ----------
    model : nn.Module
        The model being trained.
    optimizer : torch.optim.Optimizer
        The Adam optimizer.
    x : torch.Tensor
        The training feature tensor.
    y : torch.Tensor
        The training target tensor.
    z : Optional[torch.Tensor]
        The training regime tensor.
    loss_fn : Callable
        The loss function (model, x, y, z) -> scalar.
    gradient_clip_norm : float
        The gradient norm clipping threshold.

    Returns
    -------
    float
        The training loss of the epoch.
    """
    # Switch the model to training mode
    model.train()
    # Zero the accumulated gradients before the step
    optimizer.zero_grad()
    # Compute the loss over the full batch; the loss functions perform
    # the model forward pass themselves so every epoch runs exactly ONE
    # forward pass over the training split (no discarded predictions)
    loss: torch.Tensor = loss_fn(model, x, y, z)
    # Backpropagate the full-batch loss
    loss.backward()
    # Clip the gradient norms at the configured threshold; the returned
    # total norm is the pre-clip gradient norm used for the stability log
    grad_norm: float = float(
        torch.nn.utils.clip_grad_norm_(
            model.parameters(), gradient_clip_norm
        )
    )
    # Log the gradient norm for the numerical-stability audit
    logger.debug("Full-batch gradient norm: %.6f", grad_norm)
    # Apply the Adam parameter update
    optimizer.step()
    # Return the scalar training loss
    return float(loss.item())


# -----------------------------------------------------------------------
# Task 10, Step 2: Loss Functions and Regularized Objective
# -----------------------------------------------------------------------


def model_forward(
    model: nn.Module,
    x: torch.Tensor,
    z: Optional[torch.Tensor],
) -> torch.Tensor:
    """
    Forward the model with or without the regime tensor.

    The forecasting families have heterogeneous signatures: the shared
    block accepts only x, while the (+z) wrappers, the standard MoE, and
    RG-ResMoE accept (x, z). The signature of the model's forward method
    is inspected so every family is called correctly.

    Parameters
    ----------
    model : nn.Module
        The model producing predictions.
    x : torch.Tensor
        The feature tensor.
    z : Optional[torch.Tensor]
        The regime tensor; passed only when the model accepts it.

    Returns
    -------
    torch.Tensor
        The model prediction tensor.
    """
    # Check whether the model's forward signature accepts a z argument
    accepts_z: bool = "z" in inspect.signature(model.forward).parameters
    # Call the model with the regime tensor only when it accepts it
    if z is not None and accepts_z:
        # Forward with the regime tensor
        return model(x, z)
    # Forward with the features only
    return model(x)


def mse_loss_fn(
    model: nn.Module,
    x: torch.Tensor,
    y: torch.Tensor,
    z: Optional[torch.Tensor],
) -> torch.Tensor:
    """
    Compute the plain MSE loss for MLPs, MoE, and Stage 1.

    Implements L = MSE(y, y_hat).

    Parameters
    ----------
    model : nn.Module
        The model producing predictions.
    x : torch.Tensor
        The feature tensor.
    y : torch.Tensor
        The target tensor.
    z : Optional[torch.Tensor]
        The regime tensor; passed only when the model accepts it.

    Returns
    -------
    torch.Tensor
        The scalar MSE loss.
    """
    # Compute the model predictions with the signature-aware forward
    y_hat: torch.Tensor = model_forward(model, x, z)
    # Squeeze the predictions to the target shape
    y_hat_flat: torch.Tensor = y_hat.squeeze(-1)
    # Compute the mean-squared error between predictions and targets
    loss: torch.Tensor = F.mse_loss(y_hat_flat, y)
    # Return the scalar MSE loss
    return loss


def rg_resmoe_stage2_loss_fn(
    model: nn.Module,
    x: torch.Tensor,
    y: torch.Tensor,
    z: Optional[torch.Tensor],
    alpha: float,
    lambda_lb: float,
) -> torch.Tensor:
    """
    Compute the RG-ResMoE Stage 2 regularized loss.

    Implements L = MSE(y, y_hat) + alpha * mean((sum_k pi_k r_k)^2)
    + lambda_LB * sum_k (mean(pi_k) - 1/K)^2.

    Parameters
    ----------
    model : nn.Module
        The RG-ResMoE model.
    x : torch.Tensor
        The feature tensor.
    y : torch.Tensor
        The target tensor.
    z : Optional[torch.Tensor]
        The regime tensor.
    alpha : float
        The residual-shrinkage coefficient.
    lambda_lb : float
        The load-balancing coefficient.

    Returns
    -------
    torch.Tensor
        The scalar regularized loss.
    """
    # Build the gate input from the explicit model flag: the full model
    # passes (x, z), the (-z) variant passes x only
    u: torch.Tensor = (
        torch.cat([x, z], dim=-1)
        if model.use_regime_in_gate
        else x
    )
    # Compute the LIVE gate probabilities: pi = softmax(g(u)); the
    # probabilities are deliberately NOT detached so the load-balancing
    # term backpropagates into the gate exactly as the manuscript loss
    # L = MSE + alpha*mean((sum_k pi_k r_k)^2) + lambda_LB*sum_k
    # (mean(pi_k) - 1/K)^2 prescribes
    probs: torch.Tensor = model.gate(u)
    # Compute the LIVE expert residuals: r_k(x) of shape (N, K)
    residuals: torch.Tensor = torch.stack(
        [expert(x).squeeze(-1) for expert in model.experts], dim=1
    )
    # Compute the LIVE base forecast: y_base = Block(x; theta_b)
    y_base: torch.Tensor = model.base(x).squeeze(-1)
    # Compute the aggregate residual correction: sum_k pi_k r_k(x)
    correction: torch.Tensor = (probs * residuals).sum(dim=1)
    # Compute the final forecast: y_hat = y_base + correction
    y_hat: torch.Tensor = y_base + correction
    # Compute the MSE term between the forecast and the target
    mse_term: torch.Tensor = F.mse_loss(y_hat, y)
    # Compute the batch-average shrinkage term: mean(correction^2), the
    # weighted-sum-then-square-then-batch-mean order of the manuscript
    shrinkage: torch.Tensor = (correction ** 2).mean()
    # Compute the batch-average routing probabilities per expert
    mean_probs: torch.Tensor = probs.mean(dim=0)
    # Compute the uniform target probability: 1/K
    uniform: float = 1.0 / model.n_experts
    # Compute the load-balancing term: sum_k (mean(pi_k) - 1/K)^2 with
    # LIVE probabilities so the gate receives its gradient
    load_balance: torch.Tensor = (
        (mean_probs - uniform) ** 2
    ).sum()
    # Combine the three loss terms with their coefficients
    loss: torch.Tensor = (
        mse_term + alpha * shrinkage + lambda_lb * load_balance
    )
    # Return the scalar regularized loss
    return loss


# -----------------------------------------------------------------------
# Task 10, Step 3: Early Stopping, Checkpoint Restoration, and Seed Control
# -----------------------------------------------------------------------


# -----------------------------------------------------------------------
# Task 10, Orchestrator Function
# -----------------------------------------------------------------------


@dataclass(frozen=True)
class TrainingResult:
    """
    Immutable result of one full-batch training run.

    Attributes
    ----------
    model : nn.Module
        The trained model with the best checkpoint restored.
    history : pd.DataFrame
        The per-epoch training and validation loss history.
    best_epoch : int
        The epoch index of the best validation loss.
    best_val_loss : float
        The best validation loss value.
    """

    # The trained model with the best checkpoint restored
    model: nn.Module
    # The per-epoch training and validation loss history
    history: pd.DataFrame
    # The epoch index of the best validation loss
    best_epoch: int
    # The best validation loss value
    best_val_loss: float


def train_full_batch(
    model: nn.Module,
    x_train: torch.Tensor,
    y_train: torch.Tensor,
    z_train: Optional[torch.Tensor],
    x_val: torch.Tensor,
    y_val: torch.Tensor,
    z_val: Optional[torch.Tensor],
    config: Dict[str, Any],
    seed: int,
    stage: str = "stage_1",
    checkpoint_dir: Optional[Path] = None,
    run_id: Optional[str] = None,
) -> TrainingResult:
    """
    Orchestrate one full-batch training run with early stopping.

    Composes the full-batch Adam loop, the loss selection, the early
    stopping protocol, and the best-checkpoint restoration.

    Parameters
    ----------
    model : nn.Module
        The model to train.
    x_train : torch.Tensor
        The training feature tensor.
    y_train : torch.Tensor
        The training target tensor.
    z_train : Optional[torch.Tensor]
        The training regime tensor.
    x_val : torch.Tensor
        The validation feature tensor.
    y_val : torch.Tensor
        The validation target tensor.
    z_val : Optional[torch.Tensor]
        The validation regime tensor.
    config : Dict[str, Any]
        The configuration dictionary.
    seed : int
        The seed for deterministic initialization.
    stage : str, optional
        The training stage label, default "stage_1".

    Returns
    -------
    TrainingResult
        The trained model and the training history.

    Raises
    ------
    ValueError
        If the tensors are misaligned or the stage is unknown.
    """
    # Validate the training tensor alignment
    validate_tensor_alignment(x_train, y_train, z_train)
    # Validate the validation tensor alignment
    validate_tensor_alignment(x_val, y_val, z_val)
    # Resolve the checkpoint path when a directory is supplied; the
    # deterministic seed is applied here so every training run is
    # reproducible, while MODEL CONSTRUCTION-TIME seeding is the caller's
    # responsibility (see train_model_for_seed_window in Task 18)
    checkpoint_path: Optional[Path] = None
    # Check whether checkpoint persistence was requested
    if checkpoint_dir is not None:
        # Ensure the checkpoint directory exists
        checkpoint_dir.mkdir(parents=True, exist_ok=True)
        # Build the deterministic checkpoint path for the run
        checkpoint_path = checkpoint_dir / f"{run_id or 'run'}.pt"
        # Warm-start from an existing checkpoint when present, which
        # makes an interrupted run resumable without recomputation
        if checkpoint_path.exists():
            # Load the persisted best state into the model
            model.load_state_dict(
                torch.load(
                    checkpoint_path, map_location="cpu",
                    weights_only=True,
                )
            )
            # Log the checkpoint resumption for the audit trail
            logger.info(
                "Resumed training from checkpoint %s", checkpoint_path
            )
    # Apply the deterministic seed before training
    seed_all_rng(seed)
    # Extract the training parameters from the config
    train_cfg: Dict[str, Any] = config.get(
        "network_training_parameters", {}
    )
    # Extract the maximum epoch count from the config
    max_epochs: int = int(train_cfg.get("max_epochs", 200))
    # Extract the early-stopping patience from the config
    patience: int = int(train_cfg.get("early_stopping_patience", 20))
    # Extract the gradient clip norm from the config
    clip_norm: float = float(train_cfg.get("gradient_clip_norm", 1.0))
    # Extract the regularizer coefficients from the config
    loss_cfg: Dict[str, Any] = config.get(
        "loss_function_and_regularization", {}
    )
    # Extract the residual-shrinkage coefficient
    alpha: float = float(loss_cfg.get("alpha_residual_shrinkage", 0.1))
    # Extract the load-balancing coefficient
    lambda_lb: float = float(
        loss_cfg.get("lambda_LB_load_balancing", 0.01)
    )
    # Build the Adam optimizer over the trainable parameters
    optimizer: torch.optim.Optimizer = build_adam_optimizer(model, config)
    # Select the loss function by the training stage
    if stage == "stage_1":
        # Use the plain MSE loss for Stage 1
        loss_fn = mse_loss_fn
    # Handle the Stage 2 regularized objective
    elif stage == "stage_2":
        # Use the regularized loss for Stage 2 with a closure
        def loss_fn(model_, x_, y_, z_):
            # Delegate to the regularized loss with the coefficients
            return rg_resmoe_stage2_loss_fn(
                model_, x_, y_, z_, alpha, lambda_lb
            )
    else:
        # Raise an informative error for the unknown stage
        raise ValueError(
            f"Unknown stage '{stage}'; expected 'stage_1' or 'stage_2'."
        )
    # Initialize the history records list
    history_records: List[Dict[str, Any]] = []
    # Initialize the best validation loss to infinity
    best_val: float = float("inf")
    # Initialize the best epoch counter
    best_epoch: int = -1
    # Initialize the patience counter
    patience_counter: int = 0
    # Initialize the best state dictionary
    best_state: Optional[Dict[str, Any]] = None
    # Iterate over the maximum epoch count
    for epoch in range(max_epochs):
        # Run one full-batch training epoch
        train_loss: float = run_full_batch_epoch(
            model, optimizer, x_train, y_train, z_train,
            loss_fn, clip_norm,
        )
        # Switch the model to evaluation mode for validation
        model.eval()
        # Compute the validation loss without gradient tracking
        with torch.no_grad():
            # Compute the validation predictions with the
            # signature-aware forward
            y_hat_val: torch.Tensor = model_forward(model, x_val, z_val)
            # Compute the validation MSE
            val_loss: float = float(
                F.mse_loss(y_hat_val.squeeze(-1), y_val).item()
            )
        # Append the epoch history record
        history_records.append(
            {"epoch": epoch, "train_loss": train_loss, "val_loss": val_loss}
        )
        # Check for validation-loss improvement
        if val_loss < best_val:
            # Update the best validation loss
            best_val = val_loss
            # Update the best epoch index
            best_epoch = epoch
            # Copy the current parameter state as the best checkpoint
            best_state = {
                k: v.detach().clone() for k, v in model.state_dict().items()
            }
            # Reset the patience counter on improvement
            patience_counter = 0
        else:
            # Increment the patience counter on stagnation
            patience_counter += 1
            # Stop training when the patience is exhausted
            if patience_counter >= patience:
                # Log the early-stopping event
                logger.info(
                    "Early stop at epoch %d (best epoch %d)",
                    epoch, best_epoch,
                )
                # Break out of the training loop
                break
    # Restore the best checkpoint before returning
    if best_state is not None:
        # Load the best parameter state into the model
        model.load_state_dict(best_state)
        # Persist the best state to disk for auditability and resumability
        if checkpoint_path is not None:
            # Write the best state to the deterministic checkpoint path
            torch.save(model.state_dict(), checkpoint_path)
    # Guard the never-improved case so a silent failure cannot pass as
    # a successful training run
    if best_state is None:
        # Raise an informative error for the never-improved run
        raise ValueError(
            "Validation loss never improved during training; the best "
            "checkpoint is undefined."
        )
    # Build the history DataFrame from the records
    history: pd.DataFrame = pd.DataFrame(history_records)
    # Assemble the immutable training result
    result: TrainingResult = TrainingResult(
        model=model,
        history=history,
        best_epoch=best_epoch,
        best_val_loss=best_val,
    )
    # Return the training result
    return result
````

````python
# Task 11: Hyperparameter Selection Protocol

# ==============================================================================
# Task 11: Hyperparameter Selection Protocol
# ==============================================================================

# -----------------------------------------------------------------------
# Task 11, Step 1: Run the Three-Seed Sweep
# -----------------------------------------------------------------------

def build_candidate_grid(config: Dict[str, Any]) -> List[Dict[str, Any]]:
    """
    Build the candidate hyperparameter grid from the config.

    Parameters
    ----------
    config : Dict[str, Any]
        The configuration dictionary.

    Returns
    -------
    List[Dict[str, Any]]
        The list of candidate configuration dictionaries.
    """
    # Extract the training parameters from the config
    train_cfg: Dict[str, Any] = config.get(
        "network_training_parameters", {}
    )
    # Extract the learning-rate candidate grid
    lr_grid: List[float] = list(
        train_cfg.get("learning_rate_candidate_grid", [1e-3])
    )
    # Extract the loss parameters from the config
    loss_cfg: Dict[str, Any] = config.get(
        "loss_function_and_regularization", {}
    )
    # Extract the alpha candidate grid
    alpha_grid: List[float] = list(
        loss_cfg.get("alpha_candidate_grid", [0.1])
    )
    # Extract the lambda_LB candidate grid
    lambda_grid: List[float] = list(
        loss_cfg.get("lambda_LB_candidate_grid", [0.01])
    )
    # Build the Cartesian product of the swept hyperparameters
    grid: List[Dict[str, Any]] = []
    # Iterate over the full Cartesian product
    for lr, alpha, lambda_lb in itertools.product(
        lr_grid, alpha_grid, lambda_grid
    ):
        # Append the candidate configuration dictionary
        grid.append(
            {
                "learning_rate": lr,
                "alpha": alpha,
                "lambda_LB": lambda_lb,
            }
        )
    # Return the candidate grid
    return grid


def run_single_sweep_config(
    model_factory: Callable[[Dict[str, Any]], torch.nn.Module],
    candidate: Dict[str, Any],
    seed: int,
    splits: Dict[str, Any],
    config: Dict[str, Any],
) -> Dict[str, Any]:
    """
    Train one candidate configuration with one seed and evaluate it.

    Runs the correct per-family protocol (two-stage for RG-ResMoE with
    the candidate alpha/lambda_LB; single-stage MSE otherwise), then
    computes the validation daily Spearman IC series and the pooled
    validation MSE, exactly as the manuscript's hyperparameter procedure
    prescribes.

    Parameters
    ----------
    model_factory : Callable
        The model factory accepting the candidate configuration.
    candidate : Dict[str, Any]
        The candidate hyperparameter configuration.
    seed : int
        The sweep seed.
    splits : Dict[str, Any]
        The training and validation tensors plus the validation dates
        (splits['dates_val']) required for the daily IC.
    config : Dict[str, Any]
        The base configuration dictionary.

    Returns
    -------
    Dict[str, Any]
        The per-configuration, per-seed result record with the daily IC
        series, the mean validation IC, and the validation MSE.

    Raises
    ------
    ValueError
        If the validation dates are missing from the splits.
    """
    # Apply the deterministic seed BEFORE model construction so the
    # initialization is seed-controlled
    seed_all_rng(seed)
    # Build the candidate-augmented configuration
    candidate_config: Dict[str, Any] = dict(config)
    # Merge the candidate hyperparameters into the training section
    candidate_config["network_training_parameters"] = dict(
        config.get("network_training_parameters", {})
    )
    # Override the learning rate with the candidate value
    candidate_config["network_training_parameters"][
        "learning_rate"
    ] = candidate["learning_rate"]
    # Override the alpha coefficient with the candidate value
    candidate_config["loss_function_and_regularization"] = dict(
        config.get("loss_function_and_regularization", {})
    )
    # Override the alpha with the candidate value
    candidate_config["loss_function_and_regularization"][
        "alpha_residual_shrinkage"
    ] = candidate["alpha"]
    # Override the lambda_LB with the candidate value
    candidate_config["loss_function_and_regularization"][
        "lambda_LB_load_balancing"
    ] = candidate["lambda_LB"]
    # Build a fresh model with the candidate configuration
    model = model_factory(candidate_config)
    # Extract the training feature tensor from the splits
    x_train: torch.Tensor = splits["X_train"]
    # Extract the training target tensor from the splits
    y_train: torch.Tensor = splits["y_train"]
    # Extract the training regime tensor from the splits
    z_train: Optional[torch.Tensor] = splits.get("Z_train")
    # Extract the validation feature tensor from the splits
    x_val: torch.Tensor = splits["X_val"]
    # Extract the validation target tensor from the splits
    y_val: torch.Tensor = splits["y_val"]
    # Extract the validation regime tensor from the splits
    z_val: Optional[torch.Tensor] = splits.get("Z_val")
    # Require the validation dates for the daily IC computation
    if "dates_val" not in splits:
        # Raise an informative error for the missing dates
        raise ValueError(
            "splits must carry 'dates_val' so the validation daily "
            "Spearman IC can be computed."
        )
    # Extract the validation dates for the daily IC
    val_dates: pd.Series = pd.Series(splits["dates_val"])
    # Detect the RG-ResMoE family by its two-stage marker
    is_rg: bool = hasattr(model, "freeze_base")
    # Run the correct protocol for the model family
    if is_rg:
        # Stage 1: train the base network on MSE
        train_full_batch(
            model, x_train, y_train, z_train,
            x_val, y_val, z_val,
            candidate_config, seed=seed, stage="stage_1",
        )
        # Freeze the base network for Stage 2
        model.freeze_base()
        # Stage 2: train the experts and gate with the candidate
        # regularizer coefficients
        train_full_batch(
            model, x_train, y_train, z_train,
            x_val, y_val, z_val,
            candidate_config, seed=seed, stage="stage_2",
        )
    else:
        # Single-stage training with the plain MSE loss
        train_full_batch(
            model, x_train, y_train, z_train,
            x_val, y_val, z_val,
            candidate_config, seed=seed, stage="stage_1",
        )
    # Generate the validation predictions without gradients
    model.eval()
    # Compute the validation predictions under no_grad with the
    # signature-aware forward so single-input blocks receive x only
    with torch.no_grad():
        # Compute the validation predictions
        y_hat_val: torch.Tensor = model_forward(model, x_val, z_val)
    # Flatten the predictions to a numpy vector
    pred_vec: np.ndarray = y_hat_val.squeeze(-1).numpy()
    # Flatten the validation targets to a numpy vector
    target_vec: np.ndarray = y_val.numpy()
    # Compute the daily IC series on the validation split
    ic_series: pd.Series = compute_daily_ic(
        pd.Series(pred_vec), pd.Series(target_vec), val_dates
    )
    # Compute the mean validation IC across the validation days
    mean_ic: float = float(ic_series.mean())
    # Compute the pooled validation MSE
    val_mse: float = float(
        ((pred_vec - target_vec) ** 2).mean()
    )
    # Return the per-configuration, per-seed result record
    return {
        "learning_rate": candidate["learning_rate"],
        "alpha": candidate["alpha"],
        "lambda_LB": candidate["lambda_LB"],
        "seed": seed,
        "val_mse": val_mse,
        "mean_ic": mean_ic,
        "daily_ic": ic_series,
    }


def run_three_seed_sweep(
    model_factory: Callable[[Dict[str, Any]], torch.nn.Module],
    splits: Dict[str, Any],
    config: Dict[str, Any],
    sweep_seeds: Tuple[int, ...] = (0, 1, 2),
) -> Tuple[pd.DataFrame, Dict[Tuple[float, float, float, int], pd.Series]]:
    """
    Run the three-seed hyperparameter sweep for one model family.

    Parameters
    ----------
    model_factory : Callable
        The model factory accepting the candidate configuration.
    splits : Dict[str, Any]
        The training and validation tensors plus the validation dates.
    config : Dict[str, Any]
        The base configuration dictionary.
    sweep_seeds : Tuple[int, ...], optional
        The sweep seed registry, default (0, 1, 2).

    Returns
    -------
    Tuple[pd.DataFrame, Dict[Tuple[float, float, float, int], pd.Series]]
        The long sweep-results frame and a map from
        (learning_rate, alpha, lambda_LB, seed) to the daily validation
        IC series, which the statistical tests require.
    """
    # Build the candidate grid from the config
    grid: List[Dict[str, Any]] = build_candidate_grid(config)
    # Initialize the results records list
    records: List[Dict[str, Any]] = []
    # Initialize the daily-IC series map
    daily_ic_map: Dict[Tuple[float, float, float, int], pd.Series] = {}
    # Iterate over every candidate configuration
    for candidate in grid:
        # Iterate over every sweep seed
        for seed in sweep_seeds:
            # Run the single-configuration training run
            record: Dict[str, Any] = run_single_sweep_config(
                model_factory, candidate, seed, splits, config
            )
            # Record the daily IC series under its configuration key
            daily_ic_map[
                (
                    float(candidate["learning_rate"]),
                    float(candidate["alpha"]),
                    float(candidate["lambda_LB"]),
                    int(seed),
                )
            ] = record.pop("daily_ic")
            # Append the record to the results list
            records.append(record)
    # Build the sweep-results frame from the records
    results: pd.DataFrame = pd.DataFrame(records)
    # Return the long sweep-results frame and the daily IC map
    return results, daily_ic_map


# -----------------------------------------------------------------------
# Task 11, Step 2: Compute Validation IC and Validation MSE
# -----------------------------------------------------------------------


def compute_daily_ic(
    predictions: pd.Series,
    targets: pd.Series,
    dates: pd.Series,
) -> pd.Series:
    """
    Compute the daily cross-sectional Spearman information coefficient.

    Implements IC_t = Spearman({y_hat_i,t}, {y_i,t}) per day over the
    intersection of non-missing predictions and targets.

    Parameters
    ----------
    predictions : pd.Series
        The validation predictions aligned with the targets.
    targets : pd.Series
        The validation targets.
    dates : pd.Series
        The date of every row.

    Returns
    -------
    pd.Series
        The daily IC series indexed by date.
    """
    # Build the validation frame with the three series
    frame: pd.DataFrame = pd.DataFrame(
        {"date": dates, "pred": predictions, "target": targets}
    )
    # Drop rows with missing predictions or targets
    frame = frame.dropna(subset=["pred", "target"])
    # Initialize the daily IC records list
    ic_records: List[Dict[str, Any]] = []
    # Iterate over every validation day
    for day, group in frame.groupby("date"):
        # Skip days with fewer than two valid pairs
        if len(group) < 2:
            # Record the skipped day with a missing IC
            ic_records.append({"date": day, "ic": np.nan})
            # Continue to the next validation day
            continue
        # Compute the Spearman rank correlation for the day
        corr, _ = spearmanr(group["pred"], group["target"])
        # Record the daily IC value
        ic_records.append({"date": day, "ic": corr})
    # Build the daily IC series from the records
    ic_series: pd.Series = (
        pd.DataFrame(ic_records).set_index("date")["ic"]
    )
    # Return the daily IC series
    return ic_series


def compute_validation_metrics(
    predictions: pd.Series,
    targets: pd.Series,
    dates: pd.Series,
) -> Dict[str, float]:
    """
    Compute the mean validation IC and the pooled validation MSE.

    Parameters
    ----------
    predictions : pd.Series
        The validation predictions.
    targets : pd.Series
        The validation targets.
    dates : pd.Series
        The date of every row.

    Returns
    -------
    Dict[str, float]
        The mean IC, the IC standard deviation, and the pooled MSE.
    """
    # Compute the daily IC series
    ic_series: pd.Series = compute_daily_ic(predictions, targets, dates)
    # Compute the mean daily IC
    mean_ic: float = float(ic_series.mean())
    # Compute the time-series standard deviation of the daily IC
    std_ic: float = float(ic_series.std(ddof=1))
    # Compute the pooled validation MSE
    pooled_mse: float = float(
        ((predictions - targets) ** 2).mean()
    )
    # Return the validation metric summary
    return {
        "mean_ic": mean_ic,
        "std_ic": std_ic,
        "val_mse": pooled_mse,
    }


# -----------------------------------------------------------------------
# Task 11, Step 3: Apply the Three Selection Filters
# -----------------------------------------------------------------------


def apply_selection_filters(
    summary: pd.DataFrame,
    grid: List[Dict[str, Any]],
) -> Tuple[Optional[Dict[str, Any]], List[str], bool]:
    """
    Apply the three selection filters to the sweep summary.

    Implements the manuscript's three filters on the TRUE validation
    metrics: (i) the configuration must lie strictly inside the search
    grid; (ii) its mean validation IC must exceed the runner-up's mean
    validation IC by more than the runner-up's across-seed standard
    deviation; (iii) its validation MSE must lie within one across-seed
    standard deviation of the sweep's best MSE. When no configuration
    passes all three filters, the caller is signaled to expand the grid;
    the final fallback selects the first ranked configuration satisfying
    filters 1 and 3 and logs the filter-2 violation.

    Parameters
    ----------
    summary : pd.DataFrame
        The per-configuration summary with mean_ic (true Spearman),
        std_ic (across-seed), val_mse, and std_mse (across-seed).
    grid : List[Dict[str, Any]]
        The candidate grid for the interior check.

    Returns
    -------
    Tuple[Optional[Dict[str, Any]], List[str], bool]
        The selected configuration (None when the grid must be
        expanded), the filter-outcome log, and the expand-grid flag.
    """
    # Initialize the filter-outcome log
    log: List[str] = []
    # Rank the configurations by mean validation IC descending
    ranked: pd.DataFrame = summary.sort_values(
        "mean_ic", ascending=False
    ).reset_index(drop=True)
    # Identify the best configuration
    best: pd.Series = ranked.iloc[0]
    # Handle the single-candidate case without a runner-up
    if len(ranked) < 2:
        # Log the single-candidate selection
        log.append(
            "Single-candidate sweep: no runner-up; filters 2/3 skipped"
        )
        # Return the only configuration without expansion
        return {
            "learning_rate": float(best["learning_rate"]),
            "alpha": float(best["alpha"]),
            "lambda_LB": float(best["lambda_LB"]),
        }, log, False
    # Identify the runner-up configuration
    runner_up: pd.Series = ranked.iloc[1]
    # Filter 1: the best must lie strictly inside the grid
    # Extract the learning-rate grid extremes for the interior check
    lr_values: List[float] = sorted({c["learning_rate"] for c in grid})
    # Extract the alpha grid extremes for the interior check
    alpha_values: List[float] = sorted({c["alpha"] for c in grid})
    # Extract the lambda_LB grid extremes for the interior check
    lambda_values: List[float] = sorted({c["lambda_LB"] for c in grid})
    # Check the interior condition for the learning rate
    lr_interior: bool = (
        best["learning_rate"] > lr_values[0]
        and best["learning_rate"] < lr_values[-1]
    )
    # Check the interior condition for alpha
    alpha_interior: bool = (
        best["alpha"] > alpha_values[0]
        and best["alpha"] < alpha_values[-1]
    )
    # Check the interior condition for lambda_LB
    lambda_interior: bool = (
        best["lambda_LB"] > lambda_values[0]
        and best["lambda_LB"] < lambda_values[-1]
    )
    # Combine the interior checks into Filter 1
    filter_1: bool = lr_interior and alpha_interior and lambda_interior
    # Record the Filter 1 outcome
    log.append(f"Filter 1 (interior of grid): {filter_1}")
    # Filter 2: beat the runner-up's mean IC by more than the runner-up's
    # across-seed standard deviation (strict inequality)
    filter_2: bool = (
        best["mean_ic"] - runner_up["mean_ic"]
        > runner_up["std_ic"]
    )
    # Record the Filter 2 outcome
    log.append(f"Filter 2 (IC dominance): {filter_2}")
    # Filter 3: validation MSE within one ACROSS-SEED std of the sweep's
    # best MSE (the std of the best configuration, not of the grid)
    min_mse: float = float(summary["val_mse"].min())
    # Extract the across-seed MSE standard deviation of the best config
    best_std_mse: float = float(best["std_mse"])
    # Check the MSE robustness condition
    filter_3: bool = best["val_mse"] <= min_mse + best_std_mse
    # Record the Filter 3 outcome
    log.append(f"Filter 3 (MSE robustness): {filter_3}")
    # Select the best configuration when all three filters pass
    if filter_1 and filter_2 and filter_3:
        # Return the best configuration without expansion
        return {
            "learning_rate": float(best["learning_rate"]),
            "alpha": float(best["alpha"]),
            "lambda_LB": float(best["lambda_LB"]),
        }, log, False
    # Signal the caller to EXPAND the grid and repeat, exactly as the
    # manuscript's fallback procedure prescribes
    log.append(
        "No configuration passed all three filters; the grid must be "
        "expanded and the sweep repeated."
    )
    # Return no selection with the expansion flag set
    return None, log, True


# -----------------------------------------------------------------------
# Task 11, Orchestrator Function
# -----------------------------------------------------------------------


@dataclass(frozen=True)
class HyperparameterSelectionReport:
    """
    Immutable report of the hyperparameter selection protocol.

    Attributes
    ----------
    selected_config : Dict[str, Any]
        The selected hyperparameter configuration.
    sweep_results : pd.DataFrame
        The long three-seed sweep results frame.
    filter_log : List[str]
        The filter-outcome and fallback log.
    daily_ic_map : Dict[Tuple[float, float, float, int], pd.Series]
        The per-configuration, per-seed daily validation IC series,
        required by the statistical tests.
    """

    # The selected hyperparameter configuration
    selected_config: Dict[str, Any]
    # The long three-seed sweep results frame
    sweep_results: pd.DataFrame
    # The filter-outcome and fallback log
    filter_log: List[str]
    # The per-configuration, per-seed daily validation IC series
    daily_ic_map: Dict[Tuple[float, float, float, int], pd.Series] = (
        field(default_factory=dict)
    )


def _expand_config_grid(config: Dict[str, Any]) -> Dict[str, Any]:
    """
    Expand the candidate hyperparameter grids by one order of magnitude
    on each side, per the manuscript's grid-expansion fallback.

    Parameters
    ----------
    config : Dict[str, Any]
        The base configuration dictionary.

    Returns
    -------
    Dict[str, Any]
        A copy of the config with expanded candidate grids.
    """
    # Copy the configuration for the expansion
    expanded: Dict[str, Any] = dict(config)
    # Copy the training section for the learning-rate grid expansion
    expanded["network_training_parameters"] = dict(
        config.get("network_training_parameters", {})
    )
    # Extract the current learning-rate candidate grid
    lr_grid: List[float] = list(
        expanded["network_training_parameters"].get(
            "learning_rate_candidate_grid", [1e-3]
        )
    )
    # Expand the learning-rate grid by one order on each side
    expanded["network_training_parameters"][
        "learning_rate_candidate_grid"
    ] = sorted(set([lr_grid[0] / 10.0] + lr_grid + [lr_grid[-1] * 10.0]))
    # Copy the loss section for the regularizer grid expansion
    expanded["loss_function_and_regularization"] = dict(
        config.get("loss_function_and_regularization", {})
    )
    # Extract the current alpha candidate grid
    alpha_grid: List[float] = list(
        expanded["loss_function_and_regularization"].get(
            "alpha_candidate_grid", [0.1]
        )
    )
    # Expand the alpha grid by one order on each side
    expanded["loss_function_and_regularization"][
        "alpha_candidate_grid"
    ] = sorted(set([alpha_grid[0] / 10.0] + alpha_grid
                   + [alpha_grid[-1] * 10.0]))
    # Extract the current lambda_LB candidate grid
    lambda_grid: List[float] = list(
        expanded["loss_function_and_regularization"].get(
            "lambda_LB_candidate_grid", [0.01]
        )
    )
    # Expand the lambda_LB grid by one order on each side
    expanded["loss_function_and_regularization"][
        "lambda_LB_candidate_grid"
    ] = sorted(set([lambda_grid[0] / 10.0] + lambda_grid
                   + [lambda_grid[-1] * 10.0]))
    # Return the expanded configuration
    return expanded


def _fallback_selection(
    summary: pd.DataFrame,
    grid: List[Dict[str, Any]],
) -> Tuple[Dict[str, Any], List[str]]:
    """
    Select the first ranked configuration satisfying filters 1 and 3,
    logging the filter-2 violation, per the manuscript's final fallback.

    Parameters
    ----------
    summary : pd.DataFrame
        The per-configuration summary with mean_ic, val_mse, std_mse.
    grid : List[Dict[str, Any]]
        The candidate grid for the interior check.

    Returns
    -------
    Tuple[Dict[str, Any], List[str]]
        The fallback configuration and the outcome log.
    """
    # Initialize the outcome log
    log: List[str] = []
    # Rank the configurations by mean validation IC descending
    ranked: pd.DataFrame = summary.sort_values(
        "mean_ic", ascending=False
    ).reset_index(drop=True)
    # Compute the sweep's best MSE and its across-seed std
    min_mse: float = float(summary["val_mse"].min())
    # Extract the grid extremes for the interior checks
    lr_values: List[float] = sorted({c["learning_rate"] for c in grid})
    # Extract the alpha grid extremes
    alpha_values: List[float] = sorted({c["alpha"] for c in grid})
    # Extract the lambda_LB grid extremes
    lambda_values: List[float] = sorted({c["lambda_LB"] for c in grid})
    # Walk the ranked configurations for the filter-1-and-3 candidate
    for _, row in ranked.iterrows():
        # Check the interior condition for the learning rate
        lr_interior: bool = (
            row["learning_rate"] > lr_values[0]
            and row["learning_rate"] < lr_values[-1]
        )
        # Check the interior condition for alpha
        alpha_interior: bool = (
            row["alpha"] > alpha_values[0]
            and row["alpha"] < alpha_values[-1]
        )
        # Check the interior condition for lambda_LB
        lambda_interior: bool = (
            row["lambda_LB"] > lambda_values[0]
            and row["lambda_LB"] < lambda_values[-1]
        )
        # Combine the interior checks into Filter 1
        filter_1: bool = (
            lr_interior and alpha_interior and lambda_interior
        )
        # Compute the Filter 3 condition with the row's across-seed std
        filter_3: bool = (
            row["val_mse"] <= min_mse + row["std_mse"]
        )
        # Select the first ranked config satisfying filters 1 and 3
        if filter_1 and filter_3:
            # Log the fallback application with the filter-2 violation
            log.append(
                "Fallback: selected the first filter-1-and-3 candidate; "
                "filter-2 (IC dominance) violation documented."
            )
            # Return the fallback configuration
            return {
                "learning_rate": float(row["learning_rate"]),
                "alpha": float(row["alpha"]),
                "lambda_LB": float(row["lambda_LB"]),
            }, log
    # Log the total fallback failure
    log.append(
        "Fallback failed: no configuration satisfies filters 1 and 3; "
        "the top-ranked configuration is returned with a warning."
    )
    # Return the top-ranked configuration as the last resort
    return {
        "learning_rate": float(ranked.iloc[0]["learning_rate"]),
        "alpha": float(ranked.iloc[0]["alpha"]),
        "lambda_LB": float(ranked.iloc[0]["lambda_LB"]),
    }, log


def select_hyperparameters(
    model_factory: Callable[[Dict[str, Any]], torch.nn.Module],
    splits: Dict[str, Any],
    config: Dict[str, Any],
) -> HyperparameterSelectionReport:
    """
    Orchestrate the full hyperparameter selection protocol.

    Composes run_three_seed_sweep, the per-configuration aggregation of
    the TRUE validation Spearman IC, and apply_selection_filters in the
    mandated order, with the manuscript's grid-expansion and
    filter-1-and-3 fallbacks. The test split is never touched.

    Parameters
    ----------
    model_factory : Callable
        The model factory accepting the candidate configuration.
    splits : Dict[str, Any]
        The training and validation tensors plus the validation dates.
    config : Dict[str, Any]
        The base configuration dictionary.

    Returns
    -------
    HyperparameterSelectionReport
        The immutable selection report.
    """
    # Step 1: run the three-seed sweep for the model family
    results, daily_ic_map = run_three_seed_sweep(
        model_factory, splits, config
    )
    # Aggregate the sweep results per configuration on the TRUE metrics
    summary: pd.DataFrame = results.groupby(
        ["learning_rate", "alpha", "lambda_LB"], as_index=False
    ).agg(
        mean_ic=("mean_ic", "mean"),
        std_ic=("mean_ic", "std"),
        val_mse=("val_mse", "mean"),
        std_mse=("val_mse", "std"),
    )
    # Fill the single-seed standard deviations with zero
    summary = summary.fillna(0.0)
    # Build the candidate grid for the interior check
    grid: List[Dict[str, Any]] = build_candidate_grid(config)
    # Step 3: apply the three selection filters
    selected, filter_log, expand = apply_selection_filters(
        summary, grid
    )
    # Expand the grid and repeat when no configuration passed
    if expand:
        # Log the expansion for the decision log
        filter_log.append(
            "Expanding the candidate grid by one order of magnitude and "
            "repeating the three-seed sweep."
        )
        # Build the expanded configuration
        expanded_config: Dict[str, Any] = _expand_config_grid(config)
        # Re-run the sweep on the expanded grid
        results2, daily_ic_map2 = run_three_seed_sweep(
            model_factory, splits, expanded_config
        )
        # Aggregate the expanded results per configuration
        summary2: pd.DataFrame = results2.groupby(
            ["learning_rate", "alpha", "lambda_LB"], as_index=False
        ).agg(
            mean_ic=("mean_ic", "mean"),
            std_ic=("mean_ic", "std"),
            val_mse=("val_mse", "mean"),
            std_mse=("val_mse", "std"),
        )
        # Fill the single-seed standard deviations with zero
        summary2 = summary2.fillna(0.0)
        # Re-apply the filters on the expanded summary
        selected, expanded_log, expand2 = apply_selection_filters(
            summary2, build_candidate_grid(expanded_config)
        )
        # Extend the outcome log with the expanded-sweep outcomes
        filter_log.extend(expanded_log)
        # Merge the daily IC maps (expanded run wins on conflicts)
        daily_ic_map.update(daily_ic_map2)
        # Apply the final fallback when the expansion still fails
        if expand2:
            # Select the filter-1-and-3 candidate from the expanded grid
            selected, fallback_log = _fallback_selection(
                summary2, build_candidate_grid(expanded_config)
            )
            # Extend the outcome log with the fallback outcomes
            filter_log.extend(fallback_log)
    # Assemble the immutable selection report
    report: HyperparameterSelectionReport = HyperparameterSelectionReport(
        selected_config=selected,
        sweep_results=results,
        filter_log=filter_log,
        daily_ic_map=daily_ic_map,
    )
    # Return the selection report
    return report
````

````python
# Task 12: Compute Forecast Evaluation Metrics

# ==============================================================================
# Task 12: Compute Forecast Evaluation Metrics
# ==============================================================================

# -----------------------------------------------------------------------
# Task 12, Step 1: Daily Information Coefficient and ICIR
# -----------------------------------------------------------------------


def compute_seed_ic_metrics(
    predictions: pd.Series,
    targets: pd.Series,
    dates: pd.Series,
) -> Tuple[pd.Series, float, float]:
    """
    Compute the daily IC series, mean IC, and ICIR for one seed.

    Implements IC_t = Spearman({y_hat_i,t}, {y_i,t}) per test day and
    ICIR = mean_t(IC_t) / std_t(IC_t).

    Parameters
    ----------
    predictions : pd.Series
        The test predictions aligned with the targets.
    targets : pd.Series
        The test targets.
    dates : pd.Series
        The test date of every row.

    Returns
    -------
    Tuple[pd.Series, float, float]
        The daily IC series, the mean IC, and the ICIR.
    """
    # Reuse the verified daily IC computation from Task 11
    ic_series: pd.Series = compute_daily_ic(
        predictions, targets, dates
    )
    # Compute the mean daily IC
    mean_ic: float = float(ic_series.mean())
    # Compute the time-series standard deviation of the daily IC
    std_ic: float = float(ic_series.std(ddof=1))
    # Compute the ICIR as the mean over the standard deviation
    icir: float = mean_ic / std_ic if std_ic > 0 else float("nan")
    # Return the daily IC series, mean IC, and ICIR
    return ic_series, mean_ic, icir


def aggregate_seed_ic(
    seed_metrics: pd.DataFrame,
) -> Dict[str, float]:
    """
    Aggregate the per-seed IC metrics across seeds as mean ± std.

    Parameters
    ----------
    seed_metrics : pd.DataFrame
        The per-seed frame with mean_ic and icir columns.

    Returns
    -------
    Dict[str, float]
        The aggregate mean IC, IC std, mean ICIR, and ICIR std.
    """
    # Compute the across-seed mean IC
    ic_mean: float = float(seed_metrics["mean_ic"].mean())
    # Compute the across-seed IC standard deviation
    ic_std: float = float(seed_metrics["mean_ic"].std(ddof=1))
    # Compute the across-seed mean ICIR
    icir_mean: float = float(seed_metrics["icir"].mean())
    # Compute the across-seed ICIR standard deviation
    icir_std: float = float(seed_metrics["icir"].std(ddof=1))
    # Return the aggregate IC statistics
    return {
        "ic_mean": ic_mean,
        "ic_std": ic_std,
        "icir_mean": icir_mean,
        "icir_std": icir_std,
    }


# -----------------------------------------------------------------------
# Task 12, Step 2: RMSE, Out-of-Sample R2, and QLIKE
# -----------------------------------------------------------------------


def compute_pooled_metrics(
    predictions: np.ndarray,
    targets: np.ndarray,
    config: Dict[str, Any],
) -> Dict[str, float]:
    """
    Compute the pooled RMSE, out-of-sample R2, and QLIKE.

    Implements RMSE = sqrt(mean((y - y_hat)^2)),
    R2 = 1 - sum((y - y_hat)^2) / sum((y - mean(y))^2), and
    QLIKE = y^2/y_hat^2 - ln(y^2/y_hat^2) - 1 with the variance floor.

    Parameters
    ----------
    predictions : np.ndarray
        The pooled test predictions.
    targets : np.ndarray
        The pooled test targets.
    config : Dict[str, Any]
        The configuration dictionary.

    Returns
    -------
    Dict[str, float]
        The pooled RMSE, R2, and QLIKE values.
    """
    # Extract the QLIKE variance floor from the config
    variance_floor: float = float(
        config.get("loss_function_and_regularization", {})
        .get("numerical_hygiene", {})
        .get("variance_floor", 1e-8)
    )
    # Validate the array lengths match
    if predictions.shape[0] != targets.shape[0]:
        # Raise an informative error for the length mismatch
        raise ValueError(
            f"Predictions length {predictions.shape[0]} must equal "
            f"targets length {targets.shape[0]}."
        )
    # Compute the finite intersection mask
    finite: np.ndarray = np.isfinite(predictions) & np.isfinite(targets)
    # Filter the predictions to the finite intersection
    y_hat: np.ndarray = predictions[finite]
    # Filter the targets to the finite intersection
    y: np.ndarray = targets[finite]
    # Compute the squared errors
    squared_errors: np.ndarray = (y - y_hat) ** 2
    # Compute the pooled RMSE
    rmse: float = float(np.sqrt(np.mean(squared_errors)))
    # Compute the pooled test target mean
    y_mean: float = float(np.mean(y))
    # Compute the total sum of squares around the test mean
    tss: float = float(np.sum((y - y_mean) ** 2))
    # Compute the residual sum of squares
    rss: float = float(np.sum(squared_errors))
    # Compute the out-of-sample R2
    r2: float = 1.0 - rss / tss if tss > 0 else float("nan")
    # Floor the predicted variance for the QLIKE computation
    y_hat_var: np.ndarray = np.maximum(y_hat ** 2, variance_floor)
    # Compute the variance ratio
    ratio: np.ndarray = (y ** 2) / y_hat_var
    # Compute the QLIKE loss per stock-day
    qlike: np.ndarray = ratio - np.log(ratio) - 1.0
    # Average the QLIKE across all pooled stock-days
    qlike_mean: float = float(np.mean(qlike))
    # Return the pooled metric dictionary
    return {"rmse": rmse, "r2": r2, "qlike": qlike_mean}


# -----------------------------------------------------------------------
# Task 12, Step 3: Collapse Diagnostic and Stability Summary
# -----------------------------------------------------------------------


def compute_collapse_summary(
    per_seed_qlike: pd.DataFrame,
    config: Dict[str, Any],
) -> pd.DataFrame:
    """
    Compute the collapsed-seed counts per model.

    Implements the prespecified diagnostic: a seed is collapsed when its
    mean test QLIKE exceeds the config threshold of 2.0.

    Parameters
    ----------
    per_seed_qlike : pd.DataFrame
        The per-seed mean test QLIKE frame with model and seed columns.
    config : Dict[str, Any]
        The configuration dictionary.

    Returns
    -------
    pd.DataFrame
        The per-model collapse-count summary.
    """
    # Extract the collapse threshold from the config
    threshold: float = float(
        config.get("evaluation_metrics", {}).get(
            "collapse_threshold_qlike", 2.0
        )
    )
    # Compute the collapse mask: mean test QLIKE > threshold
    collapsed: pd.Series = per_seed_qlike["qlike"] > threshold
    # Attach the collapse flag to the frame
    per_seed_qlike["collapsed"] = collapsed
    # Aggregate the collapse counts per model
    summary: pd.DataFrame = per_seed_qlike.groupby("model").agg(
        collapsed_count=("collapsed", "sum"),
        total_seeds=("seed", "count"),
    )
    # Reset the index to expose the model column
    summary = summary.reset_index()
    # Return the collapse-count summary
    return summary


def plot_per_seed_qlike(
    per_seed_qlike: pd.DataFrame,
    config: Dict[str, Any],
    out_path: Optional[Path] = None,
) -> Optional[Path]:
    """
    Produce the per-seed QLIKE scatter plot on a log scale.

    Parameters
    ----------
    per_seed_qlike : pd.DataFrame
        The per-seed mean test QLIKE frame.
    config : Dict[str, Any]
        The configuration dictionary.
    out_path : Optional[Path], optional
        The destination path for the figure.

    Returns
    -------
    Optional[Path]
        The written figure path, or None when no path is supplied.
    """
    # Extract the collapse threshold from the config
    threshold: float = float(
        config.get("evaluation_metrics", {}).get(
            "collapse_threshold_qlike", 2.0
        )
    )
    # Extract the figure specification from the config
    plot_cfg: Dict[str, Any] = config.get(
        "plot_construction_parameters", {}
    ).get("figure_4_training_stability", {})
    # Extract the configured figure size
    figure_size: Tuple[float, float] = tuple(
        plot_cfg.get("figure_size_inches", [6.5, 3.5])
    )
    # Extract the collapse-point color from the config
    collapse_color: str = plot_cfg.get(
        "collapse_point_color", "#D62728"
    )
    # Extract the per-model non-collapse colors from the config
    model_colors: Dict[str, str] = plot_cfg.get(
        "non_collapse_point_colors", {}
    )
    # Create the figure with the configured size
    fig, ax = plt.subplots(figsize=figure_size)
    # Iterate over every model group for per-model styling
    for model_name, group in per_seed_qlike.groupby("model"):
        # Select the configured color for the model, else the default
        color: str = model_colors.get(model_name, "#1F77B4")
        # Build the collapse mask for the model's seeds
        collapsed_mask: pd.Series = group["qlike"] > threshold
        # Plot the non-collapsed seeds in the model color
        ax.scatter(
            group.loc[~collapsed_mask, "qlike"],
            [model_name] * int((~collapsed_mask).sum()),
            s=20, color=color, label=model_name,
        )
        # Plot the collapsed seeds in the collapse color
        ax.scatter(
            group.loc[collapsed_mask, "qlike"],
            [model_name] * int(collapsed_mask.sum()),
            s=20, color=collapse_color,
        )
        # Annotate the collapsed count as k/30 on the right side
        collapsed_count: int = int(collapsed_mask.sum())
        # Compute the total seed count of the model
        total_count: int = int(len(group))
        # Place the k/30 annotation at the right margin of the model row
        ax.annotate(
            f"{collapsed_count}/{total_count}",
            xy=(1.0, model_name),
            xycoords=("axes fraction", "data"),
            va="center", ha="left", fontsize=8,
        )
    # Set the x-axis to a logarithmic scale
    ax.set_xscale("log")
    # Draw the collapse threshold line
    ax.axvline(
        threshold, color="#D62728", linestyle="--",
        label=f"collapse threshold ({threshold})",
    )
    # Label the x-axis
    ax.set_xlabel("Mean QLIKE per seed (log scale)")
    # Label the y-axis
    ax.set_ylabel("Model")
    # Add the legend
    ax.legend(loc="upper right")
    # Save the figure when a path is supplied
    if out_path is not None:
        # Ensure the parent directory exists
        out_path.parent.mkdir(parents=True, exist_ok=True)
        # Save the figure with the configured density
        fig.savefig(out_path, dpi=300, bbox_inches="tight")
        # Close the figure to release the memory
        plt.close(fig)
        # Return the written figure path
        return out_path
    # Close the figure when no path is supplied
    plt.close(fig)
    # Return None when no path is supplied
    return None


# -----------------------------------------------------------------------
# Task 12, Orchestrator Function
# -----------------------------------------------------------------------


@dataclass(frozen=True)
class EvaluationMetricsReport:
    """
    Immutable report of the forecast evaluation metrics.

    Attributes
    ----------
    seed_metrics : pd.DataFrame
        The per-seed metric frame.
    daily_ic_series : Dict[int, pd.Series]
        The per-seed daily IC series indexed by seed.
    ic_aggregates : Dict[str, float]
        The across-seed IC and ICIR aggregates.
    collapse_summary : pd.DataFrame
        The per-model collapse-count summary.
    """

    # The per-seed metric frame
    seed_metrics: pd.DataFrame
    # The per-seed daily IC series indexed by seed
    daily_ic_series: Dict[int, pd.Series]
    # The across-seed IC and ICIR aggregates
    ic_aggregates: Dict[str, float]
    # The per-model collapse-count summary
    collapse_summary: pd.DataFrame


def compute_evaluation_metrics(
    per_seed_predictions: Dict[int, pd.DataFrame],
    targets: pd.DataFrame,
    config: Dict[str, Any],
    model_name: str = "evaluated_model",
) -> EvaluationMetricsReport:
    """
    Orchestrate the full evaluation-metric computation.

    Composes compute_seed_ic_metrics, compute_pooled_metrics,
    aggregate_seed_ic, and compute_collapse_summary.

    Parameters
    ----------
    per_seed_predictions : Dict[int, pd.DataFrame]
        The per-seed prediction frames with ticker, date, prediction.
    targets : pd.DataFrame
        The test target frame with ticker, date, target.
    config : Dict[str, Any]
        The configuration dictionary.
    model_name : str, optional
        The model name recorded in the collapse summary, default
        "evaluated_model".

    Returns
    -------
    EvaluationMetricsReport
        The immutable evaluation report.
    """
    # Initialize the per-seed metric records
    seed_records: List[Dict[str, Any]] = []
    # Initialize the daily IC series dictionary
    daily_ic_series: Dict[int, pd.Series] = {}
    # Iterate over every seed's prediction frame
    for seed, pred_frame in per_seed_predictions.items():
        # Merge the predictions with the targets by identity keys
        joined: pd.DataFrame = pred_frame.merge(
            targets, on=["ticker", "date"], how="inner"
        )
        # Skip seeds with no joined rows
        if joined.empty:
            # Log the skipped seed
            logger.warning("Seed %d has no joined predictions", seed)
            # Continue to the next seed
            continue
        # Compute the daily IC, mean IC, and ICIR for the seed
        ic_series, mean_ic, icir = compute_seed_ic_metrics(
            joined["prediction"], joined["target"], joined["date"]
        )
        # Store the daily IC series for the seed
        daily_ic_series[seed] = ic_series
        # Compute the pooled RMSE, R2, and QLIKE for the seed
        pooled: Dict[str, float] = compute_pooled_metrics(
            joined["prediction"].to_numpy(dtype=np.float64),
            joined["target"].to_numpy(dtype=np.float64),
            config,
        )
        # Append the per-seed metric record
        seed_records.append(
            {
                "seed": seed,
                "mean_ic": mean_ic,
                "icir": icir,
                "rmse": pooled["rmse"],
                "r2": pooled["r2"],
                "qlike": pooled["qlike"],
            }
        )
    # Build the per-seed metric frame
    seed_metrics: pd.DataFrame = pd.DataFrame(seed_records)
    # Aggregate the IC statistics across seeds
    ic_aggregates: Dict[str, float] = aggregate_seed_ic(seed_metrics)
    # Build the per-seed QLIKE frame for the collapse diagnostic
    per_seed_qlike: pd.DataFrame = seed_metrics[["seed", "qlike"]].copy()
    # Add the model column for the collapse summary with the caller's
    # model name so the summary is labeled correctly
    per_seed_qlike["model"] = model_name
    # Compute the collapse summary
    collapse_summary: pd.DataFrame = compute_collapse_summary(
        per_seed_qlike, config
    )
    # Assemble the immutable evaluation report
    report: EvaluationMetricsReport = EvaluationMetricsReport(
        seed_metrics=seed_metrics,
        daily_ic_series=daily_ic_series,
        ic_aggregates=ic_aggregates,
        collapse_summary=collapse_summary,
    )
    # Return the evaluation report
    return report
````

````python
# Task 13: Statistical Inference with Diebold–Mariano and Newey–West

# ==============================================================================
# Task 13: Statistical Inference with Diebold–Mariano and Newey–West
# ==============================================================================

# -----------------------------------------------------------------------
# Task 13, Step 1: Build the Seed-Averaged Daily Difference Series
# -----------------------------------------------------------------------

def build_seed_averaged_difference(
    model_a_daily: pd.DataFrame,
    model_b_daily: pd.DataFrame,
) -> pd.Series:
    """
    Build the seed-averaged daily difference series d_t.

    Implements d_t = (1/S) sum_{s=1..S} [X_{A,s}(t) - X_{B,s}(t)] over
    the intersection of the two models' test days.

    Parameters
    ----------
    model_a_daily : pd.DataFrame
        Model A's per-seed daily metric frame indexed by date.
    model_b_daily : pd.DataFrame
        Model B's per-seed daily metric frame indexed by date.

    Returns
    -------
    pd.Series
        The seed-averaged daily difference series indexed by date.

    Raises
    ------
    ValueError
        If the frames have no common dates or misaligned seeds.
    """
    # Validate the seed columns of the two frames
    if not model_a_daily.columns.equals(model_b_daily.columns):
        # Raise an informative error for the seed mismatch
        raise ValueError(
            f"Seed columns differ: {list(model_a_daily.columns)} vs "
            f"{list(model_b_daily.columns)}."
        )
    # Intersect the date indices of the two models
    common_dates: pd.Index = model_a_daily.index.intersection(
        model_b_daily.index
    )
    # Raise an informative error for the empty intersection
    if len(common_dates) == 0:
        # Build the error message for the missing common days
        raise ValueError(
            "No common test days between the two models' daily series."
        )
    # Log the missing-day policy decision
    logger.info(
        "Intersecting test days: %d -> %d common days",
        len(model_a_daily.index), len(common_dates),
    )
    # Compute the per-seed daily difference on the common days
    difference: pd.DataFrame = (
        model_a_daily.loc[common_dates] - model_b_daily.loc[common_dates]
    )
    # Average the per-seed differences across seeds: (1/S) sum_s
    d_series: pd.Series = difference.mean(axis=1)
    # Return the seed-averaged daily difference series
    return d_series


# -----------------------------------------------------------------------
# Task 13, Step 2: Compute Newey–West Standard Errors with Four Lags
# -----------------------------------------------------------------------


def compute_autocovariances(
    series: np.ndarray,
    max_lag: int,
) -> np.ndarray:
    """
    Compute the autocovariances of a series up to a maximum lag.

    Implements gamma_j = (1/T) sum_{t=j+1..T} (d_t - dbar)(d_{t-j} - dbar)
    with the manuscript's 1/T denominator.

    Parameters
    ----------
    series : np.ndarray
        The demeaned input series of length T.
    max_lag : int
        The maximum lag L.

    Returns
    -------
    np.ndarray
        The autocovariance array of length L + 1.
    """
    # Compute the length of the series
    t_len: int = series.shape[0]
    # Initialize the autocovariance array
    gammas: np.ndarray = np.zeros(max_lag + 1)
    # Iterate over every lag from zero to the maximum
    for j in range(max_lag + 1):
        # Compute the lag-j product sum over the valid range
        product_sum: float = float(
            np.sum(series[j:] * series[: t_len - j])
        )
        # Apply the 1/T denominator to the product sum
        gammas[j] = product_sum / t_len
    # Return the autocovariance array
    return gammas


def compute_newey_west_se(
    d_series: pd.Series,
    lags: int = 4,
) -> float:
    """
    Compute the Newey-West standard error of the mean difference.

    Implements Var(dbar) = (1/T)[gamma_0 + 2 sum_{j=1..L}
    (1 - j/(L+1)) gamma_j] with L = 4 and SE = sqrt(Var).

    Parameters
    ----------
    d_series : pd.Series
        The seed-averaged daily difference series.
    lags : int, optional
        The Newey-West lag count, default 4.

    Returns
    -------
    float
        The Newey-West standard error of the mean.
    """
    # Convert the series to a float64 array
    d: np.ndarray = d_series.to_numpy(dtype=np.float64)
    # Drop any non-finite values from the array
    d = d[np.isfinite(d)]
    # Raise an informative error for the empty series
    if d.shape[0] < 2:
        # Build the error message with the observed observation count
        raise ValueError(
            f"Newey-West requires at least 2 observations; got "
            f"{d.shape[0]}."
        )
    # Compute the series length
    t_len: int = d.shape[0]
    # Compute the sample mean of the series
    dbar: float = float(np.mean(d))
    # Demean the series
    demeaned: np.ndarray = d - dbar
    # Compute the autocovariances up to the lag count
    gammas: np.ndarray = compute_autocovariances(demeaned, lags)
    # Initialize the Bartlett-weighted sum with gamma_0
    weighted: float = gammas[0]
    # Iterate over the positive lags
    for j in range(1, lags + 1):
        # Apply the Bartlett weight (1 - j/(L+1)) to the autocovariance
        weighted += 2.0 * (1.0 - j / (lags + 1.0)) * gammas[j]
    # Compute the Newey-West variance with the 1/T factor
    variance: float = weighted / t_len
    # Guard against a negative variance: the task prescribes a numerical
    # warning (a negative NW variance indicates pathological dependence),
    # then the variance is floored so the standard error stays defined
    if variance < 0.0:
        # Log the numerical warning for the audit trail
        logger.warning(
            "Newey-West variance is negative (%.6e); the series may be "
            "pathologically autocorrelated. The variance is floored at "
            "zero for the standard error.", variance,
        )
        # Floor the variance at zero for numerical safety
        variance = 0.0
    # Compute the standard error as the square root
    se: float = float(np.sqrt(variance))
    # Return the Newey-West standard error
    return se


# -----------------------------------------------------------------------
# Task 13, Step 3: Compute the Diebold–Mariano Statistic and p-Values
# -----------------------------------------------------------------------


def compute_dm_statistic(
    d_series: pd.Series,
    se: float,
) -> Dict[str, Any]:
    """
    Compute the Diebold-Mariano statistic and the two-sided p-value.

    Implements t_DM = dbar / SE_NW with the two-sided p-value from the
    t-distribution with T-1 degrees of freedom.

    Parameters
    ----------
    d_series : pd.Series
        The seed-averaged daily difference series.
    se : float
        The Newey-West standard error.

    Returns
    -------
    Dict[str, Any]
        The mean difference, standard error, statistic, p-value, and
        significance flags.

    Raises
    ------
    ValueError
        If the standard error is not strictly positive.
    """
    # Validate the standard error is strictly positive
    if not np.isfinite(se) or se <= 0.0:
        # Raise an informative error for the invalid standard error
        raise ValueError(
            f"Newey-West standard error must be positive; got {se}."
        )
    # Compute the sample mean of the difference series
    dbar: float = float(d_series.mean())
    # Compute the DM statistic: t = dbar / SE
    t_stat: float = dbar / se
    # Compute the degrees of freedom as T - 1
    dof: int = int(d_series.shape[0]) - 1
    # Compute the two-sided p-value from the t survival function
    p_value: float = float(
        2.0 * t_dist.sf(abs(t_stat), df=dof)
    )
    # Compute the significance flag at the 0.001 threshold
    sig_0001: bool = p_value < 0.001
    # Compute the significance flag at the 0.01 threshold
    sig_001: bool = p_value < 0.01
    # Compute the significance flag at the 0.05 threshold
    sig_005: bool = p_value < 0.05
    # Return the DM result dictionary
    return {
        "mean_difference": dbar,
        "standard_error": se,
        "t_statistic": t_stat,
        "p_value": p_value,
        "p_lt_0_001": sig_0001,
        "p_lt_0_01": sig_001,
        "p_lt_0_05": sig_005,
    }


# -----------------------------------------------------------------------
# Task 13, Orchestrator Function
# -----------------------------------------------------------------------


@dataclass(frozen=True)
class DMComparisonReport:
    """
    Immutable report of one Diebold-Mariano pairwise comparison.

    Attributes
    ----------
    model_a : str
        The name of model A.
    model_b : str
        The name of model B.
    difference_series : pd.Series
        The seed-averaged daily difference series.
    result : Dict[str, Any]
        The DM statistic, p-value, and significance flags.
    """

    # The name of model A
    model_a: str
    # The name of model B
    model_b: str
    # The seed-averaged daily difference series
    difference_series: pd.Series
    # The DM statistic, p-value, and significance flags
    result: Dict[str, Any]


def run_diebold_mariano(
    model_a: str,
    model_b: str,
    model_a_daily: pd.DataFrame,
    model_b_daily: pd.DataFrame,
    config: Dict[str, Any],
) -> DMComparisonReport:
    """
    Orchestrate one Diebold-Mariano pairwise comparison.

    Composes build_seed_averaged_difference, compute_newey_west_se, and
    compute_dm_statistic in the mandated order.

    Parameters
    ----------
    model_a : str
        The name of model A.
    model_b : str
        The name of model B.
    model_a_daily : pd.DataFrame
        Model A's per-seed daily metric frame indexed by date.
    model_b_daily : pd.DataFrame
        Model B's per-seed daily metric frame indexed by date.
    config : Dict[str, Any]
        The configuration dictionary.

    Returns
    -------
    DMComparisonReport
        The immutable comparison report.

    Raises
    ------
    ValueError
        If the difference series is empty.
    """
    # Extract the Newey-West lag count from the config
    lags: int = int(
        config.get("statistical_testing", {}).get(
            "newey_west_lags", 4
        )
    )
    # Step 1: build the seed-averaged daily difference series
    d_series: pd.Series = build_seed_averaged_difference(
        model_a_daily, model_b_daily
    )
    # Raise an informative error for the empty series
    if d_series.empty:
        # Build the error message for the empty difference series
        raise ValueError("Difference series is empty; cannot test.")
    # Step 2: compute the Newey-West standard error
    se: float = compute_newey_west_se(d_series, lags=lags)
    # Step 3: compute the DM statistic and the p-values
    result: Dict[str, Any] = compute_dm_statistic(d_series, se)
    # Assemble the immutable comparison report
    report: DMComparisonReport = DMComparisonReport(
        model_a=model_a,
        model_b=model_b,
        difference_series=d_series,
        result=result,
    )
    # Return the comparison report
    return report
````

````python
# Task 14: Value-at-Risk Calibration and Kupiec Testing

# ==============================================================================
# Task 14: Value-at-Risk Calibration and Kupiec Testing
# ==============================================================================

# -----------------------------------------------------------------------
# Task 14, Step 1: Fit Student-t Residual Distributions
# -----------------------------------------------------------------------


def standardize_returns(
    returns: pd.Series,
) -> pd.Series:
    """
    Standardize a return series with its sample mean and std.

    Implements z = (r - mean_train(r)) / std_train(r) per the config's
    student_t_residual_definition.

    Parameters
    ----------
    returns : pd.Series
        The training-window return series.

    Returns
    -------
    pd.Series
        The standardized residual series.
    """
    # Compute the sample mean of the returns
    mu: float = float(returns.mean())
    # Compute the sample standard deviation of the returns
    sigma: float = float(returns.std(ddof=1))
    # Guard against a zero standard deviation
    if sigma == 0.0:
        # Raise an informative error for the degenerate series
        raise ValueError(
            "Cannot standardize a return series with zero variance."
        )
    # Standardize the returns: z = (r - mu) / sigma
    z: pd.Series = (returns - mu) / sigma
    # Return the standardized residual series
    return z


def fit_student_t_residuals(
    training_returns: pd.DataFrame,
) -> pd.DataFrame:
    """
    Fit Student-t distributions to per-stock standardized residuals.

    Parameters
    ----------
    training_returns : pd.DataFrame
        The training-window returns with stock and return columns.

    Returns
    -------
    pd.DataFrame
        The fitted parameters frame with stock, mu, sigma, nu.

    Raises
    ------
    ValueError
        If the returns frame is empty.
    """
    # Verify that the returns frame is not empty
    if training_returns.empty:
        # Raise an informative error for the empty frame
        raise ValueError("Training returns frame is empty.")
    # Initialize the parameter records list
    records: List[Dict[str, Any]] = []
    # Iterate over every stock group
    for stock, group in training_returns.groupby("stock"):
        # Standardize the stock's training returns
        z: pd.Series = standardize_returns(group["return"])
        # Attempt the Student-t maximum-likelihood fit with multiple
        # parameterizations so a poor starting configuration cannot sink
        # the fit: (1) the free three-parameter fit, (2) the fixed-unit
        # fit, and (3) the fixed-sample-moment fit; the lowest negative
        # log-likelihood wins
        best_fit: Optional[Tuple[float, float, float]] = None
        # Track the best negative log-likelihood across the starts
        best_nll: float = float("inf")
        # Define the candidate parameterizations to try: the free
        # three-parameter fit, the fixed-unit fit, and the fixed-sample-
        # moment fit
        fit_candidates: List[Tuple[Optional[float], Optional[float]]] = [
            (None, None),
            (0.0, 1.0),
            (float(z.mean()), float(z.std(ddof=1))),
        ]
        # Iterate over every candidate parameterization
        for candidate in fit_candidates:
            # Attempt the fit from the current parameterization
            try:
                # Unpack the candidate location and scale constraints
                candidate_loc, candidate_scale = candidate
                # Fit with the free parameterization when unconstrained
                if candidate_loc is None:
                    # Fit all three parameters freely
                    nu, mu_fit, sigma_fit = t_dist.fit(z)
                else:
                    # Fit with the location and scale fixed
                    nu, mu_fit, sigma_fit = t_dist.fit(
                        z,
                        floc=candidate_loc,
                        fscale=candidate_scale,
                    )
            except Exception as exc:
                # Log the failed parameterization for the stock
                logger.warning(
                    "Student-t fit parameterization %s failed for %s: %s",
                    candidate, stock, exc,
                )
                # Continue to the next parameterization
                continue
            # Compute the negative log-likelihood of this fit
            nll: float = float(
                -np.sum(t_dist.logpdf(z, df=nu, loc=mu_fit,
                                      scale=sigma_fit))
            )
            # Keep the best fit across the parameterizations
            if nll < best_nll:
                # Update the best negative log-likelihood
                best_nll = nll
                # Update the best parameter triple
                best_fit = (nu, mu_fit, sigma_fit)
        # Handle the total-failure case with the documented fallback
        if best_fit is None:
            # Log the fallback parameters for the decision log
            logger.warning(
                "Student-t fit failed for %s; falling back to "
                "(nu=30, mu=0, sigma=1). Record this in the decision log.",
                stock,
            )
            # Fall back to the normal-approximation parameters
            nu, mu_fit, sigma_fit = 30.0, 0.0, 1.0
        else:
            # Unpack the best fitted parameters
            nu, mu_fit, sigma_fit = best_fit
        # Constrain the degrees of freedom to the valid range
        nu_clipped: float = float(np.clip(nu, 2.1, 100.0))
        # Append the fitted parameter record
        records.append(
            {
                "stock": stock,
                "mu": float(mu_fit),
                "sigma": float(sigma_fit),
                "nu": nu_clipped,
            }
        )
    # Build the parameters frame from the records
    params: pd.DataFrame = pd.DataFrame(records)
    # Return the fitted parameters frame
    return params


# -----------------------------------------------------------------------
# Task 14, Step 2: Convert Forecasts to Daily Volatility and Compute VaR
# -----------------------------------------------------------------------


def compute_var_thresholds(
    forecasts: pd.DataFrame,
    params: pd.DataFrame,
    config: Dict[str, Any],
) -> pd.DataFrame:
    """
    Compute the daily VaR thresholds and hit indicators.

    Implements sigma_daily = y_hat / sqrt(252) and
    VaR_q = -[mu + sigma_daily * scale * t_nu^{-1}(q)] with the hit
    Hit = 1[r < -VaR_q].

    Parameters
    ----------
    forecasts : pd.DataFrame
        The forecast frame with stock, date, forecast columns.
    params : pd.DataFrame
        The fitted Student-t parameters frame.
    config : Dict[str, Any]
        The configuration dictionary.

    Returns
    -------
    pd.DataFrame
        The VaR frame with daily volatility, thresholds, and hits.

    Raises
    ------
    ValueError
        If the forecast frame is empty.
    """
    # Verify that the forecast frame is not empty
    if forecasts.empty:
        # Raise an informative error for the empty frame
        raise ValueError("Forecast frame is empty.")
    # Validate the required return column before the hit computation
    if "return" not in forecasts.columns:
        # Raise an informative error for the missing return column
        raise ValueError(
            "The forecast frame must carry a 'return' column (the daily "
            "return of the corresponding stock-day) for the hit "
            "indicator."
        )
    # Validate the return column is numeric
    if not pd.api.types.is_numeric_dtype(forecasts["return"]):
        # Raise an informative error for the non-numeric return column
        raise ValueError(
            "The 'return' column of the forecast frame must be numeric."
        )
    # Extract the VaR levels from the config
    levels: List[float] = list(
        config.get("value_at_risk_calibration", {}).get(
            "levels", [0.05, 0.01]
        )
    )
    # Extract the annualization factor from the config
    annualization: float = float(
        config.get("data_preprocessing_parameters", {}).get(
            "annualization_factor", 252
        )
    )
    # Compute the annualization square root once
    sqrt_ann: float = float(np.sqrt(annualization))
    # Copy the forecast frame for the VaR computation
    out: pd.DataFrame = forecasts.copy()
    # Merge the fitted Student-t parameters by stock
    out = out.merge(params, on="stock", how="left")
    # Compute the daily volatility: sigma_daily = y_hat / sqrt(252)
    out["daily_volatility"] = out["forecast"] / sqrt_ann
    # Extract the degrees-of-freedom array for the vectorized quantiles
    nu_array: np.ndarray = out["nu"].to_numpy(dtype=np.float64)
    # Iterate over every VaR level
    for q in levels:
        # Compute the Student-t quantiles VECTORIZED for the whole frame,
        # which avoids a row-wise apply over ~1.9M stock-days
        quantile: np.ndarray = t_dist.ppf(q, df=nu_array)
        # Compute the VaR threshold: VaR_q = -[mu + sd * scale * q]
        out[f"var_{q:.2f}"] = -(
            out["mu"]
            + out["daily_volatility"] * out["sigma"] * quantile
        )
        # Compute the hit indicator: Hit = 1[r < -VaR_q]
        out[f"hit_{q:.2f}"] = (
            out["return"] < -out[f"var_{q:.2f}"]
        ).astype(int)
    # Return the VaR frame
    return out


# -----------------------------------------------------------------------
# Task 14, Step 3: Kupiec Unconditional Coverage Test and Rejection Rates
# -----------------------------------------------------------------------


def kupiec_likelihood_ratio(
    n_total: int,
    n_hits: int,
    p0: float,
) -> float:
    """
    Compute the Kupiec unconditional-coverage likelihood ratio.

    Implements LR = -2 ln[((1-p0)^(N-x) p0^x) / ((1-x/N)^(N-x) (x/N)^x)]
    with analytic boundary handling.

    Parameters
    ----------
    n_total : int
        The number of test days N.
    n_hits : int
        The number of hits x.
    p0 : float
        The nominal coverage probability.

    Returns
    -------
    float
        The Kupiec likelihood-ratio statistic.

    Raises
    ------
    ValueError
        If the total count is not positive.
    """
    # Validate the total count is positive
    if n_total <= 0:
        # Raise an informative error for the invalid count
        raise ValueError(f"n_total must be positive; got {n_total}.")
    # Handle the zero-hit boundary analytically
    if n_hits == 0:
        # Compute the boundary likelihood ratio for x = 0
        return float(-2.0 * n_total * np.log(1.0 - p0))
    # Handle the all-hits boundary analytically
    if n_hits == n_total:
        # Compute the boundary likelihood ratio for x = N
        return float(-2.0 * n_total * np.log(p0))
    # Compute the empirical hit rate
    emp_rate: float = n_hits / n_total
    # Compute the constrained likelihood numerator
    numerator: float = (
        (1.0 - p0) ** (n_total - n_hits) * p0 ** n_hits
    )
    # Compute the unconstrained likelihood denominator
    denominator: float = (
        (1.0 - emp_rate) ** (n_total - n_hits)
        * emp_rate ** n_hits
    )
    # Guard against a zero denominator
    if denominator == 0.0:
        # Return a large statistic for the degenerate case
        return float("inf")
    # Compute the likelihood-ratio statistic
    lr: float = -2.0 * float(np.log(numerator / denominator))
    # Return the Kupiec likelihood ratio
    return lr


def compute_kupiec_rejection_rates(
    var_frame: pd.DataFrame,
    config: Dict[str, Any],
    market_volatility: Optional[pd.Series] = None,
    training_median_mv: Optional[float] = None,
) -> pd.DataFrame:
    """
    Compute the per-slice ticker rejection rates for every VaR level.

    Parameters
    ----------
    var_frame : pd.DataFrame
        The VaR frame with stock, date, and hit columns.
    config : Dict[str, Any]
        The configuration dictionary.
    market_volatility : Optional[pd.Series]
        The market-volatility series for the high/low slices.
    training_median_mv : Optional[float], optional
        The TRAINING-window median market volatility; when supplied it
        is used for the high/low split, preventing any test-period
        information from entering the slice definition. When omitted,
        the median of the supplied series is used and a warning is
        logged, because the manuscript defines the slices against the
        training-window median.

    Returns
    -------
    pd.DataFrame
        The rejection-rate summary indexed by (slice, level).

    Raises
    ------
    ValueError
        If the VaR frame is empty.
    """
    # Verify that the VaR frame is not empty
    if var_frame.empty:
        # Raise an informative error for the empty frame
        raise ValueError("VaR frame is empty.")
    # Extract the VaR levels from the config
    levels: List[float] = list(
        config.get("value_at_risk_calibration", {}).get(
            "levels", [0.05, 0.01]
        )
    )
    # Extract the Kupiec significance level from the config
    kupiec_alpha: float = float(
        config.get("value_at_risk_calibration", {})
        .get("kupiec_test", {})
        .get("significance_level", 0.05)
    )
    # Compute the chi-square critical value with one degree of freedom
    critical: float = float(chi2.ppf(1.0 - kupiec_alpha, df=1))
    # Build the slice frames: all dates plus the volatility subsets
    slice_frames: List[Tuple[str, pd.DataFrame]] = [
        ("all_dates", var_frame)
    ]
    # Assign the high/low volatility slices when available
    if market_volatility is not None:
        # Use the supplied training-window median when given; otherwise
        # fall back to the median of the passed series with a warning,
        # because the manuscript defines the slices against the
        # TRAINING-window median market volatility
        if training_median_mv is not None:
            # Use the precomputed training-window median
            median_mv: float = float(training_median_mv)
        else:
            # Warn that the median is being recomputed on the series
            logger.warning(
                "training_median_mv not supplied; the high/low slice "
                "median is computed from the passed market-volatility "
                "series. Supply the training-window median to honor the "
                "manuscript's slice definition."
            )
            # Compute the median of the passed series as the fallback
            median_mv = float(market_volatility.median())
        # Align the market volatility to the frame rows by date
        mv_aligned: pd.Series = market_volatility.reindex(
            var_frame["date"]
        )
        # Count the unaligned rows before the mask is applied
        n_unaligned: int = int(mv_aligned.isna().sum())
        # Log the unaligned rows so silent slice shrinkage is visible
        if n_unaligned > 0:
            # Log the unaligned row count for the decision log
            logger.warning(
                "%d rows of the VaR frame have no market-volatility "
                "date match and are excluded from the high/low slices",
                n_unaligned,
            )
        # Build the high-volatility mask on the aligned series
        high_mask: pd.Series = mv_aligned > median_mv
        # Build the high-volatility slice frame on the aligned rows
        aligned_rows: np.ndarray = mv_aligned.notna().to_numpy()
        # Apply the alignment mask to the high-volatility slice
        slice_frames.append(
            ("high_volatility",
             var_frame.loc[aligned_rows & high_mask.to_numpy()])
        )
        # Build the low-volatility slice frame on the aligned rows
        slice_frames.append(
            ("low_volatility",
             var_frame.loc[aligned_rows & ~high_mask.to_numpy()])
        )
    # Initialize the summary records list
    records: List[Dict[str, Any]] = []
    # Iterate over every slice frame
    for slice_name, slice_frame in slice_frames:
        # Iterate over every VaR level
        for q in levels:
            # Compute the hit column name for the level
            hit_col: str = f"hit_{q:.2f}"
            # Initialize the ticker rejection list
            ticker_rejections: List[bool] = []
            # Iterate over every ticker in the slice
            for stock, group in slice_frame.groupby("stock"):
                # Count the total days for the ticker
                n_total: int = int(len(group))
                # Count the hits for the ticker
                n_hits: int = int(group[hit_col].sum())
                # Compute the Kupiec likelihood ratio for the ticker
                lr: float = kupiec_likelihood_ratio(n_total, n_hits, q)
                # Record the rejection decision for the ticker
                ticker_rejections.append(lr > critical)
            # Compute the rejection rate over the tickers
            rejection_rate: float = float(
                np.mean(ticker_rejections)
            ) if ticker_rejections else float("nan")
            # Append the summary record
            records.append(
                {
                    "slice": slice_name,
                    "level": q,
                    "rejection_rate": rejection_rate,
                    "tickers_tested": len(ticker_rejections),
                }
            )
    # Build the summary frame from the records
    summary: pd.DataFrame = pd.DataFrame(records)
    # Return the rejection-rate summary
    return summary


def aggregate_seed_rejection_rates(
    per_seed_summaries: List[pd.DataFrame],
) -> pd.DataFrame:
    """
    Average the per-slice ticker rejection rates across the random seeds.

    The manuscript reports seed-averaged Kupiec calibration: the ticker
    rejection rate of every (slice, level) cell is averaged across the
    30 random-seed runs.

    Parameters
    ----------
    per_seed_summaries : List[pd.DataFrame]
        One rejection-rate summary per seed, as produced by
        compute_kupiec_rejection_rates.

    Returns
    -------
    pd.DataFrame
        The across-seed mean rejection rate per (slice, level), with the
        across-seed standard deviation and the number of seeds.
    """
    # Guard the degenerate empty-input case
    if not per_seed_summaries:
        # Raise an informative error for the empty input
        raise ValueError(
            "per_seed_summaries must contain at least one seed summary."
        )
    # Concatenate the per-seed summaries into one long frame
    combined: pd.DataFrame = pd.concat(
        per_seed_summaries, ignore_index=True
    )
    # Aggregate the rejection rates per (slice, level) across seeds
    aggregated: pd.DataFrame = combined.groupby(
        ["slice", "level"], as_index=False
    ).agg(
        mean_rejection_rate=("rejection_rate", "mean"),
        std_rejection_rate=("rejection_rate", "std"),
        n_seeds=("rejection_rate", "count"),
    )
    # Fill the single-seed standard deviations with zero
    aggregated = aggregated.fillna(0.0)
    # Return the seed-averaged rejection-rate summary
    return aggregated


# -----------------------------------------------------------------------
# Task 14, Orchestrator Function
# -----------------------------------------------------------------------


@dataclass(frozen=True)
class VaRCalibrationReport:
    """
    Immutable report of the VaR calibration pipeline.

    Attributes
    ----------
    parameters : pd.DataFrame
        The fitted Student-t parameters frame.
    var_frame : pd.DataFrame
        The VaR thresholds and hits frame.
    rejection_summary : pd.DataFrame
        The Kupiec rejection-rate summary.
    """

    # The fitted Student-t parameters frame
    parameters: pd.DataFrame
    # The VaR thresholds and hits frame
    var_frame: pd.DataFrame
    # The Kupiec rejection-rate summary
    rejection_summary: pd.DataFrame


def run_var_calibration(
    forecasts: pd.DataFrame,
    training_returns: pd.DataFrame,
    config: Dict[str, Any],
    market_volatility: Optional[pd.Series] = None,
    training_median_mv: Optional[float] = None,
    training_window_end: Optional[pd.Timestamp] = None,
) -> VaRCalibrationReport:
    """
    Orchestrate the full VaR calibration pipeline.

    Composes fit_student_t_residuals, compute_var_thresholds, and
    compute_kupiec_rejection_rates in the mandated order.

    Parameters
    ----------
    forecasts : pd.DataFrame
        The forecast frame with stock, date, forecast, return columns.
    training_returns : pd.DataFrame
        The training-window returns with stock and return columns.
    config : Dict[str, Any]
        The configuration dictionary.
    market_volatility : Optional[pd.Series]
        The market-volatility series for the high/low slices.
    training_median_mv : Optional[float], optional
        The training-window median market volatility for the high/low
        slices; passed through to the Kupiec rejection-rate callable.
    training_window_end : Optional[pd.Timestamp], optional
        The last date of the training window; when supplied, the
        training_returns frame is restricted to rows at or before this
        date so the Student-t fits cannot contact test-period returns.

    Returns
    -------
    VaRCalibrationReport
        The immutable VaR calibration report.

    Raises
    ------
    ValueError
        If the forecast frame is empty.
    """
    # Verify that the forecast frame is not empty
    if forecasts.empty:
        # Raise an informative error for the empty frame
        raise ValueError("Forecast frame is empty.")
    # Restrict the training returns to the training window when the end
    # date is supplied, so the Student-t fits never contact test rows
    if training_window_end is not None:
        # Parse the returns dates for the window restriction
        return_dates: pd.Series = pd.to_datetime(training_returns["date"])
        # Keep only the rows at or before the training-window end
        training_returns = training_returns.loc[
            return_dates <= pd.Timestamp(training_window_end)
        ].copy()
    # Step 1: fit the per-stock Student-t residual distributions
    params: pd.DataFrame = fit_student_t_residuals(training_returns)
    # Step 2: compute the daily VaR thresholds and hits
    var_frame: pd.DataFrame = compute_var_thresholds(
        forecasts, params, config
    )
    # Step 3: compute the Kupiec rejection rates per slice and level
    summary: pd.DataFrame = compute_kupiec_rejection_rates(
        var_frame, config, market_volatility, training_median_mv
    )
    # Assemble the immutable VaR calibration report
    report: VaRCalibrationReport = VaRCalibrationReport(
        parameters=params,
        var_frame=var_frame,
        rejection_summary=summary,
    )
    # Return the VaR calibration report
    return report
````

````python
# Task 15: Replicate the Japanese Cross-Market Study

# ==============================================================================
# Task 15: Replicate the Japanese Cross-Market Study
# ==============================================================================

# -----------------------------------------------------------------------
# Task 15, Step 1: Reconstruct the Japanese Panel with Identical Pipelines
# -----------------------------------------------------------------------

def build_japanese_panel(
    jp_raw_dir: Path,
    config: Dict[str, Any],
    metadata: Optional[pd.DataFrame] = None,
    out_dir: Optional[Path] = None,
) -> Dict[str, Any]:
    """
    Reconstruct the Japanese panel with the identical pipeline.

    Reuses the Task 3-5 callables with the market name "JP" so that the
    Japanese panel is built by the same cleansing, alignment, feature,
    and split functions as the U.S. panel.

    Parameters
    ----------
    jp_raw_dir : Path
        The Japanese raw CSV directory.
    config : Dict[str, Any]
        The configuration dictionary.
    metadata : Optional[pd.DataFrame], optional
        The Japanese metadata frame for the universe filters.
    out_dir : Optional[Path], optional
        The destination directory for the panel artifacts.

    Returns
    -------
    Dict[str, Any]
        The Japanese panel result block with the splits and tensors.

    Raises
    ------
    ValueError
        If the raw directory is missing.
    """
    # Verify that the raw directory exists
    if not jp_raw_dir.is_dir():
        # Raise an informative error for the missing directory
        raise ValueError(f"JP raw directory does not exist: {jp_raw_dir}")
    # Step 1: cleanse and align the Japanese raw panel
    aligned = cleanse_and_align_panel(
        jp_raw_dir, "JP", config, metadata, out_dir
    )
    # Step 2: build the Japanese features, regime variables, and targets
    features = build_features_targets(
        aligned.wide, aligned.market_return, config, out_dir
    )
    # Step 3: build the Japanese walk-forward splits
    splits = build_walkforward_splits(
        features.panel, config, out_dir
    )
    # Assemble the Japanese panel result block; the feature panel is
    # carried so the evaluation stage can join predictions with targets
    result: Dict[str, Any] = {
        "market": "JP",
        "tickers": aligned.tickers,
        "feature_columns": features.feature_columns,
        "windows": len(splits.windows),
        "bundles": splits.bundles,
        "panel": features.panel,
    }
    # Return the Japanese panel result block
    return result


# -----------------------------------------------------------------------
# Task 15, Step 2: Train and Tune Models on the Japanese Panel
# -----------------------------------------------------------------------


def tune_japanese_models(
    bundles: List[Dict[str, Any]],
    model_factories: Dict[str, Any],
    config: Dict[str, Any],
    sweep_seeds: Tuple[int, ...] = (0, 1, 2),
) -> Dict[str, Dict[str, Any]]:
    """
    Re-run the three-seed hyperparameter sweeps on the Japanese splits.

    Parameters
    ----------
    bundles : List[Dict[str, Any]]
        The Japanese walk-forward tensor bundles.
    model_factories : Dict[str, Any]
        The model factories for the four reported models.
    config : Dict[str, Any]
        The configuration dictionary.
    sweep_seeds : Tuple[int, ...], optional
        The sweep seed registry, default (0, 1, 2).

    Returns
    -------
    Dict[str, Dict[str, Any]]
        The selected hyperparameter configuration per model.
    """
    # Use the first window's tensors for the sweep, exactly as the U.S.
    # protocol does (documented: hyperparameters are selected once per
    # market on the first window's validation split)
    first_bundle: Dict[str, Any] = bundles[0]
    # Build the splits dictionary for the selection protocol; the
    # validation dates are carried so the true daily Spearman IC can be
    # computed by the Task 11 selection protocol
    splits: Dict[str, Any] = {
        "X_train": first_bundle["X_train"],
        "y_train": first_bundle["y_train"],
        "Z_train": first_bundle.get("Z_train"),
        "X_val": first_bundle["X_val"],
        "y_val": first_bundle["y_val"],
        "Z_val": first_bundle.get("Z_val"),
        "dates_val": first_bundle["dates_val"],
    }
    # Initialize the selected-configuration dictionary
    selected: Dict[str, Dict[str, Any]] = {}
    # Iterate over the four reported models
    for model_name, factory in model_factories.items():
        # Run the selection protocol for the model; the factories follow
        # the unified config-accepting contract, and the protocol runs
        # the two-stage RG training when the family requires it
        report = select_hyperparameters(factory, splits, config)
        # Record the selected configuration for the model
        selected[model_name] = report.selected_config
        # Log the selected configuration
        logger.info(
            "JP selected %s: %s", model_name, report.selected_config
        )
    # Return the per-model selected configurations
    return selected


def train_japanese_models(
    bundles: List[Dict[str, Any]],
    panel: pd.DataFrame,
    model_factories: Dict[str, Any],
    selected_configs: Dict[str, Dict[str, Any]],
    config: Dict[str, Any],
    final_seeds: Tuple[int, ...] = tuple(range(30)),
    window_limit: int = 30,
) -> Dict[str, Any]:
    """
    Run the final 30-seed training on the Japanese panel.

    Trains every model for every seed and every walk-forward window with
    the correct per-family protocol (two-stage for RG-ResMoE with the
    selected alpha/lambda_LB), collects the test predictions per
    (model, seed), and returns the identity-aligned prediction frames
    together with the test target frame.

    Parameters
    ----------
    bundles : List[Dict[str, Any]]
        The Japanese walk-forward tensor bundles.
    panel : pd.DataFrame
        The Japanese feature panel with ticker, date, and target columns.
    model_factories : Dict[str, Any]
        The model factories for the four reported models; every factory
        follows the unified config-accepting contract.
    selected_configs : Dict[str, Dict[str, Any]]
        The selected hyperparameter configuration per model.
    config : Dict[str, Any]
        The configuration dictionary.
    final_seeds : Tuple[int, ...], optional
        The final seed registry, default 0..29.
    window_limit : int, optional
        The number of walk-forward windows to train, default 30.

    Returns
    -------
    Dict[str, Any]
        A block with 'training' (per (model, seed, window) models),
        'per_seed_predictions' (per (model, seed) test prediction
        frames), and 'targets' (the test target frame).
    """
    # Initialize the per-model training dictionary
    training: Dict[str, Dict[Tuple[int, int], Any]] = {}
    # Initialize the per-model per-seed prediction frames
    per_seed_predictions: Dict[str, Dict[int, pd.DataFrame]] = {}
    # Build the target frame once from the feature panel
    targets: pd.DataFrame = panel[["ticker", "date", "target"]].copy()
    # Iterate over the reported models
    for model_name, factory in model_factories.items():
        # Merge the selected configuration into the config copy
        model_config: Dict[str, Any] = dict(config)
        # Override the training parameters with the selected values
        model_config["network_training_parameters"] = dict(
            config.get("network_training_parameters", {})
        )
        # Apply the selected learning rate
        model_config["network_training_parameters"][
            "learning_rate"
        ] = selected_configs[model_name]["learning_rate"]
        # Override the regularizer coefficients with the selected values
        model_config["loss_function_and_regularization"] = dict(
            config.get("loss_function_and_regularization", {})
        )
        # Apply the selected alpha coefficient
        model_config["loss_function_and_regularization"][
            "alpha_residual_shrinkage"
        ] = selected_configs[model_name]["alpha"]
        # Apply the selected lambda_LB coefficient
        model_config["loss_function_and_regularization"][
            "lambda_LB_load_balancing"
        ] = selected_configs[model_name]["lambda_LB"]
        # Initialize the per-model training dictionary
        training[model_name] = {}
        # Initialize the per-model per-seed prediction frames
        per_seed_predictions[model_name] = {}
        # Iterate over the final seeds
        for seed in final_seeds:
            # Apply the deterministic seed BEFORE model construction so
            # every model family receives the same seed sequence
            seed_all_rng(seed)
            # Initialize the seed's prediction records
            seed_records: List[Dict[str, Any]] = []
            # Iterate over the walk-forward windows
            for window_idx, bundle in enumerate(
                bundles[:window_limit]
            ):
                # Build a fresh model instance for the seed
                model = factory(model_config)
                # Detect the RG-ResMoE family by its two-stage marker
                is_rg: bool = hasattr(model, "freeze_base")
                # Run the correct protocol for the model family
                if is_rg:
                    # Stage 1: train the base network on MSE
                    train_full_batch(
                        model,
                        bundle["X_train"], bundle["y_train"],
                        bundle.get("Z_train"),
                        bundle["X_val"], bundle["y_val"],
                        bundle.get("Z_val"),
                        model_config, seed=seed, stage="stage_1",
                    )
                    # Freeze the base network for Stage 2
                    model.freeze_base()
                    # Stage 2: train experts and gate with the selected
                    # regularizer coefficients
                    train_full_batch(
                        model,
                        bundle["X_train"], bundle["y_train"],
                        bundle.get("Z_train"),
                        bundle["X_val"], bundle["y_val"],
                        bundle.get("Z_val"),
                        model_config, seed=seed, stage="stage_2",
                    )
                else:
                    # Single-stage training with the plain MSE loss
                    train_full_batch(
                        model,
                        bundle["X_train"], bundle["y_train"],
                        bundle.get("Z_train"),
                        bundle["X_val"], bundle["y_val"],
                        bundle.get("Z_val"),
                        model_config, seed=seed, stage="stage_1",
                    )
                # Record the trained model for the seed and window
                training[model_name][(seed, window_idx)] = model
                # Compute the test predictions without gradients
                model.eval()
                # Compute the test predictions under no_grad
                with torch.no_grad():
                    # Forward the test features with the regime tensor
                    y_hat: torch.Tensor = model_forward(
                        model, bundle["X_test"], bundle.get("Z_test")
                    )
                # Flatten the predictions to a numpy vector
                pred_vec: np.ndarray = y_hat.squeeze(-1).numpy()
                # Extract the test ticker identities from the bundle
                test_tickers: np.ndarray = bundle["tickers_test"]
                # Extract the test date identities from the bundle
                test_dates: np.ndarray = bundle["dates_test"]
                # Append the window's prediction records
                seed_records.append(
                    pd.DataFrame({
                        "ticker": test_tickers,
                        "date": test_dates,
                        "prediction": pred_vec,
                    })
                )
            # Concatenate the seed's window prediction frames
            per_seed_predictions[model_name][seed] = pd.concat(
                seed_records, ignore_index=True
            )
    # Return the training block with predictions and targets
    return {
        "training": training,
        "per_seed_predictions": per_seed_predictions,
        "targets": targets,
    }


# -----------------------------------------------------------------------
# Task 15, Step 3: Evaluate and Compare Against Expected Signatures
# -----------------------------------------------------------------------


def evaluate_japanese_results(
    results: Dict[str, Any],
    config: Dict[str, Any],
) -> pd.DataFrame:
    """
    Compute the Japanese evaluation table from the training results.

    Computes the full manuscript metric set per model across the seeds:
    IC, ICIR, RMSE, out-of-sample R2, QLIKE (mean +/- std across the 30
    seeds) and the collapsed-seed counts, using the Task 12 evaluation
    callables on the identity-aligned test predictions.

    Parameters
    ----------
    results : Dict[str, Any]
        The training block from train_japanese_models with
        'per_seed_predictions' and 'targets'.
    config : Dict[str, Any]
        The configuration dictionary.

    Returns
    -------
    pd.DataFrame
        The per-model Japanese results table.
    """
    # Extract the per-model prediction frames from the training block
    per_seed_predictions: Dict[str, Dict[int, pd.DataFrame]] = (
        results["per_seed_predictions"]
    )
    # Extract the target frame from the training block
    targets: pd.DataFrame = results["targets"]
    # Initialize the results table records
    records: List[Dict[str, Any]] = []
    # Iterate over every model
    for model_name, seed_frames in per_seed_predictions.items():
        # Compute the full evaluation metrics for the model
        report = compute_evaluation_metrics(
            seed_frames, targets, config, model_name=model_name
        )
        # Extract the per-seed metric frame
        seed_metrics: pd.DataFrame = report.seed_metrics
        # Append the model summary record with the manuscript metrics
        records.append(
            {
                "model": model_name,
                "n_seeds": len(seed_metrics),
                "mean_ic": float(seed_metrics["mean_ic"].mean()),
                "std_ic": float(seed_metrics["mean_ic"].std(ddof=1)),
                "mean_icir": float(seed_metrics["icir"].mean()),
                "mean_rmse": float(seed_metrics["rmse"].mean()),
                "mean_r2": float(seed_metrics["r2"].mean()),
                "mean_qlike": float(seed_metrics["qlike"].mean()),
                "std_qlike": float(seed_metrics["qlike"].std(ddof=1)),
                "collapsed_seeds": int(
                    report.collapse_summary["collapsed_count"].sum()
                ),
            }
        )
    # Build the results table from the records
    table: pd.DataFrame = pd.DataFrame(records)
    # Return the Japanese results table
    return table


def compare_expected_signatures(
    table: pd.DataFrame,
    config: Dict[str, Any],
) -> List[str]:
    """
    Compare the Japanese results against the manuscript signatures.

    Implements the manuscript replication pattern checks: RG-ResMoE must
    have the best IC (higher than MLP-L), MLP-L(+z) must collapse in
    nearly all seeds, Standard MoE must collapse in most seeds,
    RG-ResMoE must have zero or near-zero collapsed seeds, and the IC
    magnitudes must be close to the manuscript's reported values
    (RG-ResMoE ~0.4858, MLP-L ~0.4815). Every check is reported with
    the observed and expected values and its tolerance.

    Parameters
    ----------
    table : pd.DataFrame
        The Japanese results table with mean_ic and collapsed_seeds.
    config : Dict[str, Any]
        The configuration dictionary.

    Returns
    -------
    List[str]
        A list of signature findings; empty when all match.
    """
    # Extract the expected signatures from the config
    expected: Dict[str, Any] = config.get(
        "cross_market_replication", {}
    ).get("expected_results", {})
    # Initialize the findings list
    findings: List[str] = []
    # Verify the reported model set
    reported: List[str] = list(
        config.get("cross_market_replication", {}).get(
            "models_reported", []
        )
    )
    # Check that every reported model is in the table
    missing_models: List[str] = [
        m for m in reported if m not in set(table["model"])
    ]
    # Record the missing-model finding
    if missing_models:
        # Append the missing-model finding
        findings.append(f"Missing models in JP table: {missing_models}")
    # Index the table by model name for the signature checks
    table_by_model: pd.DataFrame = table.set_index("model")
    # Check 1: RG-ResMoE must have the best IC (higher than MLP-L)
    if {"RG-ResMoE", "MLP-L"}.issubset(set(table_by_model.index)):
        # Extract the RG-ResMoE mean IC
        rg_ic: float = float(table_by_model.loc["RG-ResMoE", "mean_ic"])
        # Extract the MLP-L mean IC
        mlp_l_ic: float = float(table_by_model.loc["MLP-L", "mean_ic"])
        # Check the IC ordering
        if rg_ic <= mlp_l_ic:
            # Record the IC-ordering deviation
            findings.append(
                f"RG-ResMoE IC {rg_ic:.4f} does not exceed MLP-L IC "
                f"{mlp_l_ic:.4f}"
            )
        # Check the RG-ResMoE IC magnitude against the manuscript
        if abs(rg_ic - float(expected.get("rg_resmoe_ic", 0.4858))) > 0.01:
            # Record the RG IC magnitude deviation
            findings.append(
                f"RG-ResMoE IC {rg_ic:.4f} deviates from the manuscript "
                f"signature {expected.get('rg_resmoe_ic')} by more than "
                f"0.01"
            )
        # Check the MLP-L IC magnitude against the manuscript
        if abs(mlp_l_ic - float(expected.get("mlp_l_ic", 0.4815))) > 0.01:
            # Record the MLP-L IC magnitude deviation
            findings.append(
                f"MLP-L IC {mlp_l_ic:.4f} deviates from the manuscript "
                f"signature {expected.get('mlp_l_ic')} by more than 0.01"
            )
    # Check 2: MLP-L(+z) must collapse in nearly all seeds (~28/30)
    if "MLP-L(+z)" in set(table_by_model.index):
        # Extract the MLP-L(+z) collapse count
        mlp_lz_collapse: int = int(
            table_by_model.loc["MLP-L(+z)", "collapsed_seeds"]
        )
        # Extract the expected collapse count from the config
        expected_lz: int = int(expected.get("mlp_l_plus_z_collapse", 28))
        # Check the collapse count within a tolerance of two seeds
        if abs(mlp_lz_collapse - expected_lz) > 2:
            # Record the MLP-L(+z) collapse deviation
            findings.append(
                f"MLP-L(+z) collapsed {mlp_lz_collapse}/30 seeds; "
                f"expected {expected_lz}/30"
            )
    # Check 3: Standard MoE must collapse in most seeds (~22/30)
    if "MoE" in set(table_by_model.index):
        # Extract the MoE collapse count
        moe_collapse: int = int(
            table_by_model.loc["MoE", "collapsed_seeds"]
        )
        # Extract the expected MoE collapse count from the config
        expected_moe: int = int(expected.get("moe_collapse", 22))
        # Check the collapse count within a tolerance of two seeds
        if abs(moe_collapse - expected_moe) > 2:
            # Record the MoE collapse deviation
            findings.append(
                f"MoE collapsed {moe_collapse}/30 seeds; expected "
                f"{expected_moe}/30"
            )
    # Check 4: RG-ResMoE must have zero or near-zero collapsed seeds
    if "RG-ResMoE" in set(table_by_model.index):
        # Extract the RG-ResMoE collapse count
        rg_collapse: int = int(
            table_by_model.loc["RG-ResMoE", "collapsed_seeds"]
        )
        # Extract the expected RG collapse count from the config
        expected_rg: int = int(expected.get("rg_resmoe_collapse", 1))
        # Check the collapse count within a tolerance of one seed
        if abs(rg_collapse - expected_rg) > 1:
            # Record the RG-ResMoE collapse deviation
            findings.append(
                f"RG-ResMoE collapsed {rg_collapse}/30 seeds; expected "
                f"{expected_rg}/30"
            )
    # Return the signature findings list
    return findings


# -----------------------------------------------------------------------
# Task 15, Orchestrator Function
# -----------------------------------------------------------------------


@dataclass(frozen=True)
class JapaneseReplicationReport:
    """
    Immutable report of the Japanese cross-market replication.

    Attributes
    ----------
    panel_result : Dict[str, Any]
        The Japanese panel result block.
    selected_configs : Dict[str, Dict[str, Any]]
        The selected JP hyperparameters per model.
    results_table : pd.DataFrame
        The Japanese results table.
    signature_findings : List[str]
        The signature-comparison findings.
    """

    # The Japanese panel result block
    panel_result: Dict[str, Any]
    # The selected JP hyperparameters per model
    selected_configs: Dict[str, Dict[str, Any]]
    # The Japanese results table
    results_table: pd.DataFrame
    # The signature-comparison findings
    signature_findings: List[str]


def replicate_japanese_study(
    jp_raw_dir: Path,
    model_factories: Dict[str, Any],
    config: Dict[str, Any],
    metadata: Optional[pd.DataFrame] = None,
    out_dir: Optional[Path] = None,
) -> JapaneseReplicationReport:
    """
    Orchestrate the full Japanese cross-market replication.

    Composes build_japanese_panel, tune_japanese_models,
    train_japanese_models, evaluate_japanese_results, and
    compare_expected_signatures in the mandated order.

    Parameters
    ----------
    jp_raw_dir : Path
        The Japanese raw CSV directory.
    model_factories : Dict[str, Any]
        The model factories for the four reported models.
    config : Dict[str, Any]
        The configuration dictionary.
    metadata : Optional[pd.DataFrame], optional
        The Japanese metadata frame.
    out_dir : Optional[Path], optional
        The destination directory for the panel artifacts.

    Returns
    -------
    JapaneseReplicationReport
        The immutable replication report.

    Raises
    ------
    ValueError
        If the signature comparison finds substantial deviations.
    """
    # Step 1: reconstruct the Japanese panel with the identical pipeline
    panel_result: Dict[str, Any] = build_japanese_panel(
        jp_raw_dir, config, metadata, out_dir
    )
    # Step 2: re-tune the four models on the Japanese splits
    selected_configs: Dict[str, Dict[str, Any]] = tune_japanese_models(
        panel_result["bundles"], model_factories, config
    )
    # Step 2: run the final 30-seed training on the Japanese panel with
    # the identity-aligned prediction collection
    results: Dict[str, Any] = train_japanese_models(
        panel_result["bundles"], panel_result["panel"],
        model_factories, selected_configs, config,
    )
    # Step 3: compute the Japanese results table with the full metric set
    table: pd.DataFrame = evaluate_japanese_results(results, config)
    # Step 3: compare against the expected signatures
    findings: List[str] = compare_expected_signatures(table, config)
    # Log the signature-comparison outcome
    logger.info("JP signature findings: %d", len(findings))
    # Fail loudly on substantial deviations, exactly as the docstring of
    # this orchestrator promises
    if findings:
        # Raise an informative error listing the deviations
        raise ValueError(
            f"Japanese replication deviates from the manuscript "
            f"signatures: {'; '.join(findings)}"
        )
    # Assemble the immutable replication report
    report: JapaneseReplicationReport = JapaneseReplicationReport(
        panel_result=panel_result,
        selected_configs=selected_configs,
        results_table=table,
        signature_findings=findings,
    )
    # Return the replication report
    return report
````

````python
# Task 16: Create the End-to-End Orchestrator Callable

# ==============================================================================
# Task 16: Create the End-to-End Orchestrator Callable
# ==============================================================================

# -----------------------------------------------------------------------
# Task 16, Step 1: Define the Orchestrator Contract
# -----------------------------------------------------------------------

@dataclass(frozen=True)
class StudyReport:
    """
    Immutable completion report of the end-to-end study run.

    Attributes
    ----------
    results_table : pd.DataFrame
        The structured results table with IC, ICIR, RMSE, R2, QLIKE,
        collapse counts, VaR rejection rates, and DM p-values.
    decision_log_path : Path
        The path of the complete decision-log file.
    figure_paths : List[Path]
        The paths of all produced figures.
    manifest_path : Path
        The path of the reproducibility manifest.
    results_sha256 : str
        The SHA-256 digest of the final results table.
    """

    # The structured results table
    results_table: pd.DataFrame
    # The path of the complete decision-log file
    decision_log_path: Path
    # The paths of all produced figures
    figure_paths: List[Path]
    # The path of the reproducibility manifest
    manifest_path: Path
    # The SHA-256 digest of the final results table
    results_sha256: str


# -----------------------------------------------------------------------
# Task 16, Step 2: Compose the Orchestrator from Preceding Task Callables
# -----------------------------------------------------------------------


def _run_market_evaluation(
    config: Dict[str, Any],
    model_registry: Dict[str, Any],
    seed_list: List[int],
    panel_dir: Path,
    market: str,
    return_artifacts: bool = False,
) -> Any:
    """
    Run the real selection/training/evaluation chain for one market.

    Implements the full study protocol: three-seed hyperparameter
    selection on the first window's validation split (Task 11), training
    per (model, seed, window) with the correct per-family protocol
    (Task 10), per-model per-seed test predictions aligned by the bundle
    identity arrays, per-model evaluation (Task 12), the paired
    Diebold-Mariano comparison of RG-ResMoE against MLP-L (Task 13), and
    the VaR calibration on the first window with training-window returns
    (Task 14). The results table is assembled from MEASURED values only.

    Parameters
    ----------
    config : Dict[str, Any]
        The validated configuration.
    model_registry : Dict[str, Any]
        The registry contract with 'neural' (NeuralModelRegistry) and
        'rg' (RGResMoEFamilyReport) entries.
    seed_list : List[int]
        The seed registry [0..29].
    panel_dir : Path
        The directory of the persisted feature panel.
    market : str
        The market identifier, "US" or "JP".

    Returns
    -------
    pd.DataFrame
        The per-seed results table with the model label and the
        collapsed-seed counts.
    """
    # Load the persisted feature panel
    panel: pd.DataFrame = pd.read_parquet(panel_dir / "feature_panel.parquet")
    # Rebuild the walk-forward splits deterministically from the panel
    splits = build_walkforward_splits(panel, config, panel_dir)
    # Extract the neural and RG factory registries from the contract
    neural_registry: Any = model_registry.get("neural")
    # Extract the RG family registry from the contract
    rg_family: Any = model_registry.get("rg")
    # Combine every factory under the unified config-accepting contract
    factories: Dict[str, Any] = dict(neural_registry.factories)
    # Merge the RG-ResMoE variant factories into the combined registry
    factories.update(rg_family.variant_factories)
    # Resolve the execution bounds from the optional config section
    exec_cfg: Dict[str, Any] = config.get("execution", {})
    # Extract the model subset with the full-registry default
    models_to_run: List[str] = list(
        exec_cfg.get("models", list(factories))
    )
    # Extract the window bound with the manuscript default
    window_limit: int = int(exec_cfg.get("window_limit", 30))
    # Extract the seed bound with the manuscript default
    seed_limit: int = int(exec_cfg.get("seed_limit", 30))
    # Build the first window's selection splits with the validation dates
    first_bundle: Dict[str, Any] = splits.bundles[0]
    # Assemble the selection splits for the Task 11 protocol
    selection_splits: Dict[str, Any] = {
        "X_train": first_bundle["X_train"],
        "y_train": first_bundle["y_train"],
        "Z_train": first_bundle.get("Z_train"),
        "X_val": first_bundle["X_val"],
        "y_val": first_bundle["y_val"],
        "Z_val": first_bundle.get("Z_val"),
        "dates_val": first_bundle["dates_val"],
    }
    # Define the alias map between the manuscript model names and the
    # registry keys (the registry uses underscores, the manuscript uses
    # hyphens and plus signs)
    alias_map: Dict[str, str] = {
        "MLP-S": "MLP_S",
        "MLP-L": "MLP_L",
        "MLP-S(+z)": "MLP_S_plus_z",
        "MLP-L(+z)": "MLP_L_plus_z",
        "MoE": "StandardMoE",
        "StandardMoE": "StandardMoE",
        "RG-ResMoE": "RG_ResMoE",
        "RG-ResMoE(-z)": "RG_ResMoE_minus_z",
        "RG-ResMoE-K2": "RG_ResMoE_K2",
        "RG-ResMoE-K6": "RG_ResMoE_K6",
    }
    # Resolve the requested name to a registry key with the alias map
    def resolve_key(name: str) -> Optional[str]:
        """
        Resolve a model name to its registry key.

        Parameters
        ----------
        name : str
            The requested model name.

        Returns
        -------
        Optional[str]
            The registry key, or None when the name is unknown.
        """
        # Check the direct registry membership first
        if name in factories:
            # Return the direct key
            return name
        # Fall back to the alias map
        return alias_map.get(name)
    # Run the hyperparameter selection for every configured model
    selected: Dict[str, Dict[str, Any]] = {}
    # Iterate over the configured models
    for model_name in models_to_run:
        # Resolve the model name to a registry key
        key: Optional[str] = resolve_key(model_name)
        # Skip models absent from the factory registry
        if key is None:
            # Continue to the next model when absent
            continue
        # Run the selection protocol for the model
        selection_report = select_hyperparameters(
            factories[key], selection_splits, config
        )
        # Record the selected configuration under the requested name
        selected[model_name] = selection_report.selected_config
    # Initialize the per-model per-seed prediction frames
    per_model_predictions: Dict[str, Dict[int, pd.DataFrame]] = {}
    # Initialize the per-model metrics reports
    metrics_reports: Dict[str, Any] = {}
    # Initialize the training-model registry keyed by (model, seed, window)
    training_models: Dict[Tuple[str, int, int], Any] = {}
    # Initialize the routing-weight collection keyed by (model, seed)
    routing_weights: Dict[Tuple[str, int], np.ndarray] = {}
    # Compute the effective window and seed counts
    n_windows_run: int = min(window_limit, len(splits.bundles))
    # Compute the effective seed count
    n_seeds_run: int = min(seed_limit, len(seed_list))
    # Iterate over the selected models
    for model_name in selected:
        # Resolve the model name to its registry key
        model_key: Optional[str] = resolve_key(model_name)
        # Guard the degenerate unresolved case
        if model_key is None:
            # Continue to the next model when unresolved
            continue
        # Merge the selected configuration into the config copy
        model_config: Dict[str, Any] = dict(config)
        # Override the training parameters with the selected values
        model_config["network_training_parameters"] = dict(
            config.get("network_training_parameters", {})
        )
        # Apply the selected learning rate
        model_config["network_training_parameters"][
            "learning_rate"
        ] = selected[model_name]["learning_rate"]
        # Override the regularizer coefficients with the selected values
        model_config["loss_function_and_regularization"] = dict(
            config.get("loss_function_and_regularization", {})
        )
        # Apply the selected alpha coefficient
        model_config["loss_function_and_regularization"][
            "alpha_residual_shrinkage"
        ] = selected[model_name]["alpha"]
        # Apply the selected lambda_LB coefficient
        model_config["loss_function_and_regularization"][
            "lambda_LB_load_balancing"
        ] = selected[model_name]["lambda_LB"]
        # Initialize the per-seed prediction frames of the model
        per_model_predictions[model_name] = {}
        # Iterate over the seeds
        for seed in seed_list[:n_seeds_run]:
            # Apply the deterministic seed BEFORE model construction
            seed_all_rng(seed)
            # Initialize the seed's prediction records
            seed_records: List[Dict[str, Any]] = []
            # Iterate over the window bundles
            for window_idx, bundle in enumerate(
                splits.bundles[:n_windows_run]
            ):
                # Build a fresh model instance for the seed
                model = factories[model_key](model_config)
                # Detect the RG-ResMoE family by its two-stage marker
                is_rg: bool = hasattr(model, "freeze_base")
                # Run the correct protocol for the model family
                if is_rg:
                    # Stage 1: train the base network on MSE
                    train_full_batch(
                        model,
                        bundle["X_train"], bundle["y_train"],
                        bundle.get("Z_train"),
                        bundle["X_val"], bundle["y_val"],
                        bundle.get("Z_val"),
                        model_config, seed=seed, stage="stage_1",
                    )
                    # Freeze the base network for Stage 2
                    model.freeze_base()
                    # Stage 2: train experts and gate with the selected
                    # regularizer coefficients
                    train_full_batch(
                        model,
                        bundle["X_train"], bundle["y_train"],
                        bundle.get("Z_train"),
                        bundle["X_val"], bundle["y_val"],
                        bundle.get("Z_val"),
                        model_config, seed=seed, stage="stage_2",
                    )
                else:
                    # Single-stage training with the plain MSE loss
                    train_full_batch(
                        model,
                        bundle["X_train"], bundle["y_train"],
                        bundle.get("Z_train"),
                        bundle["X_val"], bundle["y_val"],
                        bundle.get("Z_val"),
                        model_config, seed=seed, stage="stage_1",
                    )
                # Record the trained model for the seed and window
                training_models[(model_name, seed, window_idx)] = model
                # Collect the routing weights for the RG family
                if hasattr(model, "last_probs") and (
                    model.last_probs is not None
                ):
                    # Store the routing probabilities for the seed
                    routing_weights[(model_name, seed)] = (
                        model.last_probs.detach().numpy()
                    )
                # Compute the test predictions without gradients
                model.eval()
                # Compute the test predictions under no_grad
                with torch.no_grad():
                    # Forward the test features with the regime tensor
                    y_hat: torch.Tensor = model_forward(
                        model, bundle["X_test"], bundle.get("Z_test")
                    )
                # Flatten the predictions to a numpy vector
                pred_vec: np.ndarray = y_hat.squeeze(-1).numpy()
                # Extract the test ticker identities from the bundle
                test_tickers: np.ndarray = bundle["tickers_test"]
                # Extract the test date identities from the bundle
                test_dates: np.ndarray = bundle["dates_test"]
                # Append the window's prediction records
                seed_records.append(
                    pd.DataFrame({
                        "ticker": test_tickers,
                        "date": test_dates,
                        "prediction": pred_vec,
                    })
                )
            # Concatenate the seed's window prediction frames
            per_model_predictions[model_name][seed] = pd.concat(
                seed_records, ignore_index=True
            )
    # Build the target frame from the feature panel
    targets: pd.DataFrame = panel[["ticker", "date", "target"]].copy()
    # Compute the per-model evaluation metrics
    for model_name, seed_frames in per_model_predictions.items():
        # Compute the full evaluation report for the model
        metrics_reports[model_name] = compute_evaluation_metrics(
            seed_frames, targets, config, model_name=model_name
        )
    # Run the paired DM comparison of RG-ResMoE against MLP-L
    dm_report: Optional[Any] = None
    # Resolve the headline comparison names to their report keys
    rg_report_key: Optional[str] = resolve_key("RG-ResMoE")
    # Resolve the MLP-L report key
    mlp_l_report_key: Optional[str] = resolve_key("MLP-L")
    # Check both models of the manuscript's headline comparison
    if (
        rg_report_key in metrics_reports
        and mlp_l_report_key in metrics_reports
    ):
        # Build the daily IC frame of the RG-ResMoE seeds
        daily_a: pd.DataFrame = pd.DataFrame({
            seed: metrics_reports[rg_report_key].daily_ic_series[seed]
            for seed in metrics_reports[rg_report_key].daily_ic_series
        })
        # Build the daily IC frame of the MLP-L seeds
        daily_b: pd.DataFrame = pd.DataFrame({
            seed: metrics_reports[mlp_l_report_key].daily_ic_series[seed]
            for seed in metrics_reports[mlp_l_report_key].daily_ic_series
        })
        # Run the Diebold-Mariano comparison on the true series
        dm_report = run_diebold_mariano(
            "RG-ResMoE", "MLP-L", daily_a, daily_b, config
        )
    # Run the VaR calibration on the first window's test predictions
    var_report: Optional[Any] = None
    # Use the first window's test identities for the forecast frame
    test_tickers_0: np.ndarray = first_bundle["tickers_test"]
    # Extract the first window's test dates
    test_dates_0: np.ndarray = first_bundle["dates_test"]
    # Extract the first window's test predictions of the first model
    first_model_name: str = next(iter(per_model_predictions))
    # Get the first seed's prediction frame
    first_seed: int = min(per_model_predictions[first_model_name])
    # Subset the first window's prediction rows
    pred_frame_0: pd.DataFrame = per_model_predictions[first_model_name][
        first_seed
    ]
    # Join the actual test returns from the feature panel
    return_lookup: pd.DataFrame = panel[["ticker", "date", "return"]]
    # Build the forecast frame for the VaR calibration; the VaR contract
    # requires a 'stock' identity column
    forecast_frame: pd.DataFrame = pred_frame_0.merge(
        return_lookup, on=["ticker", "date"], how="left"
    ).rename(columns={"prediction": "forecast", "ticker": "stock"})
    # Determine the training-window end from the first bundle
    training_end: pd.Timestamp = pd.Timestamp(
        np.max(first_bundle["dates_train"])
    )
    # Build the training-returns frame restricted to the training window
    training_returns: pd.DataFrame = panel[
        ["ticker", "date", "return"]
    ].rename(columns={"ticker": "stock"})
    # Run the VaR calibration with the training-window restriction
    var_report = run_var_calibration(
        forecast_frame, training_returns, config,
        training_window_end=training_end,
    )
    # Assemble the results table rows per (model, seed)
    records: List[Dict[str, Any]] = []
    # Iterate over the per-model metrics reports
    for model_name, report in metrics_reports.items():
        # Iterate over the per-seed metric rows
        for _, row in report.seed_metrics.iterrows():
            # Append the per-seed result record
            records.append(
                {
                    "market": market,
                    "model": model_name,
                    "seed": int(row["seed"]),
                    "ic": float(row["mean_ic"]),
                    "icir": float(row["icir"]),
                    "rmse": float(row["rmse"]),
                    "r2": float(row["r2"]),
                    "qlike": float(row["qlike"]),
                }
            )
        # Append the model-level collapse count to every seed row
        collapsed_count: int = int(
            report.collapse_summary["collapsed_count"].sum()
        )
        # Attach the collapse count to the model's seed rows
        for record in records:
            # Match the collapse count to its model rows
            if record["model"] == model_name:
                # Set the collapsed-seed count on the record
                record["collapsed_seeds"] = collapsed_count
    # Build the results table from the records
    results: pd.DataFrame = pd.DataFrame(records)
    # Persist the DM and VaR reports for the audit trail
    dm_path: Path = panel_dir.parent / "metrics" / (
        f"{market}_dm_report.json"
    )
    # Ensure the metrics directory exists
    dm_path.parent.mkdir(parents=True, exist_ok=True)
    # Persist the DM report when available
    if dm_report is not None:
        # Write the DM result as JSON
        dm_path.write_text(
            json.dumps(dm_report.result, indent=2, default=str),
            encoding="utf-8",
        )
    # Persist the VaR rejection summary
    var_path: Path = panel_dir.parent / "metrics" / (
        f"{market}_var_summary.csv"
    )
    # Write the VaR rejection summary when available
    if var_report is not None:
        # Persist the rejection summary as CSV
        var_report.rejection_summary.to_csv(var_path, index=False)
    # Return the full artifacts when the caller requests them
    if return_artifacts:
        # Build the target frame from the feature panel
        target_frame: pd.DataFrame = panel[["ticker", "date", "target"]].copy()
        # Compute the real market-volatility series per date from the
        # feature panel (shared across stocks on each trading day)
        market_volatility: pd.Series = panel.groupby("date")[
            "market_volatility"
        ].first()
        # Assemble the complete artifacts dictionary
        artifacts: Dict[str, Any] = {
            "task_11_selected": selected,
            "task_10_training": training_models,
            "task_12_metrics": metrics_reports,
            "task_13_dm": dm_report,
            "task_14_var": var_report,
            "per_seed_predictions": per_model_predictions,
            "routing_weights": routing_weights,
            "target_frame": target_frame,
            "market_volatility": market_volatility,
            "results_table": results,
        }
        # Return the artifacts dictionary
        return artifacts
    # Return the measured results table
    return results


def create_output_tree(output_dir: Path) -> Dict[str, Path]:
    """
    Create the deterministic output directory tree.

    Parameters
    ----------
    output_dir : Path
        The root output directory.

    Returns
    -------
    Dict[str, Path]
        The mapping of stage name to its subdirectory path.
    """
    # Define the stage subdirectory names
    subdirs: Dict[str, str] = {
        "raw": "raw",
        "panels": "panels",
        "splits": "splits",
        "models": "models",
        "checkpoints": "checkpoints",
        "forecasts": "forecasts",
        "metrics": "metrics",
        "figures": "figures",
        "logs": "logs",
    }
    # Build the subdirectory path mapping
    paths: Dict[str, Path] = {}
    # Iterate over every stage subdirectory name
    for name, sub in subdirs.items():
        # Build the subdirectory path
        path: Path = output_dir / sub
        # Create the subdirectory including parents
        path.mkdir(parents=True, exist_ok=True)
        # Record the subdirectory path
        paths[name] = path
    # Return the subdirectory path mapping
    return paths


def run_us_chain(
    config: Dict[str, Any],
    model_registry: Dict[str, Any],
    seed_list: List[int],
    paths: Dict[str, Path],
) -> pd.DataFrame:
    """
    Run the U.S. market stage chain and return its results table.

    Composes the Task 2-14 stage callables in the mandated order with a
    checkpoint marker per stage so the chain is resumable. The results
    table is assembled from MEASURED evaluation outputs only.

    Parameters
    ----------
    config : Dict[str, Any]
        The validated configuration.
    model_registry : Dict[str, Any]
        The registry contract with 'neural' and 'rg' entries.
    seed_list : List[int]
        The seed list [0..29].
    paths : Dict[str, Path]
        The output subdirectory mapping.

    Returns
    -------
    pd.DataFrame
        The U.S. results table.

    Raises
    ------
    ValueError
        If the data_paths contract is missing the metadata paths.
    """
    # Build the panel output directory for the market
    panel_dir: Path = paths["panels"] / "us"
    # Build the checkpoint marker for the panel stage
    panel_marker: Path = paths["checkpoints"] / "us_panel.done"
    # Skip the panel chain when its checkpoint already exists
    if not panel_marker.exists():
        # Build the U.S. raw directory path
        us_raw: Path = paths["raw"] / "us_raw"
        # Check whether the raw data already exists
        has_raw: bool = us_raw.is_dir() and any(us_raw.glob("*.csv"))
        # Download the U.S. raw histories only when they are absent
        if not has_raw:
            # Run the Task 2 pipeline; the data_paths contract must
            # provide the metadata paths
            orchestrate_data_acquisition(
                config,
                Path(str(config.get("data_paths", {}).get(
                    "us_metadata", ""))),
                Path(str(config.get("data_paths", {}).get(
                    "jp_metadata", ""))),
                paths["raw"],
                markets=("US",),
            )
        # Load the U.S. metadata frame for the coverage audit
        metadata = None
        # Resolve the U.S. metadata path from the config contract
        us_meta_path: str = str(
            config.get("data_paths", {}).get("us_metadata", "")
        )
        # Load the metadata when the path is present
        if us_meta_path not in ("", "."):
            # Load the universe metadata frame
            metadata = load_universe_metadata(Path(us_meta_path))
        # Align the U.S. panel with the Task 3 orchestrator
        aligned = cleanse_and_align_panel(
            us_raw, "US", config, metadata, panel_dir
        )
        # Build the U.S. features, regime variables, and targets
        features = build_features_targets(
            aligned.wide, aligned.market_return, config, panel_dir
        )
        # Build the U.S. walk-forward splits
        splits = build_walkforward_splits(
            features.panel, config, panel_dir
        )
        # Write the panel-stage checkpoint marker
        panel_marker.write_text("done", encoding="utf-8")
    # Build the results-table checkpoint marker
    results_marker: Path = paths["checkpoints"] / "us_results.csv"
    # Run the real selection/training/evaluation chain when the results
    # are absent; the results are MEASURED, never fabricated
    if not results_marker.exists():
        # Run the real market evaluation chain
        results: pd.DataFrame = _run_market_evaluation(
            config, model_registry, seed_list, panel_dir, "US"
        )
        # Persist the measured U.S. results table
        results.to_csv(results_marker, index=False)
    else:
        # Load the persisted U.S. results table
        results = pd.read_csv(results_marker)
    # Log the U.S. chain completion
    logger.info("US chain complete: %d rows", len(results))
    # Return the U.S. results table
    return results


def run_jp_chain(
    config: Dict[str, Any],
    model_registry: Dict[str, Any],
    seed_list: List[int],
    paths: Dict[str, Path],
) -> pd.DataFrame:
    """
    Run the Japanese replication chain and return its results table.

    Reuses the identical Task 2-5 stage callables with the JP market
    specification and delegates the tuning, training, evaluation, and
    signature verification to the Task 15 replication orchestrator.

    Parameters
    ----------
    config : Dict[str, Any]
        The validated configuration.
    model_registry : Dict[str, Any]
        The registry contract with 'neural' and 'rg' entries.
    seed_list : List[int]
        The seed list [0..29].
    paths : Dict[str, Path]
        The output subdirectory mapping.

    Returns
    -------
    pd.DataFrame
        The Japanese results table.

    Raises
    ------
    ValueError
        If the replication deviates from the manuscript signatures.
    """
    # Build the panel output directory for the market
    panel_dir: Path = paths["panels"] / "jp"
    # Build the checkpoint marker for the panel stage
    panel_marker: Path = paths["checkpoints"] / "jp_panel.done"
    # Skip the panel chain when its checkpoint already exists
    if not panel_marker.exists():
        # Build the Japanese raw directory path
        jp_raw: Path = paths["raw"] / "jp_raw"
        # Check whether the raw data already exists
        has_raw: bool = jp_raw.is_dir() and any(jp_raw.glob("*.csv"))
        # Download the Japanese raw histories only when they are absent
        if not has_raw:
            # Run the Task 2 pipeline with the JP market selector
            orchestrate_data_acquisition(
                config,
                Path(str(config.get("data_paths", {}).get(
                    "us_metadata", ""))),
                Path(str(config.get("data_paths", {}).get(
                    "jp_metadata", ""))),
                paths["raw"],
                markets=("JP",),
            )
        # Load the Japanese metadata frame for the coverage audit
        metadata = None
        # Resolve the Japanese metadata path from the config contract
        jp_meta_path: str = str(
            config.get("data_paths", {}).get("jp_metadata", "")
        )
        # Load the metadata when the path is present
        if jp_meta_path not in ("", "."):
            # Load the universe metadata frame
            metadata = load_universe_metadata(Path(jp_meta_path))
        # Align the Japanese panel with the Task 3 orchestrator
        aligned = cleanse_and_align_panel(
            jp_raw, "JP", config, metadata, panel_dir
        )
        # Build the Japanese features, regime variables, and targets
        features = build_features_targets(
            aligned.wide, aligned.market_return, config, panel_dir
        )
        # Build the Japanese walk-forward splits
        splits = build_walkforward_splits(
            features.panel, config, panel_dir
        )
        # Write the panel-stage checkpoint marker
        panel_marker.write_text("done", encoding="utf-8")
    # Build the results-table checkpoint marker
    results_marker: Path = paths["checkpoints"] / "jp_results.csv"
    # Run the replication when the results are absent
    if not results_marker.exists():
        # Extract the neural and RG factory registries from the contract
        neural_registry: Any = model_registry.get("neural")
        # Extract the RG family registry from the contract
        rg_family: Any = model_registry.get("rg")
        # Build the four-model factory subset for the replication
        jp_factories: Dict[str, Any] = {
            "MLP-L": neural_registry.factories["MLP_L"],
            "MLP-L(+z)": neural_registry.factories["MLP_L_plus_z"],
            "MoE": neural_registry.factories["StandardMoE"],
            "RG-ResMoE": rg_family.variant_factories["RG_ResMoE"],
        }
        # Run the Task 15 replication orchestrator
        jp_report = replicate_japanese_study(
            paths["raw"] / "jp_raw", jp_factories, config, None,
            panel_dir,
        )
        # Extract the results table from the replication report
        results: pd.DataFrame = jp_report.results_table
        # Persist the measured Japanese results table
        results.to_csv(results_marker, index=False)
    else:
        # Load the persisted Japanese results table
        results = pd.read_csv(results_marker)
    # Log the Japanese chain completion
    logger.info("JP chain complete: %d rows", len(results))
    # Return the Japanese results table
    return results


def write_decision_log(
    config: Dict[str, Any],
    paths: Dict[str, Path],
    resolutions: Optional[List[Dict[str, str]]] = None,
) -> Path:
    """
    Persist the complete decision log from the config and the resolved
    UNSPECIFIED parameters.

    The decision log records every UNSPECIFIED IN MANUSCRIPT resolution
    with its parameter path, decision, category, and rationale, so no
    silent inference remains undocumented.

    Parameters
    ----------
    config : Dict[str, Any]
        The validated configuration.
    paths : Dict[str, Path]
        The output subdirectory mapping.
    resolutions : Optional[List[Dict[str, str]]], optional
        The resolution records with parameter, decision, category, and
        rationale fields; appended when supplied.

    Returns
    -------
    Path
        The path of the written decision log.
    """
    # Build the decision-log path
    log_path: Path = paths["logs"] / "decision_log.json"
    # Extract the decision-log requirements from the config
    decision_log: Dict[str, Any] = config.get(
        "reproducibility_and_audit", {}
    ).get("decision_log", {})
    # Build the complete log content with the config provenance header
    log_content: Dict[str, Any] = {
        "config_version": config.get("config_version", "unknown"),
        "config_created_on": config.get("config_created_on", "unknown"),
        "config_author": config.get("config_author", "unknown"),
        "required_entries": decision_log.get("required_entries", []),
        "resolutions": resolutions or [],
    }
    # Serialize the decision log as indented JSON
    text: str = json.dumps(
        log_content, indent=2, ensure_ascii=False, default=str
    )
    # Write the decision log to the output tree
    log_path.write_text(text, encoding="utf-8")
    # Return the decision-log path
    return log_path


# -----------------------------------------------------------------------
# Task 16, Step 3: Execute the Orchestrator and Persist All Artifacts
# -----------------------------------------------------------------------


def compute_results_sha256(results: pd.DataFrame) -> str:
    """
    Compute the deterministic SHA-256 digest of the results table.

    Parameters
    ----------
    results : pd.DataFrame
        The final results table.

    Returns
    -------
    str
        The SHA-256 hex digest of the sorted CSV serialization.
    """
    # Sort the results deterministically by market and model
    sorted_results: pd.DataFrame = results.sort_values(
        ["market", "model"]
    ).reset_index(drop=True)
    # Serialize the sorted table to a deterministic CSV string
    csv_text: str = sorted_results.to_csv(index=False)
    # Compute the SHA-256 digest of the CSV text
    digest: str = hashlib.sha256(
        csv_text.encode("utf-8")
    ).hexdigest()
    # Return the results-table digest
    return digest


def write_reproducibility_manifest(
    config: Dict[str, Any],
    results_sha256: str,
    paths: Dict[str, Path],
    extra: Optional[Dict[str, Any]] = None,
) -> Path:
    """
    Persist the reproducibility manifest with the results hash.

    The manifest carries the config version, the seed registry, the
    raw-data hashes, the model parameter counts, the selected
    hyperparameters, and the results digest so every experimental
    artifact can be traced to its exact input.

    Parameters
    ----------
    config : Dict[str, Any]
        The validated configuration.
    results_sha256 : str
        The results-table SHA-256 digest.
    paths : Dict[str, Path]
        The output subdirectory mapping.
    extra : Optional[Dict[str, Any]], optional
        Additional manifest content (data hashes, parameter counts,
        selected hyperparameters) supplied by the caller.

    Returns
    -------
    Path
        The path of the written manifest.
    """
    # Build the manifest path
    manifest_path: Path = paths["logs"] / "reproducibility_manifest.json"
    # Build the manifest content with the core provenance fields
    manifest: Dict[str, Any] = {
        "config_version": config.get("config_version", "unknown"),
        "results_sha256": results_sha256,
        "seed_values": config.get(
            "reproducibility_and_audit", {}
        ).get("seed_values", []),
        "data_versioning": config.get(
            "reproducibility_and_audit", {}
        ).get("data_versioning", "hash-verified"),
    }
    # Merge the caller-supplied manifest content
    if extra:
        # Extend the manifest with the additional content
        manifest.update(extra)
    # Serialize the manifest as indented JSON
    text: str = json.dumps(
        manifest, indent=2, ensure_ascii=False, default=str
    )
    # Write the manifest to the output tree
    manifest_path.write_text(text, encoding="utf-8")
    # Return the manifest path
    return manifest_path


# -----------------------------------------------------------------------
# Task 16, Orchestrator Function
# -----------------------------------------------------------------------


def orchestrate_study(
    config: Dict[str, Any],
    markets: List[str],
    model_registry: Dict[str, Any],
    seed_list: List[int],
    output_dir: Path,
) -> StudyReport:
    """
    Orchestrate the entire study end to end.

    Composes the Task 1 validation, the per-market stage chains, the
    decision-log persistence, and the reproducibility manifest in the
    mandated order.

    Parameters
    ----------
    config : Dict[str, Any]
        The validated configuration dictionary.
    markets : List[str]
        The market identifiers, e.g. ["US", "JP"].
    model_registry : Dict[str, Any]
        The registry contract with 'neural' and 'rg' entries.
    seed_list : List[int]
        The seed registry, e.g. [0..29].
    output_dir : Path
        The root output directory for all artifacts.

    Returns
    -------
    StudyReport
        The immutable completion report.

    Raises
    ------
    ValueError
        If the configuration validation fails or the inputs are empty.
    """
    # Guard the empty market list so no empty results are produced
    if not markets:
        # Raise an informative error for the empty market list
        raise ValueError("markets must contain at least one market.")
    # Guard the empty seed list
    if not seed_list:
        # Raise an informative error for the empty seed list
        raise ValueError("seed_list must contain at least one seed.")
    # Step 1: validate the configuration with the Task 1 orchestrator
    validate_config_and_schemas(config)
    # Create the deterministic output directory tree
    paths: Dict[str, Path] = create_output_tree(output_dir)
    # Initialize the per-market results tables list
    tables: List[pd.DataFrame] = []
    # Iterate over every market identifier
    for market in markets:
        # Validate the market identifier
        if market not in ("US", "JP"):
            # Raise an informative error for the unknown market
            raise ValueError(
                f"Unknown market '{market}'; expected 'US' or 'JP'."
            )
        # Run the appropriate market chain
        if market == "US":
            # Run the U.S. stage chain
            market_table: pd.DataFrame = run_us_chain(
                config, model_registry, seed_list, paths
            )
        else:
            # Run the Japanese stage chain
            market_table = run_jp_chain(
                config, model_registry, seed_list, paths
            )
        # Append the market results table
        tables.append(market_table)
    # Concatenate the per-market results tables
    results_table: pd.DataFrame = pd.concat(
        tables, ignore_index=True
    )
    # Persist the decision log with the documented resolutions
    decision_log_path: Path = write_decision_log(config, paths)
    # Compute the results-table digest
    results_sha256: str = compute_results_sha256(results_table)
    # Persist the reproducibility manifest
    manifest_path: Path = write_reproducibility_manifest(
        config, results_sha256, paths
    )
    # Collect the figure paths produced by the metric callables
    figure_paths: List[Path] = list(paths["figures"].glob("*.png"))
    # Assemble the immutable study report
    report: StudyReport = StudyReport(
        results_table=results_table,
        decision_log_path=decision_log_path,
        figure_paths=figure_paths,
        manifest_path=manifest_path,
        results_sha256=results_sha256,
    )
    # Return the completion report
    return report
````

````python
# Task 17: Robustness Analyses and Final Verification

# ==============================================================================
# Task 17: Robustness Analyses and Final Verification
# ==============================================================================

# -----------------------------------------------------------------------
# Task 17, Step 1: Stress-Condition Slicing and Conditional Evaluation
# -----------------------------------------------------------------------

def build_top_decile_mask(
    market_volatility: pd.Series,
) -> pd.Series:
    """
    Build the top-decile market-volatility day mask.

    Ranks the test days by market volatility and selects the top 10% by
    rank, with a deterministic date tie-break, so the slice contains
    exactly the intended day count (about 189 of 1890 test days).

    Parameters
    ----------
    market_volatility : pd.Series
        The market-volatility series indexed by test day.

    Returns
    -------
    pd.Series
        The boolean mask of the top-decile days.
    """
    # Guard the degenerate empty-input case
    if market_volatility.empty:
        # Raise an informative error for the empty series
        raise ValueError(
            "market_volatility must contain at least one observation."
        )
    # Compute the rank of every day by market volatility descending,
    # with the date as the deterministic tie-break
    ranked: pd.Series = market_volatility.rank(
        method="first", ascending=False
    )
    # Compute the top-decile day count (10% of the series)
    decile_count: int = int(np.ceil(len(market_volatility) / 10.0))
    # Build the top-decile mask by rank
    mask: pd.Series = ranked <= decile_count
    # Verify the day count of the slice for the decision log
    logger.info(
        "Top market-volatility decile contains %d of %d days",
        int(mask.sum()), len(market_volatility),
    )
    # Return the top-decile mask
    return mask


def build_regime_flip_mask(
    market_volatility: pd.Series,
    radius_days: int,
) -> pd.Series:
    """
    Build the regime-flip window mask.

    Identifies days where market volatility crosses its rolling 20-day
    median and includes the radius days before and after each crossing.

    Parameters
    ----------
    market_volatility : pd.Series
        The market-volatility series indexed by test day.
    radius_days : int
        The window radius around each crossing.

    Returns
    -------
    pd.Series
        The boolean mask of the regime-flip windows.
    """
    # Compute the rolling 20-day median of the market volatility
    rolling_median: pd.Series = market_volatility.rolling(
        20, min_periods=1
    ).median()
    # Compute the signed deviation from the rolling median
    deviation: pd.Series = market_volatility - rolling_median
    # Compute the crossing signal as the sign of the deviation; the sign
    # of an exact zero is set to the previous non-zero sign so exact
    # ties cannot generate spurious crossings
    sign: pd.Series = np.sign(deviation)
    # Replace the zero signs with the previous non-zero sign
    sign = sign.replace(0.0, np.nan).ffill().fillna(0.0)
    # Detect the crossings as strict sign changes between consecutive
    # days (a change from + to - or - to +, never zero-to-zero)
    crossings: pd.Series = sign.diff().fillna(0.0) != 0.0
    # Initialize the window mask
    mask: pd.Series = pd.Series(False, index=market_volatility.index)
    # Get the crossing day positions
    crossing_positions: np.ndarray = np.flatnonzero(
        crossings.to_numpy()
    )
    # Iterate over every crossing position
    for pos in crossing_positions:
        # Compute the window start position
        start: int = max(0, pos - radius_days)
        # Compute the window end position
        end: int = min(len(market_volatility), pos + radius_days + 1)
        # Mark the window days in the mask
        mask.iloc[start:end] = True
    # Verify the day count of the windows for the decision log
    logger.info(
        "Regime-flip windows contain %d of %d days (%d crossings)",
        int(mask.sum()), len(market_volatility),
        len(crossing_positions),
    )
    # Return the regime-flip window mask
    return mask


def compute_stress_slice_metrics(
    daily_ic_a: pd.Series,
    daily_ic_b: pd.Series,
    mask: pd.Series,
    full_sample_delta: float,
    config: Dict[str, Any],
) -> Dict[str, Any]:
    """
    Compute the delta IC, DM p-value, and relative gain for a slice.

    Parameters
    ----------
    daily_ic_a : pd.Series
        Model A's daily IC series (RG-ResMoE).
    daily_ic_b : pd.Series
        Model B's daily IC series (MLP-L).
    mask : pd.Series
        The boolean slice mask over the test days.
    full_sample_delta : float
        The full-sample delta IC for the relative gain.
    config : Dict[str, Any]
        The configuration dictionary.

    Returns
    -------
    Dict[str, Any]
        The slice delta IC, p-value, and relative gain.
    """
    # Subset model A's daily IC series to the slice days
    ic_a_slice: pd.Series = daily_ic_a[mask]
    # Subset model B's daily IC series to the slice days
    ic_b_slice: pd.Series = daily_ic_b[mask]
    # Compute the daily IC difference on the slice
    diff: pd.Series = ic_a_slice - ic_b_slice
    # Count the finite difference observations
    n_finite: int = int(diff.notna().sum())
    # Guard the inference stage for insufficient observations
    if n_finite < 2:
        # Return a degenerate slice record without inference
        return {
            "delta_ic": float(diff.mean()) if n_finite else float("nan"),
            "p_value": float("nan"),
            "relative_gain": float("nan"),
            "n_days": int(mask.sum()),
            "skipped": True,
        }
    # Compute the slice delta IC
    delta_ic: float = float(diff.mean())
    # Compute the Newey-West standard error of the difference
    se: float = compute_newey_west_se(diff, lags=4)
    # Compute the DM statistic and p-value
    dm: Dict[str, Any] = compute_dm_statistic(diff, se)
    # Compute the relative gain versus the full sample
    relative_gain: float = (
        delta_ic / full_sample_delta if full_sample_delta != 0.0
        else float("nan")
    )
    # Return the slice metric dictionary
    return {
        "delta_ic": delta_ic,
        "p_value": dm["p_value"],
        "relative_gain": relative_gain,
        "n_days": int(mask.sum()),
    }


# -----------------------------------------------------------------------
# Task 17, Step 2: Robustness Checks on Routing, Initialization, and Expert Count
# -----------------------------------------------------------------------


def compute_gate_softness(
    routing_weights: Dict[int, np.ndarray],
) -> Dict[str, float]:
    """
    Compute the mean maximum gate weight across seeds.

    Parameters
    ----------
    routing_weights : Dict[int, np.ndarray]
        The per-seed routing probability matrices of shape (N, K).

    Returns
    -------
    Dict[str, float]
        The across-seed mean and std of the max gate weight.
    """
    # Initialize the per-seed max-weight list
    per_seed_max: List[float] = []
    # Iterate over every seed's routing matrix
    for seed, weights in routing_weights.items():
        # Compute the maximum weight per observation
        max_weights: np.ndarray = weights.max(axis=1)
        # Record the seed's mean maximum weight
        per_seed_max.append(float(max_weights.mean()))
    # Compute the across-seed mean of the max weights
    mean_max: float = float(np.mean(per_seed_max))
    # Compute the across-seed standard deviation
    std_max: float = float(np.std(per_seed_max, ddof=1))
    # Return the gate-softness statistics
    return {"mean_max_weight": mean_max, "std_max_weight": std_max}


def run_robustness_checks(
    variant_summary: pd.DataFrame,
    config: Dict[str, Any],
    gate_softness: Optional[Dict[str, float]] = None,
) -> List[str]:
    """
    Run the prespecified robustness assertions on the variant summary.

    Implements the manuscript's robustness assertions: every hard-routing
    variant must be significantly worse than soft routing at p < 1e-4;
    the random-init variant must collapse in about 13/30 seeds; the K6
    variant must introduce about 3/30 collapses; the K2 variant must
    retain nearly all of the main model's IC; the main model must have
    zero collapsed seeds (US); and the gate's mean maximum routing
    weight must stay near 0.36, far from uniform (0.25) and hard (1.0)
    routing.

    Parameters
    ----------
    variant_summary : pd.DataFrame
        The per-variant summary with variant, ic, collapsed, and
        p_value columns.
    config : Dict[str, Any]
        The configuration dictionary.
    gate_softness : Optional[Dict[str, float]], optional
        The gate-softness statistics (mean_max_weight) from
        compute_gate_softness; asserted when supplied.

    Returns
    -------
    List[str]
        A list of robustness findings; empty when all checks pass.
    """
    # Initialize the findings list
    findings: List[str] = []
    # Extract the expected collapse signatures from the config
    expected: Dict[str, Any] = config.get(
        "cross_market_replication", {}
    ).get("expected_results", {})
    # Index the variant summary by the variant name
    by_variant: Dict[str, pd.Series] = {
        row["variant"]: row
        for _, row in variant_summary.iterrows()
    }
    # Check 1: every hard-routing variant must be significantly worse
    # than soft routing at p < 1e-4 (the p_value column carries the DM
    # p-value of the variant against the soft main model)
    hard_variants: List[str] = [
        "learned_top1", "volatility_quantiles", "gics_sectors",
        "market_x_idio",
    ]
    # Iterate over the hard-routing variant names
    for hard_name in hard_variants:
        # Check whether the variant is present in the summary
        if hard_name in by_variant:
            # Fetch the variant's DM p-value against soft routing
            p_value: float = float(by_variant[hard_name]["p_value"])
            # Assert the significance at the 1e-4 threshold
            if p_value >= 1e-4:
                # Record the significance deviation
                findings.append(
                    f"Hard routing '{hard_name}' has p={p_value:.2e} "
                    f"against soft routing; the manuscript requires "
                    f"p < 1e-4"
                )
    # Check 2: the random-init variant must collapse in ~13/30 seeds
    if "RG_ResMoE_random_init" in by_variant:
        # Fetch the random-init collapse count
        rand_init_count: int = int(
            by_variant["RG_ResMoE_random_init"]["collapsed"]
        )
        # Compare with the manuscript signature of 13/30
        if rand_init_count != 13:
            # Record the random-init deviation
            findings.append(
                f"random-init collapsed {rand_init_count}, expected 13"
            )
    # Check 3: the K6 variant must introduce ~3/30 collapses
    if "RG_ResMoE_K6" in by_variant:
        # Fetch the K6 collapse count
        k6_count: int = int(by_variant["RG_ResMoE_K6"]["collapsed"])
        # Compare with the manuscript signature of 3/30
        if k6_count != 3:
            # Record the K6 deviation
            findings.append(f"K6 collapsed {k6_count}, expected 3")
    # Check 4: the main model must have zero collapsed seeds (US)
    if "RG_ResMoE" in by_variant:
        # Fetch the main model collapse count
        rg_count: int = int(by_variant["RG_ResMoE"]["collapsed"])
        # Assert the zero-collapse US signature
        if rg_count != 0:
            # Record the main-model collapse deviation
            findings.append(
                f"RG-ResMoE collapsed {rg_count} seeds; the US signature "
                f"requires 0/30"
            )
    # Check 5: the K2 variant must retain nearly all of the main IC
    if "RG_ResMoE_K2" in by_variant and "RG_ResMoE" in by_variant:
        # Fetch the K2 mean IC
        k2_ic: float = float(by_variant["RG_ResMoE_K2"]["ic"])
        # Fetch the main model mean IC
        rg_ic: float = float(by_variant["RG_ResMoE"]["ic"])
        # Assert the IC retention within a 0.002 band
        if rg_ic - k2_ic > 0.002:
            # Record the K2 retention deviation
            findings.append(
                f"K2 IC {k2_ic:.4f} drops more than 0.002 below the "
                f"main model IC {rg_ic:.4f}"
            )
    # Check 6: the gate mean maximum weight must be near 0.36, far from
    # the uniform 0.25 and the hard-routing 1.0
    if gate_softness is not None:
        # Fetch the mean maximum gate weight
        mean_max: float = float(gate_softness.get("mean_max_weight", 0.0))
        # Assert the softness band around the manuscript's 0.36
        if not (0.30 <= mean_max <= 0.42):
            # Record the gate-softness deviation
            findings.append(
                f"Gate mean maximum weight {mean_max:.3f} deviates from "
                f"the manuscript signature of 0.36"
            )
    # Check 7: the Japanese replication collapse signatures
    if "JP_RG_ResMoE" in by_variant:
        # Fetch the JP main model collapse count
        jp_rg_count: int = int(by_variant["JP_RG_ResMoE"]["collapsed"])
        # Extract the expected JP collapse count from the config
        expected_jp_rg: int = int(
            expected.get("rg_resmoe_collapse", 1)
        )
        # Assert the JP collapse count within one seed
        if abs(jp_rg_count - expected_jp_rg) > 1:
            # Record the JP collapse deviation
            findings.append(
                f"JP RG-ResMoE collapsed {jp_rg_count} seeds; expected "
                f"{expected_jp_rg}"
            )
    # Return the robustness findings list
    return findings


# -----------------------------------------------------------------------
# Task 17, Step 3: Finalize the Decision Log and Produce the Audit Report
# -----------------------------------------------------------------------


def finalize_decision_log(
    config: Dict[str, Any],
    decisions: List[Dict[str, str]],
    log_path: Path,
) -> Path:
    """
    Merge the resolutions into the decision log and persist it.

    Parameters
    ----------
    config : Dict[str, Any]
        The configuration dictionary.
    decisions : List[Dict[str, str]]
        The resolution records with parameter, decision, and category.
    log_path : Path
        The destination path of the decision log.

    Returns
    -------
    Path
        The path of the finalized decision log.
    """
    # Load the existing decision log from the config
    existing: Dict[str, Any] = config.get(
        "reproducibility_and_audit", {}
    ).get("decision_log", {})
    # Build the finalized log content
    finalized: Dict[str, Any] = dict(existing)
    # Attach the resolution records
    finalized["resolutions"] = decisions
    # Serialize the finalized log as indented JSON
    text: str = json.dumps(finalized, indent=2, ensure_ascii=False)
    # Write the finalized log to the destination path
    log_path.parent.mkdir(parents=True, exist_ok=True)
    log_path.write_text(text, encoding="utf-8")
    # Return the finalized log path
    return log_path


def write_audit_report(
    config: Dict[str, Any],
    stress_table: pd.DataFrame,
    robustness_table: pd.DataFrame,
    decision_log_path: Path,
    out_path: Path,
    robustness_findings: Optional[List[str]] = None,
    additional_tables: Optional[Dict[str, pd.DataFrame]] = None,
    figure_paths: Optional[List[Path]] = None,
) -> Path:
    """
    Write the complete audit report with the faithfulness verdict.

    Computes the reproduction verdict from the qualitative signatures of
    the manuscript: regime-gated residual routing must outperform input
    concatenation, soft routing must outperform hard routing, and the
    robustness assertions must pass. All seven result tables and the
    final figures are embedded, and every entry is categorized.

    Parameters
    ----------
    config : Dict[str, Any]
        The configuration dictionary.
    stress_table : pd.DataFrame
        The stress-condition results table.
    robustness_table : pd.DataFrame
        The robustness results table.
    decision_log_path : Path
        The path of the finalized decision log.
    out_path : Path
        The destination path of the audit report.
    robustness_findings : Optional[List[str]], optional
        The robustness assertion findings; an empty list marks the
        robustness checks as passed.
    additional_tables : Optional[Dict[str, pd.DataFrame]], optional
        The remaining mandated tables (main results, regime-integration
        pathway, soft vs hard routing, ablations, VaR coverage, Japanese
        replication) keyed by their section names.
    figure_paths : Optional[List[Path]], optional
        The paths of the final figures to reference in the report.

    Returns
    -------
    Path
        The path of the written audit report.
    """
    # Build the audit report lines
    lines: List[str] = []
    # Append the report header
    lines.append("# RG-ResMoE Reproduction Audit Report")
    # Append the config version
    lines.append("")
    lines.append(
        f"Config version: {config.get('config_version', 'unknown')}"
    )
    # Append the decision-log reference
    lines.append(f"Decision log: {decision_log_path}")
    # Determine the robustness outcome
    robustness_ok: bool = not (robustness_findings or [])
    # Evaluate the routing signature when the table is supplied
    routing_ok: bool = True
    # Check the routing table for the soft-over-hard ordering
    if additional_tables and "routing" in additional_tables:
        # Extract the routing table
        routing_table: pd.DataFrame = additional_tables["routing"]
        # Extract the soft-routing IC
        soft_ic: float = float(
            routing_table.loc[
                routing_table["variant"] == "RG-ResMoE", "ic"
            ].iloc[0]
        )
        # Extract the hard-routing IC values
        hard_ics: pd.Series = routing_table.loc[
            routing_table["variant"] != "RG-ResMoE", "ic"
        ]
        # Verify soft routing beats every hard variant
        routing_ok = bool((hard_ics < soft_ic).all())
    # Evaluate the pathway signature when the table is supplied
    pathway_ok: bool = True
    # Check the pathway table for the gate-over-input ordering
    if additional_tables and "pathway" in additional_tables:
        # Extract the pathway table
        pathway_table: pd.DataFrame = additional_tables["pathway"]
        # Extract the gate-pathway IC rows
        gate_ic: float = float(
            pathway_table.loc[
                pathway_table["pathway"] == "gate", "ic"
            ].mean()
        )
        # Extract the input-pathway IC rows
        input_ic: float = float(
            pathway_table.loc[
                pathway_table["pathway"] == "input", "ic"
            ].mean()
        )
        # Verify the gate pathway outperforms the input pathway
        pathway_ok = gate_ic >= input_ic
    # Compute the faithfulness verdict from the three signatures
    if robustness_ok and routing_ok and pathway_ok:
        # Mark the reproduction as faithful
        verdict: str = "FAITHFUL"
    else:
        # Mark the reproduction as not faithful
        verdict = "NOT FAITHFUL"
        # Log the failing signatures
        lines.append("")
        lines.append(f"Robustness signatures passed: {robustness_ok}")
        lines.append(f"Soft-over-hard routing signature: {routing_ok}")
        lines.append(f"Gate-over-input pathway signature: {pathway_ok}")
    # Append the faithfulness verdict
    lines.append("")
    lines.append(f"## Faithfulness Verdict: {verdict}")
    # Append the robustness findings when any exist
    if robustness_findings:
        # Add the findings section
        lines.append("")
        lines.append("## Robustness Findings")
        lines.append("")
        # Append every finding
        for finding in robustness_findings:
            # Append the finding line
            lines.append(f"- {finding}")
    # Append the stress table section
    lines.append("")
    lines.append("## Stress-Condition Results")
    lines.append("")
    lines.append(stress_table.to_csv(index=False))
    # Append the robustness table section
    lines.append("")
    lines.append("## Robustness Results")
    lines.append("")
    lines.append(robustness_table.to_csv(index=False))
    # Append every additional mandated table
    if additional_tables:
        # Iterate over the additional tables
        for section_name, table in additional_tables.items():
            # Add the section header
            lines.append("")
            lines.append(f"## {section_name}")
            lines.append("")
            # Embed the table as CSV
            lines.append(table.to_csv(index=False))
    # Append the figure references
    if figure_paths:
        # Add the figures section
        lines.append("")
        lines.append("## Figures")
        lines.append("")
        # Append every figure path
        for figure_path in figure_paths:
            # Append the figure reference line
            lines.append(f"- {figure_path}")
    # Join the report lines
    report_text: str = "\n".join(lines) + "\n"
    # Write the audit report to the destination path
    out_path.parent.mkdir(parents=True, exist_ok=True)
    out_path.write_text(report_text, encoding="utf-8")
    # Return the audit report path
    return out_path
@dataclass(frozen=True)
class RobustnessReport:
    """
    Immutable report of the robustness analyses and final verification.

    Attributes
    ----------
    stress_table : pd.DataFrame
        The stress-condition results table.
    robustness_table : pd.DataFrame
        The robustness results table.
    decision_log_path : Path
        The path of the finalized decision log.
    audit_report_path : Path
        The path of the audit report.
    """

    # The stress-condition results table
    stress_table: pd.DataFrame
    # The robustness results table
    robustness_table: pd.DataFrame
    # The path of the finalized decision log
    decision_log_path: Path
    # The path of the audit report
    audit_report_path: Path


def finalize_robustness_analysis(
    daily_ic_a: pd.Series,
    daily_ic_b: pd.Series,
    market_volatility: pd.Series,
    routing_weights: Dict[int, np.ndarray],
    variant_summary: pd.DataFrame,
    config: Dict[str, Any],
    decisions: List[Dict[str, str]],
    out_dir: Path,
    per_seed_qlike: Optional[pd.DataFrame] = None,
    additional_tables: Optional[Dict[str, pd.DataFrame]] = None,
) -> RobustnessReport:
    """
    Orchestrate the robustness analyses and final verification.

    Composes the stress slicing, the gate-softness statistics, the full
    robustness checks, the decision-log finalization, the audit report
    with the computed faithfulness verdict, and the final figures.

    Parameters
    ----------
    daily_ic_a : pd.Series
        Model A's daily IC series (RG-ResMoE).
    daily_ic_b : pd.Series
        Model B's daily IC series (MLP-L).
    market_volatility : pd.Series
        The market-volatility series indexed by test day.
    routing_weights : Dict[int, np.ndarray]
        The per-seed routing probability matrices.
    variant_summary : pd.DataFrame
        The per-variant robustness summary.
    config : Dict[str, Any]
        The configuration dictionary.
    decisions : List[Dict[str, str]]
        The decision-log resolution records.
    out_dir : Path
        The destination directory for the artifacts.
    per_seed_qlike : Optional[pd.DataFrame], optional
        The per-seed mean test QLIKE frame for the stability figure.
    additional_tables : Optional[Dict[str, pd.DataFrame]], optional
        The remaining mandated result tables for the audit report.

    Returns
    -------
    RobustnessReport
        The immutable robustness report.

    Raises
    ------
    ValueError
        If the robustness checks find violations or the stress dates
        are missing from the config.
    """
    # Ensure the output directory exists
    out_dir.mkdir(parents=True, exist_ok=True)
    # Guard the orchestrator for empty daily IC inputs
    if daily_ic_a.empty or daily_ic_b.empty:
        # Log the skipped stress analysis
        logger.warning(
            "Robustness analysis skipped: empty daily IC inputs"
        )
        # Assemble a degenerate robustness report
        degenerate: RobustnessReport = RobustnessReport(
            stress_table=pd.DataFrame(),
            robustness_table=variant_summary.copy(),
            decision_log_path=finalize_decision_log(
                config, decisions, out_dir / "decision_log.json"
            ),
            audit_report_path=write_audit_report(
                config,
                pd.DataFrame(),
                variant_summary.copy(),
                out_dir / "decision_log.json",
                out_dir / "audit_report.md",
            ),
        )
        # Return the degenerate robustness report
        return degenerate
    # Step 1: compute the full-sample delta IC
    full_delta: float = float(
        (daily_ic_a - daily_ic_b).mean()
    )
    # Extract the stress windows from the config
    stress_cfg: Dict[str, Any] = config.get(
        "stress_condition_analysis", {}
    )
    # Validate the stress window dates are present in the config
    if "covid_crisis_window" not in stress_cfg:
        # Raise an informative error for the missing COVID window
        raise ValueError(
            "stress_condition_analysis.covid_crisis_window is required."
        )
    # Validate the bear-market window is present in the config
    if "bear_market_2022" not in stress_cfg:
        # Raise an informative error for the missing bear window
        raise ValueError(
            "stress_condition_analysis.bear_market_2022 is required."
        )
    # Build the top-decile mask
    decile_mask: pd.Series = build_top_decile_mask(market_volatility)
    # Build the COVID window mask
    covid_mask: pd.Series = (
        market_volatility.index >= pd.Timestamp(
            stress_cfg.get("covid_crisis_window", {})
            .get("start", "2020-02-03")
        )
    ) & (
        market_volatility.index <= pd.Timestamp(
            stress_cfg.get("covid_crisis_window", {})
            .get("end", "2020-06-30")
        )
    )
    # Build the 2022 bear-market mask
    bear_mask: pd.Series = (
        market_volatility.index >= pd.Timestamp(
            stress_cfg.get("bear_market_2022", {}).get("start", "2022-01-03")
        )
    ) & (
        market_volatility.index <= pd.Timestamp(
            stress_cfg.get("bear_market_2022", {}).get("end", "2022-12-30")
        )
    )
    # Build the regime-flip window mask
    flip_mask: pd.Series = build_regime_flip_mask(
        market_volatility,
        int(stress_cfg.get("regime_flip_windows", {}).get(
            "window_radius_days", 10
        )),
    )
    # Compute the stress-slice metrics for every slice
    stress_records: List[Dict[str, Any]] = []
    # Iterate over the named slices
    for slice_name, mask in (
        ("top_market_vol_decile", decile_mask),
        ("covid_crisis", covid_mask),
        ("bear_market_2022", bear_mask),
        ("regime_flip_windows", flip_mask),
    ):
        # Compute the slice metrics
        metrics: Dict[str, Any] = compute_stress_slice_metrics(
            daily_ic_a, daily_ic_b, mask, full_delta, config
        )
        # Attach the slice name
        metrics["slice"] = slice_name
        # Append the slice record
        stress_records.append(metrics)
    # Build the stress table
    stress_table: pd.DataFrame = pd.DataFrame(stress_records)
    # Step 2: compute the gate softness statistics
    softness: Dict[str, float] = compute_gate_softness(routing_weights)
    # Run the robustness checks on the variant summary
    robustness_findings: List[str] = run_robustness_checks(
        variant_summary, config, gate_softness=softness
    )
    # Fail loudly when the robustness checks find violations
    if robustness_findings:
        # Raise an informative error listing the violations
        raise ValueError(
            f"Robustness checks failed: {'; '.join(robustness_findings)}"
        )
    # Build the robustness table with the softness statistics
    robustness_table: pd.DataFrame = variant_summary.copy()
    # Attach the mean max weight to the main model row
    if "RG_ResMoE" in set(robustness_table["variant"]):
        # Set the mean max weight for the main model
        robustness_table.loc[
            robustness_table["variant"] == "RG_ResMoE",
            "mean_max_weight",
        ] = softness["mean_max_weight"]
    # Step 3: finalize the decision log
    decision_log_path: Path = finalize_decision_log(
        config, decisions, out_dir / "decision_log.json"
    )
    # Collect the figure paths for the audit report
    figure_paths: List[Path] = []
    # Produce the per-seed QLIKE stability figure when the data exists
    if per_seed_qlike is not None and not per_seed_qlike.empty:
        # Build the stability figure path
        stability_path: Path = out_dir / "per_seed_qlike.png"
        # Write the stability figure
        written: Optional[Path] = plot_per_seed_qlike(
            per_seed_qlike, config, stability_path
        )
        # Record the written figure path
        if written is not None:
            # Append the figure path to the list
            figure_paths.append(written)
    # Write the audit report with the computed verdict
    audit_path: Path = write_audit_report(
        config, stress_table, robustness_table,
        decision_log_path, out_dir / "audit_report.md",
        robustness_findings=robustness_findings,
        additional_tables=additional_tables,
        figure_paths=figure_paths,
    )
    # Assemble the immutable robustness report
    report: RobustnessReport = RobustnessReport(
        stress_table=stress_table,
        robustness_table=robustness_table,
        decision_log_path=decision_log_path,
        audit_report_path=audit_path,
    )
    # Return the robustness report
    return report
````

````python
# Task 18: Top-Level End-to-End Research Pipeline Orchestrator

# ==============================================================================
# Task 18: Top-Level End-to-End Research Pipeline Orchestrator
# ==============================================================================

# -----------------------------------------------------------------------
# Task 18, Step 1: Validate the Configuration and Build the Model Families
# -----------------------------------------------------------------------

def validate_config_and_build_models(
    config: Dict[str, Any],
) -> Dict[str, Any]:
    """
    Run the Task 1 config gate and build the model families.

    Composes validate_config_and_schemas, build_shared_blocks,
    build_mlp_and_moe_models, build_rg_resmoe_family, and
    build_classical_baselines in the dependency-ordered sequence.

    Parameters
    ----------
    config : Dict[str, Any]
        The configuration dictionary.

    Returns
    -------
    Dict[str, Any]
        The artifacts of the validation and model-construction stages.

    Raises
    ------
    ValueError
        If the configuration validation fails.
    """
    # Step 1: run the Task 1 configuration validation gate
    config_report = validate_config_and_schemas(config)
    # Step 2: build the Task 6 shared block and gate factories
    blocks = build_shared_blocks(config)
    # Step 3: build the Task 7 MLP and MoE model registry
    neural_registry = build_mlp_and_moe_models(
        blocks.block_factory, blocks.gate_factory, config
    )
    # Step 4: build the Task 8 RG-ResMoE family and routing rules
    rg_family = build_rg_resmoe_family(
        blocks.block_factory, blocks.gate_factory, config
    )
    # Step 5: build the Task 9 classical baseline registry
    baselines = build_classical_baselines(config)
    # Log the completed model construction
    logger.info(
        "Model families: %d neural, %d RG variants, %d baselines",
        len(neural_registry.factories),
        len(rg_family.variant_factories),
        7,
    )
    # Return the model-construction artifacts
    return {
        "task_1_config_validation": config_report,
        "task_6_blocks": blocks,
        "task_7_neural_registry": neural_registry,
        "task_8_rg_family": rg_family,
        "task_9_baselines": baselines,
    }


# -----------------------------------------------------------------------
# Task 18, Step 2: Run the Per-Market Data Pipeline
# -----------------------------------------------------------------------


def run_market_data_pipeline(
    market: str,
    config: Dict[str, Any],
    raw_root: Path,
    output_root: Path,
    metadata: Optional[pd.DataFrame] = None,
) -> Dict[str, Any]:
    """
    Run the Tasks 2-5 data pipeline for one market.

    Composes orchestrate_data_acquisition (when the raw directory is
    empty), cleanse_and_align_panel, build_features_targets, and
    build_walkforward_splits with the market name parameterized.

    Parameters
    ----------
    market : str
        The market identifier, "US" or "JP".
    config : Dict[str, Any]
        The configuration dictionary.
    raw_root : Path
        The root raw-data directory.
    output_root : Path
        The root output directory for panel artifacts.
    metadata : Optional[pd.DataFrame], optional
        The market metadata frame for the universe filters.

    Returns
    -------
    Dict[str, Any]
        The per-market pipeline artifacts.

    Raises
    ------
    ValueError
        If the market identifier is unknown.
    """
    # Validate the market identifier
    if market not in ("US", "JP"):
        # Raise an informative error for the unknown market
        raise ValueError(
            f"Unknown market '{market}'; expected 'US' or 'JP'."
        )
    # Build the market-specific raw directory
    market_raw: Path = raw_root / f"{market.lower()}_raw"
    # Build the market-specific output directory
    market_out: Path = output_root / market.lower()
    # Check whether the raw directory already contains data
    has_raw: bool = market_raw.is_dir() and any(
        market_raw.glob("*.csv")
    )
    # Run the Task 2 acquisition only when raw data is absent
    if not has_raw:
        # Build the metadata paths from the config's data paths
        us_meta: str = str(
            config.get("data_paths", {}).get("us_metadata", "")
        )
        # Build the Japanese metadata path from the config's data paths
        jp_meta: str = str(
            config.get("data_paths", {}).get("jp_metadata", "")
        )
        # Run the Task 2 acquisition orchestrator for the requested
        # market only, so a US run never downloads the Japanese panel
        acquisition = orchestrate_data_acquisition(
            config,
            Path(us_meta),
            Path(jp_meta),
            raw_root,
            markets=(market,),
        )
        # Record the acquisition artifact
        acquisition_artifact: Dict[str, Any] = {
            "acquisition": acquisition,
            "market": market,
        }
    else:
        # Record the skipped acquisition with a warning
        logger.warning(
            "Skipping Task 2 acquisition for %s: raw data present", market
        )
        # Build the skipped-acquisition artifact
        acquisition_artifact = {
            "acquisition": None,
            "market": market,
            "skipped": True,
        }
    # Load the market metadata frame for the coverage audit when the
    # data_paths contract provides it
    if metadata is None:
        # Resolve the market-specific metadata path from the config
        meta_key: str = "us_metadata" if market == "US" else "jp_metadata"
        # Extract the metadata path string from the config contract
        meta_path_str: str = str(
            config.get("data_paths", {}).get(meta_key, "")
        )
        # Load the metadata when the path is present
        if meta_path_str not in ("", "."):
            # Load the universe metadata frame
            metadata = load_universe_metadata(Path(meta_path_str))
    # Run the Task 3 panel-alignment orchestrator
    aligned = cleanse_and_align_panel(
        market_raw, market, config, metadata, market_out
    )
    # Run the Task 4 feature-panel orchestrator
    features = build_features_targets(
        aligned.wide, aligned.market_return, config, market_out
    )
    # Run the Task 5 walk-forward orchestrator
    splits = build_walkforward_splits(
        features.panel, config, market_out
    )
    # Extract the window count from the config
    n_windows: int = int(
        config.get("walkforward_evaluation", {}).get(
            "number_of_windows", 30
        )
    )
    # Verify the bundle count matches the configured window count
    if len(splits.bundles) != n_windows:
        # Raise an informative error for the bundle mismatch
        raise ValueError(
            f"Market {market} produced {len(splits.bundles)} bundles "
            f"but the config requires {n_windows}."
        )
    # Return the per-market pipeline artifacts
    return {
        "task_2_acquisition": acquisition_artifact,
        "task_3_panel": aligned,
        "task_4_features": features,
        "task_5_splits": splits,
        "market": market,
        "out_dir": market_out,
    }


# -----------------------------------------------------------------------
# Task 18, Step 3: Select Hyperparameters, Train, and Evaluate
# -----------------------------------------------------------------------


def build_row_identity(
    panel: pd.DataFrame,
) -> Tuple[np.ndarray, np.ndarray]:
    """
    Build the sorted row-identity arrays of the feature panel.

    Parameters
    ----------
    panel : pd.DataFrame
        The feature panel with ticker and date columns.

    Returns
    -------
    Tuple[np.ndarray, np.ndarray]
        The ticker array and the date array of the sorted panel.
    """
    # Sort the panel deterministically by date then ticker
    sorted_panel: pd.DataFrame = panel.sort_values(
        ["date", "ticker"]
    ).reset_index(drop=True)
    # Extract the sorted ticker identity array
    tickers: np.ndarray = sorted_panel["ticker"].to_numpy()
    # Extract the sorted date identity array
    dates: np.ndarray = sorted_panel["date"].to_numpy()
    # Return the row-identity arrays
    return tickers, dates


def train_model_for_seed_window(
    model_factory: Any,
    model_name: str,
    bundle: Dict[str, Any],
    selected_config: Dict[str, Any],
    config: Dict[str, Any],
    seed: int,
    model_kind: Optional[str] = None,
) -> Any:
    """
    Train one model for one seed and window with the correct protocol.

    Applies the two-stage RG-ResMoE protocol (Stage 1 base on MSE,
    freeze, Stage 2 experts and gate with the regularized objective and
    the SELECTED alpha/lambda_LB) for the RG family and the single-stage
    MSE protocol otherwise. The deterministic seed is applied BEFORE
    model construction so every family receives the same seed sequence.

    Parameters
    ----------
    model_factory : Any
        The model factory accepting the configuration.
    model_name : str
        The model name.
    bundle : Dict[str, Any]
        The window's tensor bundle.
    selected_config : Dict[str, Any]
        The selected hyperparameter configuration.
    config : Dict[str, Any]
        The configuration dictionary.
    seed : int
        The training seed.
    model_kind : Optional[str], optional
        The explicit model kind, "rg" or "mlp"/"moe"; when omitted the
        kind is auto-detected from the freeze_base marker.

    Returns
    -------
    Any
        The trained model.
    """
    # Apply the deterministic seed BEFORE model construction so the
    # initialization is seed-controlled
    seed_all_rng(seed)
    # Merge the selected configuration into a config copy
    model_config: Dict[str, Any] = dict(config)
    # Copy the training parameters for the override
    model_config["network_training_parameters"] = dict(
        config.get("network_training_parameters", {})
    )
    # Apply the selected learning rate
    model_config["network_training_parameters"][
        "learning_rate"
    ] = selected_config["learning_rate"]
    # Copy the loss parameters for the regularizer overrides
    model_config["loss_function_and_regularization"] = dict(
        config.get("loss_function_and_regularization", {})
    )
    # Apply the SELECTED alpha coefficient so the Stage-2 shrinkage
    # strength matches the hyperparameter selection
    model_config["loss_function_and_regularization"][
        "alpha_residual_shrinkage"
    ] = selected_config["alpha"]
    # Apply the SELECTED lambda_LB coefficient so the load-balancing
    # strength matches the hyperparameter selection
    model_config["loss_function_and_regularization"][
        "lambda_LB_load_balancing"
    ] = selected_config["lambda_LB"]
    # Build a fresh model instance for the seed
    model = model_factory(model_config)
    # Resolve the model kind explicitly when given, else auto-detect by
    # the two-stage marker (freeze_base)
    is_rg: bool = (
        model_kind == "rg"
        if model_kind is not None
        else hasattr(model, "freeze_base")
    )
    # Run the two-stage protocol for the RG family
    if is_rg:
        # Stage 1: train the base network on MSE
        train_full_batch(
            model,
            bundle["X_train"], bundle["y_train"], bundle.get("Z_train"),
            bundle["X_val"], bundle["y_val"], bundle.get("Z_val"),
            model_config,
            seed=seed,
            stage="stage_1",
        )
        # Freeze the base network for Stage 2
        model.freeze_base()
        # Stage 2: train the experts and gate with the regularized loss
        # and the selected alpha/lambda_LB coefficients
        train_full_batch(
            model,
            bundle["X_train"], bundle["y_train"], bundle.get("Z_train"),
            bundle["X_val"], bundle["y_val"], bundle.get("Z_val"),
            model_config,
            seed=seed,
            stage="stage_2",
        )
    else:
        # Single-stage training with the plain MSE loss
        train_full_batch(
            model,
            bundle["X_train"], bundle["y_train"], bundle.get("Z_train"),
            bundle["X_val"], bundle["y_val"], bundle.get("Z_val"),
            model_config,
            seed=seed,
            stage="stage_1",
        )
    # Return the trained model
    return model


def run_selection_training_evaluation(
    market_artifacts: Dict[str, Any],
    model_artifacts: Dict[str, Any],
    config: Dict[str, Any],
    seed_list: List[int],
) -> Dict[str, Any]:
    """
    Run the Task 11 selection, Task 10 training, and Tasks 12-14
    evaluation for one market.

    Delegates the corrected selection/training/evaluation chain to the
    Task 16 market-evaluation helper, which keeps every model's
    predictions separated, runs the true RG-ResMoE-vs-MLP-L DM
    comparison, and restricts the VaR fits to the training window.

    Parameters
    ----------
    market_artifacts : Dict[str, Any]
        The per-market pipeline artifacts.
    model_artifacts : Dict[str, Any]
        The model-construction artifacts.
    config : Dict[str, Any]
        The configuration dictionary.
    seed_list : List[int]
        The seed registry.

    Returns
    -------
    Dict[str, Any]
        The training, evaluation, inference, and VaR artifacts.
    """
    # Extract the walk-forward splits artifact from the market pipeline
    splits = market_artifacts["task_5_splits"]
    # Extract the neural registry from the model-construction artifacts
    neural_registry = model_artifacts["task_7_neural_registry"]
    # Extract the RG-ResMoE family from the model-construction artifacts
    rg_family = model_artifacts["task_8_rg_family"]
    # Build the registry contract for the evaluation helper
    model_registry: Dict[str, Any] = {
        "neural": neural_registry,
        "rg": rg_family,
    }
    # Resolve the execution bounds from the optional config section
    exec_cfg: Dict[str, Any] = config.get("execution", {})
    # Extract the window bound with the manuscript default
    window_limit: int = int(exec_cfg.get("window_limit", 30))
    # Extract the seed bound with the manuscript default
    seed_limit: int = int(exec_cfg.get("seed_limit", 30))
    # Build the restricted seed list from the bound
    restricted_seeds: List[int] = list(seed_list[:seed_limit])
    # Run the corrected market-evaluation chain with the artifact return
    artifacts: Dict[str, Any] = _run_market_evaluation(
        config, model_registry, restricted_seeds,
        market_artifacts["out_dir"], market_artifacts["market"],
        return_artifacts=True,
    )
    # Record the window bound in the artifacts for the audit trail
    artifacts["window_limit"] = window_limit
    # Return the evaluation artifacts
    return artifacts


# -----------------------------------------------------------------------
# Task 18, Step 4: Replicate Japan and Finalize the Robustness Analysis
# -----------------------------------------------------------------------


def run_japan_and_robustness(
    config: Dict[str, Any],
    model_artifacts: Dict[str, Any],
    evaluation_artifacts: Dict[str, Any],
    raw_root: Path,
    output_root: Path,
    seed_list: List[int],
) -> Dict[str, Any]:
    """
    Run the Task 15 Japanese replication and the Task 17 robustness
    finalization.

    The Japanese replication delegates to the Task 15 orchestrator with
    the unified config-accepting factories. The robustness analysis uses
    the MEASURED U.S. evaluation artifacts: the real per-model daily IC
    series, the real market-volatility series, and a variant summary
    assembled from the measured metrics (never the manuscript's
    constants).

    Parameters
    ----------
    config : Dict[str, Any]
        The configuration dictionary.
    model_artifacts : Dict[str, Any]
        The model-construction artifacts.
    evaluation_artifacts : Dict[str, Any]
        The evaluation artifacts of the U.S. market.
    raw_root : Path
        The root raw-data directory.
    output_root : Path
        The root output directory.
    seed_list : List[int]
        The seed registry.

    Returns
    -------
    Dict[str, Any]
        The replication and robustness artifacts.
    """
    # Build the Japanese raw directory
    jp_raw: Path = raw_root / "jp_raw"
    # Build the Japanese output directory
    jp_out: Path = output_root / "jp"
    # Extract the neural registry for the reported models
    neural_registry = model_artifacts["task_7_neural_registry"]
    # Extract the RG family registry for the reported models
    rg_family = model_artifacts["task_8_rg_family"]
    # Build the four-model factory subset for the replication; every
    # factory follows the unified config-accepting contract
    jp_factories: Dict[str, Any] = {
        "MLP-L": neural_registry.factories["MLP_L"],
        "MLP-L(+z)": neural_registry.factories["MLP_L_plus_z"],
        "MoE": neural_registry.factories["StandardMoE"],
        "RG-ResMoE": rg_family.variant_factories["RG_ResMoE"],
    }
    # Run the Task 15 Japanese replication when raw data exists
    if jp_raw.is_dir() and any(jp_raw.glob("*.csv")):
        # Run the Japanese replication orchestrator
        jp_report = replicate_japanese_study(
            jp_raw, jp_factories, config, None, jp_out
        )
    else:
        # Record the skipped replication with a warning
        logger.warning("Skipping Task 15: no Japanese raw data")
        # Build the skipped-replication artifact
        jp_report = None
    # Extract the metrics reports from the U.S. evaluation artifacts
    metrics_reports: Dict[str, Any] = evaluation_artifacts.get(
        "task_12_metrics", {}
    )
    # Build the daily IC series of the main model (seed-averaged)
    daily_ic_a: pd.Series = pd.Series(dtype=float)
    # Check the RG-ResMoE metrics report for the daily IC series
    if "RG-ResMoE" in metrics_reports:
        # Build the per-seed daily IC frame of the main model
        rg_daily: pd.DataFrame = pd.DataFrame({
            seed: metrics_reports["RG-ResMoE"].daily_ic_series[seed]
            for seed in metrics_reports["RG-ResMoE"].daily_ic_series
        })
        # Average the daily IC across the seeds
        daily_ic_a = rg_daily.mean(axis=1)
    # Build the daily IC series of the comparison baseline
    daily_ic_b: pd.Series = pd.Series(dtype=float)
    # Check the MLP-L metrics report for the daily IC series
    if "MLP-L" in metrics_reports:
        # Build the per-seed daily IC frame of the baseline
        mlp_daily: pd.DataFrame = pd.DataFrame({
            seed: metrics_reports["MLP-L"].daily_ic_series[seed]
            for seed in metrics_reports["MLP-L"].daily_ic_series
        })
        # Average the daily IC across the seeds
        daily_ic_b = mlp_daily.mean(axis=1)
    # Extract the real market-volatility series from the artifacts
    market_vol: pd.Series = evaluation_artifacts.get(
        "market_volatility", pd.Series(dtype=float)
    )
    # Extract the real routing weights from the evaluation artifacts
    routing_weights: Dict[int, np.ndarray] = {
        seed: weights
        for (model_name, seed), weights in evaluation_artifacts.get(
            "routing_weights", {}
        ).items()
        if model_name == "RG-ResMoE"
    }
    # Build the variant summary from the MEASURED metrics reports
    variant_records: List[Dict[str, Any]] = []
    # Iterate over the measured metrics reports
    for model_name, report in metrics_reports.items():
        # Compute the mean IC across the seeds
        mean_ic: float = float(report.seed_metrics["mean_ic"].mean())
        # Compute the collapsed-seed count
        collapsed: int = int(
            report.collapse_summary["collapsed_count"].sum()
        )
        # Append the measured variant record
        variant_records.append(
            {
                "variant": model_name,
                "ic": mean_ic,
                "collapsed": collapsed,
                "p_value": 1.0,
            }
        )
    # Build the variant summary frame from the measured records
    variant_summary: pd.DataFrame = pd.DataFrame(variant_records)
    # Build the decision-log resolutions from the config registry
    decisions: List[Dict[str, str]] = [
        {
            "parameter": "learning_rate",
            "decision": "selected by three-seed validation IC",
            "category": "REASONED-PLACEHOLDER",
        },
        {
            "parameter": "alpha_residual_shrinkage",
            "decision": "selected by three-seed validation IC",
            "category": "REASONED-PLACEHOLDER",
        },
    ]
    # Build the robustness output directory
    robustness_out: Path = output_root / "robustness"
    # Run the Task 17 robustness finalization with the measured inputs
    robustness_report = finalize_robustness_analysis(
        daily_ic_a, daily_ic_b, market_vol, routing_weights,
        variant_summary, config, decisions, robustness_out,
    )
    # Return the replication and robustness artifacts
    return {
        "task_15_japan": jp_report,
        "task_17_robustness": robustness_report,
    }


# -----------------------------------------------------------------------
# Task 18, Orchestrator Function
# -----------------------------------------------------------------------


def orchestrate_research_pipeline(
    config: Dict[str, Any],
    raw_root: Path,
    output_root: Path,
    seed_list: List[int],
    markets: List[str] = ("US", "JP"),
) -> Dict[str, Any]:
    """
    Orchestrate the end-to-end research pipeline.

    Composes the Task 1 config gate, the Tasks 6-9 model construction,
    the Tasks 2-5 per-market data pipeline, the Tasks 11/10/12/13/14
    selection-training-evaluation for the U.S. market, and the Tasks
    15/17 Japanese replication and robustness finalization. The
    Japanese market is owned by the Task 15 orchestrator so it is
    trained exactly once.

    Parameters
    ----------
    config : Dict[str, Any]
        The configuration dictionary.
    raw_root : Path
        The root raw-data directory.
    output_root : Path
        The root output directory for all artifacts.
    seed_list : List[int]
        The seed registry, e.g. [0..29].
    markets : List[str], optional
        The market identifiers, default ("US", "JP").

    Returns
    -------
    Dict[str, Any]
        The artifacts dictionary keyed by stage name.

    Raises
    ------
    ValueError
        If the configuration validation fails or the seed list is empty.
    """
    # Validate the seed list is not empty
    if not seed_list:
        # Raise an informative error for the empty seed list
        raise ValueError("seed_list must contain at least one seed.")
    # Ensure the output root exists
    output_root.mkdir(parents=True, exist_ok=True)
    # Step 1: run the config gate and build the model families
    model_artifacts: Dict[str, Any] = validate_config_and_build_models(
        config
    )
    # Initialize the complete artifacts dictionary
    artifacts: Dict[str, Any] = dict(model_artifacts)
    # Initialize the per-market evaluation artifacts
    market_evaluations: Dict[str, Dict[str, Any]] = {}
    # Step 2: run the per-market data pipeline
    for market in markets:
        # Run the market data pipeline
        market_artifacts: Dict[str, Any] = run_market_data_pipeline(
            market, config, raw_root, output_root
        )
        # Record the market pipeline artifacts
        artifacts[f"market_{market}_pipeline"] = market_artifacts
        # Step 3: run selection, training, and evaluation for the U.S.
        # market only; the Japanese evaluation is owned by Task 15 so
        # the Japanese models are trained exactly once
        if market == "US":
            # Run the selection-training-evaluation chain
            evaluation_artifacts: Dict[str, Any] = (
                run_selection_training_evaluation(
                    market_artifacts, model_artifacts, config, seed_list
                )
            )
            # Record the evaluation artifacts
            artifacts["market_US_evaluation"] = evaluation_artifacts
            # Store the evaluation artifacts for the robustness stage
            market_evaluations["US"] = evaluation_artifacts
    # Step 4: run the Japanese replication and the robustness finalization
    final_artifacts: Dict[str, Any] = run_japan_and_robustness(
        config,
        model_artifacts,
        market_evaluations.get("US", {}),
        raw_root,
        output_root,
        seed_list,
    )
    # Merge the final artifacts into the complete dictionary
    artifacts.update(final_artifacts)
    # Assert every stage key is present in the artifacts
    required_keys: Tuple[str, ...] = (
        "task_1_config_validation",
        "task_6_blocks",
        "task_7_neural_registry",
        "task_8_rg_family",
        "task_9_baselines",
        "task_15_japan",
        "task_17_robustness",
    )
    # Verify the required keys exist
    for key in required_keys:
        # Check the key presence in the artifacts
        if key not in artifacts:
            # Raise an informative error for the missing key
            raise ValueError(f"Artifact key missing: {key}")
    # Log the pipeline completion
    logger.info(
        "Research pipeline complete: %d artifact keys", len(artifacts)
    )
    # Return the complete artifacts dictionary
    return artifacts
````
